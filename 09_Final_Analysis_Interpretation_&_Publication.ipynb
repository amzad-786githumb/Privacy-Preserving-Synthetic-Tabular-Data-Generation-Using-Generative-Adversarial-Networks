```python
# ==============================================================================
# B2 — PROJECT ROOT DISCOVERY & CONFIGURATION CANDIDATE RESOLUTION
# ==============================================================================

print("=" * 100)
print("B2 — PROJECT ROOT DISCOVERY & CONFIGURATION CANDIDATE RESOLUTION")
print("=" * 100)

# ==============================================================================
# B2.1 — DEFINE GOOGLE DRIVE SEARCH ROOT
# ==============================================================================

GOOGLE_MYDRIVE = pathlib.Path(
    "/content/drive/MyDrive"
).resolve()

if not GOOGLE_MYDRIVE.exists():

    raise FileNotFoundError(
        f"Google Drive MyDrive was not found:\n{GOOGLE_MYDRIVE}"
    )

print()
print(f"Google Drive search root:")
print(f"  {GOOGLE_MYDRIVE}")


# ==============================================================================
# B2.2 — SEARCH FOR ALL CONFIGURATION FILES
# ==============================================================================

CONFIG_DISCOVERY_RESULTS = sorted(
    [
        path.resolve()
        for path in GOOGLE_MYDRIVE.rglob("config.yaml")
        if path.is_file()
    ],
    key=lambda path: str(path).lower()
)

print()
print("-" * 100)
print("DISCOVERED CONFIGURATION FILES")
print("-" * 100)

if not CONFIG_DISCOVERY_RESULTS:

    raise FileNotFoundError(
        "No config.yaml file was found anywhere inside Google Drive."
    )

for index, config_path in enumerate(
    CONFIG_DISCOVERY_RESULTS,
    start=1
):

    print(f"[{index}] {config_path}")


# ==============================================================================
# B2.3 — SCORE CONFIGURATION CANDIDATES USING RESEARCH PROJECT STRUCTURE
# ==============================================================================

# These names are based on the established synthetic-data research pipeline.
# They are used ONLY to identify the most likely research project root.

RESEARCH_STRUCTURE_INDICATORS = [
    "config.yaml",
    "data",
    "processed_data",
    "processed",
    "synthetic_data",
    "synthetic",
    "evaluation",
    "evaluations",
    "results",
    "reports",
    "figures",
    "tables",
    "artifacts"
]

RESEARCH_FILE_INDICATORS = [
    "adult_income_processed.csv",
    "bank_marketing_processed.csv",
    "diabetes_130us_processed.csv",
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
    "SPP_GAN",
    "SPP-GAN",
    "CTGAN",
    "TVAE",
    "DP-CTGAN"
]


def score_research_project_root(config_path):

    project_root = config_path.parent

    score = 0
    indicators_found = []

    # --------------------------------------------------------------------------
    # Configuration file
    # --------------------------------------------------------------------------

    if config_path.exists():

        score += 10
        indicators_found.append(
            "config.yaml"
        )

    # --------------------------------------------------------------------------
    # Directory indicators
    # --------------------------------------------------------------------------

    for indicator in RESEARCH_STRUCTURE_INDICATORS:

        candidate = project_root / indicator

        if candidate.exists():

            score += 2
            indicators_found.append(
                indicator
            )

    # --------------------------------------------------------------------------
    # Recursive file/directory indicators
    # --------------------------------------------------------------------------

    for indicator in RESEARCH_FILE_INDICATORS:

        try:

            matches = list(
                project_root.rglob(
                    f"*{indicator}*"
                )
            )

            if matches:

                score += 3
                indicators_found.append(
                    indicator
                )

        except Exception:
            pass

    return score, sorted(
        set(indicators_found)
    )


# ==============================================================================
# B2.4 — SCORE ALL CANDIDATES
# ==============================================================================

PROJECT_ROOT_CANDIDATES = []

for config_path in CONFIG_DISCOVERY_RESULTS:

    score, indicators = score_research_project_root(
        config_path
    )

    PROJECT_ROOT_CANDIDATES.append(
        {
            "Config File": config_path,
            "Project Root": config_path.parent,
            "Research Score": score,
            "Indicators": indicators
        }
    )


PROJECT_ROOT_CANDIDATES = sorted(
    PROJECT_ROOT_CANDIDATES,
    key=lambda item: (
        -item["Research Score"],
        str(item["Project Root"]).lower()
    )
)


# ==============================================================================
# B2.5 — DISPLAY CANDIDATE SCORES
# ==============================================================================

print()
print("-" * 100)
print("PROJECT ROOT CANDIDATE ANALYSIS")
print("-" * 100)

for index, candidate in enumerate(
    PROJECT_ROOT_CANDIDATES,
    start=1
):

    print()
    print(f"Candidate {index}")
    print(f"  Score       : {candidate['Research Score']}")
    print(f"  Project Root: {candidate['Project Root']}")
    print(f"  Config File : {candidate['Config File']}")

    if candidate["Indicators"]:

        print(
            "  Indicators  : "
            + ", ".join(candidate["Indicators"])
        )


# ==============================================================================
# B2.6 — RESOLVE BEST PROJECT ROOT
# ==============================================================================

BEST_CANDIDATE = PROJECT_ROOT_CANDIDATES[0]

BEST_SCORE = BEST_CANDIDATE["Research Score"]

TOP_SCORE_CANDIDATES = [
    candidate
    for candidate in PROJECT_ROOT_CANDIDATES
    if candidate["Research Score"] == BEST_SCORE
]


# ==============================================================================
# B2.7 — HANDLE AMBIGUOUS TOP CANDIDATES
# ==============================================================================

if len(TOP_SCORE_CANDIDATES) > 1:

    print()
    print("=" * 100)
    print("AMBIGUOUS PROJECT ROOT DETECTED")
    print("=" * 100)

    print(
        "Multiple configuration files have the same highest "
        "research-project score."
    )

    print()
    print("Top candidates:")

    for index, candidate in enumerate(
        TOP_SCORE_CANDIDATES,
        start=1
    ):

        print()
        print(f"[{index}]")
        print(
            f"Project Root: {candidate['Project Root']}"
        )
        print(
            f"Config File : {candidate['Config File']}"
        )
        print(
            f"Score       : {candidate['Research Score']}"
        )

    raise RuntimeError(
        "Project root selection remains ambiguous. "
        "Use the candidate list above to identify the established "
        "research project root."
    )


# ==============================================================================
# B2.8 — FINAL PROJECT ROOT
# ==============================================================================

PROJECT_ROOT = BEST_CANDIDATE["Project Root"].resolve()

CONFIG_FILE = BEST_CANDIDATE["Config File"].resolve()


# ==============================================================================
# B2.9 — FINAL VALIDATION
# ==============================================================================

if not PROJECT_ROOT.exists():

    raise FileNotFoundError(
        f"Resolved project root does not exist:\n{PROJECT_ROOT}"
    )

if not CONFIG_FILE.exists():

    raise FileNotFoundError(
        f"Resolved config.yaml does not exist:\n{CONFIG_FILE}"
    )


# ==============================================================================
# B2.10 — PROJECT ROOT SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("RESOLVED RESEARCH PROJECT")
print("=" * 100)

print()
print(f"Project Root:")
print(f"  {PROJECT_ROOT}")

print()
print(f"Configuration File:")
print(f"  {CONFIG_FILE}")

print()
print(f"Research Project Score:")
print(f"  {BEST_SCORE}")

print()
print("B2 completed successfully.")
```

```python
# ==============================================================================
# B3 — CONFIGURATION FILE DISCOVERY
# ==============================================================================

print("=" * 100)
print("B3 — CONFIGURATION FILE DISCOVERY")
print("=" * 100)

CONFIG_FILE_CANDIDATES = [
    PROJECT_ROOT / "config.yaml",
    PROJECT_ROOT / "CONFIG.yaml",
    PROJECT_ROOT / "configuration.yaml",
    PROJECT_ROOT / "Configuration" / "config.yaml",
    PROJECT_ROOT / "config" / "config.yaml",
]

CONFIG_FILES_FOUND = [
    path
    for path in CONFIG_FILE_CANDIDATES
    if path.exists() and path.is_file()
]

# --------------------------------------------------------------------------
# Fallback recursive search inside project root
# --------------------------------------------------------------------------

if not CONFIG_FILES_FOUND:

    CONFIG_FILES_FOUND = [
        path
        for path in PROJECT_ROOT.rglob("config.yaml")
        if path.is_file()
    ]

# --------------------------------------------------------------------------
# Validate configuration file
# --------------------------------------------------------------------------

if len(CONFIG_FILES_FOUND) == 0:

    raise FileNotFoundError(
        f"No config.yaml was found inside project root:\n"
        f"{PROJECT_ROOT}"
    )

if len(CONFIG_FILES_FOUND) > 1:

    raise RuntimeError(
        "Multiple configuration files were found. "
        "Notebook 9 will not arbitrarily select one:\n"
        + "\n".join(
            str(path)
            for path in CONFIG_FILES_FOUND
        )
    )

CONFIG_FILE = CONFIG_FILES_FOUND[0].resolve()

print()
print(f"Configuration file:")
print(f"  {CONFIG_FILE}")

print()
print("Configuration file discovery completed successfully.")
```

```python
# ==============================================================================
# B4 — LOAD PROJECT CONFIGURATION
# ==============================================================================

print("=" * 100)
print("B4 — LOAD PROJECT CONFIGURATION")
print("=" * 100)

try:
    import yaml
except ImportError as error:

    raise ImportError(
        "PyYAML is required to load config.yaml."
    ) from error

with open(
    CONFIG_FILE,
    "r",
    encoding="utf-8"
) as file:

    PROJECT_CONFIG = yaml.safe_load(file)

if PROJECT_CONFIG is None:

    raise ValueError(
        "config.yaml was loaded but contains no configuration."
    )

if not isinstance(PROJECT_CONFIG, dict):

    raise TypeError(
        "config.yaml must contain a dictionary-like configuration."
    )

print()
print("Project configuration loaded successfully.")
print()
print("Top-level configuration keys:")

for key in PROJECT_CONFIG.keys():
    print(f"  - {key}")
```

```python
# ==============================================================================
# D1 — FINAL REFERENCE DATA DISCOVERY
# ==============================================================================

print("=" * 100)
print("D1 — FINAL REFERENCE DATA DISCOVERY")
print("=" * 100)


# ==============================================================================
# D1.1 — VALIDATE DATA DIRECTORY
# ==============================================================================

if "DATA_DIRECTORY" not in globals():
    raise RuntimeError(
        "DATA_DIRECTORY is not available. Run B6 before D1."
    )

if not DATA_DIRECTORY.exists():
    raise FileNotFoundError(
        f"Reference data directory does not exist:\n{DATA_DIRECTORY}"
    )


# ==============================================================================
# D1.2 — FINAL DATASET REGISTRY
# ==============================================================================

REFERENCE_DATASET_NAMES = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]


# ==============================================================================
# D1.3 — DATASET NAME MATCHING TOKENS
# ==============================================================================

REFERENCE_DATASET_TOKENS = {

    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income",
        "adultincome"
    ],

    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing",
        "bankmarketing"
    ],

    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us",
        "diabetes130us",
        "diabetes"
    ]
}


# ==============================================================================
# D1.4 — DISCOVER ALL CSV FILES
# ==============================================================================

ALL_REFERENCE_CSV_FILES = sorted(
    [
        path.resolve()
        for path in DATA_DIRECTORY.rglob("*")
        if path.is_file()
        and path.suffix.lower() == ".csv"
    ],
    key=lambda path: str(path).lower()
)


print()
print(
    f"CSV files discovered under data directory: "
    f"{len(ALL_REFERENCE_CSV_FILES)}"
)


# ==============================================================================
# D1.5 — DISCOVER DATASET CANDIDATES
# ==============================================================================

REFERENCE_DATA_CANDIDATES = {
    dataset: []
    for dataset in REFERENCE_DATASET_NAMES
}


for dataset in REFERENCE_DATASET_NAMES:

    tokens = REFERENCE_DATASET_TOKENS[dataset]

    for filepath in ALL_REFERENCE_CSV_FILES:

        normalized_path = str(
            filepath
        ).lower().replace(
            "\\",
            "/"
        )

        filename = filepath.name.lower()

        if any(
            token.lower() in normalized_path
            or token.lower() in filename
            for token in tokens
        ):

            REFERENCE_DATA_CANDIDATES[
                dataset
            ].append(filepath)


# ==============================================================================
# D1.6 — REMOVE DUPLICATE PATHS
# ==============================================================================

for dataset in REFERENCE_DATASET_NAMES:

    REFERENCE_DATA_CANDIDATES[dataset] = sorted(
        set(
            REFERENCE_DATA_CANDIDATES[dataset]
        ),
        key=lambda path: str(path).lower()
    )


# ==============================================================================
# D1.7 — DISPLAY DISCOVERY RESULTS
# ==============================================================================

print()
print("-" * 100)
print("REFERENCE DATASET CANDIDATES")
print("-" * 100)

for dataset in REFERENCE_DATASET_NAMES:

    candidates = REFERENCE_DATA_CANDIDATES[
        dataset
    ]

    print()
    print(f"{dataset}:")

    if candidates:

        for index, filepath in enumerate(
            candidates,
            start=1
        ):

            print(
                f"  [{index}] {filepath}"
            )

    else:

        print(
            "  NOT FOUND"
        )


# ==============================================================================
# D1.8 — DISCOVERY STATUS
# ==============================================================================

REFERENCE_DATA_DISCOVERY_STATUS = {

    dataset: len(
        REFERENCE_DATA_CANDIDATES[dataset]
    ) > 0

    for dataset in REFERENCE_DATASET_NAMES
}


# ==============================================================================
# D1.9 — DISPLAY STATUS
# ==============================================================================

print()
print("-" * 100)
print("REFERENCE DATA DISCOVERY STATUS")
print("-" * 100)

for dataset, status in (
    REFERENCE_DATA_DISCOVERY_STATUS.items()
):

    print(
        f"{dataset:<25}: "
        f"{'FOUND' if status else 'NOT FOUND'}"
    )


# ==============================================================================
# D1.10 — DO NOT FAIL YET ON AMBIGUITY
# ==============================================================================

# D1 intentionally does not require exactly one file.
#
# Zero candidates will be investigated in D1/D2.
# Multiple candidates will be resolved in D2.
#
# This prevents Notebook 9 from silently selecting an incorrect file.


# ==============================================================================
# D1.11 — SECTION STATUS
# ==============================================================================

NOTEBOOK_9_D1_COMPLETED = True

print()
print("=" * 100)
print("D1 — REFERENCE DATA DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# D2 — REFERENCE DATASET RESOLUTION
# ==============================================================================

print("=" * 100)
print("D2 — REFERENCE DATASET RESOLUTION")
print("=" * 100)

REFERENCE_DATA_FILES = {}

for dataset in REFERENCE_DATASET_NAMES:

    candidates = REFERENCE_DATA_CANDIDATES.get(
        dataset,
        []
    )

    if len(candidates) == 0:

        raise FileNotFoundError(
            f"No reference dataset was found for '{dataset}'. "
            f"Expected filename: "
            f"{REFERENCE_DATASET_FILENAMES[dataset]}"
        )

    if len(candidates) > 1:

        raise RuntimeError(
            f"Multiple reference datasets were found for '{dataset}':\n"
            + "\n".join(
                f" - {path}"
                for path in candidates
            )
            + "\n\nNotebook 9 will not select one arbitrarily."
        )

    REFERENCE_DATA_FILES[dataset] = candidates[0]


# ==============================================================================
# DISPLAY RESOLVED DATASETS
# ==============================================================================

print()

for dataset, filepath in REFERENCE_DATA_FILES.items():

    print(
        f"{dataset:<20}: {filepath}"
    )

print()
print("All three reference datasets resolved successfully.")
```

```python
# ==============================================================================
# D3 — LOAD REFERENCE DATASETS
# ==============================================================================

print("=" * 100)
print("D3 — LOAD REFERENCE DATASETS")
print("=" * 100)

import pandas as pd

REFERENCE_DATASETS = {}

for dataset in REFERENCE_DATASET_NAMES:

    filepath = REFERENCE_DATA_FILES[dataset]

    print()
    print(f"Loading {dataset}...")
    print(f"File: {filepath}")

    dataframe = pd.read_csv(
        filepath,
        low_memory=False
    )

    if dataframe.empty:

        raise ValueError(
            f"Reference dataset '{dataset}' is empty."
        )

    REFERENCE_DATASETS[dataset] = dataframe

    print(
        f"Loaded successfully: "
        f"{dataframe.shape[0]:,} rows × "
        f"{dataframe.shape[1]:,} columns"
    )


# ==============================================================================
# VALIDATE MEMORY REGISTRY
# ==============================================================================

if set(REFERENCE_DATASETS.keys()) != set(
    REFERENCE_DATASET_NAMES
):

    raise RuntimeError(
        "Reference dataset registry is incomplete."
    )

print()
print("All final reference datasets loaded successfully.")
```

```python
# ==============================================================================
# D4 — REFERENCE DATASET STRUCTURAL VALIDATION
# ==============================================================================

print("=" * 100)
print("D4 — REFERENCE DATASET STRUCTURAL VALIDATION")
print("=" * 100)

REFERENCE_STRUCTURE_SUMMARY = []

for dataset in REFERENCE_DATASET_NAMES:

    df = REFERENCE_DATASETS[dataset]

    numeric_columns = df.select_dtypes(
        include="number"
    ).columns.tolist()

    categorical_columns = df.select_dtypes(
        exclude="number"
    ).columns.tolist()

    missing_cells = int(
        df.isna().sum().sum()
    )

    duplicate_rows = int(
        df.duplicated().sum()
    )

    REFERENCE_STRUCTURE_SUMMARY.append(
        {
            "Dataset": dataset,
            "Rows": int(df.shape[0]),
            "Columns": int(df.shape[1]),
            "Numeric Columns": len(numeric_columns),
            "Categorical Columns": len(categorical_columns),
            "Missing Cells": missing_cells,
            "Duplicate Rows": duplicate_rows,
            "Column Names Unique": (
                len(df.columns)
                == len(set(df.columns))
            )
        }
    )


REFERENCE_STRUCTURE_SUMMARY = pd.DataFrame(
    REFERENCE_STRUCTURE_SUMMARY
)

print()

display(
    REFERENCE_STRUCTURE_SUMMARY
)


# ==============================================================================
# STRUCTURAL HARD CHECKS
# ==============================================================================

for dataset in REFERENCE_DATASET_NAMES:

    df = REFERENCE_DATASETS[dataset]

    if len(df.columns) == 0:

        raise ValueError(
            f"{dataset} contains no columns."
        )

    if len(df) == 0:

        raise ValueError(
            f"{dataset} contains no rows."
        )

    if len(df.columns) != len(
        set(df.columns)
    ):

        raise ValueError(
            f"{dataset} contains duplicate column names."
        )

print()
print("Reference dataset structural validation completed.")
```

```python
# ==============================================================================
# D5 — REFERENCE DATASET TARGET VALIDATION
# ==============================================================================

print("=" * 100)
print("D5 — REFERENCE DATASET TARGET VALIDATION")
print("=" * 100)

# ==============================================================================
# TARGET CANDIDATE DEFINITIONS
# ==============================================================================

TARGET_COLUMN_CANDIDATES = {

    "adult_income": [
        "income",
        "income_class",
        "target",
        "class"
    ],

    "bank_marketing": [
        "y",
        "target",
        "class"
    ],

    "diabetes_130us": [
        "readmitted",
        "target",
        "class"
    ]
}


# ==============================================================================
# DISCOVER TARGET COLUMNS
# ==============================================================================

REFERENCE_TARGET_COLUMNS = {}

for dataset in REFERENCE_DATASET_NAMES:

    df = REFERENCE_DATASETS[dataset]

    discovered_targets = [
        column
        for column in TARGET_COLUMN_CANDIDATES[dataset]
        if column in df.columns
    ]

    REFERENCE_TARGET_COLUMNS[dataset] = discovered_targets


# ==============================================================================
# DISPLAY
# ==============================================================================

print()

for dataset in REFERENCE_DATASET_NAMES:

    print(
        f"{dataset:<20}: "
        f"{REFERENCE_TARGET_COLUMNS[dataset]}"
    )


# ==============================================================================
# TARGET VALIDATION
# ==============================================================================

REFERENCE_TARGET_VALIDATION = {}

for dataset in REFERENCE_DATASET_NAMES:

    targets = REFERENCE_TARGET_COLUMNS[dataset]

    REFERENCE_TARGET_VALIDATION[dataset] = (
        len(targets) == 1
    )


# ==============================================================================
# DISPLAY VALIDATION
# ==============================================================================

print()
print("-" * 100)
print("TARGET VALIDATION")
print("-" * 100)

for dataset, status in REFERENCE_TARGET_VALIDATION.items():

    print(
        f"{dataset:<20}: "
        f"{'PASS' if status else 'FAIL'}"
    )


# ==============================================================================
# HARD VALIDATION
# ==============================================================================

FAILED_TARGET_DATASETS = [
    dataset
    for dataset, status
    in REFERENCE_TARGET_VALIDATION.items()
    if not status
]

if FAILED_TARGET_DATASETS:

    raise RuntimeError(
        "Target validation failed for:\n"
        + "\n".join(
            f" - {dataset}"
            for dataset in FAILED_TARGET_DATASETS
        )
        + "\n\nInspect the actual processed-data column names "
        "before continuing."
    )


# ==============================================================================
# FINAL TARGET REGISTRY
# ==============================================================================

REFERENCE_TARGET_COLUMNS = {
    dataset: columns[0]
    for dataset, columns
    in REFERENCE_TARGET_COLUMNS.items()
}

print()
print("Reference target variables resolved successfully.")
```

```python
# ==============================================================================
# D6 — REFERENCE DATASET REGISTRY
# ==============================================================================

print("=" * 100)
print("D6 — REFERENCE DATASET REGISTRY")
print("=" * 100)

REFERENCE_DATASET_REGISTRY = {}

for dataset in REFERENCE_DATASET_NAMES:

    df = REFERENCE_DATASETS[dataset]

    REFERENCE_DATASET_REGISTRY[dataset] = {

        "Dataset": dataset,

        "File": str(
            REFERENCE_DATA_FILES[dataset]
        ),

        "Rows": int(
            df.shape[0]
        ),

        "Columns": int(
            df.shape[1]
        ),

        "Column_Names": df.columns.tolist(),

        "Data_Types": {
            column: str(dtype)
            for column, dtype
            in df.dtypes.items()
        },

        "Numeric_Columns": (
            df.select_dtypes(
                include="number"
            ).columns.tolist()
        ),

        "Categorical_Columns": (
            df.select_dtypes(
                exclude="number"
            ).columns.tolist()
        ),

        "Target_Column": (
            REFERENCE_TARGET_COLUMNS[dataset]
        ),

        "Missing_Cells": int(
            df.isna().sum().sum()
        ),

        "Duplicate_Rows": int(
            df.duplicated().sum()
        )
    }


# ==============================================================================
# DISPLAY SUMMARY
# ==============================================================================

REFERENCE_DATASET_SUMMARY = pd.DataFrame(
    [
        {
            "Dataset": dataset,
            "Rows": registry["Rows"],
            "Columns": registry["Columns"],
            "Target": registry["Target_Column"],
            "Missing Cells": registry["Missing_Cells"],
            "Duplicate Rows": registry["Duplicate_Rows"]
        }
        for dataset, registry
        in REFERENCE_DATASET_REGISTRY.items()
    ]
)

print()

display(
    REFERENCE_DATASET_SUMMARY
)
```

```python
# ==============================================================================
# D7 — REFERENCE DATA READINESS VALIDATION
# ==============================================================================

print("=" * 100)
print("D7 — REFERENCE DATA READINESS VALIDATION")
print("=" * 100)

D7_VALIDATION_RESULTS = {}

# ==============================================================================
# DATASET COUNT
# ==============================================================================

D7_VALIDATION_RESULTS[
    "Three Reference Datasets Loaded"
] = (
    len(REFERENCE_DATASETS) == 3
)


# ==============================================================================
# DATASET REGISTRY
# ==============================================================================

D7_VALIDATION_RESULTS[
    "Dataset Registry Complete"
] = (
    set(REFERENCE_DATASETS.keys())
    == set(REFERENCE_DATASET_NAMES)
)


# ==============================================================================
# FILE REGISTRY
# ==============================================================================

D7_VALIDATION_RESULTS[
    "Reference File Registry Complete"
] = (
    set(REFERENCE_DATA_FILES.keys())
    == set(REFERENCE_DATASET_NAMES)
)


# ==============================================================================
# TARGET REGISTRY
# ==============================================================================

D7_VALIDATION_RESULTS[
    "Target Registry Complete"
] = (
    set(REFERENCE_TARGET_COLUMNS.keys())
    == set(REFERENCE_DATASET_NAMES)
)


# ==============================================================================
# NON-EMPTY DATASETS
# ==============================================================================

D7_VALIDATION_RESULTS[
    "All Datasets Non-Empty"
] = all(
    not REFERENCE_DATASETS[dataset].empty
    for dataset in REFERENCE_DATASET_NAMES
)


# ==============================================================================
# UNIQUE COLUMN NAMES
# ==============================================================================

D7_VALIDATION_RESULTS[
    "All Column Names Unique"
] = all(
    len(df.columns)
    == len(set(df.columns))
    for df in REFERENCE_DATASETS.values()
)


# ==============================================================================
# DISPLAY VALIDATION
# ==============================================================================

print()

for check_name, status in D7_VALIDATION_RESULTS.items():

    print(
        f"{check_name:<40}: "
        f"{'PASS' if status else 'FAIL'}"
    )


# ==============================================================================
# FAILURE CHECK
# ==============================================================================

D7_FAILED_CHECKS = [
    check
    for check, status
    in D7_VALIDATION_RESULTS.items()
    if not status
]

if D7_FAILED_CHECKS:

    raise RuntimeError(
        "Section D reference-data readiness validation failed:\n"
        + "\n".join(
            f" - {check}"
            for check in D7_FAILED_CHECKS
        )
    )


# ==============================================================================
# FINAL SECTION STATUS
# ==============================================================================

NOTEBOOK_9_SECTION_D_COMPLETED = True

print()
print("=" * 100)
print("SECTION D — LOAD FINAL REFERENCE DATA COMPLETED SUCCESSFULLY")
print("=" * 100)

print()
print(
    "Reference datasets ready for Section E:"
)

for dataset in REFERENCE_DATASET_NAMES:

    df = REFERENCE_DATASETS[dataset]

    print(
        f"  {dataset:<20} "
        f"{df.shape[0]:>8,} rows × "
        f"{df.shape[1]:>3} columns"
    )
```

```python
# ==============================================================================
# E1 — SYNTHETIC DATA DISCOVERY
# ==============================================================================

print("=" * 100)
print("E1 — SYNTHETIC DATA DISCOVERY")
print("=" * 100)


# ==============================================================================
# E1.1 — VALIDATE SYNTHETIC DATA DIRECTORY
# ==============================================================================

if "SYNTHETIC_DATA_DIRECTORY" not in globals():

    raise RuntimeError(
        "SYNTHETIC_DATA_DIRECTORY is not available. "
        "Run B6 before E1."
    )

if not SYNTHETIC_DATA_DIRECTORY.exists():

    raise FileNotFoundError(
        "Synthetic data directory does not exist:\n"
        f"{SYNTHETIC_DATA_DIRECTORY}"
    )


# ==============================================================================
# E1.2 — FINAL DATASETS AND METHODS
# ==============================================================================

if "RESEARCH_DATASETS" not in globals():

    raise RuntimeError(
        "RESEARCH_DATASETS is not available. "
        "Run C1 before E1."
    )

if "RESEARCH_METHODS" not in globals():

    raise RuntimeError(
        "RESEARCH_METHODS is not available. "
        "Run C1 before E1."
    )


# ==============================================================================
# E1.3 — DISCOVER ALL SYNTHETIC CSV FILES
# ==============================================================================

ALL_SYNTHETIC_CSV_FILES = sorted(
    [
        path.resolve()
        for path in SYNTHETIC_DATA_DIRECTORY.rglob("*")
        if path.is_file()
        and path.suffix.lower() == ".csv"
    ],
    key=lambda path: str(path).lower()
)


print()
print(
    f"Synthetic CSV files discovered: "
    f"{len(ALL_SYNTHETIC_CSV_FILES)}"
)


# ==============================================================================
# E1.4 — DISPLAY DISCOVERED FILES
# ==============================================================================

print()
print("-" * 100)
print("DISCOVERED SYNTHETIC CSV FILES")
print("-" * 100)

if ALL_SYNTHETIC_CSV_FILES:

    for filepath in ALL_SYNTHETIC_CSV_FILES:

        print(
            f"  {filepath}"
        )

else:

    print(
        "No synthetic CSV files were discovered."
    )
```

```python
# ==============================================================================
# E1.1 — VALIDATE DISCOVERED SYNTHETIC FILES AGAINST FINAL RESEARCH SCOPE
# ==============================================================================

print("=" * 100)
print("E1.1 — VALIDATE DISCOVERED SYNTHETIC FILES AGAINST FINAL RESEARCH SCOPE")
print("=" * 100)

VALID_RESEARCH_DATASET_TOKENS = {
    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income"
    ],
    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing"
    ],
    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us"
    ]
}

VALID_RESEARCH_METHOD_TOKENS = {
    "CTGAN": [
        "ctgan"
    ],
    "TVAE": [
        "tvae"
    ],
    "DP-CTGAN": [
        "dp-ctgan",
        "dp_ctgan",
        "dpctgan"
    ],
    "SPP-GAN": [
        "spp-gan",
        "spp_gan",
        "sppgan"
    ]
}


# ==============================================================================
# CLASSIFY DISCOVERED FILES
# ==============================================================================

VALID_SYNTHETIC_FILES = []
OUT_OF_SCOPE_SYNTHETIC_FILES = []

for filepath in ALL_SYNTHETIC_CSV_FILES:

    normalized_name = filepath.name.lower()

    dataset_match = any(
        token.lower() in normalized_name
        for tokens in VALID_RESEARCH_DATASET_TOKENS.values()
        for token in tokens
    )

    method_match = any(
        token.lower() in normalized_name
        for tokens in VALID_RESEARCH_METHOD_TOKENS.values()
        for token in tokens
    )

    if dataset_match and method_match:

        VALID_SYNTHETIC_FILES.append(filepath)

    else:

        OUT_OF_SCOPE_SYNTHETIC_FILES.append(filepath)


# ==============================================================================
# DISPLAY VALID FILES
# ==============================================================================

print()
print("-" * 100)
print("VALID FINAL-RESEARCH SYNTHETIC FILES")
print("-" * 100)

if VALID_SYNTHETIC_FILES:

    for filepath in VALID_SYNTHETIC_FILES:
        print(f"  {filepath}")

else:

    print("  NONE")


# ==============================================================================
# DISPLAY OUT-OF-SCOPE FILES
# ==============================================================================

print()
print("-" * 100)
print("OUT-OF-SCOPE SYNTHETIC FILES")
print("-" * 100)

if OUT_OF_SCOPE_SYNTHETIC_FILES:

    for filepath in OUT_OF_SCOPE_SYNTHETIC_FILES:

        print(
            f"  {filepath}"
        )

else:

    print("  NONE")


# ==============================================================================
# FINAL STATUS
# ==============================================================================

print()
print("-" * 100)
print("E1.1 STATUS")
print("-" * 100)

print(
    f"Valid final-research files : "
    f"{len(VALID_SYNTHETIC_FILES)}"
)

print(
    f"Out-of-scope files         : "
    f"{len(OUT_OF_SCOPE_SYNTHETIC_FILES)}"
)

print()

if OUT_OF_SCOPE_SYNTHETIC_FILES:

    print(
        "NOTE: Out-of-scope files are ignored by Notebook 9."
    )

print()
print(
    "E1.1 — Synthetic file scope validation completed."
)
```

```python
# ==============================================================================
# E1.2 — SYNTHETIC ARTIFACT RECOVERY DISCOVERY
# ==============================================================================

print("=" * 100)
print("E1.2 — SYNTHETIC ARTIFACT RECOVERY DISCOVERY")
print("=" * 100)


# ==============================================================================
# E1.2.1 — REQUIRED RESEARCH SCOPE
# ==============================================================================

RECOVERY_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RECOVERY_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# E1.2.2 — DATASET TOKENS
# ==============================================================================

RECOVERY_DATASET_TOKENS = {

    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income",
        "adultincome"
    ],

    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing",
        "bankmarketing"
    ],

    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us",
        "diabetes130us"
    ]
}


# ==============================================================================
# E1.2.3 — METHOD TOKENS
# ==============================================================================

RECOVERY_METHOD_TOKENS = {

    "CTGAN": [
        "ctgan"
    ],

    "TVAE": [
        "tvae"
    ],

    "DP-CTGAN": [
        "dp-ctgan",
        "dp_ctgan",
        "dpctgan"
    ],

    "SPP-GAN": [
        "spp-gan",
        "spp_gan",
        "sppgan"
    ]
}


# ==============================================================================
# E1.2.4 — SEARCH ROOTS
# ==============================================================================

RECOVERY_SEARCH_ROOTS = [
    SYNTHETIC_DATA_DIRECTORY,
    RESULTS_DIRECTORY,
    EVALUATION_DIRECTORY,
    REPORTS_DIRECTORY
]


# ==============================================================================
# E1.2.5 — SEARCHABLE ARTIFACT TYPES
# ==============================================================================

RECOVERY_EXTENSIONS = {
    ".csv",
    ".parquet",
    ".pkl",
    ".pickle",
    ".pt",
    ".pth",
    ".joblib",
    ".json",
    ".npz"
}


# ==============================================================================
# E1.2.6 — DISCOVER PROJECT ARTIFACTS
# ==============================================================================

RECOVERY_ARTIFACTS = []

for root in RECOVERY_SEARCH_ROOTS:

    if not root.exists():
        continue

    for filepath in root.rglob("*"):

        if not filepath.is_file():
            continue

        if filepath.suffix.lower() not in RECOVERY_EXTENSIONS:
            continue

        RECOVERY_ARTIFACTS.append(
            filepath.resolve()
        )


RECOVERY_ARTIFACTS = sorted(
    set(RECOVERY_ARTIFACTS),
    key=lambda path: str(path).lower()
)


# ==============================================================================
# E1.2.7 — BUILD ARTIFACT TABLE
# ==============================================================================

RECOVERY_RECORDS = []

for filepath in RECOVERY_ARTIFACTS:

    filename = filepath.name.lower()
    full_path = str(filepath).lower()

    matched_datasets = []

    for dataset in RECOVERY_DATASETS:

        if any(
            token.lower() in filename
            or token.lower() in full_path
            for token in RECOVERY_DATASET_TOKENS[dataset]
        ):

            matched_datasets.append(
                dataset
            )

    matched_methods = []

    for method in RECOVERY_METHODS:

        if any(
            token.lower() in filename
            or token.lower() in full_path
            for token in RECOVERY_METHOD_TOKENS[method]
        ):

            matched_methods.append(
                method
            )

    RECOVERY_RECORDS.append(
        {
            "File": str(filepath),
            "Filename": filepath.name,
            "Extension": filepath.suffix.lower(),
            "Matched Datasets": ", ".join(
                matched_datasets
            ),
            "Matched Methods": ", ".join(
                matched_methods
            ),
            "Size MB": round(
                filepath.stat().st_size / (1024 ** 2),
                3
            )
        }
    )


RECOVERY_SEARCH_RESULTS = pd.DataFrame(
    RECOVERY_RECORDS
)


# ==============================================================================
# E1.2.8 — DISPLAY OVERALL DISCOVERY
# ==============================================================================

print()
print("-" * 100)
print("ARTIFACT DISCOVERY SUMMARY")
print("-" * 100)

print(
    f"Potential research artifacts discovered : "
    f"{len(RECOVERY_SEARCH_RESULTS)}"
)


# ==============================================================================
# E1.2.9 — IDENTIFY DATASET-LIKE ARTIFACTS
# ==============================================================================

DATASET_LIKE_EXTENSIONS = {
    ".csv",
    ".parquet"
}

DATASET_LIKE_ARTIFACTS = (
    RECOVERY_SEARCH_RESULTS[
        RECOVERY_SEARCH_RESULTS["Extension"].isin(
            DATASET_LIKE_EXTENSIONS
        )
        &
        (
            RECOVERY_SEARCH_RESULTS[
                "Matched Datasets"
            ].str.len() > 0
        )
        &
        (
            RECOVERY_SEARCH_RESULTS[
                "Matched Methods"
            ].str.len() > 0
        )
    ]
    .copy()
)


# ==============================================================================
# E1.2.10 — DISPLAY DATASET-LIKE ARTIFACTS
# ==============================================================================

print()
print("-" * 100)
print("DATASET-LIKE SYNTHETIC ARTIFACTS")
print("-" * 100)

if DATASET_LIKE_ARTIFACTS.empty:

    print(
        "No dataset-like synthetic artifacts were discovered."
    )

else:

    display(
        DATASET_LIKE_ARTIFACTS[
            [
                "File",
                "Filename",
                "Matched Datasets",
                "Matched Methods",
                "Size MB"
            ]
        ]
    )


# ==============================================================================
# E1.2.11 — BUILD TARGETED RECOVERY MATRIX
# ==============================================================================

TARGETED_RECOVERY_MATRIX = []

for dataset in RECOVERY_DATASETS:

    for method in RECOVERY_METHODS:

        matches = []

        for _, row in (
            DATASET_LIKE_ARTIFACTS.iterrows()
        ):

            matched_datasets = [
                item.strip()
                for item in row[
                    "Matched Datasets"
                ].split(",")
                if item.strip()
            ]

            matched_methods = [
                item.strip()
                for item in row[
                    "Matched Methods"
                ].split(",")
                if item.strip()
            ]

            if (
                dataset in matched_datasets
                and method in matched_methods
            ):

                matches.append(
                    row["File"]
                )

        TARGETED_RECOVERY_MATRIX.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Candidate Count": len(matches),
                "Candidates": matches
            }
        )


TARGETED_RECOVERY_MATRIX = pd.DataFrame(
    TARGETED_RECOVERY_MATRIX
)


# ==============================================================================
# E1.2.12 — DISPLAY TARGETED RECOVERY STATUS
# ==============================================================================

print()
print("-" * 100)
print("TARGETED SYNTHETIC DATA RECOVERY STATUS")
print("-" * 100)

for _, row in TARGETED_RECOVERY_MATRIX.iterrows():

    dataset = row["Dataset"]
    method = row["Method"]
    count = row["Candidate Count"]

    print()
    print(
        f"{dataset} × {method}:"
    )

    if count == 0:

        print(
            "  NOT FOUND"
        )

    else:

        print(
            f"  {count} candidate(s) found:"
        )

        for candidate in row["Candidates"]:

            print(
                f"    {candidate}"
            )


# ==============================================================================
# E1.2.13 — FINAL RECOVERY COUNTS
# ==============================================================================

FOUND_EXPERIMENTS = (
    TARGETED_RECOVERY_MATRIX[
        TARGETED_RECOVERY_MATRIX[
            "Candidate Count"
        ] > 0
    ]
)

MISSING_EXPERIMENTS = (
    TARGETED_RECOVERY_MATRIX[
        TARGETED_RECOVERY_MATRIX[
            "Candidate Count"
        ] == 0
    ]
)


print()
print("-" * 100)
print("RECOVERY SUMMARY")
print("-" * 100)

print(
    f"Expected Dataset × Method combinations : "
    f"{len(RECOVERY_DATASETS) * len(RECOVERY_METHODS)}"
)

print(
    f"Combinations with candidates            : "
    f"{len(FOUND_EXPERIMENTS)}"
)

print(
    f"Combinations still missing              : "
    f"{len(MISSING_EXPERIMENTS)}"
)


# ==============================================================================
# E1.2.14 — SECTION STATUS
# ==============================================================================

NOTEBOOK_9_E1_2_COMPLETED = True

print()
print("=" * 100)
print("E1.2 — SYNTHETIC ARTIFACT RECOVERY DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# E2 — FINAL SYNTHETIC CANDIDATE MAPPING
# ==============================================================================

print("=" * 100)
print("E2 — FINAL SYNTHETIC CANDIDATE MAPPING")
print("=" * 100)


# ==============================================================================
# E2.1 — FINAL RESEARCH SCOPE
# ==============================================================================

RESEARCH_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RESEARCH_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# E2.2 — NORMALIZATION TOKENS
# ==============================================================================

E2_DATASET_TOKENS = {

    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income",
        "adultincome"
    ],

    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing",
        "bankmarketing"
    ],

    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us",
        "diabetes130us"
    ]
}


E2_METHOD_TOKENS = {

    "CTGAN": [
        "ctgan"
    ],

    "TVAE": [
        "tvae"
    ],

    "DP-CTGAN": [
        "dp-ctgan",
        "dp_ctgan",
        "dpctgan"
    ],

    "SPP-GAN": [
        "spp-gan",
        "spp_gan",
        "sppgan"
    ]
}


# ==============================================================================
# E2.3 — INITIALIZE CANDIDATE MAP
# ==============================================================================

SYNTHETIC_CANDIDATE_MAP = {

    dataset: {
        method: []
        for method in RESEARCH_METHODS
    }
    for dataset in RESEARCH_DATASETS
}


# ==============================================================================
# E2.4 — RECOVER FROM E1.2 ARTIFACT SEARCH
# ==============================================================================

if "DATASET_LIKE_ARTIFACTS" in globals():

    for _, row in DATASET_LIKE_ARTIFACTS.iterrows():

        filepath = pathlib.Path(
            row["File"]
        )

        matched_datasets = [
            item.strip()
            for item in str(
                row["Matched Datasets"]
            ).split(",")
            if item.strip()
        ]

        matched_methods = [
            item.strip()
            for item in str(
                row["Matched Methods"]
            ).split(",")
            if item.strip()
        ]


        for dataset in matched_datasets:

            if dataset not in RESEARCH_DATASETS:
                continue

            for method in matched_methods:

                if method not in RESEARCH_METHODS:
                    continue

                if filepath not in (
                    SYNTHETIC_CANDIDATE_MAP[
                        dataset
                    ][method]
                ):

                    SYNTHETIC_CANDIDATE_MAP[
                        dataset
                    ][method
                    ].append(
                        filepath
                    )


# ==============================================================================
# E2.5 — ALSO RECOVER FROM E1.1 VALID FILES
# ==============================================================================

if "VALID_SYNTHETIC_FILES" in globals():

    for filepath in VALID_SYNTHETIC_FILES:

        filepath = pathlib.Path(
            filepath
        )

        filename = filepath.name.lower()


        for dataset in RESEARCH_DATASETS:

            dataset_match = any(
                token.lower() in filename
                for token in E2_DATASET_TOKENS[
                    dataset
                ]
            )

            if not dataset_match:
                continue


            for method in RESEARCH_METHODS:

                method_match = any(
                    token.lower() in filename
                    for token in E2_METHOD_TOKENS[
                        method
                    ]
                )

                if not method_match:
                    continue


                if filepath not in (
                    SYNTHETIC_CANDIDATE_MAP[
                        dataset
                    ][method]
                ):

                    SYNTHETIC_CANDIDATE_MAP[
                        dataset
                    ][method
                    ].append(
                        filepath
                    )


# ==============================================================================
# E2.6 — REMOVE DUPLICATES
# ==============================================================================

for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        SYNTHETIC_CANDIDATE_MAP[
            dataset
        ][method] = sorted(
            set(
                SYNTHETIC_CANDIDATE_MAP[
                    dataset
                ][method]
            ),
            key=lambda path: str(path).lower()
        )


# ==============================================================================
# E2.7 — BUILD CANDIDATE TABLE
# ==============================================================================

E2_CANDIDATE_ROWS = []


for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        candidates = SYNTHETIC_CANDIDATE_MAP[
            dataset
        ][method]


        E2_CANDIDATE_ROWS.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Candidate Count": len(candidates),
                "Candidates": "; ".join(
                    str(path)
                    for path in candidates
                )
            }
        )


SYNTHETIC_CANDIDATE_TABLE = pd.DataFrame(
    E2_CANDIDATE_ROWS
)


# ==============================================================================
# E2.8 — DISPLAY
# ==============================================================================

print()
print("-" * 100)
print("FINAL SYNTHETIC CANDIDATE MATRIX")
print("-" * 100)

display(
    SYNTHETIC_CANDIDATE_TABLE
)


# ==============================================================================
# E2.9 — SUMMARY
# ==============================================================================

E2_FOUND_CELLS = int(
    (
        SYNTHETIC_CANDIDATE_TABLE[
            "Candidate Count"
        ] > 0
    ).sum()
)

E2_MISSING_CELLS = (
    len(RESEARCH_DATASETS)
    * len(RESEARCH_METHODS)
    - E2_FOUND_CELLS
)


print()
print("-" * 100)
print("E2 SUMMARY")
print("-" * 100)

print(
    f"Expected Dataset × Method cells : "
    f"{len(RESEARCH_DATASETS) * len(RESEARCH_METHODS)}"
)

print(
    f"Cells with candidates            : "
    f"{E2_FOUND_CELLS}"
)

print(
    f"Cells without candidates         : "
    f"{E2_MISSING_CELLS}"
)


NOTEBOOK_9_E2_COMPLETED = True

print()
print("=" * 100)
print("E2 — FINAL SYNTHETIC CANDIDATE MAPPING COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# E3 — RESOLVE FINAL SYNTHETIC DATASETS
# ==============================================================================

print("=" * 100)
print("E3 — RESOLVE FINAL SYNTHETIC DATASETS")
print("=" * 100)


# ==============================================================================
# E3.1 — FINAL RESEARCH SCOPE
# ==============================================================================

RESEARCH_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RESEARCH_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# E3.2 — DATASET TOKENS
# ==============================================================================

E3_DATASET_TOKENS = {

    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income",
        "adultincome"
    ],

    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing",
        "bankmarketing"
    ],

    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us",
        "diabetes130us"
    ]
}


# ==============================================================================
# E3.3 — METHOD TOKENS
# ==============================================================================

E3_METHOD_TOKENS = {

    "CTGAN": [
        "ctgan"
    ],

    "TVAE": [
        "tvae"
    ],

    "DP-CTGAN": [
        "dp-ctgan",
        "dp_ctgan",
        "dpctgan"
    ],

    "SPP-GAN": [
        "spp-gan",
        "spp_gan",
        "sppgan"
    ]
}


# ==============================================================================
# E3.4 — REQUIRED SYNTHETIC-DATA TOKENS
# ==============================================================================

E3_SYNTHETIC_TOKENS = [
    "synthetic",
    "generated",
    "generation",
    "samples"
]


# ==============================================================================
# E3.5 — EXCLUSION TOKENS
# ==============================================================================
# These indicate that a file is an artifact/report rather than a final
# synthetic dataset.

E3_EXCLUSION_TOKENS = [

    "training_history",
    "training-history",
    "training history",

    "training_summary",
    "training-summary",
    "training summary",

    "metadata",

    "evaluation",
    "evaluated",

    "validation",
    "validated",

    "summary",

    "report",
    "reports",

    "checkpoint",

    "model",
    "models",

    "generator",
    "discriminator",

    "architecture",

    "parameter",

    "parameters",

    "loss",

    "history",

    "metric",
    "metrics",

    "privacy_accounting",
    "privacy-accounting",

    "structure",

    "missing_value",

    "missing-value",

    "distribution",

    "correlation",

    "comparison"
]


# ==============================================================================
# E3.6 — SEARCH ALL CANDIDATE FILES
# ==============================================================================

E3_ALL_CANDIDATES = []


E3_SEARCH_ROOTS = [
    SYNTHETIC_DATA_DIRECTORY,
    RESULTS_DIRECTORY,
    EVALUATION_DIRECTORY,
    REPORTS_DIRECTORY
]


for root in E3_SEARCH_ROOTS:

    if not root.exists():
        continue

    for filepath in root.rglob("*"):

        if not filepath.is_file():
            continue

        if filepath.suffix.lower() not in {
            ".csv",
            ".parquet"
        }:
            continue

        E3_ALL_CANDIDATES.append(
            filepath.resolve()
        )


E3_ALL_CANDIDATES = sorted(
    set(E3_ALL_CANDIDATES),
    key=lambda path: str(path).lower()
)


# ==============================================================================
# E3.7 — STRICT CANDIDATE CLASSIFICATION FUNCTION
# ==============================================================================

def classify_final_synthetic_candidate(
    filepath,
    dataset,
    method
):

    filename = filepath.name.lower()
    full_path = str(filepath).lower()


    # --------------------------------------------------------------------------
    # Dataset match
    # --------------------------------------------------------------------------

    dataset_match = any(
        token.lower() in filename
        for token in E3_DATASET_TOKENS[
            dataset
        ]
    )

    if not dataset_match:
        return False, "Dataset identifier not present"


    # --------------------------------------------------------------------------
    # Method match
    # --------------------------------------------------------------------------

    method_match = any(
        token.lower() in filename
        for token in E3_METHOD_TOKENS[
            method
        ]
    )

    if not method_match:
        return False, "Method identifier not present"


    # --------------------------------------------------------------------------
    # Synthetic-data semantics
    # --------------------------------------------------------------------------

    synthetic_match = any(
        token.lower() in filename
        for token in E3_SYNTHETIC_TOKENS
    )

    if not synthetic_match:
        return False, "No synthetic-data identifier"


    # --------------------------------------------------------------------------
    # Exclude research artifacts
    # --------------------------------------------------------------------------

    for exclusion_token in E3_EXCLUSION_TOKENS:

        if exclusion_token.lower() in filename:

            return (
                False,
                f"Excluded artifact type: {exclusion_token}"
            )


    # --------------------------------------------------------------------------
    # Candidate accepted
    # --------------------------------------------------------------------------

    return True, "Valid synthetic-data candidate"


# ==============================================================================
# E3.8 — BUILD STRICT CANDIDATE MATRIX
# ==============================================================================

E3_STRICT_CANDIDATE_ROWS = []


for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        accepted_candidates = []


        for filepath in E3_ALL_CANDIDATES:

            accepted, reason = (
                classify_final_synthetic_candidate(
                    filepath,
                    dataset,
                    method
                )
            )


            if accepted:

                accepted_candidates.append(
                    filepath
                )


        # ----------------------------------------------------------------------
        # PRIORITIZE SYNTHETIC_DATA DIRECTORY
        # ----------------------------------------------------------------------

        accepted_candidates = sorted(
            set(accepted_candidates),
            key=lambda path: (
                0
                if SYNTHETIC_DATA_DIRECTORY
                in path.parents
                else 1,
                str(path).lower()
            )
        )


        for filepath in accepted_candidates:

            E3_STRICT_CANDIDATE_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Candidate File": str(filepath),
                    "Priority": (
                        "PRIMARY"
                        if SYNTHETIC_DATA_DIRECTORY
                        in filepath.parents
                        else "SECONDARY"
                    )
                }
            )


E3_STRICT_CANDIDATES = pd.DataFrame(
    E3_STRICT_CANDIDATE_ROWS
)


# ==============================================================================
# E3.9 — DISPLAY STRICT CANDIDATES
# ==============================================================================

print()
print("-" * 100)
print("STRICT FINAL SYNTHETIC-DATA CANDIDATES")
print("-" * 100)


if E3_STRICT_CANDIDATES.empty:

    print(
        "No strict final synthetic-data candidates were found."
    )

else:

    display(
        E3_STRICT_CANDIDATES
    )


# ==============================================================================
# E3.10 — RESOLVE FINAL DATASETS
# ==============================================================================

SYNTHETIC_DATA_FILES = {}

E3_RESOLUTION_ROWS = []


for dataset in RESEARCH_DATASETS:

    SYNTHETIC_DATA_FILES[
        dataset
    ] = {}


    for method in RESEARCH_METHODS:

        if E3_STRICT_CANDIDATES.empty:

            candidates = []

        else:

            candidates = (
                E3_STRICT_CANDIDATES[
                    (
                        E3_STRICT_CANDIDATES[
                            "Dataset"
                        ]
                        == dataset
                    )
                    &
                    (
                        E3_STRICT_CANDIDATES[
                            "Method"
                        ]
                        == method
                    )
                ]
                .sort_values(
                    by=[
                        "Priority",
                        "Candidate File"
                    ]
                )[
                    "Candidate File"
                ]
                .tolist()
            )


        # ----------------------------------------------------------------------
        # EXACTLY ONE CANDIDATE
        # ----------------------------------------------------------------------

        if len(candidates) == 1:

            resolved_file = pathlib.Path(
                candidates[0]
            )

            status = "RESOLVED"


        # ----------------------------------------------------------------------
        # MULTIPLE CANDIDATES
        # ----------------------------------------------------------------------

        elif len(candidates) > 1:

            # Prefer a single candidate inside synthetic_data.
            primary_candidates = [
                pathlib.Path(path)
                for path in candidates
                if SYNTHETIC_DATA_DIRECTORY
                in pathlib.Path(path).parents
            ]


            if len(primary_candidates) == 1:

                resolved_file = (
                    primary_candidates[0]
                )

                status = "RESOLVED"


            else:

                resolved_file = None

                status = "AMBIGUOUS"


        # ----------------------------------------------------------------------
        # NO CANDIDATE
        # ----------------------------------------------------------------------

        else:

            resolved_file = None

            status = "MISSING"


        SYNTHETIC_DATA_FILES[
            dataset
        ][method] = resolved_file


        E3_RESOLUTION_ROWS.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Candidate Count": len(candidates),
                "Status": status,
                "Resolved File": (
                    str(resolved_file)
                    if resolved_file is not None
                    else None
                )
            }
        )


# ==============================================================================
# E3.11 — FINAL RESOLUTION TABLE
# ==============================================================================

SYNTHETIC_RESOLUTION_RESULTS = pd.DataFrame(
    E3_RESOLUTION_ROWS
)


print()
print("-" * 100)
print("FINAL SYNTHETIC DATASET RESOLUTION")
print("-" * 100)

display(
    SYNTHETIC_RESOLUTION_RESULTS
)


# ==============================================================================
# E3.12 — SUMMARY
# ==============================================================================

E3_RESOLVED_COUNT = int(
    (
        SYNTHETIC_RESOLUTION_RESULTS[
            "Status"
        ]
        == "RESOLVED"
    ).sum()
)


E3_MISSING_COUNT = int(
    (
        SYNTHETIC_RESOLUTION_RESULTS[
            "Status"
        ]
        == "MISSING"
    ).sum()
)


E3_AMBIGUOUS_COUNT = int(
    (
        SYNTHETIC_RESOLUTION_RESULTS[
            "Status"
        ]
        == "AMBIGUOUS"
    ).sum()
)


print()
print("-" * 100)
print("E3 SUMMARY")
print("-" * 100)

print(
    f"Expected combinations : "
    f"{len(RESEARCH_DATASETS) * len(RESEARCH_METHODS)}"
)

print(
    f"Resolved               : "
    f"{E3_RESOLVED_COUNT}"
)

print(
    f"Missing                : "
    f"{E3_MISSING_COUNT}"
)

print(
    f"Ambiguous              : "
    f"{E3_AMBIGUOUS_COUNT}"
)


# ==============================================================================
# E3.13 — DISPLAY RESOLVED FILES
# ==============================================================================

print()
print("-" * 100)
print("RESOLVED FINAL SYNTHETIC DATASETS")
print("-" * 100)


resolved_rows = (
    SYNTHETIC_RESOLUTION_RESULTS[
        SYNTHETIC_RESOLUTION_RESULTS[
            "Status"
        ] == "RESOLVED"
    ]
)


if resolved_rows.empty:

    print("  NONE")

else:

    for _, row in resolved_rows.iterrows():

        print(
            f"  {row['Dataset']} × "
            f"{row['Method']}"
        )

        print(
            f"    {row['Resolved File']}"
        )


# ==============================================================================
# E3.14 — DISPLAY MISSING DATASETS
# ==============================================================================

missing_rows = (
    SYNTHETIC_RESOLUTION_RESULTS[
        SYNTHETIC_RESOLUTION_RESULTS[
            "Status"
        ] == "MISSING"
    ]
)


print()
print("-" * 100)
print("MISSING FINAL SYNTHETIC DATASETS")
print("-" * 100)


if missing_rows.empty:

    print("  NONE")

else:

    for _, row in missing_rows.iterrows():

        print(
            f"  {row['Dataset']} × "
            f"{row['Method']}"
        )


# ==============================================================================
# E3.15 — FINAL STATUS
# ==============================================================================

NOTEBOOK_9_E3_COMPLETED = True


print()
print("=" * 100)
print("E3 — FINAL SYNTHETIC DATASET RESOLUTION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# E4 — LOAD ALL FINAL SYNTHETIC DATASETS
# ==============================================================================

print("=" * 100)
print("E4 — LOAD ALL FINAL SYNTHETIC DATASETS")
print("=" * 100)


# ==============================================================================
# E4.1 — FINAL RESEARCH SCOPE
# ==============================================================================

RESEARCH_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RESEARCH_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# E4.2 — ENSURE FINAL RESOLUTION OBJECT EXISTS
# ==============================================================================

if "SYNTHETIC_DATA_FILES" not in globals():

    raise RuntimeError(
        "SYNTHETIC_DATA_FILES is not available. "
        "Run E3 — RESOLVE FINAL SYNTHETIC DATASETS first."
    )


# ==============================================================================
# E4.3 — INITIALIZE FINAL SYNTHETIC DATA CONTAINERS
# ==============================================================================

SYNTHETIC_DATASETS = {

    dataset: {
        method: None
        for method in RESEARCH_METHODS
    }

    for dataset in RESEARCH_DATASETS
}


SYNTHETIC_LOAD_ROWS = []


# ==============================================================================
# E4.4 — LOAD ALL RESOLVED FINAL SYNTHETIC DATASETS
# ==============================================================================

for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        filepath = SYNTHETIC_DATA_FILES[
            dataset
        ][method]


        # ----------------------------------------------------------------------
        # NO FINAL DATASET RESOLVED
        # ----------------------------------------------------------------------

        if filepath is None:

            SYNTHETIC_LOAD_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "MISSING",
                    "Rows": None,
                    "Columns": None,
                    "File": None
                }
            )

            continue


        filepath = pathlib.Path(
            filepath
        )


        # ----------------------------------------------------------------------
        # VERIFY FILE EXISTS
        # ----------------------------------------------------------------------

        if not filepath.exists():

            SYNTHETIC_LOAD_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "FILE_NOT_FOUND",
                    "Rows": None,
                    "Columns": None,
                    "File": str(filepath)
                }
            )

            continue


        # ----------------------------------------------------------------------
        # LOAD CSV
        # ----------------------------------------------------------------------

        try:

            if filepath.suffix.lower() == ".csv":

                dataframe = pd.read_csv(
                    filepath,
                    low_memory=False
                )

            elif filepath.suffix.lower() == ".parquet":

                dataframe = pd.read_parquet(
                    filepath
                )

            else:

                raise ValueError(
                    f"Unsupported synthetic-data format: "
                    f"{filepath.suffix}"
                )


            # ------------------------------------------------------------------
            # EMPTY DATASET CHECK
            # ------------------------------------------------------------------

            if dataframe.empty:

                SYNTHETIC_LOAD_ROWS.append(
                    {
                        "Dataset": dataset,
                        "Method": method,
                        "Status": "EMPTY",
                        "Rows": 0,
                        "Columns": 0,
                        "File": str(filepath)
                    }
                )

                continue


            # ------------------------------------------------------------------
            # STORE FINAL DATASET
            # ------------------------------------------------------------------

            SYNTHETIC_DATASETS[
                dataset
            ][method] = dataframe


            SYNTHETIC_LOAD_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "LOADED",
                    "Rows": int(
                        dataframe.shape[0]
                    ),
                    "Columns": int(
                        dataframe.shape[1]
                    ),
                    "File": str(filepath)
                }
            )


        except Exception as error:

            SYNTHETIC_LOAD_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "LOAD_ERROR",
                    "Rows": None,
                    "Columns": None,
                    "File": str(filepath)
                }
            )

            print()
            print(
                f"LOAD ERROR — {dataset} × {method}"
            )

            print(
                f"  {error}"
            )


# ==============================================================================
# E4.5 — BUILD LOAD RESULTS
# ==============================================================================

SYNTHETIC_LOAD_RESULTS = pd.DataFrame(
    SYNTHETIC_LOAD_ROWS
)


# ==============================================================================
# E4.6 — DISPLAY FINAL LOAD RESULTS
# ==============================================================================

print()
print("-" * 100)
print("FINAL SYNTHETIC DATASET LOAD RESULTS")
print("-" * 100)

display(
    SYNTHETIC_LOAD_RESULTS
)


# ==============================================================================
# E4.7 — LOAD STATUS SUMMARY
# ==============================================================================

print()
print("-" * 100)
print("LOAD SUMMARY")
print("-" * 100)

print(
    SYNTHETIC_LOAD_RESULTS[
        "Status"
    ].value_counts()
)


# ==============================================================================
# E4.8 — SUCCESSFULLY LOADED DATASETS
# ==============================================================================

LOADED_SYNTHETIC_RESULTS = (
    SYNTHETIC_LOAD_RESULTS[
        SYNTHETIC_LOAD_RESULTS[
            "Status"
        ] == "LOADED"
    ]
)


print()
print("-" * 100)
print("SUCCESSFULLY LOADED SYNTHETIC DATASETS")
print("-" * 100)


if LOADED_SYNTHETIC_RESULTS.empty:

    print("  NONE")

else:

    for _, row in (
        LOADED_SYNTHETIC_RESULTS.iterrows()
    ):

        print(
            f"  {row['Dataset']} × "
            f"{row['Method']} — "
            f"{row['Rows']:,} rows × "
            f"{row['Columns']:,} columns"
        )

        print(
            f"    {row['File']}"
        )


# ==============================================================================
# E4.9 — DATASET COVERAGE
# ==============================================================================

E4_DATASET_COVERAGE_ROWS = []


for dataset in RESEARCH_DATASETS:

    dataset_results = (
        SYNTHETIC_LOAD_RESULTS[
            SYNTHETIC_LOAD_RESULTS[
                "Dataset"
            ] == dataset
        ]
    )


    loaded_count = int(
        (
            dataset_results[
                "Status"
            ] == "LOADED"
        ).sum()
    )


    E4_DATASET_COVERAGE_ROWS.append(
        {
            "Dataset": dataset,
            "Expected Methods": len(
                RESEARCH_METHODS
            ),
            "Loaded Methods": loaded_count,
            "Missing Methods": (
                len(RESEARCH_METHODS)
                - loaded_count
            ),
            "Coverage (%)": round(
                (
                    loaded_count
                    / len(RESEARCH_METHODS)
                ) * 100,
                2
            )
        }
    )


E4_DATASET_COVERAGE = pd.DataFrame(
    E4_DATASET_COVERAGE_ROWS
)


print()
print("-" * 100)
print("DATASET-LEVEL SYNTHETIC COVERAGE")
print("-" * 100)

display(
    E4_DATASET_COVERAGE
)


# ==============================================================================
# E4.10 — METHOD COVERAGE
# ==============================================================================

E4_METHOD_COVERAGE_ROWS = []


for method in RESEARCH_METHODS:

    method_results = (
        SYNTHETIC_LOAD_RESULTS[
            SYNTHETIC_LOAD_RESULTS[
                "Method"
            ] == method
        ]
    )


    loaded_count = int(
        (
            method_results[
                "Status"
            ] == "LOADED"
        ).sum()
    )


    E4_METHOD_COVERAGE_ROWS.append(
        {
            "Method": method,
            "Expected Datasets": len(
                RESEARCH_DATASETS
            ),
            "Loaded Datasets": loaded_count,
            "Missing Datasets": (
                len(RESEARCH_DATASETS)
                - loaded_count
            ),
            "Coverage (%)": round(
                (
                    loaded_count
                    / len(RESEARCH_DATASETS)
                ) * 100,
                2
            )
        }
    )


E4_METHOD_COVERAGE = pd.DataFrame(
    E4_METHOD_COVERAGE_ROWS
)


print()
print("-" * 100)
print("METHOD-LEVEL SYNTHETIC COVERAGE")
print("-" * 100)

display(
    E4_METHOD_COVERAGE
)


# ==============================================================================
# E4.11 — FINAL COVERAGE
# ==============================================================================

E4_EXPECTED_DATASET_COUNT = (
    len(RESEARCH_DATASETS)
    * len(RESEARCH_METHODS)
)


E4_LOADED_DATASET_COUNT = int(
    (
        SYNTHETIC_LOAD_RESULTS[
            "Status"
        ] == "LOADED"
    ).sum()
)


E4_MISSING_DATASET_COUNT = (
    E4_EXPECTED_DATASET_COUNT
    - E4_LOADED_DATASET_COUNT
)


E4_COVERAGE_PERCENT = round(
    (
        E4_LOADED_DATASET_COUNT
        / E4_EXPECTED_DATASET_COUNT
    ) * 100,
    2
)


print()
print("-" * 100)
print("FINAL SYNTHETIC DATA COVERAGE")
print("-" * 100)

print(
    f"Expected synthetic datasets : "
    f"{E4_EXPECTED_DATASET_COUNT}"
)

print(
    f"Successfully loaded         : "
    f"{E4_LOADED_DATASET_COUNT}"
)

print(
    f"Still unavailable           : "
    f"{E4_MISSING_DATASET_COUNT}"
)

print(
    f"Overall coverage            : "
    f"{E4_COVERAGE_PERCENT}%"
)


# ==============================================================================
# E4.12 — RESEARCH SAFETY CHECK
# ==============================================================================

if E4_LOADED_DATASET_COUNT == 0:

    E4_STATUS = "FAIL"

elif E4_LOADED_DATASET_COUNT < E4_EXPECTED_DATASET_COUNT:

    E4_STATUS = "PARTIAL"

else:

    E4_STATUS = "COMPLETE"


print()
print("-" * 100)
print("E4 RESEARCH STATUS")
print("-" * 100)

print(
    f"Status : {E4_STATUS}"
)


# ==============================================================================
# E4.13 — FINAL COMPLETION FLAG
# ==============================================================================

NOTEBOOK_9_E4_COMPLETED = True


print()
print("=" * 100)
print("E4 — ALL AVAILABLE FINAL SYNTHETIC DATASETS LOADED")
print("=" * 100)
```

```python
# ==============================================================================
# E5 — SYNTHETIC DATASET STRUCTURAL VALIDATION
# ==============================================================================

print("=" * 100)
print("E5 — SYNTHETIC DATASET STRUCTURAL VALIDATION")
print("=" * 100)


SYNTHETIC_STRUCTURE_RESULTS = []


for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        df = SYNTHETIC_DATASETS[
            dataset
        ][method]


        if df is None:

            SYNTHETIC_STRUCTURE_RESULTS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "NOT AVAILABLE",
                    "Rows": None,
                    "Columns": None,
                    "Non-Empty": False,
                    "Unique Column Names": False,
                    "Missing Cells": None,
                    "Duplicate Rows": None
                }
            )

            continue


        SYNTHETIC_STRUCTURE_RESULTS.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Status": "VALID",
                "Rows": int(df.shape[0]),
                "Columns": int(df.shape[1]),
                "Non-Empty": not df.empty,
                "Unique Column Names": (
                    len(df.columns)
                    == len(set(df.columns))
                ),
                "Missing Cells": int(
                    df.isna().sum().sum()
                ),
                "Duplicate Rows": int(
                    df.duplicated().sum()
                )
            }
        )


SYNTHETIC_STRUCTURE_RESULTS = pd.DataFrame(
    SYNTHETIC_STRUCTURE_RESULTS
)


print()
display(
    SYNTHETIC_STRUCTURE_RESULTS
)


# ==============================================================================
# INVALID LOADED DATASETS
# ==============================================================================

STRUCTURAL_FAILURES = (
    SYNTHETIC_STRUCTURE_RESULTS[
        (
            SYNTHETIC_STRUCTURE_RESULTS[
                "Status"
            ] == "VALID"
        )
        &
        (
            (
                ~SYNTHETIC_STRUCTURE_RESULTS[
                    "Non-Empty"
                ]
            )
            |
            (
                ~SYNTHETIC_STRUCTURE_RESULTS[
                    "Unique Column Names"
                ]
            )
        )
    ]
)


if not STRUCTURAL_FAILURES.empty:

    print()
    print(
        "Structural validation failures:"
    )

    display(
        STRUCTURAL_FAILURES
    )

    raise RuntimeError(
        "One or more loaded synthetic datasets failed structural validation."
    )


NOTEBOOK_9_E5_COMPLETED = True

print()
print("=" * 100)
print("E5 — SYNTHETIC STRUCTURAL VALIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# E6 — SYNTHETIC–REFERENCE SCHEMA VALIDATION
# ==============================================================================

print("=" * 100)
print("E6 — SYNTHETIC–REFERENCE SCHEMA VALIDATION")
print("=" * 100)


if "REFERENCE_DATASETS" not in globals():

    raise RuntimeError(
        "REFERENCE_DATASETS is not available. "
        "Complete Section D before E6."
    )


SCHEMA_VALIDATION_RESULTS = []


for dataset in RESEARCH_DATASETS:

    reference_df = REFERENCE_DATASETS[
        dataset
    ]

    reference_columns = list(
        reference_df.columns
    )


    for method in RESEARCH_METHODS:

        synthetic_df = SYNTHETIC_DATASETS[
            dataset
        ][method]


        if synthetic_df is None:

            SCHEMA_VALIDATION_RESULTS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Status": "NOT AVAILABLE",
                    "Column Count Match": False,
                    "Column Order Match": False,
                    "Column Set Match": False
                }
            )

            continue


        synthetic_columns = list(
            synthetic_df.columns
        )


        SCHEMA_VALIDATION_RESULTS.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Status": "VALIDATED",
                "Column Count Match": (
                    len(reference_columns)
                    == len(synthetic_columns)
                ),
                "Column Order Match": (
                    reference_columns
                    == synthetic_columns
                ),
                "Column Set Match": (
                    set(reference_columns)
                    == set(synthetic_columns)
                )
            }
        )


SCHEMA_VALIDATION_RESULTS = pd.DataFrame(
    SCHEMA_VALIDATION_RESULTS
)


print()
display(
    SCHEMA_VALIDATION_RESULTS
)


# ==============================================================================
# VALIDATION FAILURES
# ==============================================================================

SCHEMA_FAILURES = (
    SCHEMA_VALIDATION_RESULTS[
        (
            SCHEMA_VALIDATION_RESULTS[
                "Status"
            ] == "VALIDATED"
        )
        &
        (
            ~SCHEMA_VALIDATION_RESULTS[
                "Column Count Match"
            ]
            |
            ~SCHEMA_VALIDATION_RESULTS[
                "Column Set Match"
            ]
        )
    ]
)


if not SCHEMA_FAILURES.empty:

    print()
    print(
        "Schema validation failures:"
    )

    display(
        SCHEMA_FAILURES
    )

    raise RuntimeError(
        "Synthetic–reference schema validation failed."
    )


NOTEBOOK_9_E6_COMPLETED = True

print()
print("=" * 100)
print("E6 — SYNTHETIC–REFERENCE SCHEMA VALIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# E7 — FINAL SYNTHETIC DATASET REGISTRY & RECOVERY
# ==============================================================================

print("=" * 100)
print("E7 — FINAL SYNTHETIC DATASET REGISTRY & RECOVERY")
print("=" * 100)


# ==============================================================================
# E7.1 — FINAL RESEARCH SCOPE
# ==============================================================================

RESEARCH_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RESEARCH_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# E7.2 — IDENTIFIER TOKENS
# ==============================================================================

E7_DATASET_TOKENS = {
    "adult_income": [
        "adult_income",
        "adult-income",
        "adult income",
        "adultincome"
    ],
    "bank_marketing": [
        "bank_marketing",
        "bank-marketing",
        "bank marketing",
        "bankmarketing"
    ],
    "diabetes_130us": [
        "diabetes_130us",
        "diabetes-130us",
        "diabetes 130us",
        "diabetes130us"
    ]
}


E7_METHOD_TOKENS = {
    "CTGAN": [
        "ctgan"
    ],
    "TVAE": [
        "tvae"
    ],
    "DP-CTGAN": [
        "dp_ctgan",
        "dp-ctgan",
        "dpctgan"
    ],
    "SPP-GAN": [
        "spp_gan",
        "spp-gan",
        "sppgan"
    ]
}


# ==============================================================================
# E7.3 — ARTIFACT EXCLUSION TOKENS
# ==============================================================================

E7_EXCLUSION_TOKENS = [
    "training_history",
    "training-history",
    "training history",
    "training_summary",
    "training-summary",
    "training summary",
    "metadata",
    "evaluation",
    "validation",
    "validated",
    "summary",
    "report",
    "reports",
    "checkpoint",
    "checkpoints",
    "model",
    "models",
    "generator",
    "discriminator",
    "architecture",
    "parameter",
    "parameters",
    "loss",
    "history",
    "metric",
    "metrics",
    "privacy_accounting",
    "privacy-accounting",
    "structure",
    "missing_value",
    "missing-value",
    "distribution",
    "correlation",
    "comparison"
]


# ==============================================================================
# E7.4 — DISCOVER ALL CSV/PARQUET FILES
# ==============================================================================

E7_SEARCH_ROOTS = [
    PROJECT_ROOT
]

E7_ALL_DATA_FILES = []


for root in E7_SEARCH_ROOTS:

    if not root.exists():
        continue

    for filepath in root.rglob("*"):

        if not filepath.is_file():
            continue

        if filepath.suffix.lower() not in {
            ".csv",
            ".parquet"
        }:
            continue

        E7_ALL_DATA_FILES.append(
            filepath.resolve()
        )


E7_ALL_DATA_FILES = sorted(
    set(E7_ALL_DATA_FILES),
    key=lambda x: str(x).lower()
)


print()
print("-" * 100)
print("E7 DATA FILE DISCOVERY")
print("-" * 100)

print(
    f"CSV/Parquet files discovered : "
    f"{len(E7_ALL_DATA_FILES)}"
)


# ==============================================================================
# E7.5 — CANDIDATE MATCHING
# ==============================================================================

def e7_match_candidate(
    filepath,
    dataset,
    method
):

    filename = filepath.name.lower()

    full_path = str(
        filepath
    ).lower()


    # --------------------------------------------------------------------------
    # Dataset identification
    # --------------------------------------------------------------------------

    dataset_match = any(
        token.lower() in filename
        or token.lower() in full_path
        for token in E7_DATASET_TOKENS[
            dataset
        ]
    )

    if not dataset_match:
        return False


    # --------------------------------------------------------------------------
    # Method identification
    # --------------------------------------------------------------------------

    method_match = any(
        token.lower() in filename
        or token.lower() in full_path
        for token in E7_METHOD_TOKENS[
            method
        ]
    )

    if not method_match:
        return False


    # --------------------------------------------------------------------------
    # Exclude known non-dataset artifacts
    # --------------------------------------------------------------------------

    for token in E7_EXCLUSION_TOKENS:

        if token.lower() in filename:

            return False


    return True


# ==============================================================================
# E7.6 — BUILD CANDIDATE REGISTRY
# ==============================================================================

E7_CANDIDATE_ROWS = []


for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        candidates = []


        for filepath in E7_ALL_DATA_FILES:

            if e7_match_candidate(
                filepath,
                dataset,
                method
            ):

                candidates.append(
                    filepath
                )


        for filepath in candidates:

            E7_CANDIDATE_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Candidate File": str(filepath),
                    "Synthetic Directory":
                        "synthetic_data"
                        in str(filepath).lower()
                }
            )


E7_CANDIDATES = pd.DataFrame(
    E7_CANDIDATE_ROWS
)


# ==============================================================================
# E7.7 — DISPLAY RECOVERED CANDIDATES
# ==============================================================================

print()
print("-" * 100)
print("RECOVERED SYNTHETIC DATASET CANDIDATES")
print("-" * 100)


if E7_CANDIDATES.empty:

    print(
        "No candidate synthetic datasets were discovered."
    )

else:

    display(
        E7_CANDIDATES
    )


# ==============================================================================
# E7.8 — DATASET VALIDATION FUNCTION
# ==============================================================================

def e7_validate_dataframe(
    filepath
):

    try:

        if filepath.suffix.lower() == ".csv":

            df = pd.read_csv(
                filepath,
                low_memory=False
            )

        elif filepath.suffix.lower() == ".parquet":

            df = pd.read_parquet(
                filepath
            )

        else:

            return None, "Unsupported format"


        if df.empty:

            return None, "Empty dataset"


        # Remove completely unnamed index columns only
        unnamed_columns = [
            col
            for col in df.columns
            if str(col).lower().startswith(
                "unnamed:"
            )
        ]


        if unnamed_columns:

            df = df.drop(
                columns=unnamed_columns
            )


        if df.shape[1] == 0:

            return None, "No usable columns"


        missing_cells = int(
            df.isna().sum().sum()
        )


        duplicate_rows = int(
            df.duplicated().sum()
        )


        return {
            "dataframe": df,
            "rows": int(df.shape[0]),
            "columns": int(df.shape[1]),
            "missing_cells": missing_cells,
            "duplicate_rows": duplicate_rows
        }, "VALID"


    except Exception as error:

        return None, str(error)


# ==============================================================================
# E7.9 — RESOLVE DATASETS
# ==============================================================================

E7_REGISTRY_ROWS = []

SYNTHETIC_DATASETS = {
    dataset: {
        method: None
        for method in RESEARCH_METHODS
    }
    for dataset in RESEARCH_DATASETS
}


for dataset in RESEARCH_DATASETS:

    for method in RESEARCH_METHODS:

        if E7_CANDIDATES.empty:

            candidate_paths = []

        else:

            candidate_paths = (
                E7_CANDIDATES[
                    (
                        E7_CANDIDATES[
                            "Dataset"
                        ] == dataset
                    )
                    &
                    (
                        E7_CANDIDATES[
                            "Method"
                        ] == method
                    )
                ][
                    "Candidate File"
                ]
                .tolist()
            )


        valid_candidates = []


        # ----------------------------------------------------------------------
        # VALIDATE EVERY CANDIDATE
        # ----------------------------------------------------------------------

        for candidate in candidate_paths:

            filepath = pathlib.Path(
                candidate
            )


            validation, validation_status = (
                e7_validate_dataframe(
                    filepath
                )
            )


            if validation is None:

                continue


            valid_candidates.append(
                (
                    filepath,
                    validation
                )
            )


        # ----------------------------------------------------------------------
        # PRIORITIZE FINAL SYNTHETIC DIRECTORY
        # ----------------------------------------------------------------------

        primary_candidates = [

            item
            for item in valid_candidates
            if "synthetic_data"
            in str(item[0]).lower()
        ]


        if len(primary_candidates) == 1:

            selected_filepath = (
                primary_candidates[0][0]
            )

            selected_validation = (
                primary_candidates[0][1]
            )

            status = "READY"


        elif len(primary_candidates) > 1:

            # Prefer the shortest filename/path as the most
            # direct generated-data artifact.
            primary_candidates = sorted(
                primary_candidates,
                key=lambda item: (
                    len(str(item[0])),
                    str(item[0]).lower()
                )
            )


            selected_filepath = (
                primary_candidates[0][0]
            )

            selected_validation = (
                primary_candidates[0][1]
            )

            status = "READY"


        elif len(valid_candidates) == 1:

            selected_filepath = (
                valid_candidates[0][0]
            )

            selected_validation = (
                valid_candidates[0][1]
            )

            status = "RECOVERED"


        elif len(valid_candidates) > 1:

            selected_filepath = None
            selected_validation = None
            status = "AMBIGUOUS"


        else:

            selected_filepath = None
            selected_validation = None
            status = "MISSING"


        # ----------------------------------------------------------------------
        # STORE DATASET
        # ----------------------------------------------------------------------

        if selected_filepath is not None:

            SYNTHETIC_DATASETS[
                dataset
            ][method] = (
                selected_validation[
                    "dataframe"
                ]
            )


        # ----------------------------------------------------------------------
        # REGISTRY ROW
        # ----------------------------------------------------------------------

        E7_REGISTRY_ROWS.append(
            {
                "Dataset": dataset,
                "Method": method,
                "Status": status,
                "File": (
                    str(selected_filepath)
                    if selected_filepath is not None
                    else None
                ),
                "Rows": (
                    selected_validation["rows"]
                    if selected_validation is not None
                    else None
                ),
                "Columns": (
                    selected_validation["columns"]
                    if selected_validation is not None
                    else None
                ),
                "Missing Cells": (
                    selected_validation[
                        "missing_cells"
                    ]
                    if selected_validation is not None
                    else None
                ),
                "Duplicate Rows": (
                    selected_validation[
                        "duplicate_rows"
                    ]
                    if selected_validation is not None
                    else None
                )
            }
        )


# ==============================================================================
# E7.10 — FINAL REGISTRY
# ==============================================================================

SYNTHETIC_DATASET_REGISTRY_TABLE = pd.DataFrame(
    E7_REGISTRY_ROWS
)


print()
print("-" * 100)
print("FINAL SYNTHETIC DATASET REGISTRY")
print("-" * 100)

display(
    SYNTHETIC_DATASET_REGISTRY_TABLE
)


# ==============================================================================
# E7.11 — COVERAGE SUMMARY
# ==============================================================================

E7_READY_COUNT = int(
    (
        SYNTHETIC_DATASET_REGISTRY_TABLE[
            "Status"
        ].isin(
            [
                "READY",
                "RECOVERED"
            ]
        )
    ).sum()
)


E7_AMBIGUOUS_COUNT = int(
    (
        SYNTHETIC_DATASET_REGISTRY_TABLE[
            "Status"
        ] == "AMBIGUOUS"
    ).sum()
)


E7_MISSING_COUNT = int(
    (
        SYNTHETIC_DATASET_REGISTRY_TABLE[
            "Status"
        ] == "MISSING"
    ).sum()
)


E7_EXPECTED_COUNT = (
    len(RESEARCH_DATASETS)
    * len(RESEARCH_METHODS)
)


E7_COVERAGE = round(
    (
        E7_READY_COUNT
        / E7_EXPECTED_COUNT
    ) * 100,
    2
)


print()
print("-" * 100)
print("E7 COVERAGE SUMMARY")
print("-" * 100)

print(
    f"Expected datasets : {E7_EXPECTED_COUNT}"
)

print(
    f"Ready/Recovered   : {E7_READY_COUNT}"
)

print(
    f"Ambiguous         : {E7_AMBIGUOUS_COUNT}"
)

print(
    f"Missing           : {E7_MISSING_COUNT}"
)

print(
    f"Coverage          : {E7_COVERAGE}%"
)


# ==============================================================================
# E7.12 — FINAL STATUS
# ==============================================================================

if E7_READY_COUNT == E7_EXPECTED_COUNT:

    E7_STATUS = "COMPLETE"

elif E7_READY_COUNT > 0:

    E7_STATUS = "PARTIAL"

else:

    E7_STATUS = "FAIL"


print()
print("-" * 100)
print("E7 RESEARCH STATUS")
print("-" * 100)

print(
    f"Status : {E7_STATUS}"
)


NOTEBOOK_9_E7_COMPLETED = True


print()
print("=" * 100)
print("E7 — FINAL SYNTHETIC DATASET REGISTRY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# F1 — EVALUATION ARTIFACT DISCOVERY
# ==============================================================================

print("=" * 100)
print("F1 — EVALUATION ARTIFACT DISCOVERY")
print("=" * 100)


# ==============================================================================
# F1.1 — FINAL RESEARCH SCOPE
# ==============================================================================

RESEARCH_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

RESEARCH_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]


# ==============================================================================
# F1.2 — EVALUATION CATEGORIES
# ==============================================================================

EVALUATION_CATEGORIES = [
    "fidelity",
    "utility",
    "privacy",
    "structural",
    "statistical",
    "downstream"
]


# ==============================================================================
# F1.3 — EVALUATION ROOTS
# ==============================================================================

F1_SEARCH_ROOTS = []

for directory_name in [
    "EVALUATION_DIRECTORY",
    "REPORTS_DIRECTORY",
    "RESULTS_DIRECTORY"
]:

    if directory_name in globals():

        directory = globals()[directory_name]

        if directory is not None and pathlib.Path(directory).exists():

            F1_SEARCH_ROOTS.append(
                pathlib.Path(directory)
            )


F1_SEARCH_ROOTS = list(
    dict.fromkeys(
        F1_SEARCH_ROOTS
    )
)


# ==============================================================================
# F1.4 — DISCOVER EVALUATION FILES
# ==============================================================================

F1_DISCOVERED_FILES = []


for root in F1_SEARCH_ROOTS:

    for filepath in root.rglob("*"):

        if not filepath.is_file():
            continue

        if filepath.suffix.lower() not in {
            ".csv",
            ".json",
            ".xlsx",
            ".parquet"
        }:
            continue

        F1_DISCOVERED_FILES.append(
            filepath.resolve()
        )


F1_DISCOVERED_FILES = sorted(
    set(F1_DISCOVERED_FILES),
    key=lambda path: str(path).lower()
)


print()
print("-" * 100)
print("DISCOVERED EVALUATION ARTIFACTS")
print("-" * 100)

print(
    f"Evaluation files discovered : "
    f"{len(F1_DISCOVERED_FILES)}"
)
```

```python
# ==============================================================================
# F2 — EVALUATION SCOPE CLASSIFICATION
# ==============================================================================

print()
print("=" * 100)
print("F2 — EVALUATION SCOPE CLASSIFICATION")
print("=" * 100)


# ==============================================================================
# F2.1 — CATEGORY TOKENS
# ==============================================================================

F2_CATEGORY_TOKENS = {

    "fidelity": [
        "fidelity",
        "distribution",
        "wasserstein",
        "ks",
        "js",
        "kl",
        "correlation",
        "similarity"
    ],

    "utility": [
        "utility",
        "tstr",
        "trts",
        "utility_score"
    ],

    "privacy": [
        "privacy",
        "epsilon",
        "delta",
        "dp",
        "privacy_accounting",
        "privacy_loss"
    ],

    "structural": [
        "structural",
        "structure",
        "schema",
        "missing_value",
        "duplicate"
    ],

    "statistical": [
        "statistical",
        "descriptive",
        "statistics",
        "mean",
        "variance",
        "distribution"
    ],

    "downstream": [
        "downstream",
        "classification",
        "prediction",
        "classifier",
        "model_performance",
        "auc",
        "f1",
        "accuracy"
    ]
}


# ==============================================================================
# F2.2 — EXCLUDE NON-EVALUATION ARTIFACTS
# ==============================================================================

F2_EXCLUSION_TOKENS = [
    "training_history",
    "training-summary",
    "training_summary",
    "checkpoint",
    "model",
    "generator",
    "discriminator",
    "architecture",
    "parameter_summary",
    "saved_models",
    "synthetic_dataset_summary"
]


# ==============================================================================
# F2.3 — CLASSIFY FILE
# ==============================================================================

def classify_evaluation_file(filepath):

    filename = filepath.name.lower()
    full_path = str(filepath).lower()


    for exclusion in F2_EXCLUSION_TOKENS:

        if exclusion in filename:

            return None


    matched_categories = []


    for category, tokens in F2_CATEGORY_TOKENS.items():

        if any(
            token in filename
            or token in full_path
            for token in tokens
        ):

            matched_categories.append(
                category
            )


    if not matched_categories:

        return None


    return matched_categories


# ==============================================================================
# F2.4 — BUILD EVALUATION REGISTRY
# ==============================================================================

F2_REGISTRY_ROWS = []


for filepath in F1_DISCOVERED_FILES:

    categories = classify_evaluation_file(
        filepath
    )


    if categories is None:
        continue


    for category in categories:

        F2_REGISTRY_ROWS.append(
            {
                "File": str(filepath),
                "Filename": filepath.name,
                "Category": category,
                "Extension": filepath.suffix.lower()
            }
        )


EVALUATION_ARTIFACT_REGISTRY = pd.DataFrame(
    F2_REGISTRY_ROWS
)


print()
print("-" * 100)
print("CLASSIFIED EVALUATION ARTIFACTS")
print("-" * 100)


if EVALUATION_ARTIFACT_REGISTRY.empty:

    print(
        "No classified evaluation artifacts found."
    )

else:

    display(
        EVALUATION_ARTIFACT_REGISTRY
    )


# ==============================================================================
# F2.5 — CATEGORY COUNTS
# ==============================================================================

print()
print("-" * 100)
print("EVALUATION CATEGORY COUNTS")
print("-" * 100)

if EVALUATION_ARTIFACT_REGISTRY.empty:

    print("NONE")

else:

    print(
        EVALUATION_ARTIFACT_REGISTRY[
            "Category"
        ].value_counts()
    )
```

```python
# ==============================================================================
# F9 — EVALUATION REPORT VALIDATION
# ==============================================================================

print()
print("=" * 100)
print("F9 — EVALUATION REPORT VALIDATION")
print("=" * 100)


F9_VALIDATION_ROWS = []


for _, row in EVALUATION_ARTIFACT_REGISTRY.iterrows():

    filepath = pathlib.Path(
        row["File"]
    )


    if not filepath.exists():

        status = "FILE_NOT_FOUND"
        rows = None
        columns = None


    else:

        try:

            if filepath.suffix.lower() == ".csv":

                df = pd.read_csv(
                    filepath,
                    low_memory=False
                )

                rows = int(
                    df.shape[0]
                )

                columns = int(
                    df.shape[1]
                )

                status = (
                    "VALID"
                    if not df.empty
                    else "EMPTY"
                )


            elif filepath.suffix.lower() == ".parquet":

                df = pd.read_parquet(
                    filepath
                )

                rows = int(
                    df.shape[0]
                )

                columns = int(
                    df.shape[1]
                )

                status = (
                    "VALID"
                    if not df.empty
                    else "EMPTY"
                )


            elif filepath.suffix.lower() == ".json":

                with open(
                    filepath,
                    "r",
                    encoding="utf-8"
                ) as handle:

                    json.load(handle)

                rows = None
                columns = None
                status = "VALID"


            elif filepath.suffix.lower() == ".xlsx":

                excel = pd.ExcelFile(
                    filepath
                )

                rows = None
                columns = len(
                    excel.sheet_names
                )

                status = "VALID"


            else:

                rows = None
                columns = None
                status = "UNSUPPORTED"


        except Exception as error:

            rows = None
            columns = None
            status = "INVALID"


    F9_VALIDATION_ROWS.append(
        {
            "File": row["File"],
            "Category": row["Category"],
            "Status": status,
            "Rows": rows,
            "Columns": columns
        }
    )


EVALUATION_VALIDATION_RESULTS = pd.DataFrame(
    F9_VALIDATION_ROWS
)


print()
print("-" * 100)
print("EVALUATION VALIDATION RESULTS")
print("-" * 100)

display(
    EVALUATION_VALIDATION_RESULTS
)


print()
print("-" * 100)
print("VALIDATION STATUS SUMMARY")
print("-" * 100)

print(
    EVALUATION_VALIDATION_RESULTS[
        "Status"
    ].value_counts()
)
```

```python
# ==============================================================================
# F10 — FINAL EVALUATION REPORT REGISTRY
# ==============================================================================

print()
print("=" * 100)
print("F10 — FINAL EVALUATION REPORT REGISTRY")
print("=" * 100)


FINAL_EVALUATION_REPORT_REGISTRY = (
    EVALUATION_ARTIFACT_REGISTRY
    .merge(
        EVALUATION_VALIDATION_RESULTS[
            [
                "File",
                "Status",
                "Rows",
                "Columns"
            ]
        ],
        on="File",
        how="left"
    )
    .drop_duplicates(
        subset=[
            "File",
            "Category"
        ]
    )
    .reset_index(drop=True)
)


print()
print("-" * 100)
print("FINAL VALIDATED EVALUATION REPORT REGISTRY")
print("-" * 100)

display(
    FINAL_EVALUATION_REPORT_REGISTRY
)
```

```python
# ==============================================================================
# F11 — EVALUATION COMPLETENESS SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("F11 — EVALUATION COMPLETENESS SUMMARY")
print("=" * 100)


F11_CATEGORY_ROWS = []


for category in EVALUATION_CATEGORIES:

    category_results = (
        FINAL_EVALUATION_REPORT_REGISTRY[
            FINAL_EVALUATION_REPORT_REGISTRY[
                "Category"
            ] == category
        ]
    )


    valid_count = int(
        (
            category_results[
                "Status"
            ] == "VALID"
        ).sum()
    )


    total_count = int(
        category_results.shape[0]
    )


    F11_CATEGORY_ROWS.append(
        {
            "Category": category,
            "Validated Reports": valid_count,
            "Discovered Reports": total_count,
            "Status": (
                "AVAILABLE"
                if valid_count > 0
                else "MISSING"
            )
        }
    )


EVALUATION_COMPLETENESS_SUMMARY = pd.DataFrame(
    F11_CATEGORY_ROWS
)


print()
print("-" * 100)
print("EVALUATION COMPLETENESS BY CATEGORY")
print("-" * 100)

display(
    EVALUATION_COMPLETENESS_SUMMARY
)


# ==============================================================================
# F11.1 — FINAL STATUS
# ==============================================================================

F11_MISSING_CATEGORIES = (
    EVALUATION_COMPLETENESS_SUMMARY[
        EVALUATION_COMPLETENESS_SUMMARY[
            "Status"
        ] == "MISSING"
    ][
        "Category"
    ].tolist()
)


if not F11_MISSING_CATEGORIES:

    F11_STATUS = "COMPLETE"

else:

    F11_STATUS = "PARTIAL"


print()
print("-" * 100)
print("F11 RESEARCH STATUS")
print("-" * 100)

print(
    f"Status : {F11_STATUS}"
)


if F11_MISSING_CATEGORIES:

    print(
        "Missing evaluation categories : "
        + ", ".join(
            F11_MISSING_CATEGORIES
        )
    )


NOTEBOOK_9_F_COMPLETED = True


print()
print("=" * 100)
print("F — FINAL EVALUATION REPORT LOADING COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# G3 — SYNTHETIC DATASET AVAILABILITY AUDIT
# ==============================================================================

print()
print("=" * 100)
print("G3 — SYNTHETIC DATASET AVAILABILITY AUDIT")
print("=" * 100)

G3_ROWS = []

for dataset, method in G_EXPECTED_COMBINATIONS:

    match = pd.DataFrame()

    if "SYNTHETIC_DATASET_REGISTRY_TABLE" in globals():

        registry = SYNTHETIC_DATASET_REGISTRY_TABLE.copy()

        match = registry[
            (registry["Dataset"] == dataset) &
            (registry["Method"] == method)
        ]

    if match.empty:

        status = "MISSING"
        file_path = None
        rows = None
        columns = None

    else:

        record = match.iloc[0]

        status = str(
            record.get("Status", "UNKNOWN")
        )

        file_path = record.get(
            "File",
            None
        )

        rows = record.get(
            "Rows",
            None
        )

        columns = record.get(
            "Columns",
            None
        )

    G3_ROWS.append({
        "Dataset": dataset,
        "Method": method,
        "Synthetic Status": status,
        "Synthetic File": file_path,
        "Rows": rows,
        "Columns": columns
    })

G3_SYNTHETIC_AUDIT = pd.DataFrame(
    G3_ROWS
)

print()
print("-" * 100)
print("SYNTHETIC DATASET AVAILABILITY")
print("-" * 100)

display(
    G3_SYNTHETIC_AUDIT
)

print()
print("-" * 100)
print("SYNTHETIC AVAILABILITY SUMMARY")
print("-" * 100)

print(
    G3_SYNTHETIC_AUDIT[
        "Synthetic Status"
    ].value_counts()
)
```

```python
# ==============================================================================
# G4 — EVALUATION REPORT AVAILABILITY AUDIT
# ==============================================================================

print()
print("=" * 100)
print("G4 — EVALUATION REPORT AVAILABILITY AUDIT")
print("=" * 100)

G4_ROWS = []

for dataset, method in G_EXPECTED_COMBINATIONS:

    if "FINAL_EVALUATION_REPORT_REGISTRY" not in globals():

        G4_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Evaluation Files": 0,
            "Evaluation Categories": 0,
            "Categories Present": "",
            "Evaluation Status": "REGISTRY_UNAVAILABLE"
        })

        continue

    registry = FINAL_EVALUATION_REPORT_REGISTRY.copy()

    subset = registry[
        registry["Status"] == "VALID"
    ]

    path_text = subset[
        "File"
    ].astype(str).str.lower()

    dataset_token = dataset.lower()

    if method == "DP-CTGAN":
        method_tokens = [
            "dp_ctgan",
            "dp-ctgan",
            "dpctgan"
        ]
    else:
        method_tokens = [
            method.lower(),
            method.lower().replace("-", "_")
        ]

    dataset_mask = path_text.str.contains(
        dataset_token,
        regex=False
    )

    method_mask = pd.Series(
        False,
        index=subset.index
    )

    for token in method_tokens:

        method_mask = (
            method_mask |
            path_text.str.contains(
                token,
                regex=False
            )
        )

    matched = subset[
        dataset_mask &
        method_mask
    ]

    categories = sorted(
        matched["Category"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    G4_ROWS.append({
        "Dataset": dataset,
        "Method": method,
        "Evaluation Files": int(
            matched["File"].nunique()
        ),
        "Evaluation Categories": len(
            categories
        ),
        "Categories Present": ", ".join(
            categories
        ),
        "Evaluation Status": (
            "AVAILABLE"
            if len(categories) > 0
            else "MISSING"
        )
    })

G4_EVALUATION_AUDIT = pd.DataFrame(
    G4_ROWS
)

print()
print("-" * 100)
print("EVALUATION AVAILABILITY")
print("-" * 100)

display(
    G4_EVALUATION_AUDIT
)
```

```python
# ==============================================================================
# G6 — FAILED / INCOMPLETE EXPERIMENT AUDIT
# ==============================================================================

print()
print("=" * 100)
print("G6 — FAILED / INCOMPLETE EXPERIMENT AUDIT")
print("=" * 100)

G6_ROWS = []

for _, row in G3_SYNTHETIC_AUDIT.iterrows():

    dataset = row["Dataset"]
    method = row["Method"]
    synthetic_status = row["Synthetic Status"]

    if synthetic_status in [
        "READY",
        "LOADED",
        "VALID"
    ]:

        experiment_status = "AVAILABLE"

    elif synthetic_status == "MISSING":

        experiment_status = (
            "INCOMPLETE — SYNTHETIC DATA UNAVAILABLE"
        )

    else:

        experiment_status = (
            "REQUIRES VERIFICATION"
        )

    G6_ROWS.append({
        "Dataset": dataset,
        "Method": method,
        "Synthetic Status": synthetic_status,
        "Experiment Status": experiment_status
    })

G6_EXPERIMENT_AUDIT = pd.DataFrame(
    G6_ROWS
)

print()
print("-" * 100)
print("EXPERIMENT STATUS")
print("-" * 100)

display(
    G6_EXPERIMENT_AUDIT
)

print()
print("-" * 100)
print("IMPORTANT")
print("-" * 100)

print(
    "Missing synthetic output is classified as INCOMPLETE, "
    "not automatically as FAILED."
)
```

```python
# ==============================================================================
# G7 — DATA INTEGRITY AUDIT
# ==============================================================================

print()
print("=" * 100)
print("G7 — DATA INTEGRITY AUDIT")
print("=" * 100)

G7_ROWS = []

for _, row in G3_SYNTHETIC_AUDIT.iterrows():

    filepath = row["Synthetic File"]

    if not filepath or not isinstance(
        filepath,
        str
    ):

        G7_ROWS.append({
            "Dataset": row["Dataset"],
            "Method": row["Method"],
            "File Exists": False,
            "Rows": None,
            "Columns": None,
            "Missing Cells": None,
            "Duplicate Rows": None,
            "Integrity Status": "UNAVAILABLE"
        })

        continue

    filepath = pathlib.Path(
        filepath
    )

    if not filepath.exists():

        G7_ROWS.append({
            "Dataset": row["Dataset"],
            "Method": row["Method"],
            "File Exists": False,
            "Rows": None,
            "Columns": None,
            "Missing Cells": None,
            "Duplicate Rows": None,
            "Integrity Status": "FILE_NOT_FOUND"
        })

        continue

    try:

        df = pd.read_csv(
            filepath,
            low_memory=False
        )

        rows = int(
            df.shape[0]
        )

        columns = int(
            df.shape[1]
        )

        missing_cells = int(
            df.isna().sum().sum()
        )

        duplicate_rows = int(
            df.duplicated().sum()
        )

        integrity_status = (
            "VALID"
            if rows > 0 and columns > 0
            else "INVALID"
        )

    except Exception:

        rows = None
        columns = None
        missing_cells = None
        duplicate_rows = None
        integrity_status = "READ_ERROR"

    G7_ROWS.append({
        "Dataset": row["Dataset"],
        "Method": row["Method"],
        "File Exists": True,
        "Rows": rows,
        "Columns": columns,
        "Missing Cells": missing_cells,
        "Duplicate Rows": duplicate_rows,
        "Integrity Status": integrity_status
    })

G7_DATA_INTEGRITY_AUDIT = pd.DataFrame(
    G7_ROWS
)

print()
print("-" * 100)
print("DATA INTEGRITY RESULTS")
print("-" * 100)

display(
    G7_DATA_INTEGRITY_AUDIT
)
```

```python
# ==============================================================================
# G8 — CROSS-SOURCE CONSISTENCY AUDIT
# ==============================================================================

print()
print("=" * 100)
print("G8 — CROSS-SOURCE CONSISTENCY AUDIT")
print("=" * 100)

G8_ROWS = []

for dataset, method in G_EXPECTED_COMBINATIONS:

    e7_match = G3_SYNTHETIC_AUDIT[
        (G3_SYNTHETIC_AUDIT["Dataset"] == dataset) &
        (G3_SYNTHETIC_AUDIT["Method"] == method)
    ]

    g7_match = G7_DATA_INTEGRITY_AUDIT[
        (G7_DATA_INTEGRITY_AUDIT["Dataset"] == dataset) &
        (G7_DATA_INTEGRITY_AUDIT["Method"] == method)
    ]

    if e7_match.empty or g7_match.empty:

        consistency = "UNVERIFIABLE"

    else:

        e7_status = str(
            e7_match.iloc[0]["Synthetic Status"]
        )

        g7_status = str(
            g7_match.iloc[0]["Integrity Status"]
        )

        if e7_status == "MISSING":

            consistency = (
                "CONSISTENTLY UNAVAILABLE"
                if g7_status == "UNAVAILABLE"
                else "CHECK REQUIRED"
            )

        elif g7_status == "VALID":

            consistency = "CONSISTENT"

        else:

            consistency = "INCONSISTENT"

    G8_ROWS.append({
        "Dataset": dataset,
        "Method": method,
        "Cross-Source Status": consistency
    })

G8_CONSISTENCY_AUDIT = pd.DataFrame(
    G8_ROWS
)

print()
print("-" * 100)
print("CROSS-SOURCE CONSISTENCY")
print("-" * 100)

display(
    G8_CONSISTENCY_AUDIT
)
```

```python
# ==============================================================================
# G9 — FINAL COMPLETENESS MATRIX
# ==============================================================================

print()
print("=" * 100)
print("G9 — FINAL COMPLETENESS MATRIX")
print("=" * 100)

G9_FINAL_ROWS = []

for dataset, method in G_EXPECTED_COMBINATIONS:

    synthetic = G3_SYNTHETIC_AUDIT[
        (G3_SYNTHETIC_AUDIT["Dataset"] == dataset) &
        (G3_SYNTHETIC_AUDIT["Method"] == method)
    ].iloc[0]

    evaluation = G4_EVALUATION_AUDIT[
        (G4_EVALUATION_AUDIT["Dataset"] == dataset) &
        (G4_EVALUATION_AUDIT["Method"] == method)
    ].iloc[0]

    integrity = G7_DATA_INTEGRITY_AUDIT[
        (G7_DATA_INTEGRITY_AUDIT["Dataset"] == dataset) &
        (G7_DATA_INTEGRITY_AUDIT["Method"] == method)
    ].iloc[0]

    consistency = G8_CONSISTENCY_AUDIT[
        (G8_CONSISTENCY_AUDIT["Dataset"] == dataset) &
        (G8_CONSISTENCY_AUDIT["Method"] == method)
    ].iloc[0]

    synthetic_ready = synthetic[
        "Synthetic Status"
    ] in [
        "READY",
        "LOADED",
        "VALID"
    ]

    evaluation_available = (
        evaluation["Evaluation Status"]
        == "AVAILABLE"
    )

    integrity_valid = (
        integrity["Integrity Status"]
        == "VALID"
    )

    cross_source_consistent = (
        consistency["Cross-Source Status"]
        in [
            "CONSISTENT",
            "CONSISTENTLY UNAVAILABLE"
        ]
    )

    if (
        synthetic_ready
        and evaluation_available
        and integrity_valid
    ):

        final_status = "RESEARCH-READY"

    elif not synthetic_ready:

        final_status = (
            "INCOMPLETE — SYNTHETIC DATA"
        )

    elif not evaluation_available:

        final_status = (
            "INCOMPLETE — EVALUATION"
        )

    elif not integrity_valid:

        final_status = (
            "INTEGRITY FAILURE"
        )

    else:

        final_status = "REQUIRES REVIEW"

    G9_FINAL_ROWS.append({
        "Dataset": dataset,
        "Method": method,
        "Synthetic Data": (
            "AVAILABLE"
            if synthetic_ready
            else "MISSING"
        ),
        "Evaluation Evidence": (
            "AVAILABLE"
            if evaluation_available
            else "MISSING"
        ),
        "Data Integrity": (
            "VALID"
            if integrity_valid
            else "INVALID / UNAVAILABLE"
        ),
        "Cross-Source Check": (
            "CONSISTENT"
            if cross_source_consistent
            else "CHECK REQUIRED"
        ),
        "Final Audit Status": final_status
    })

G_FINAL_COMPLETENESS_MATRIX = pd.DataFrame(
    G9_FINAL_ROWS
)

print()
print("-" * 100)
print("FINAL DATASET × METHOD COMPLETENESS MATRIX")
print("-" * 100)

display(
    G_FINAL_COMPLETENESS_MATRIX
)

print()
print("-" * 100)
print("FINAL AUDIT STATUS COUNTS")
print("-" * 100)

print(
    G_FINAL_COMPLETENESS_MATRIX[
        "Final Audit Status"
    ].value_counts()
)
```

```python
# ==============================================================================
# H6 — RESULT STATUS LOGIC
# ==============================================================================

print()
print("=" * 100)
print("H6 — RESULT STATUS LOGIC")
print("=" * 100)

H_VALID_RESULT_STATUSES = [
    "AVAILABLE",
    "UNAVAILABLE",
    "FAILED",
    "INVALID",
    "NOT_APPLICABLE"
]

def classify_result_status(
    value,
    source_exists=True,
    failure_reason=None
):

    if failure_reason:

        return "FAILED"

    if not source_exists:

        return "UNAVAILABLE"

    if value is None:

        return "UNAVAILABLE"

    try:

        numeric_value = float(value)

        if not np.isfinite(
            numeric_value
        ):

            return "INVALID"

    except (
        TypeError,
        ValueError
    ):

        return "INVALID"

    return "AVAILABLE"


print()
print("Valid result statuses:")

for status in H_VALID_RESULT_STATUSES:

    print(f"  - {status}")
```

```python
# ==============================================================================
# H7 — POPULATE AVAILABLE RESULTS
# ==============================================================================

print()
print("=" * 100)
print("H7 — POPULATE AVAILABLE RESULTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# SAFE POPULATION FROM EXISTING NORMALIZED RESULT TABLES
# ------------------------------------------------------------------------------

H_RESULT_TABLE_CANDIDATES = []

for variable_name in [
    "FINAL_METRIC_RESULTS",
    "FINAL_EVALUATION_RESULTS",
    "EVALUATION_METRICS",
    "METRIC_RESULTS",
    "MASTER_METRIC_RESULTS"
]:

    if variable_name in globals():

        candidate = globals()[variable_name]

        if isinstance(
            candidate,
            pd.DataFrame
        ):

            H_RESULT_TABLE_CANDIDATES.append(
                (
                    variable_name,
                    candidate.copy()
                )
            )

print()
print(
    f"Existing normalized result tables found : "
    f"{len(H_RESULT_TABLE_CANDIDATES)}"
)

if H_RESULT_TABLE_CANDIDATES:

    print()
    print("Available result tables:")

    for name, table in H_RESULT_TABLE_CANDIDATES:

        print(
            f"  {name} : "
            f"{table.shape[0]} rows × "
            f"{table.shape[1]} columns"
        )

else:

    print()
    print(
        "No normalized metric result table is currently "
        "available. H matrix remains explicitly UNAVAILABLE "
        "until results are recovered from evaluation reports."
    )

H7_STATUS = (
    "RESULTS_AVAILABLE_FOR_MAPPING"
    if H_RESULT_TABLE_CANDIDATES
    else "NO_NORMALIZED_RESULTS_AVAILABLE"
)

print()
print(f"H7 STATUS : {H7_STATUS}")
```

```python
# ==============================================================================
# I10 — FINAL METRIC DICTIONARY AND VALIDATION
# ==============================================================================

print()
print("=" * 100)
print("I10 — FINAL METRIC DICTIONARY AND VALIDATION")
print("=" * 100)

I_ALL_DEFINITIONS = {}

I_ALL_DEFINITIONS.update(
    I_FIDELITY_DEFINITIONS
)

I_ALL_DEFINITIONS.update(
    I_UTILITY_DEFINITIONS
)

I_ALL_DEFINITIONS.update(
    I_PRIVACY_DEFINITIONS
)

I_ALL_DEFINITIONS.update(
    I_STRUCTURAL_STATISTICAL_DEFINITIONS
)

I_ALL_DEFINITIONS.update(
    I_DOWNSTREAM_DEFINITIONS
)

I_METRIC_DICTIONARY_ROWS = []

for metric in I_EXPECTED_METRICS:

    definition = I_ALL_DEFINITIONS.get(
        metric
    )

    I_METRIC_DICTIONARY_ROWS.append({

        "Metric": metric,

        "Category":
            next(
                (
                    category
                    for category, metrics
                    in I_METRIC_FRAMEWORK.items()
                    if metric in metrics
                ),
                "UNDEFINED"
            ),

        "Meaning":
            definition["Meaning"]
            if definition else None,

        "Ideal Direction":
            definition["Ideal Direction"]
            if definition else "UNDEFINED",

        "Ideal Value":
            definition["Ideal Value"]
            if definition else None,

        "Normalization":
            definition["Normalization"]
            if definition else "UNDEFINED",

        "Interpretation":
            definition["Interpretation"]
            if definition else None
    })

I_FINAL_METRIC_DICTIONARY = pd.DataFrame(
    I_METRIC_DICTIONARY_ROWS
)

display(
    I_FINAL_METRIC_DICTIONARY
)

# ------------------------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------------------------

I_UNDEFINED_METRICS = (
    I_FINAL_METRIC_DICTIONARY[
        I_FINAL_METRIC_DICTIONARY[
            "Ideal Direction"
        ] == "UNDEFINED"
    ]["Metric"]
    .tolist()
)

I_DUPLICATE_METRICS = (
    I_FINAL_METRIC_DICTIONARY[
        "Metric"
    ]
    .duplicated()
    .sum()
)

print()
print("-" * 100)
print("METRIC DICTIONARY VALIDATION")
print("-" * 100)

print(
    f"Expected metrics       : "
    f"{len(I_EXPECTED_METRICS)}"
)

print(
    f"Defined metrics        : "
    f"{len(I_FINAL_METRIC_DICTIONARY)}"
)

print(
    f"Undefined metrics      : "
    f"{len(I_UNDEFINED_METRICS)}"
)

print(
    f"Duplicate metric names : "
    f"{I_DUPLICATE_METRICS}"
)

if I_UNDEFINED_METRICS:

    print()
    print("Undefined metrics:")
    for metric in I_UNDEFINED_METRICS:
        print(f"  - {metric}")

if (
    not I_UNDEFINED_METRICS
    and I_DUPLICATE_METRICS == 0
):

    I_METRIC_DICTIONARY_STATUS = "READY"

else:

    I_METRIC_DICTIONARY_STATUS = "REQUIRES_REVIEW"

print()
print(
    f"I10 STATUS : "
    f"{I_METRIC_DICTIONARY_STATUS}"
)

print()
print("=" * 100)
print("I — METRIC DEFINITION & DIRECTIONALITY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# J1 — VALIDATE UPSTREAM RESEARCH OBJECTS
# ==============================================================================

print("=" * 100)
print("J1 — VALIDATE UPSTREAM RESEARCH OBJECTS")
print("=" * 100)

J_REQUIRED_OBJECTS = {
    "H_MASTER_RESULTS_TABLE": "Master Dataset × Method × Metric matrix",
    "I_FINAL_METRIC_DICTIONARY": "Canonical metric definitions"
}

J_OBJECT_STATUS = []

for object_name, description in J_REQUIRED_OBJECTS.items():

    exists = (
        object_name in globals()
        and isinstance(
            globals()[object_name],
            pd.DataFrame
        )
    )

    J_OBJECT_STATUS.append({
        "Object": object_name,
        "Description": description,
        "Available": exists
    })

J_UPSTREAM_STATUS = pd.DataFrame(
    J_OBJECT_STATUS
)

display(J_UPSTREAM_STATUS)

J_MISSING_OBJECTS = (
    J_UPSTREAM_STATUS.loc[
        ~J_UPSTREAM_STATUS["Available"],
        "Object"
    ]
    .tolist()
)

if J_MISSING_OBJECTS:

    raise RuntimeError(
        "Required upstream objects are missing: "
        f"{J_MISSING_OBJECTS}"
    )

print()
print("J1 STATUS : READY")
```

```python
# ==============================================================================
# J2 — COLLECT VALIDATED EVALUATION METRICS
# ==============================================================================

print()
print("=" * 100)
print("J2 — COLLECT VALIDATED EVALUATION METRICS")
print("=" * 100)

J_SOURCE_TABLES = []

# --------------------------------------------------------------------------
# PRIMARY SOURCE
# --------------------------------------------------------------------------

J_SOURCE_TABLES.append(
    (
        "H_MASTER_RESULTS_TABLE",
        H_MASTER_RESULTS_TABLE.copy()
    )
)

# --------------------------------------------------------------------------
# OPTIONAL VALIDATED EVALUATION TABLE FROM SECTION F
# --------------------------------------------------------------------------

for variable_name in [
    "FINAL_EVALUATION_RESULTS",
    "FINAL_METRIC_RESULTS",
    "VALIDATED_EVALUATION_RESULTS"
]:

    if (
        variable_name in globals()
        and isinstance(
            globals()[variable_name],
            pd.DataFrame
        )
    ):

        J_SOURCE_TABLES.append(
            (
                variable_name,
                globals()[variable_name].copy()
            )
        )

print()
print("-" * 100)
print("VALIDATED RESULT SOURCES")
print("-" * 100)

for source_name, source_table in J_SOURCE_TABLES:

    print(
        f"{source_name:<35} "
        f"{source_table.shape[0]:>8} rows × "
        f"{source_table.shape[1]:>3} columns"
    )

J_PRIMARY_RESULTS = (
    H_MASTER_RESULTS_TABLE.copy()
)

print()
print(
    f"Primary consolidation source rows : "
    f"{len(J_PRIMARY_RESULTS)}"
)

print()
print("J2 STATUS : READY")
```

```python
# ==============================================================================
# J3 — NORMALIZE METRIC NAMES
# ==============================================================================

print()
print("=" * 100)
print("J3 — NORMALIZE METRIC NAMES")
print("=" * 100)

J_METRIC_NAME_MAP = {

    "KS": "KS Statistic",
    "KS Test": "KS Statistic",
    "KS Statistic Mean": "KS Statistic",

    "Wasserstein": "Wasserstein Distance",
    "Wasserstein Mean": "Wasserstein Distance",

    "JS": "JS Divergence",
    "Jensen-Shannon Divergence": "JS Divergence",

    "KL": "KL Divergence",
    "Kullback-Leibler Divergence": "KL Divergence",

    "Correlation Preservation": "Correlation Similarity",
    "Correlation Score": "Correlation Similarity",

    "Train on Synthetic Test on Real":
        "TSTR",

    "Train on Real Test on Synthetic":
        "TRTS",

    "ε": "Epsilon",
    "Privacy Budget": "Epsilon",
    "Target Epsilon": "Epsilon",
    "Achieved Epsilon": "Epsilon",

    "δ": "Delta",
    "Target Delta": "Delta",

    "Row Count": "Rows",
    "Synthetic Rows": "Rows",

    "Column Count": "Columns",

    "Missing Values": "Missing Cells",
    "Missing Count": "Missing Cells",

    "Duplicates": "Duplicate Rows",
    "Duplicate Count": "Duplicate Rows",

    "Mean Absolute Error": "Mean Error",
    "Mean Difference": "Mean Error",

    "Std Error": "Standard Deviation Error",
    "Standard Deviation Difference":
        "Standard Deviation Error",

    "Test Accuracy": "Accuracy",
    "F1": "F1 Score",
    "Test F1": "F1 Score",
    "AUC": "ROC AUC",
    "ROC_AUC": "ROC AUC"
}

J_PRIMARY_RESULTS["Metric"] = (
    J_PRIMARY_RESULTS["Metric"]
    .replace(J_METRIC_NAME_MAP)
)

print()
print(
    f"Metric names normalized : "
    f"{J_PRIMARY_RESULTS['Metric'].nunique()}"
)

print()
print("J3 STATUS : READY")
```

```python
# ==============================================================================
# J4 — MERGE WITH FINAL DATASET × METHOD SCOPE
# ==============================================================================

print()
print("=" * 100)
print("J4 — MERGE WITH FINAL DATASET × METHOD SCOPE")
print("=" * 100)

J_SCOPE_ROWS = []

for dataset in H_DATASETS:

    for method in H_METHODS:

        J_SCOPE_ROWS.append({
            "Dataset": dataset,
            "Method": method
        })

J_EXPERIMENT_SCOPE = pd.DataFrame(
    J_SCOPE_ROWS
)

J_METRIC_SCOPE = (
    I_FINAL_METRIC_DICTIONARY[
        [
            "Metric",
            "Category",
            "Ideal Direction",
            "Ideal Value",
            "Normalization",
            "Meaning",
            "Interpretation"
        ]
    ]
    .drop_duplicates(
        subset=["Metric"]
    )
)

J_FULL_SCOPE = (
    J_EXPERIMENT_SCOPE
    .merge(
        J_METRIC_SCOPE,
        how="cross"
    )
)

print()
print(
    f"Expected Dataset × Method cells : "
    f"{len(J_EXPERIMENT_SCOPE)}"
)

print(
    f"Expected final metric cells : "
    f"{len(J_FULL_SCOPE)}"
)

display(
    J_FULL_SCOPE.head(20)
)

print()
print("J4 STATUS : READY")
```

```python
# ==============================================================================
# J5 — MERGE RESULTS WITHOUT LOSING MISSING/FAILED STATUS
# ==============================================================================

print()
print("=" * 100)
print("J5 — MERGE RESULTS WITHOUT LOSING MISSING/FAILED STATUS")
print("=" * 100)

J_RESULT_COLUMNS = [
    "Dataset",
    "Method",
    "Metric",
    "Value",
    "Result Status",
    "Source File",
    "Source Column",
    "Failure Reason"
]

J_AVAILABLE_RESULT_COLUMNS = [
    column
    for column in J_RESULT_COLUMNS
    if column in J_PRIMARY_RESULTS.columns
]

J_RESULTS_FOR_MERGE = (
    J_PRIMARY_RESULTS[
        J_AVAILABLE_RESULT_COLUMNS
    ]
    .copy()
)

# Ensure required fields exist
for column in J_RESULT_COLUMNS:

    if column not in J_RESULTS_FOR_MERGE.columns:

        if column == "Result Status":

            J_RESULTS_FOR_MERGE[column] = "UNAVAILABLE"

        elif column == "Value":

            J_RESULTS_FOR_MERGE[column] = np.nan

        else:

            J_RESULTS_FOR_MERGE[column] = None

# Keep only canonical research scope
J_RESULTS_FOR_MERGE = (
    J_RESULTS_FOR_MERGE[
        J_RESULTS_FOR_MERGE["Dataset"]
        .isin(H_DATASETS)
        &
        J_RESULTS_FOR_MERGE["Method"]
        .isin(H_METHODS)
        &
        J_RESULTS_FOR_MERGE["Metric"]
        .isin(
            I_EXPECTED_METRICS
        )
    ]
    .copy()
)

# Remove accidental duplicate metric records.
# Prefer AVAILABLE over unavailable records.
J_STATUS_PRIORITY = {
    "AVAILABLE": 0,
    "FAILED": 1,
    "INVALID": 2,
    "NOT_APPLICABLE": 3,
    "UNAVAILABLE": 4
}

J_RESULTS_FOR_MERGE["_STATUS_PRIORITY"] = (
    J_RESULTS_FOR_MERGE["Result Status"]
    .map(J_STATUS_PRIORITY)
    .fillna(99)
)

J_RESULTS_FOR_MERGE = (
    J_RESULTS_FOR_MERGE
    .sort_values(
        [
            "Dataset",
            "Method",
            "Metric",
            "_STATUS_PRIORITY"
        ]
    )
    .drop_duplicates(
        subset=[
            "Dataset",
            "Method",
            "Metric"
        ],
        keep="first"
    )
    .drop(
        columns="_STATUS_PRIORITY"
    )
)

J_CONSOLIDATED_RESULTS = (
    J_FULL_SCOPE
    .merge(
        J_RESULTS_FOR_MERGE[
            [
                "Dataset",
                "Method",
                "Metric",
                "Value",
                "Result Status",
                "Source File",
                "Source Column",
                "Failure Reason"
            ]
        ],
        on=[
            "Dataset",
            "Method",
            "Metric"
        ],
        how="left"
    )
)

# Anything not present in the validated evaluation results
# remains explicitly UNAVAILABLE.
J_CONSOLIDATED_RESULTS["Result Status"] = (
    J_CONSOLIDATED_RESULTS["Result Status"]
    .fillna("UNAVAILABLE")
)

print()
print(
    f"Consolidated result rows : "
    f"{len(J_CONSOLIDATED_RESULTS)}"
)

print()
print("J5 STATUS : READY")
```

```python
# ==============================================================================
# J6 — ATTACH METRIC DEFINITIONS AND DIRECTIONALITY
# ==============================================================================

print()
print("=" * 100)
print("J6 — ATTACH METRIC DEFINITIONS AND DIRECTIONALITY")
print("=" * 100)

J_CONSOLIDATED_RESULTS["Directionality Valid"] = (
    J_CONSOLIDATED_RESULTS["Ideal Direction"]
    .isin([
        "LOWER",
        "HIGHER",
        "REFERENCE_MATCH"
    ])
)

J_CONSOLIDATED_RESULTS["Metric Category"] = (
    J_CONSOLIDATED_RESULTS["Category"]
)

J_CONSOLIDATED_RESULTS["Normalization Rule"] = (
    J_CONSOLIDATED_RESULTS["Normalization"]
)

print()
print("-" * 100)
print("CANONICAL METADATA ATTACHED")
print("-" * 100)

display(
    J_CONSOLIDATED_RESULTS[
        [
            "Dataset",
            "Method",
            "Category",
            "Metric",
            "Value",
            "Result Status",
            "Ideal Direction",
            "Normalization"
        ]
    ].head(20)
)

print()
print("J6 STATUS : READY")
```

```python
# ==============================================================================
# J7 — VALIDATE FINAL CONSOLIDATED DATAFRAME
# ==============================================================================

print()
print("=" * 100)
print("J7 — VALIDATE FINAL CONSOLIDATED DATAFRAME")
print("=" * 100)

J_EXPECTED_ROWS = (
    len(H_DATASETS)
    * len(H_METHODS)
    * len(I_EXPECTED_METRICS)
)

J_ACTUAL_ROWS = (
    len(J_CONSOLIDATED_RESULTS)
)

J_DUPLICATE_KEYS = (
    J_CONSOLIDATED_RESULTS
    .duplicated(
        subset=[
            "Dataset",
            "Method",
            "Metric"
        ]
    )
    .sum()
)

J_INVALID_DATASETS = (
    ~J_CONSOLIDATED_RESULTS["Dataset"]
    .isin(H_DATASETS)
).sum()

J_INVALID_METHODS = (
    ~J_CONSOLIDATED_RESULTS["Method"]
    .isin(H_METHODS)
).sum()

J_INVALID_METRICS = (
    ~J_CONSOLIDATED_RESULTS["Metric"]
    .isin(I_EXPECTED_METRICS)
).sum()

J_INVALID_STATUSES = (
    ~J_CONSOLIDATED_RESULTS["Result Status"]
    .isin(H_VALID_RESULT_STATUSES)
).sum()

print()
print("-" * 100)
print("CONSOLIDATION VALIDATION")
print("-" * 100)

print(
    f"Expected rows          : {J_EXPECTED_ROWS}"
)

print(
    f"Actual rows            : {J_ACTUAL_ROWS}"
)

print(
    f"Duplicate experiment keys : "
    f"{J_DUPLICATE_KEYS}"
)

print(
    f"Invalid datasets       : {J_INVALID_DATASETS}"
)

print(
    f"Invalid methods        : {J_INVALID_METHODS}"
)

print(
    f"Invalid metrics        : {J_INVALID_METRICS}"
)

print(
    f"Invalid statuses       : {J_INVALID_STATUSES}"
)

J_VALIDATION_ERRORS = (
    J_EXPECTED_ROWS != J_ACTUAL_ROWS
    or J_DUPLICATE_KEYS > 0
    or J_INVALID_DATASETS > 0
    or J_INVALID_METHODS > 0
    or J_INVALID_METRICS > 0
    or J_INVALID_STATUSES > 0
)

if J_VALIDATION_ERRORS:

    J7_STATUS = "REQUIRES_REVIEW"

else:

    J7_STATUS = "VALID"

print()
print(f"J7 STATUS : {J7_STATUS}")
```

```python
# ==============================================================================
# K1 — VALIDATE SECTION J INPUT
# ==============================================================================

print("=" * 100)
print("K1 — VALIDATE SECTION J INPUT")
print("=" * 100)

if (
    "J_FINAL_RESEARCH_RESULTS" not in globals()
    or not isinstance(J_FINAL_RESEARCH_RESULTS, pd.DataFrame)
):
    raise RuntimeError(
        "J_FINAL_RESEARCH_RESULTS is not available. "
        "Complete Section J before running Section K."
    )

K_RESULTS = J_FINAL_RESEARCH_RESULTS.copy()

K_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Metric",
    "Value",
    "Result Status",
    "Ideal Direction"
]

K_MISSING_COLUMNS = [
    column
    for column in K_REQUIRED_COLUMNS
    if column not in K_RESULTS.columns
]

if K_MISSING_COLUMNS:
    raise KeyError(
        "Section J results are missing required columns: "
        f"{K_MISSING_COLUMNS}"
    )

print()
print(f"Input rows    : {len(K_RESULTS):,}")
print(f"Input columns : {len(K_RESULTS.columns)}")

print()
print("K1 STATUS : READY")
```

```python
# ==============================================================================
# K2 — MISSING-VALUE QUALITY CONTROL
# ==============================================================================

print()
print("=" * 100)
print("K2 — MISSING-VALUE QUALITY CONTROL")
print("=" * 100)

K_RESULTS["QC_Missing_Value"] = (
    K_RESULTS["Value"].isna()
)

K_RESULTS["QC_Missing_Value_Severity"] = "NONE"

K_RESULTS.loc[
    K_RESULTS["QC_Missing_Value"]
    & K_RESULTS["Result Status"].eq("AVAILABLE"),
    "QC_Missing_Value_Severity"
] = "CRITICAL"

K_RESULTS.loc[
    K_RESULTS["QC_Missing_Value"]
    & K_RESULTS["Result Status"].eq("FAILED"),
    "QC_Missing_Value_Severity"
] = "EXPECTED_FAILED"

K_RESULTS.loc[
    K_RESULTS["QC_Missing_Value"]
    & K_RESULTS["Result Status"].eq("UNAVAILABLE"),
    "QC_Missing_Value_Severity"
] = "EXPECTED_UNAVAILABLE"

K_AVAILABLE_MISSING = int(
    (
        K_RESULTS["QC_Missing_Value"]
        & K_RESULTS["Result Status"].eq("AVAILABLE")
    ).sum()
)

K_EXPECTED_UNAVAILABLE = int(
    (
        K_RESULTS["QC_Missing_Value"]
        & K_RESULTS["Result Status"].eq("UNAVAILABLE")
    ).sum()
)

K_EXPECTED_FAILED = int(
    (
        K_RESULTS["QC_Missing_Value"]
        & K_RESULTS["Result Status"].eq("FAILED")
    ).sum()
)

print()
print("-" * 100)
print("MISSING-VALUE SUMMARY")
print("-" * 100)

print(
    f"Available results with missing values : "
    f"{K_AVAILABLE_MISSING}"
)

print(
    f"Expected unavailable values            : "
    f"{K_EXPECTED_UNAVAILABLE}"
)

print(
    f"Expected failed values                 : "
    f"{K_EXPECTED_FAILED}"
)

if K_AVAILABLE_MISSING > 0:
    print()
    print(
        "WARNING: AVAILABLE results contain missing metric values."
    )

print()
print("K2 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K3 — INVALID-VALUE QUALITY CONTROL
# ==============================================================================

print()
print("=" * 100)
print("K3 — INVALID-VALUE QUALITY CONTROL")
print("=" * 100)

K_RESULTS["QC_Numeric_Value"] = pd.to_numeric(
    K_RESULTS["Value"],
    errors="coerce"
)

K_RESULTS["QC_NonNumeric_Value"] = (
    K_RESULTS["Value"].notna()
    &
    K_RESULTS["QC_Numeric_Value"].isna()
)

K_RESULTS["QC_Infinite_Value"] = False

K_NUMERIC_MASK = (
    K_RESULTS["QC_Numeric_Value"].notna()
)

K_RESULTS.loc[
    K_NUMERIC_MASK,
    "QC_Infinite_Value"
] = np.isinf(
    K_RESULTS.loc[
        K_NUMERIC_MASK,
        "QC_Numeric_Value"
    ]
)

K_RESULTS["QC_Invalid_Value"] = (
    K_RESULTS["QC_NonNumeric_Value"]
    |
    K_RESULTS["QC_Infinite_Value"]
)

# --------------------------------------------------------------------------
# METRIC-SPECIFIC DOMAIN RULES
# --------------------------------------------------------------------------

K_RESULTS["QC_Domain_Violation"] = False

# Metrics that must be non-negative
K_NON_NEGATIVE_METRICS = {
    "Wasserstein Distance",
    "JS Divergence",
    "KL Divergence",
    "Missing Cells",
    "Duplicate Rows",
    "Rows",
    "Columns",
    "Mean Error",
    "Standard Deviation Error"
}

K_MASK = (
    K_RESULTS["Metric"].isin(K_NON_NEGATIVE_METRICS)
    &
    K_RESULTS["QC_Numeric_Value"].notna()
    &
    (
        K_RESULTS["QC_Numeric_Value"] < 0
    )
)

K_RESULTS.loc[
    K_MASK,
    "QC_Domain_Violation"
] = True

# --------------------------------------------------------------------------
# PROBABILITY / SCORE METRICS
# --------------------------------------------------------------------------

K_BOUNDED_ZERO_ONE_METRICS = {
    "Accuracy",
    "F1 Score",
    "ROC AUC",
    "Correlation Similarity"
}

K_MASK = (
    K_RESULTS["Metric"]
    .isin(K_BOUNDED_ZERO_ONE_METRICS)
    &
    K_RESULTS["QC_Numeric_Value"].notna()
    &
    (
        (K_RESULTS["QC_Numeric_Value"] < 0)
        |
        (K_RESULTS["QC_Numeric_Value"] > 1)
    )
)

K_RESULTS.loc[
    K_MASK,
    "QC_Domain_Violation"
] = True

# --------------------------------------------------------------------------
# EPSILON / DELTA
# --------------------------------------------------------------------------

K_MASK = (
    K_RESULTS["Metric"].isin(
        ["Epsilon", "Delta"]
    )
    &
    K_RESULTS["QC_Numeric_Value"].notna()
    &
    (
        K_RESULTS["QC_Numeric_Value"] < 0
    )
)

K_RESULTS.loc[
    K_MASK,
    "QC_Domain_Violation"
] = True

K_INVALID_COUNT = int(
    K_RESULTS[
        [
            "QC_Invalid_Value",
            "QC_Domain_Violation"
        ]
    ]
    .any(axis=1)
    .sum()
)

print()
print("-" * 100)
print("INVALID-VALUE SUMMARY")
print("-" * 100)

print(
    f"Non-numeric values : "
    f"{int(K_RESULTS['QC_NonNumeric_Value'].sum())}"
)

print(
    f"Infinite values    : "
    f"{int(K_RESULTS['QC_Infinite_Value'].sum())}"
)

print(
    f"Domain violations  : "
    f"{int(K_RESULTS['QC_Domain_Violation'].sum())}"
)

print(
    f"Total invalid rows : "
    f"{K_INVALID_COUNT}"
)

print()
print("K3 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K4 — DUPLICATE-RESULT QUALITY CONTROL
# ==============================================================================

print()
print("=" * 100)
print("K4 — DUPLICATE-RESULT QUALITY CONTROL")
print("=" * 100)

K_DUPLICATE_KEY_COLUMNS = [
    "Dataset",
    "Method",
    "Metric"
]

K_RESULTS["QC_Duplicate_Result"] = (
    K_RESULTS
    .duplicated(
        subset=K_DUPLICATE_KEY_COLUMNS,
        keep=False
    )
)

K_DUPLICATE_ROWS = int(
    K_RESULTS["QC_Duplicate_Result"].sum()
)

K_DUPLICATE_GROUPS = int(
    K_RESULTS.loc[
        K_RESULTS["QC_Duplicate_Result"],
        K_DUPLICATE_KEY_COLUMNS
    ]
    .drop_duplicates()
    .shape[0]
)

print()
print("-" * 100)
print("DUPLICATE-RESULT SUMMARY")
print("-" * 100)

print(
    f"Duplicate result rows  : "
    f"{K_DUPLICATE_ROWS}"
)

print(
    f"Duplicate result groups : "
    f"{K_DUPLICATE_GROUPS}"
)

if K_DUPLICATE_ROWS > 0:

    print()
    print("WARNING: Duplicate Dataset × Method × Metric records detected.")

    display(
        K_RESULTS.loc[
            K_RESULTS["QC_Duplicate_Result"],
            K_DUPLICATE_KEY_COLUMNS + [
                "Value",
                "Result Status"
            ]
        ]
    )

print()
print("K4 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K5 — METRIC-SCALE CONSISTENCY
# ==============================================================================

print()
print("=" * 100)
print("K5 — METRIC-SCALE CONSISTENCY")
print("=" * 100)

K_RESULTS["QC_Scale_Status"] = "NOT_ASSESSED"

K_SCALE_SUMMARY = []

for metric, group in K_RESULTS.groupby(
    "Metric",
    dropna=False
):

    values = pd.to_numeric(
        group["Value"],
        errors="coerce"
    ).dropna()

    if len(values) == 0:

        status = "NO_VALID_VALUES"
        minimum = np.nan
        maximum = np.nan
        median = np.nan

    else:

        minimum = float(values.min())
        maximum = float(values.max())
        median = float(values.median())

        # A robust scale diagnostic.
        # This is a flag, not an automatic rejection.
        if median != 0:

            scale_ratio = (
                abs(maximum)
                /
                max(abs(median), np.finfo(float).eps)
            )

        else:

            scale_ratio = np.inf

        if scale_ratio > 1e6:

            status = "POTENTIAL_SCALE_ANOMALY"

        else:

            status = "CONSISTENT"

    K_SCALE_SUMMARY.append({
        "Metric": metric,
        "Valid Values": len(values),
        "Minimum": minimum,
        "Median": median,
        "Maximum": maximum,
        "Scale Status": status
    })

K_SCALE_SUMMARY = pd.DataFrame(
    K_SCALE_SUMMARY
)

display(K_SCALE_SUMMARY)

K_SCALE_ANOMALIES = int(
    (
        K_SCALE_SUMMARY["Scale Status"]
        == "POTENTIAL_SCALE_ANOMALY"
    ).sum()
)

print()
print(
    f"Potential scale anomalies : "
    f"{K_SCALE_ANOMALIES}"
)

print()
print("K5 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K6 — ANOMALOUS-OBSERVATION DETECTION
# ==============================================================================

print()
print("=" * 100)
print("K6 — ANOMALOUS-OBSERVATION DETECTION")
print("=" * 100)

K_RESULTS["QC_Anomaly"] = False
K_RESULTS["QC_Anomaly_Score"] = np.nan

K_ANOMALY_ROWS = []

for metric, indices in K_RESULTS.groupby(
    "Metric"
).groups.items():

    metric_values = pd.to_numeric(
        K_RESULTS.loc[
            indices,
            "Value"
        ],
        errors="coerce"
    )

    valid_mask = metric_values.notna()

    if valid_mask.sum() < 4:
        continue

    valid_values = metric_values[
        valid_mask
    ]

    median = float(
        valid_values.median()
    )

    mad = float(
        np.median(
            np.abs(
                valid_values - median
            )
        )
    )

    if mad == 0:
        continue

    robust_z = (
        0.6745
        *
        (
            metric_values - median
        )
        /
        mad
    )

    for idx, score in robust_z.items():

        if pd.notna(score):

            K_RESULTS.loc[
                idx,
                "QC_Anomaly_Score"
            ] = float(score)

            if abs(score) > 3.5:

                K_RESULTS.loc[
                    idx,
                    "QC_Anomaly"
                ] = True

K_ANOMALY_COUNT = int(
    K_RESULTS["QC_Anomaly"].sum()
)

print()
print("-" * 100)
print("ANOMALY SUMMARY")
print("-" * 100)

print(
    f"Robust anomalies detected : "
    f"{K_ANOMALY_COUNT}"
)

if K_ANOMALY_COUNT > 0:

    display(
        K_RESULTS.loc[
            K_RESULTS["QC_Anomaly"],
            [
                "Dataset",
                "Method",
                "Metric",
                "Value",
                "Result Status",
                "QC_Anomaly_Score"
            ]
        ]
        .sort_values(
            "QC_Anomaly_Score",
            key=lambda x: abs(x),
            ascending=False
        )
    )

print()
print(
    "NOTE: Anomalies are flagged for review and are not removed."
)

print()
print("K6 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K7 — DATASET × METHOD QC SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("K7 — DATASET × METHOD QC SUMMARY")
print("=" * 100)

K_METHOD_SUMMARY = []

for (dataset, method), group in K_RESULTS.groupby(
    [
        "Dataset",
        "Method"
    ]
):

    K_METHOD_SUMMARY.append({

        "Dataset": dataset,

        "Method": method,

        "Total Metrics":
            len(group),

        "Available Metrics":
            int(
                group["Result Status"]
                .eq("AVAILABLE")
                .sum()
            ),

        "Unavailable Metrics":
            int(
                group["Result Status"]
                .eq("UNAVAILABLE")
                .sum()
            ),

        "Failed Metrics":
            int(
                group["Result Status"]
                .eq("FAILED")
                .sum()
            ),

        "Missing Available Values":
            int(
                (
                    group["QC_Missing_Value"]
                    &
                    group["Result Status"].eq(
                        "AVAILABLE"
                    )
                ).sum()
            ),

        "Invalid Values":
            int(
                (
                    group[
                        [
                            "QC_Invalid_Value",
                            "QC_Domain_Violation"
                        ]
                    ]
                    .any(axis=1)
                ).sum()
            ),

        "Duplicate Results":
            int(
                group["QC_Duplicate_Result"].sum()
            ),

        "Anomalies":
            int(
                group["QC_Anomaly"].sum()
            )
    })

K_METHOD_SUMMARY = pd.DataFrame(
    K_METHOD_SUMMARY
)

display(K_METHOD_SUMMARY)

print()
print("K7 STATUS : COMPLETED")
```

```python
# ==============================================================================
# K8 — FINAL METRIC QC STATUS
# ==============================================================================

print()
print("=" * 100)
print("K8 — FINAL METRIC QC STATUS")
print("=" * 100)

K_RESULTS["QC_Status"] = "PASS"

# --------------------------------------------------------------------------
# AVAILABLE + MISSING VALUE
# --------------------------------------------------------------------------

K_RESULTS.loc[
    K_RESULTS["QC_Missing_Value"]
    &
    K_RESULTS["Result Status"].eq("AVAILABLE"),
    "QC_Status"
] = "FAIL_MISSING"

# --------------------------------------------------------------------------
# INVALID VALUE
# --------------------------------------------------------------------------

K_RESULTS.loc[
    K_RESULTS["QC_Invalid_Value"],
    "QC_Status"
] = "FAIL_INVALID"

# --------------------------------------------------------------------------
# DOMAIN VIOLATION
# --------------------------------------------------------------------------

K_RESULTS.loc[
    K_RESULTS["QC_Domain_Violation"],
    "QC_Status"
] = "FAIL_DOMAIN"

# --------------------------------------------------------------------------
# DUPLICATE
# --------------------------------------------------------------------------

K_RESULTS.loc[
    K_RESULTS["QC_Duplicate_Result"],
    "QC_Status"
] = "FAIL_DUPLICATE"

# --------------------------------------------------------------------------
# ANOMALY
# --------------------------------------------------------------------------

K_RESULTS.loc[
    (
        K_RESULTS["QC_Anomaly"]
        &
        K_RESULTS["QC_Status"].eq("PASS")
    ),
    "QC_Status"
] = "REVIEW_ANOMALY"

# --------------------------------------------------------------------------
# EXPECTED UNAVAILABLE / FAILED
# --------------------------------------------------------------------------

K_RESULTS.loc[
    K_RESULTS["Result Status"].eq("UNAVAILABLE")
    &
    K_RESULTS["QC_Status"].eq("PASS"),
    "QC_Status"
] = "UNAVAILABLE"

K_RESULTS.loc[
    K_RESULTS["Result Status"].eq("FAILED")
    &
    K_RESULTS["QC_Status"].eq("PASS"),
    "QC_Status"
] = "FAILED"

K_QC_STATUS_COUNTS = (
    K_RESULTS["QC_Status"]
    .value_counts()
    .rename_axis("QC Status")
    .reset_index(name="Count")
)

display(K_QC_STATUS_COUNTS)

print()
print("K8 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L1 — FIDELITY ANALYSIS INPUT VALIDATION
# ==============================================================================

print("=" * 100)
print("L1 — FIDELITY ANALYSIS INPUT VALIDATION")
print("=" * 100)

import os
import pathlib
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# --------------------------------------------------------------------------
# REQUIRED NOTEBOOK 9 OBJECTS
# --------------------------------------------------------------------------

if "FINAL_DATASETS" in globals():
    L_DATASETS = list(FINAL_DATASETS)
elif "DATASETS" in globals():
    L_DATASETS = list(DATASETS)
else:
    L_DATASETS = [
        "adult_income",
        "bank_marketing",
        "diabetes_130us"
    ]

if "METHODS" in globals():
    L_METHODS = list(METHODS)
else:
    L_METHODS = [
        "CTGAN",
        "TVAE",
        "DP-CTGAN",
        "SPP-GAN"
    ]

# --------------------------------------------------------------------------
# REAL DATA REGISTRY
# --------------------------------------------------------------------------

L_REFERENCE_REGISTRY = None

for candidate in [
    "REFERENCE_DATA_REGISTRY",
    "FINAL_REFERENCE_DATA_REGISTRY",
    "REFERENCE_DATASET_REGISTRY"
]:
    if candidate in globals():
        if isinstance(globals()[candidate], pd.DataFrame):
            L_REFERENCE_REGISTRY = globals()[candidate].copy()
            break

# --------------------------------------------------------------------------
# SYNTHETIC DATA REGISTRY
# --------------------------------------------------------------------------

L_SYNTHETIC_REGISTRY = None

for candidate in [
    "SYNTHETIC_DATASET_REGISTRY_TABLE",
    "SYNTHETIC_DATASET_REGISTRY",
    "SYNTHETIC_LOAD_RESULTS"
]:
    if candidate in globals():
        if isinstance(globals()[candidate], pd.DataFrame):
            L_SYNTHETIC_REGISTRY = globals()[candidate].copy()
            break

if L_SYNTHETIC_REGISTRY is None:
    raise RuntimeError(
        "No validated synthetic dataset registry was found. "
        "Complete Section E before running Section L."
    )

print()
print("-" * 100)
print("FIDELITY INPUT STATUS")
print("-" * 100)

print(f"Datasets : {L_DATASETS}")
print(f"Methods  : {L_METHODS}")

print()
print(
    f"Synthetic registry rows : "
    f"{len(L_SYNTHETIC_REGISTRY):,}"
)

if L_REFERENCE_REGISTRY is not None:
    print(
        f"Reference registry rows : "
        f"{len(L_REFERENCE_REGISTRY):,}"
    )
else:
    print(
        "Reference registry : NOT FOUND"
    )

print()
print("L1 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L2 — NUMERICAL FEATURE IDENTIFICATION
# ==============================================================================

print()
print("=" * 100)
print("L2 — NUMERICAL FEATURE IDENTIFICATION")
print("=" * 100)

L_REFERENCE_DATA = {}
L_FEATURE_REGISTRY = []

# --------------------------------------------------------------------------
# RECOVER REFERENCE DATA
# --------------------------------------------------------------------------

if "REFERENCE_DATASETS" in globals():
    if isinstance(REFERENCE_DATASETS, dict):
        L_REFERENCE_DATA = {
            key: value.copy()
            for key, value in REFERENCE_DATASETS.items()
            if isinstance(value, pd.DataFrame)
        }

if not L_REFERENCE_DATA:

    for candidate in [
        "FINAL_REFERENCE_DATASETS",
        "REFERENCE_DATA"
    ]:

        if candidate in globals():

            obj = globals()[candidate]

            if isinstance(obj, dict):

                L_REFERENCE_DATA = {
                    key: value.copy()
                    for key, value in obj.items()
                    if isinstance(value, pd.DataFrame)
                }

                if L_REFERENCE_DATA:
                    break

# --------------------------------------------------------------------------
# FALLBACK: LOAD FROM REFERENCE REGISTRY
# --------------------------------------------------------------------------

if not L_REFERENCE_DATA and L_REFERENCE_REGISTRY is not None:

    registry = L_REFERENCE_REGISTRY.copy()

    if "Dataset" in registry.columns:

        for dataset in L_DATASETS:

            rows = registry[
                registry["Dataset"].astype(str)
                == str(dataset)
            ]

            if len(rows) == 0:
                continue

            file_column = next(
                (
                    column
                    for column in [
                        "File",
                        "Reference File",
                        "Path",
                        "Resolved File"
                    ]
                    if column in rows.columns
                ),
                None
            )

            if file_column is None:
                continue

            file_values = rows[file_column].dropna().tolist()

            if not file_values:
                continue

            file_path = pathlib.Path(
                str(file_values[0])
            )

            if file_path.exists():

                try:
                    L_REFERENCE_DATA[dataset] = pd.read_csv(
                        file_path
                    )
                except Exception as exc:
                    print(
                        f"WARNING: Could not load {dataset}: {exc}"
                    )

# --------------------------------------------------------------------------
# IDENTIFY FEATURES
# --------------------------------------------------------------------------

for dataset, df in L_REFERENCE_DATA.items():

    numerical_columns = (
        df.select_dtypes(
            include=[np.number]
        )
        .columns
        .tolist()
    )

    categorical_columns = [
        column
        for column in df.columns
        if column not in numerical_columns
    ]

    L_FEATURE_REGISTRY.append({

        "Dataset": dataset,

        "Rows": len(df),

        "Columns": len(df.columns),

        "Numerical Features":
            len(numerical_columns),

        "Categorical Features":
            len(categorical_columns),

        "Numerical Columns":
            numerical_columns,

        "Categorical Columns":
            categorical_columns
    })

L_FEATURE_REGISTRY = pd.DataFrame(
    L_FEATURE_REGISTRY
)

display(
    L_FEATURE_REGISTRY[
        [
            "Dataset",
            "Rows",
            "Columns",
            "Numerical Features",
            "Categorical Features"
        ]
    ]
)

print()
print("L2 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L3 — NUMERICAL DISTRIBUTION COMPARISON
# ==============================================================================

print()
print("=" * 100)
print("L3 — NUMERICAL DISTRIBUTION COMPARISON")
print("=" * 100)

L_NUMERICAL_RESULTS = []

def L_GET_SYNTHETIC_PATH(dataset, method):

    registry = L_SYNTHETIC_REGISTRY.copy()

    required = {"Dataset", "Method"}

    if not required.issubset(registry.columns):
        return None

    rows = registry[
        registry["Dataset"].astype(str).eq(str(dataset))
        &
        registry["Method"].astype(str).eq(str(method))
    ]

    if len(rows) == 0:
        return None

    # Prefer READY status
    if "Status" in rows.columns:

        ready = rows[
            rows["Status"].astype(str).str.upper().isin(
                ["READY", "LOADED", "AVAILABLE"]
            )
        ]

        if len(ready) > 0:
            rows = ready

    file_column = next(
        (
            column
            for column in [
                "File",
                "Resolved File",
                "Synthetic File",
                "Path"
            ]
            if column in rows.columns
        ),
        None
    )

    if file_column is None:
        return None

    paths = rows[file_column].dropna().tolist()

    for path in paths:

        candidate = pathlib.Path(str(path))

        if candidate.exists():
            return candidate

    return None


for dataset in L_DATASETS:

    if dataset not in L_REFERENCE_DATA:
        continue

    reference_df = L_REFERENCE_DATA[dataset]

    reference_numeric = (
        reference_df
        .select_dtypes(include=[np.number])
        .columns
        .tolist()
    )

    for method in L_METHODS:

        synthetic_path = L_GET_SYNTHETIC_PATH(
            dataset,
            method
        )

        if synthetic_path is None:

            for feature in reference_numeric:

                L_NUMERICAL_RESULTS.append({
                    "Dataset": dataset,
                    "Method": method,
                    "Feature": feature,
                    "Metric": "Wasserstein Distance",
                    "Value": np.nan,
                    "Status": "UNAVAILABLE"
                })

            continue

        try:
            synthetic_df = pd.read_csv(
                synthetic_path
            )
        except Exception:
            continue

        for feature in reference_numeric:

            if feature not in synthetic_df.columns:

                L_NUMERICAL_RESULTS.append({
                    "Dataset": dataset,
                    "Method": method,
                    "Feature": feature,
                    "Metric": "Wasserstein Distance",
                    "Value": np.nan,
                    "Status": "MISSING_FEATURE"
                })

                continue

            real_values = pd.to_numeric(
                reference_df[feature],
                errors="coerce"
            ).dropna()

            synthetic_values = pd.to_numeric(
                synthetic_df[feature],
                errors="coerce"
            ).dropna()

            if (
                len(real_values) == 0
                or
                len(synthetic_values) == 0
            ):

                distance = np.nan
                status = "INVALID"

            else:

                from scipy.stats import wasserstein_distance

                distance = float(
                    wasserstein_distance(
                        real_values.to_numpy(),
                        synthetic_values.to_numpy()
                    )
                )

                status = "AVAILABLE"

            L_NUMERICAL_RESULTS.append({

                "Dataset": dataset,

                "Method": method,

                "Feature": feature,

                "Metric":
                    "Wasserstein Distance",

                "Value":
                    distance,

                "Status":
                    status
            })

L_NUMERICAL_RESULTS = pd.DataFrame(
    L_NUMERICAL_RESULTS
)

print()
print(
    f"Numerical fidelity records : "
    f"{len(L_NUMERICAL_RESULTS):,}"
)

if len(L_NUMERICAL_RESULTS) > 0:

    print()
    print(
        L_NUMERICAL_RESULTS["Status"]
        .value_counts()
    )

print()
print("L3 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L4 — CATEGORICAL DISTRIBUTION COMPARISON
# ==============================================================================

print()
print("=" * 100)
print("L4 — CATEGORICAL DISTRIBUTION COMPARISON")
print("=" * 100)

L_CATEGORICAL_RESULTS = []

for dataset in L_DATASETS:

    if dataset not in L_REFERENCE_DATA:
        continue

    reference_df = L_REFERENCE_DATA[dataset]

    categorical_columns = [
        column
        for column in reference_df.columns
        if column not in
        reference_df.select_dtypes(
            include=[np.number]
        ).columns
    ]

    for method in L_METHODS:

        synthetic_path = L_GET_SYNTHETIC_PATH(
            dataset,
            method
        )

        if synthetic_path is None:

            for feature in categorical_columns:

                L_CATEGORICAL_RESULTS.append({

                    "Dataset": dataset,
                    "Method": method,
                    "Feature": feature,
                    "Metric": "Total Variation Distance",
                    "Value": np.nan,
                    "Status": "UNAVAILABLE"
                })

            continue

        try:
            synthetic_df = pd.read_csv(
                synthetic_path
            )
        except Exception:
            continue

        for feature in categorical_columns:

            if feature not in synthetic_df.columns:

                L_CATEGORICAL_RESULTS.append({

                    "Dataset": dataset,
                    "Method": method,
                    "Feature": feature,
                    "Metric": "Total Variation Distance",
                    "Value": np.nan,
                    "Status": "MISSING_FEATURE"
                })

                continue

            real_distribution = (
                reference_df[feature]
                .astype(str)
                .value_counts(
                    normalize=True
                )
            )

            synthetic_distribution = (
                synthetic_df[feature]
                .astype(str)
                .value_counts(
                    normalize=True
                )
            )

            categories = (
                set(real_distribution.index)
                |
                set(synthetic_distribution.index)
            )

            real_probabilities = np.array([
                real_distribution.get(
                    category,
                    0.0
                )
                for category in categories
            ])

            synthetic_probabilities = np.array([
                synthetic_distribution.get(
                    category,
                    0.0
                )
                for category in categories
            ])

            tvd = float(
                0.5
                *
                np.abs(
                    real_probabilities
                    -
                    synthetic_probabilities
                ).sum()
            )

            L_CATEGORICAL_RESULTS.append({

                "Dataset": dataset,

                "Method": method,

                "Feature": feature,

                "Metric":
                    "Total Variation Distance",

                "Value": tvd,

                "Status":
                    "AVAILABLE"
            })

L_CATEGORICAL_RESULTS = pd.DataFrame(
    L_CATEGORICAL_RESULTS
)

print()
print(
    f"Categorical fidelity records : "
    f"{len(L_CATEGORICAL_RESULTS):,}"
)

if len(L_CATEGORICAL_RESULTS) > 0:

    print()
    print(
        L_CATEGORICAL_RESULTS["Status"]
        .value_counts()
    )

print()
print("L4 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L5 — DESCRIPTIVE STATISTICS COMPARISON
# ==============================================================================

print()
print("=" * 100)
print("L5 — DESCRIPTIVE STATISTICS COMPARISON")
print("=" * 100)

L_DESCRIPTIVE_RESULTS = []

for dataset in L_DATASETS:

    if dataset not in L_REFERENCE_DATA:
        continue

    reference_df = L_REFERENCE_DATA[dataset]

    numerical_columns = (
        reference_df
        .select_dtypes(include=[np.number])
        .columns
        .tolist()
    )

    for method in L_METHODS:

        synthetic_path = L_GET_SYNTHETIC_PATH(
            dataset,
            method
        )

        if synthetic_path is None:
            continue

        try:
            synthetic_df = pd.read_csv(
                synthetic_path
            )
        except Exception:
            continue

        for feature in numerical_columns:

            if feature not in synthetic_df.columns:
                continue

            real_values = pd.to_numeric(
                reference_df[feature],
                errors="coerce"
            ).dropna()

            synthetic_values = pd.to_numeric(
                synthetic_df[feature],
                errors="coerce"
            ).dropna()

            if (
                len(real_values) == 0
                or
                len(synthetic_values) == 0
            ):
                continue

            real_stats = {
                "Mean": real_values.mean(),
                "Std": real_values.std(),
                "Median": real_values.median(),
                "Min": real_values.min(),
                "Max": real_values.max()
            }

            synthetic_stats = {
                "Mean": synthetic_values.mean(),
                "Std": synthetic_values.std(),
                "Median": synthetic_values.median(),
                "Min": synthetic_values.min(),
                "Max": synthetic_values.max()
            }

            for statistic in real_stats:

                real_value = float(
                    real_stats[statistic]
                )

                synthetic_value = float(
                    synthetic_stats[statistic]
                )

                absolute_error = abs(
                    real_value
                    -
                    synthetic_value
                )

                denominator = max(
                    abs(real_value),
                    np.finfo(float).eps
                )

                relative_error = (
                    absolute_error
                    /
                    denominator
                )

                L_DESCRIPTIVE_RESULTS.append({

                    "Dataset": dataset,

                    "Method": method,

                    "Feature": feature,

                    "Statistic": statistic,

                    "Reference Value":
                        real_value,

                    "Synthetic Value":
                        synthetic_value,

                    "Absolute Error":
                        absolute_error,

                    "Relative Error":
                        relative_error,

                    "Status":
                        "AVAILABLE"
                })

L_DESCRIPTIVE_RESULTS = pd.DataFrame(
    L_DESCRIPTIVE_RESULTS
)

print()
print(
    f"Descriptive statistic records : "
    f"{len(L_DESCRIPTIVE_RESULTS):,}"
)

print()
print("L5 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L6 — TARGET-DISTRIBUTION COMPARISON
# ==============================================================================

print()
print("=" * 100)
print("L6 — TARGET-DISTRIBUTION COMPARISON")
print("=" * 100)

L_TARGET_CANDIDATES = {
    "adult_income": [
        "income",
        "target",
        "class"
    ],
    "bank_marketing": [
        "y",
        "target",
        "class"
    ],
    "diabetes_130us": [
        "readmitted",
        "target",
        "class"
    ]
}

L_TARGET_RESULTS = []

for dataset in L_DATASETS:

    if dataset not in L_REFERENCE_DATA:
        continue

    reference_df = L_REFERENCE_DATA[dataset]

    target_column = None

    for candidate in L_TARGET_CANDIDATES.get(
        dataset,
        []
    ):

        if candidate in reference_df.columns:
            target_column = candidate
            break

    # fallback to last column
    if target_column is None and len(reference_df.columns) > 0:

        target_column = reference_df.columns[-1]

    if target_column is None:
        continue

    for method in L_METHODS:

        synthetic_path = L_GET_SYNTHETIC_PATH(
            dataset,
            method
        )

        if synthetic_path is None:
            continue

        try:
            synthetic_df = pd.read_csv(
                synthetic_path
            )
        except Exception:
            continue

        if target_column not in synthetic_df.columns:
            continue

        real_distribution = (
            reference_df[target_column]
            .astype(str)
            .value_counts(
                normalize=True
            )
        )

        synthetic_distribution = (
            synthetic_df[target_column]
            .astype(str)
            .value_counts(
                normalize=True
            )
        )

        categories = (
            set(real_distribution.index)
            |
            set(synthetic_distribution.index)
        )

        real_probabilities = np.array([
            real_distribution.get(
                category,
                0.0
            )
            for category in categories
        ])

        synthetic_probabilities = np.array([
            synthetic_distribution.get(
                category,
                0.0
            )
            for category in categories
        ])

        target_tvd = float(
            0.5
            *
            np.abs(
                real_probabilities
                -
                synthetic_probabilities
            ).sum()
        )

        L_TARGET_RESULTS.append({

            "Dataset": dataset,

            "Method": method,

            "Target":
                target_column,

            "Metric":
                "Target TVD",

            "Value":
                target_tvd,

            "Status":
                "AVAILABLE"
        })

L_TARGET_RESULTS = pd.DataFrame(
    L_TARGET_RESULTS
)

display(L_TARGET_RESULTS)

print()
print("L6 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L7 — CORRELATION-STRUCTURE COMPARISON
# ==============================================================================

print()
print("=" * 100)
print("L7 — CORRELATION-STRUCTURE COMPARISON")
print("=" * 100)

L_CORRELATION_RESULTS = []

for dataset in L_DATASETS:

    if dataset not in L_REFERENCE_DATA:
        continue

    reference_df = L_REFERENCE_DATA[dataset]

    numerical_columns = (
        reference_df
        .select_dtypes(include=[np.number])
        .columns
        .tolist()
    )

    if len(numerical_columns) < 2:
        continue

    reference_corr = (
        reference_df[numerical_columns]
        .corr(method="pearson")
    )

    for method in L_METHODS:

        synthetic_path = L_GET_SYNTHETIC_PATH(
            dataset,
            method
        )

        if synthetic_path is None:
            continue

        try:
            synthetic_df = pd.read_csv(
                synthetic_path
            )
        except Exception:
            continue

        common_columns = [
            column
            for column in numerical_columns
            if column in synthetic_df.columns
        ]

        if len(common_columns) < 2:
            continue

        synthetic_corr = (
            synthetic_df[common_columns]
            .corr(method="pearson")
        )

        reference_matrix = (
            reference_corr.loc[
                common_columns,
                common_columns
            ]
        )

        difference_matrix = (
            reference_matrix
            -
            synthetic_corr
        ).abs()

        # Exclude diagonal
        upper_triangle = np.triu(
            np.ones(
                difference_matrix.shape,
                dtype=bool
            ),
            k=1
        )

        values = difference_matrix.values[
            upper_triangle
        ]

        if len(values) == 0:
            continue

        mean_absolute_error = float(
            np.mean(values)
        )

        max_absolute_error = float(
            np.max(values)
        )

        L_CORRELATION_RESULTS.append({

            "Dataset": dataset,

            "Method": method,

            "Metric":
                "Mean Absolute Correlation Error",

            "Value":
                mean_absolute_error,

            "Maximum Correlation Error":
                max_absolute_error,

            "Features Compared":
                len(common_columns),

            "Status":
                "AVAILABLE"
        })

L_CORRELATION_RESULTS = pd.DataFrame(
    L_CORRELATION_RESULTS
)

display(L_CORRELATION_RESULTS)

print()
print("L7 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L8 — DISTRIBUTIONAL SIMILARITY METRICS
# ==============================================================================

print()
print("=" * 100)
print("L8 — DISTRIBUTIONAL SIMILARITY METRICS")
print("=" * 100)

L_DISTRIBUTIONAL_RESULTS = []

# --------------------------------------------------------------------------
# NUMERICAL
# --------------------------------------------------------------------------

if len(L_NUMERICAL_RESULTS) > 0:

    for (dataset, method), group in (
        L_NUMERICAL_RESULTS
        .groupby(
            ["Dataset", "Method"]
        )
    ):

        values = pd.to_numeric(
            group.loc[
                group["Status"].eq("AVAILABLE"),
                "Value"
            ],
            errors="coerce"
        ).dropna()

        if len(values) > 0:

            L_DISTRIBUTIONAL_RESULTS.append({

                "Dataset": dataset,

                "Method": method,

                "Metric":
                    "Mean Wasserstein Distance",

                "Value":
                    float(values.mean()),

                "Features Evaluated":
                    len(values),

                "Direction":
                    "LOWER_IS_BETTER",

                "Status":
                    "AVAILABLE"
            })

# --------------------------------------------------------------------------
# CATEGORICAL
# --------------------------------------------------------------------------

if len(L_CATEGORICAL_RESULTS) > 0:

    for (dataset, method), group in (
        L_CATEGORICAL_RESULTS
        .groupby(
            ["Dataset", "Method"]
        )
    ):

        values = pd.to_numeric(
            group.loc[
                group["Status"].eq("AVAILABLE"),
                "Value"
            ],
            errors="coerce"
        ).dropna()

        if len(values) > 0:

            L_DISTRIBUTIONAL_RESULTS.append({

                "Dataset": dataset,

                "Method": method,

                "Metric":
                    "Mean Total Variation Distance",

                "Value":
                    float(values.mean()),

                "Features Evaluated":
                    len(values),

                "Direction":
                    "LOWER_IS_BETTER",

                "Status":
                    "AVAILABLE"
            })

L_DISTRIBUTIONAL_RESULTS = pd.DataFrame(
    L_DISTRIBUTIONAL_RESULTS
)

display(
    L_DISTRIBUTIONAL_RESULTS
)

print()
print("L8 STATUS : COMPLETED")
```

```python
# ==============================================================================
# L9 — DATASET × METHOD FIDELITY SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("L9 — DATASET × METHOD FIDELITY SUMMARY")
print("=" * 100)

L_FIDELITY_SUMMARY = []

for dataset in L_DATASETS:

    for method in L_METHODS:

        numerical_values = pd.Series(
            dtype=float
        )

        categorical_values = pd.Series(
            dtype=float
        )

        correlation_values = pd.Series(
            dtype=float
        )

        target_values = pd.Series(
            dtype=float
        )

        if len(L_NUMERICAL_RESULTS) > 0:

            numerical_values = pd.to_numeric(
                L_NUMERICAL_RESULTS.loc[
                    (
                        L_NUMERICAL_RESULTS["Dataset"]
                        == dataset
                    )
                    &
                    (
                        L_NUMERICAL_RESULTS["Method"]
                        == method
                    )
                    &
                    (
                        L_NUMERICAL_RESULTS["Status"]
                        == "AVAILABLE"
                    ),
                    "Value"
                ],
                errors="coerce"
            ).dropna()

        if len(L_CATEGORICAL_RESULTS) > 0:

            categorical_values = pd.to_numeric(
                L_CATEGORICAL_RESULTS.loc[
                    (
                        L_CATEGORICAL_RESULTS["Dataset"]
                        == dataset
                    )
                    &
                    (
                        L_CATEGORICAL_RESULTS["Method"]
                        == method
                    )
                    &
                    (
                        L_CATEGORICAL_RESULTS["Status"]
                        == "AVAILABLE"
                    ),
                    "Value"
                ],
                errors="coerce"
            ).dropna()

        if len(L_CORRELATION_RESULTS) > 0:

            correlation_values = pd.to_numeric(
                L_CORRELATION_RESULTS.loc[
                    (
                        L_CORRELATION_RESULTS["Dataset"]
                        == dataset
                    )
                    &
                    (
                        L_CORRELATION_RESULTS["Method"]
                        == method
                    ),
                    "Value"
                ],
                errors="coerce"
            ).dropna()

        if len(L_TARGET_RESULTS) > 0:

            target_values = pd.to_numeric(
                L_TARGET_RESULTS.loc[
                    (
                        L_TARGET_RESULTS["Dataset"]
                        == dataset
                    )
                    &
                    (
                        L_TARGET_RESULTS["Method"]
                        == method
                    ),
                    "Value"
                ],
                errors="coerce"
            ).dropna()

        L_FIDELITY_SUMMARY.append({

            "Dataset": dataset,

            "Method": method,

            "Mean Wasserstein Distance":
                (
                    numerical_values.mean()
                    if len(numerical_values) > 0
                    else np.nan
                ),

            "Mean TVD":
                (
                    categorical_values.mean()
                    if len(categorical_values) > 0
                    else np.nan
                ),

            "Correlation Error":
                (
                    correlation_values.mean()
                    if len(correlation_values) > 0
                    else np.nan
                ),

            "Target TVD":
                (
                    target_values.mean()
                    if len(target_values) > 0
                    else np.nan
                ),

            "Numerical Features Evaluated":
                len(numerical_values),

            "Categorical Features Evaluated":
                len(categorical_values),

            "Fidelity Status":
                (
                    "AVAILABLE"
                    if (
                        len(numerical_values)
                        +
                        len(categorical_values)
                        +
                        len(correlation_values)
                        +
                        len(target_values)
                    ) > 0
                    else "UNAVAILABLE"
                )
        })

L_FIDELITY_SUMMARY = pd.DataFrame(
    L_FIDELITY_SUMMARY
)

display(
    L_FIDELITY_SUMMARY
)

print()
print("L9 STATUS : COMPLETED")
```

```python
# ==============================================================================
# M1 — CORRELATION ANALYSIS INITIALIZATION
# ==============================================================================

print("=" * 100)
print("M1 — CORRELATION ANALYSIS INITIALIZATION")
print("=" * 100)

import os
import re
import pathlib
import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# FINAL RESEARCH SCOPE
# ------------------------------------------------------------------------------

FINAL_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

FINAL_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]

print()
print("-" * 100)
print("FINAL CORRELATION ANALYSIS SCOPE")
print("-" * 100)

print(f"Datasets : {FINAL_DATASETS}")
print(f"Methods  : {FINAL_METHODS}")

# ------------------------------------------------------------------------------
# DATASET DISPLAY NAMES
# ------------------------------------------------------------------------------

DATASET_DISPLAY_NAMES = {
    "adult_income": "Adult Income",
    "bank_marketing": "Bank Marketing",
    "diabetes_130us": "Diabetes 130US"
}

# ------------------------------------------------------------------------------
# METHOD DISPLAY NAMES
# ------------------------------------------------------------------------------

METHOD_DISPLAY_NAMES = {
    "CTGAN": "CTGAN",
    "TVAE": "TVAE",
    "DP-CTGAN": "DP-CTGAN",
    "SPP-GAN": "SPP-GAN"
}

# ------------------------------------------------------------------------------
# CORRELATION METRIC DEFINITIONS
# ------------------------------------------------------------------------------

CORRELATION_METRIC_DEFINITIONS = {
    "correlation_difference": {
        "meaning": "Difference between real and synthetic correlation structures",
        "direction": "LOWER_IS_BETTER",
        "ideal": 0.0
    },
    "correlation_similarity": {
        "meaning": "Similarity between real and synthetic correlation structures",
        "direction": "HIGHER_IS_BETTER",
        "ideal": 1.0
    },
    "correlation_preservation": {
        "meaning": "Degree to which feature relationships are preserved",
        "direction": "HIGHER_IS_BETTER",
        "ideal": 1.0
    },
    "pearson_correlation": {
        "meaning": "Linear relationship preservation",
        "direction": "CLOSER_TO_REAL_IS_BETTER",
        "ideal": None
    },
    "spearman_correlation": {
        "meaning": "Rank-based dependency preservation",
        "direction": "CLOSER_TO_REAL_IS_BETTER",
        "ideal": None
    }
}

# ------------------------------------------------------------------------------
# CONTAINERS
# ------------------------------------------------------------------------------

CORRELATION_RAW_RESULTS = []
CORRELATION_RESOLVED_RESULTS = []
CORRELATION_VALIDATED_RESULTS = []

CORRELATION_ANALYSIS_MATRIX = pd.DataFrame()
CORRELATION_DATASET_SUMMARY = pd.DataFrame()
CORRELATION_METHOD_SUMMARY = pd.DataFrame()
CORRELATION_ROBUSTNESS_SUMMARY = pd.DataFrame()

print()
print("Correlation analysis containers initialized successfully.")

print()
print("=" * 100)
print("M1 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M2 — DISCOVER CORRELATION/DEPENDENCY EVALUATION RESULTS
# ==============================================================================

print("=" * 100)
print("M2 — DISCOVER CORRELATION/DEPENDENCY EVALUATION RESULTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# PROJECT DIRECTORIES
# ------------------------------------------------------------------------------

SEARCH_DIRECTORIES = []

for variable_name in [
    "EVALUATION_DIRECTORY",
    "RESULTS_DIRECTORY",
    "REPORTS_DIRECTORY"
]:
    if variable_name in globals():
        directory = pathlib.Path(globals()[variable_name])
        if directory.exists():
            SEARCH_DIRECTORIES.append(directory)

# Fallback to known project root if required
if not SEARCH_DIRECTORIES and "PROJECT_ROOT" in globals():
    project_root = pathlib.Path(PROJECT_ROOT)

    for folder_name in ["evaluation", "results", "reports"]:
        directory = project_root / folder_name

        if directory.exists():
            SEARCH_DIRECTORIES.append(directory)

SEARCH_DIRECTORIES = list(dict.fromkeys(SEARCH_DIRECTORIES))

print()
print("-" * 100)
print("SEARCH DIRECTORIES")
print("-" * 100)

for directory in SEARCH_DIRECTORIES:
    print(f"  {directory}")

# ------------------------------------------------------------------------------
# CORRELATION SEARCH TERMS
# ------------------------------------------------------------------------------

CORRELATION_TERMS = [
    "correlation",
    "correlations",
    "correlation_matrix",
    "correlation_difference",
    "correlation_similarity",
    "correlation_preservation",
    "dependency",
    "dependency_structure",
    "feature_relationship",
    "feature_relationships",
    "pearson",
    "spearman"
]

# ------------------------------------------------------------------------------
# METHOD TOKENS
# ------------------------------------------------------------------------------

METHOD_SEARCH_TOKENS = {
    "CTGAN": ["ctgan"],
    "TVAE": ["tvae"],
    "DP-CTGAN": ["dp_ctgan", "dp-ctgan", "dpctgan"],
    "SPP-GAN": ["spp_gan", "spp-gan", "sppgan"]
}

# ------------------------------------------------------------------------------
# DATASET TOKENS
# ------------------------------------------------------------------------------

DATASET_SEARCH_TOKENS = {
    "adult_income": [
        "adult_income",
        "adult income",
        "adultincome"
    ],
    "bank_marketing": [
        "bank_marketing",
        "bank marketing",
        "bankmarketing"
    ],
    "diabetes_130us": [
        "diabetes_130us",
        "diabetes 130us",
        "diabetes130us"
    ]
}

# ------------------------------------------------------------------------------
# DISCOVERY
# ------------------------------------------------------------------------------

CORRELATION_DISCOVERY_RESULTS = []

for root_directory in SEARCH_DIRECTORIES:

    for file_path in root_directory.rglob("*"):

        if not file_path.is_file():
            continue

        filename_lower = file_path.name.lower()

        if file_path.suffix.lower() not in [
            ".csv",
            ".json",
            ".xlsx",
            ".xls"
        ]:
            continue

        matched_terms = [
            term
            for term in CORRELATION_TERMS
            if term.lower() in filename_lower
        ]

        if not matched_terms:
            continue

        dataset_matches = []

        for dataset, tokens in DATASET_SEARCH_TOKENS.items():
            if any(token.lower() in filename_lower for token in tokens):
                dataset_matches.append(dataset)

        method_matches = []

        for method, tokens in METHOD_SEARCH_TOKENS.items():
            if any(token.lower() in filename_lower for token in tokens):
                method_matches.append(method)

        CORRELATION_DISCOVERY_RESULTS.append({
            "File": str(file_path),
            "Filename": file_path.name,
            "Extension": file_path.suffix.lower(),
            "Matched Terms": ", ".join(matched_terms),
            "Dataset Matches": ", ".join(dataset_matches),
            "Method Matches": ", ".join(method_matches),
            "Size MB": round(file_path.stat().st_size / (1024 ** 2), 3)
        })

CORRELATION_DISCOVERY_TABLE = pd.DataFrame(
    CORRELATION_DISCOVERY_RESULTS
)

if not CORRELATION_DISCOVERY_TABLE.empty:
    CORRELATION_DISCOVERY_TABLE = (
        CORRELATION_DISCOVERY_TABLE
        .sort_values(
            by=["Dataset Matches", "Method Matches", "Filename"]
        )
        .reset_index(drop=True)
    )

print()
print("-" * 100)
print("CORRELATION/DEPENDENCY ARTIFACT DISCOVERY")
print("-" * 100)

print(
    f"Potential correlation/dependency artifacts discovered: "
    f"{len(CORRELATION_DISCOVERY_TABLE)}"
)

if not CORRELATION_DISCOVERY_TABLE.empty:
    display(CORRELATION_DISCOVERY_TABLE)
else:
    print("No correlation/dependency evaluation files were discovered.")

print()
print("=" * 100)
print("M2 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M3 — RESOLVE DATASET × METHOD CORRELATION RESULTS
# ==============================================================================

print("=" * 100)
print("M3 — RESOLVE DATASET × METHOD CORRELATION RESULTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# HELPER
# ------------------------------------------------------------------------------

def normalize_text(value):
    return re.sub(
        r"[^a-z0-9]+",
        "_",
        str(value).lower()
    ).strip("_")

# ------------------------------------------------------------------------------
# RESOLUTION
# ------------------------------------------------------------------------------

CORRELATION_RESOLUTION_ROWS = []

for dataset in FINAL_DATASETS:

    dataset_tokens = [
        normalize_text(token)
        for token in DATASET_SEARCH_TOKENS[dataset]
    ]

    for method in FINAL_METHODS:

        method_tokens = [
            normalize_text(token)
            for token in METHOD_SEARCH_TOKENS[method]
        ]

        candidates = []

        if not CORRELATION_DISCOVERY_TABLE.empty:

            for _, row in CORRELATION_DISCOVERY_TABLE.iterrows():

                filename_normalized = normalize_text(
                    row["Filename"]
                )

                path_normalized = normalize_text(
                    row["File"]
                )

                dataset_match = any(
                    token in filename_normalized
                    or token in path_normalized
                    for token in dataset_tokens
                )

                method_match = any(
                    token in filename_normalized
                    or token in path_normalized
                    for token in method_tokens
                )

                if dataset_match and method_match:
                    candidates.append(row["File"])

        candidates = list(dict.fromkeys(candidates))

        # --------------------------------------------------------------
        # Resolution policy
        # --------------------------------------------------------------

        if len(candidates) == 1:

            status = "RESOLVED"
            resolved_file = candidates[0]

        elif len(candidates) == 0:

            status = "MISSING"
            resolved_file = None

        else:

            status = "AMBIGUOUS"
            resolved_file = None

        CORRELATION_RESOLUTION_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Candidate Count": len(candidates),
            "Status": status,
            "Resolved File": resolved_file,
            "Candidate Files": candidates
        })

CORRELATION_RESOLUTION_TABLE = pd.DataFrame(
    CORRELATION_RESOLUTION_ROWS
)

print()
print("-" * 100)
print("CORRELATION RESULT RESOLUTION")
print("-" * 100)

display(
    CORRELATION_RESOLUTION_TABLE[
        [
            "Dataset",
            "Method",
            "Candidate Count",
            "Status",
            "Resolved File"
        ]
    ]
)

print()
print("-" * 100)
print("RESOLUTION SUMMARY")
print("-" * 100)

print(
    CORRELATION_RESOLUTION_TABLE["Status"]
    .value_counts()
)

print()
print("=" * 100)
print("M3 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M4 — VALIDATE CORRELATION METRICS
# ==============================================================================

print("=" * 100)
print("M4 — VALIDATE CORRELATION METRICS")
print("=" * 100)

# ------------------------------------------------------------------------------
# ACCEPTED CORRELATION COLUMN PATTERNS
# ------------------------------------------------------------------------------

CORRELATION_COLUMN_PATTERNS = [
    "correlation",
    "correlation_difference",
    "correlation_similarity",
    "correlation_preservation",
    "dependency",
    "dependency_difference",
    "dependency_similarity",
    "pearson",
    "spearman"
]

# ------------------------------------------------------------------------------
# VALIDATE RESOLVED FILES
# ------------------------------------------------------------------------------

CORRELATION_VALIDATION_ROWS = []

for _, row in CORRELATION_RESOLUTION_TABLE.iterrows():

    dataset = row["Dataset"]
    method = row["Method"]
    status = row["Status"]
    file_path = row["Resolved File"]

    if status != "RESOLVED":
        CORRELATION_VALIDATION_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Status": status,
            "File": file_path,
            "Metric Columns": None,
            "Valid": False,
            "Reason": status
        })
        continue

    try:

        path = pathlib.Path(file_path)

        if path.suffix.lower() == ".csv":
            df_temp = pd.read_csv(path)

        elif path.suffix.lower() in [".xlsx", ".xls"]:
            df_temp = pd.read_excel(path)

        elif path.suffix.lower() == ".json":
            df_temp = pd.read_json(path)

        else:
            raise ValueError(
                f"Unsupported file type: {path.suffix}"
            )

        metric_columns = []

        for column in df_temp.columns:

            column_normalized = normalize_text(column)

            if any(
                pattern in column_normalized
                for pattern in CORRELATION_COLUMN_PATTERNS
            ):
                metric_columns.append(column)

        if metric_columns:

            validation_status = "VALID"
            reason = "Correlation/dependency metric columns identified."

        else:

            validation_status = "INVALID"
            reason = "No correlation/dependency metric columns identified."

        CORRELATION_VALIDATION_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Status": validation_status,
            "File": file_path,
            "Metric Columns": ", ".join(
                map(str, metric_columns)
            ) if metric_columns else None,
            "Valid": validation_status == "VALID",
            "Reason": reason
        })

    except Exception as exc:

        CORRELATION_VALIDATION_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Status": "ERROR",
            "File": file_path,
            "Metric Columns": None,
            "Valid": False,
            "Reason": str(exc)
        })

CORRELATION_VALIDATION_TABLE = pd.DataFrame(
    CORRELATION_VALIDATION_ROWS
)

print()
print("-" * 100)
print("CORRELATION METRIC VALIDATION")
print("-" * 100)

display(CORRELATION_VALIDATION_TABLE)

print()
print("=" * 100)
print("M4 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M5 — BUILD CORRELATION PRESERVATION MATRIX
# ==============================================================================

print("=" * 100)
print("M5 — BUILD CORRELATION PRESERVATION MATRIX")
print("=" * 100)

M5_ROWS = []

for dataset in FINAL_DATASETS:

    for method in FINAL_METHODS:

        validation_match = CORRELATION_VALIDATION_TABLE[
            (CORRELATION_VALIDATION_TABLE["Dataset"] == dataset) &
            (CORRELATION_VALIDATION_TABLE["Method"] == method)
        ]

        if validation_match.empty:

            status = "MISSING"
            file_path = None
            metric_columns = None

        else:

            validation_row = validation_match.iloc[0]

            status = validation_row["Status"]
            file_path = validation_row["File"]
            metric_columns = validation_row["Metric Columns"]

        M5_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Status": status,
            "File": file_path,
            "Metric Columns": metric_columns
        })

CORRELATION_ANALYSIS_MATRIX = pd.DataFrame(M5_ROWS)

print()
display(CORRELATION_ANALYSIS_MATRIX)

print()
print("-" * 100)
print("CORRELATION MATRIX COVERAGE")
print("-" * 100)

total_expected = len(FINAL_DATASETS) * len(FINAL_METHODS)

valid_count = int(
    (CORRELATION_ANALYSIS_MATRIX["Status"] == "VALID").sum()
)

print(f"Expected Dataset × Method combinations : {total_expected}")
print(f"Validated correlation combinations     : {valid_count}")
print(f"Unavailable/unvalidated combinations   : {total_expected - valid_count}")

print()
print("=" * 100)
print("M5 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M6 — COMPARE FEATURE-RELATIONSHIP PRESERVATION
# ==============================================================================

print("=" * 100)
print("M6 — COMPARE FEATURE-RELATIONSHIP PRESERVATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# CONTAINERS
# ------------------------------------------------------------------------------

FEATURE_RELATIONSHIP_RESULTS = []

# ------------------------------------------------------------------------------
# REAL DATA CONTAINERS
# ------------------------------------------------------------------------------

REAL_DATA_DICTIONARY = {}

for variable_name in [
    "REFERENCE_DATASETS",
    "FINAL_REFERENCE_DATASETS",
    "REFERENCE_DATA"
]:
    if variable_name in globals():

        candidate = globals()[variable_name]

        if isinstance(candidate, dict):

            for dataset, dataframe in candidate.items():

                if isinstance(dataframe, pd.DataFrame):
                    REAL_DATA_DICTIONARY[dataset] = dataframe

# ------------------------------------------------------------------------------
# SYNTHETIC DATA CONTAINERS
# ------------------------------------------------------------------------------

SYNTHETIC_DATA_DICTIONARY = {}

for variable_name in [
    "FINAL_SYNTHETIC_DATASETS",
    "SYNTHETIC_DATASETS",
    "LOADED_SYNTHETIC_DATASETS"
]:
    if variable_name in globals():

        candidate = globals()[variable_name]

        if isinstance(candidate, dict):

            for key, dataframe in candidate.items():

                if not isinstance(dataframe, pd.DataFrame):
                    continue

                if isinstance(key, tuple) and len(key) == 2:

                    dataset, method = key
                    SYNTHETIC_DATA_DICTIONARY[
                        (dataset, method)
                    ] = dataframe

# ------------------------------------------------------------------------------
# DIRECT CORRELATION COMPARISON
# ------------------------------------------------------------------------------

for dataset in FINAL_DATASETS:

    real_df = REAL_DATA_DICTIONARY.get(dataset)

    for method in FINAL_METHODS:

        synthetic_df = SYNTHETIC_DATA_DICTIONARY.get(
            (dataset, method)
        )

        result = {
            "Dataset": dataset,
            "Method": method,
            "Status": "UNAVAILABLE",
            "Compared Features": None,
            "Mean Absolute Correlation Difference": None,
            "Maximum Absolute Correlation Difference": None
        }

        if real_df is None or synthetic_df is None:
            FEATURE_RELATIONSHIP_RESULTS.append(result)
            continue

        real_numeric = real_df.select_dtypes(
            include=np.number
        )

        synthetic_numeric = synthetic_df.select_dtypes(
            include=np.number
        )

        common_features = [
            column
            for column in real_numeric.columns
            if column in synthetic_numeric.columns
        ]

        if len(common_features) < 2:
            FEATURE_RELATIONSHIP_RESULTS.append(result)
            continue

        real_corr = real_numeric[
            common_features
        ].corr(method="pearson")

        synthetic_corr = synthetic_numeric[
            common_features
        ].corr(method="pearson")

        difference_matrix = (
            real_corr - synthetic_corr
        ).abs()

        # --------------------------------------------------------------
        # Exclude diagonal
        # --------------------------------------------------------------

        mask = ~np.eye(
            len(common_features),
            dtype=bool
        )

        differences = difference_matrix.values[mask]

        differences = differences[
            np.isfinite(differences)
        ]

        if len(differences) == 0:
            FEATURE_RELATIONSHIP_RESULTS.append(result)
            continue

        result.update({
            "Status": "VALID",
            "Compared Features": len(common_features),
            "Mean Absolute Correlation Difference": float(
                np.mean(differences)
            ),
            "Maximum Absolute Correlation Difference": float(
                np.max(differences)
            )
        })

        FEATURE_RELATIONSHIP_RESULTS.append(result)

FEATURE_RELATIONSHIP_TABLE = pd.DataFrame(
    FEATURE_RELATIONSHIP_RESULTS
)

print()
display(FEATURE_RELATIONSHIP_TABLE)

print()
print("=" * 100)
print("M6 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M7 — DATASET-WISE DEPENDENCY ANALYSIS
# ==============================================================================

print("=" * 100)
print("M7 — DATASET-WISE DEPENDENCY ANALYSIS")
print("=" * 100)

M7_ROWS = []

for dataset in FINAL_DATASETS:

    dataset_results = FEATURE_RELATIONSHIP_TABLE[
        FEATURE_RELATIONSHIP_TABLE["Dataset"] == dataset
    ].copy()

    valid_results = dataset_results[
        dataset_results["Status"] == "VALID"
    ]

    if valid_results.empty:

        M7_ROWS.append({
            "Dataset": dataset,
            "Validated Methods": 0,
            "Mean Correlation Difference": None,
            "Best Method": None,
            "Worst Method": None,
            "Status": "INSUFFICIENT_RESULTS"
        })

        continue

    method_scores = (
        valid_results
        .groupby("Method")[
            "Mean Absolute Correlation Difference"
        ]
        .mean()
        .sort_values()
    )

    M7_ROWS.append({
        "Dataset": dataset,
        "Validated Methods": len(method_scores),
        "Mean Correlation Difference": float(
            method_scores.mean()
        ),
        "Best Method": method_scores.index[0],
        "Worst Method": method_scores.index[-1],
        "Status": "VALID"
    })

CORRELATION_DATASET_SUMMARY = pd.DataFrame(
    M7_ROWS
)

display(CORRELATION_DATASET_SUMMARY)

print()
print("=" * 100)
print("M7 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M8 — METHOD-WISE DEPENDENCY ANALYSIS
# ==============================================================================

print("=" * 100)
print("M8 — METHOD-WISE DEPENDENCY ANALYSIS")
print("=" * 100)

M8_ROWS = []

for method in FINAL_METHODS:

    method_results = FEATURE_RELATIONSHIP_TABLE[
        FEATURE_RELATIONSHIP_TABLE["Method"] == method
    ].copy()

    valid_results = method_results[
        method_results["Status"] == "VALID"
    ]

    if valid_results.empty:

        M8_ROWS.append({
            "Method": method,
            "Validated Datasets": 0,
            "Mean Correlation Difference": None,
            "Best Dataset": None,
            "Worst Dataset": None,
            "Status": "INSUFFICIENT_RESULTS"
        })

        continue

    dataset_scores = (
        valid_results
        .groupby("Dataset")[
            "Mean Absolute Correlation Difference"
        ]
        .mean()
        .sort_values()
    )

    M8_ROWS.append({
        "Method": method,
        "Validated Datasets": len(dataset_scores),
        "Mean Correlation Difference": float(
            dataset_scores.mean()
        ),
        "Best Dataset": dataset_scores.index[0],
        "Worst Dataset": dataset_scores.index[-1],
        "Status": "VALID"
    })

CORRELATION_METHOD_SUMMARY = pd.DataFrame(
    M8_ROWS
)

display(CORRELATION_METHOD_SUMMARY)

print()
print("=" * 100)
print("M8 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M9 — CROSS-DATASET CORRELATION ROBUSTNESS
# ==============================================================================

print("=" * 100)
print("M9 — CROSS-DATASET CORRELATION ROBUSTNESS")
print("=" * 100)

M9_ROWS = []

for method in FINAL_METHODS:

    method_results = FEATURE_RELATIONSHIP_TABLE[
        (FEATURE_RELATIONSHIP_TABLE["Method"] == method) &
        (FEATURE_RELATIONSHIP_TABLE["Status"] == "VALID")
    ]

    if method_results.empty:

        M9_ROWS.append({
            "Method": method,
            "Datasets Evaluated": 0,
            "Mean Correlation Difference": None,
            "Std Correlation Difference": None,
            "Minimum": None,
            "Maximum": None,
            "Robustness Status": "INSUFFICIENT_RESULTS"
        })

        continue

    scores = method_results[
        "Mean Absolute Correlation Difference"
    ].dropna()

    if len(scores) == 0:

        robustness_status = "INSUFFICIENT_RESULTS"

    elif len(scores) == 1:

        robustness_status = "SINGLE_DATASET"

    else:

        robustness_status = "VALID"

    M9_ROWS.append({
        "Method": method,
        "Datasets Evaluated": len(scores),
        "Mean Correlation Difference": float(scores.mean()),
        "Std Correlation Difference": float(scores.std(ddof=1))
            if len(scores) > 1 else 0.0,
        "Minimum": float(scores.min()),
        "Maximum": float(scores.max()),
        "Robustness Status": robustness_status
    })

CORRELATION_ROBUSTNESS_SUMMARY = pd.DataFrame(
    M9_ROWS
)

display(CORRELATION_ROBUSTNESS_SUMMARY)

print()
print("=" * 100)
print("M9 — COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# M10 — CORRELATION/DEPENDENCY RESEARCH SUMMARY
# ==============================================================================

print("=" * 100)
print("M10 — CORRELATION/DEPENDENCY RESEARCH SUMMARY")
print("=" * 100)

print()
print("-" * 100)
print("CORRELATION ANALYSIS COVERAGE")
print("-" * 100)

expected_combinations = (
    len(FINAL_DATASETS) *
    len(FINAL_METHODS)
)

validated_combinations = int(
    (
        FEATURE_RELATIONSHIP_TABLE["Status"] == "VALID"
    ).sum()
)

print(
    f"Expected Dataset × Method combinations : "
    f"{expected_combinations}"
)

print(
    f"Validated combinations                 : "
    f"{validated_combinations}"
)

print(
    f"Unavailable combinations               : "
    f"{expected_combinations - validated_combinations}"
)

# ------------------------------------------------------------------------------
# BEST METHOD
# ------------------------------------------------------------------------------

valid_method_summary = CORRELATION_METHOD_SUMMARY[
    CORRELATION_METHOD_SUMMARY["Status"] == "VALID"
].copy()

print()
print("-" * 100)
print("METHOD-LEVEL CORRELATION PRESERVATION")
print("-" * 100)

if valid_method_summary.empty:

    print("No sufficient validated method-level correlation results.")

else:

    display(
        valid_method_summary.sort_values(
            "Mean Correlation Difference"
        )
    )

# ------------------------------------------------------------------------------
# RESEARCH INTERPRETATION
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("RESEARCH INTERPRETATION RULE")
print("-" * 100)

print(
    "For correlation-difference metrics, lower values indicate closer "
    "preservation of the real-data dependency structure."
)

print(
    "A method is not considered superior merely because it has missing "
    "results. Missing or unavailable experiments remain explicitly "
    "unavailable."
)

print(
    "Cross-dataset robustness is interpreted only when the method has "
    "validated results across multiple datasets."
)

print()
print("=" * 100)
print("M10 — CORRELATION/DEPENDENCY ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O2 — IDENTIFY PRIVACY-RELATED METRICS
# ==============================================================================

print("=" * 100)
print("O2 — IDENTIFY PRIVACY-RELATED METRICS")
print("=" * 100)

# ------------------------------------------------------------------------------
# PRIVACY KEYWORDS
# ------------------------------------------------------------------------------

O_PRIVACY_KEYWORDS = [
    "privacy",
    "epsilon",
    "eps",
    "delta",
    "privacy_budget",
    "privacy_loss",
    "membership",
    "reidentification",
    "disclosure",
    "nearest_neighbor",
    "distance_to_closest",
    "attribute_disclosure",
    "identity_disclosure",
    "dcr",
    "nndr",
]

# ------------------------------------------------------------------------------
# IDENTIFY COLUMNS
# ------------------------------------------------------------------------------

O_PRIVACY_COLUMNS = []

if "FINAL_RESULTS_DF" in globals() and isinstance(FINAL_RESULTS_DF, pd.DataFrame):

    for column in FINAL_RESULTS_DF.columns:
        column_text = str(column).lower()

        if any(keyword in column_text for keyword in O_PRIVACY_KEYWORDS):
            O_PRIVACY_COLUMNS.append(column)

elif "FINAL_RESEARCH_RESULTS_DF" in globals() and isinstance(
    FINAL_RESEARCH_RESULTS_DF,
    pd.DataFrame
):

    for column in FINAL_RESEARCH_RESULTS_DF.columns:
        column_text = str(column).lower()

        if any(keyword in column_text for keyword in O_PRIVACY_KEYWORDS):
            O_PRIVACY_COLUMNS.append(column)

elif "MASTER_RESULTS_DF" in globals() and isinstance(
    MASTER_RESULTS_DF,
    pd.DataFrame
):

    for column in MASTER_RESULTS_DF.columns:
        column_text = str(column).lower()

        if any(keyword in column_text for keyword in O_PRIVACY_KEYWORDS):
            O_PRIVACY_COLUMNS.append(column)

print()
print("-" * 100)
print("PRIVACY-RELATED COLUMNS IDENTIFIED")
print("-" * 100)

if O_PRIVACY_COLUMNS:
    for column in O_PRIVACY_COLUMNS:
        print(f"  {column}")
else:
    print("  No privacy-related metric columns identified.")

print()
print("O2 — Privacy metric identification completed.")
```

```python
# ==============================================================================
# O3 — BUILD PRIVACY RESULTS SUBSET
# ==============================================================================

print("=" * 100)
print("O3 — BUILD PRIVACY RESULTS SUBSET")
print("=" * 100)

# ------------------------------------------------------------------------------
# FINAL RESEARCH SCOPE
# ------------------------------------------------------------------------------

O_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

O_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

# ------------------------------------------------------------------------------
# PRIVACY KEYWORDS
# ------------------------------------------------------------------------------

O_PRIVACY_KEYWORDS = [
    "privacy",
    "epsilon",
    "eps",
    "delta",
    "privacy_budget",
    "privacy budget",
    "privacy_loss",
    "membership",
    "reidentification",
    "re-identification",
    "disclosure",
    "nearest_neighbor",
    "nearest neighbor",
    "distance_to_closest",
    "attribute_disclosure",
    "identity_disclosure",
    "dcr",
    "nndr",
]

# ------------------------------------------------------------------------------
# CREATE SAFE SNAPSHOT OF GLOBAL NAMESPACE
# ------------------------------------------------------------------------------

O_GLOBAL_ITEMS = list(globals().items())

# ------------------------------------------------------------------------------
# DISCOVER DATAFRAMES
# ------------------------------------------------------------------------------

O_AVAILABLE_DATAFRAMES = {}

for variable_name, variable_value in O_GLOBAL_ITEMS:

    try:
        if isinstance(variable_value, pd.DataFrame):
            O_AVAILABLE_DATAFRAMES[variable_name] = variable_value
    except Exception:
        continue

print()
print("-" * 100)
print("AVAILABLE RESEARCH DATAFRAMES")
print("-" * 100)

if not O_AVAILABLE_DATAFRAMES:

    print("No pandas DataFrames are currently available.")

else:

    for variable_name, dataframe in O_AVAILABLE_DATAFRAMES.items():

        print(
            f"{variable_name:<45} "
            f"{len(dataframe):>10,} rows × "
            f"{len(dataframe.columns):>4,} columns"
        )

# ------------------------------------------------------------------------------
# PRIORITIZED NOTEBOOK 9 RESULTS DATAFRAMES
# ------------------------------------------------------------------------------

O_PRIORITY_NAMES = [

    # --------------------------------------------------------------------------
    # SECTION J
    # --------------------------------------------------------------------------

    "J_FINAL_RESULTS_DF",
    "J_METRIC_CONSOLIDATION_DF",
    "J_FINAL_METRIC_RESULTS_DF",
    "J_CONSOLIDATED_RESULTS_DF",

    # --------------------------------------------------------------------------
    # SECTION H
    # --------------------------------------------------------------------------

    "H_RESULTS_MATRIX_DF",
    "H_EXPERIMENTAL_RESULTS_MATRIX_DF",
    "RESULTS_MATRIX_DF",
    "MASTER_RESULTS_MATRIX_DF",

    # --------------------------------------------------------------------------
    # SECTION F/G
    # --------------------------------------------------------------------------

    "F_FINAL_EVALUATION_REPORTS_DF",
    "G_EVALUATION_COMPLETENESS_DF",

    # --------------------------------------------------------------------------
    # COMMON FALLBACK NAMES
    # --------------------------------------------------------------------------

    "FINAL_RESULTS_DF",
    "FINAL_RESEARCH_RESULTS_DF",
    "MASTER_RESULTS_DF",
    "CONSOLIDATED_RESULTS_DF",
    "RESEARCH_RESULTS_DF",
]

# ------------------------------------------------------------------------------
# RESOLVE SOURCE DATAFRAME
# ------------------------------------------------------------------------------

O_SOURCE_DF = None
O_SOURCE_DF_NAME = None

for candidate_name in O_PRIORITY_NAMES:

    candidate = O_AVAILABLE_DATAFRAMES.get(
        candidate_name
    )

    if isinstance(candidate, pd.DataFrame):

        O_SOURCE_DF = candidate.copy()
        O_SOURCE_DF_NAME = candidate_name
        break

# ------------------------------------------------------------------------------
# AUTOMATIC FALLBACK
# ------------------------------------------------------------------------------

if O_SOURCE_DF is None:

    O_CANDIDATE_SCORES = []

    for variable_name, dataframe in O_AVAILABLE_DATAFRAMES.items():

        column_text = " ".join(
            str(column).lower()
            for column in dataframe.columns
        )

        score = 0

        # Dataset identifier
        if "dataset" in column_text:
            score += 5

        # Method identifier
        if "method" in column_text:
            score += 5

        # Metric identifier
        if "metric" in column_text:
            score += 5

        # Result/value identifier
        if "value" in column_text:
            score += 3

        if "result" in column_text:
            score += 3

        # Privacy information
        if any(
            keyword in column_text
            for keyword in O_PRIVACY_KEYWORDS
        ):
            score += 3

        O_CANDIDATE_SCORES.append(
            (
                score,
                variable_name,
                dataframe
            )
        )

    O_CANDIDATE_SCORES.sort(
        key=lambda item: (
            item[0],
            len(item[2])
        ),
        reverse=True
    )

    if O_CANDIDATE_SCORES:

        best_score = O_CANDIDATE_SCORES[0][0]

        if best_score >= 8:

            O_SOURCE_DF = (
                O_CANDIDATE_SCORES[0][2].copy()
            )

            O_SOURCE_DF_NAME = (
                O_CANDIDATE_SCORES[0][1]
            )

# ------------------------------------------------------------------------------
# SOURCE DATAFRAME STATUS
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("SELECTED SOURCE DATAFRAME")
print("-" * 100)

if O_SOURCE_DF is None:

    print("NO SUITABLE CONSOLIDATED RESULTS DATAFRAME FOUND.")

    print()
    print("Candidate DataFrames available:")

    for variable_name, dataframe in (
        O_AVAILABLE_DATAFRAMES.items()
    ):

        print(
            f"  {variable_name}: "
            f"{dataframe.shape}"
        )

    raise RuntimeError(
        "O3 could not identify the consolidated research-results "
        "DataFrame. The Section J output must be connected to Section O."
    )

print(f"Source dataframe : {O_SOURCE_DF_NAME}")
print(f"Rows             : {len(O_SOURCE_DF):,}")
print(f"Columns          : {len(O_SOURCE_DF.columns):,}")

# ------------------------------------------------------------------------------
# SOURCE COLUMNS
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("SOURCE DATAFRAME COLUMNS")
print("-" * 100)

for column in O_SOURCE_DF.columns:
    print(f"  {column}")

# ------------------------------------------------------------------------------
# IDENTIFY PRIVACY-RELATED COLUMNS
# ------------------------------------------------------------------------------

O_PRIVACY_COLUMNS = []

for column in O_SOURCE_DF.columns:

    column_text = str(column).lower()

    if any(
        keyword in column_text
        for keyword in O_PRIVACY_KEYWORDS
    ):

        O_PRIVACY_COLUMNS.append(column)

print()
print("-" * 100)
print("PRIVACY-RELATED COLUMNS")
print("-" * 100)

if O_PRIVACY_COLUMNS:

    for column in O_PRIVACY_COLUMNS:
        print(f"  {column}")

else:

    print(
        "No explicitly named privacy-related columns were found."
    )

# ------------------------------------------------------------------------------
# IDENTIFY PRIVACY-RELATED ROWS
# ------------------------------------------------------------------------------

O_PRIVACY_MASK = pd.Series(
    False,
    index=O_SOURCE_DF.index
)

# Only search textual columns.
# This avoids converting the entire dataframe unnecessarily.

for column in O_SOURCE_DF.select_dtypes(
    include=["object", "string"]
).columns:

    column_values = (
        O_SOURCE_DF[column]
        .astype(str)
        .str.lower()
    )

    for keyword in O_PRIVACY_KEYWORDS:

        O_PRIVACY_MASK |= (
            column_values.str.contains(
                str(keyword).lower(),
                regex=False,
                na=False
            )
        )

# ------------------------------------------------------------------------------
# BUILD PRIVACY RESULTS DATAFRAME
# ------------------------------------------------------------------------------

O_PRIVACY_RESULTS_DF = O_SOURCE_DF.loc[
    O_PRIVACY_MASK
].copy()

# ------------------------------------------------------------------------------
# DISPLAY PRIVACY RESULTS
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("PRIVACY RESULTS SUBSET")
print("-" * 100)

print(
    f"Privacy-related rows identified : "
    f"{len(O_PRIVACY_RESULTS_DF):,}"
)

print(
    f"Privacy-related columns         : "
    f"{len(O_PRIVACY_COLUMNS):,}"
)

if not O_PRIVACY_RESULTS_DF.empty:

    display(
        O_PRIVACY_RESULTS_DF.head(25)
    )

else:

    print(
        "No privacy-specific rows were identified "
        "in the consolidated results dataframe."
    )

# ------------------------------------------------------------------------------
# DATASET × METHOD PRIVACY COVERAGE
# ------------------------------------------------------------------------------

O_PRIVACY_COVERAGE_ROWS = []

for dataset in O_DATASETS:

    for method in O_METHODS:

        subset = O_PRIVACY_RESULTS_DF.copy()

        # ----------------------------------------------------------------------
        # DATASET FILTER
        # ----------------------------------------------------------------------

        if "Dataset" in subset.columns:

            subset = subset[
                subset["Dataset"]
                .astype(str)
                .str.strip()
                .str.lower()
                == dataset.lower()
            ]

        # ----------------------------------------------------------------------
        # METHOD FILTER
        # ----------------------------------------------------------------------

        if "Method" in subset.columns:

            subset = subset[
                subset["Method"]
                .astype(str)
                .str.strip()
                .str.lower()
                == method.lower()
            ]

        O_PRIVACY_COVERAGE_ROWS.append({

            "Dataset": dataset,

            "Method": method,

            "Privacy Result Rows": len(subset),

            "Privacy Evidence Status": (
                "AVAILABLE"
                if len(subset) > 0
                else "NOT AVAILABLE"
            ),
        })

# ------------------------------------------------------------------------------
# CREATE COVERAGE DATAFRAME
# ------------------------------------------------------------------------------

O_PRIVACY_COVERAGE_DF = pd.DataFrame(
    O_PRIVACY_COVERAGE_ROWS
)

print()
print("-" * 100)
print("PRIVACY DATASET × METHOD COVERAGE")
print("-" * 100)

display(
    O_PRIVACY_COVERAGE_DF
)

# ------------------------------------------------------------------------------
# COVERAGE SUMMARY
# ------------------------------------------------------------------------------

O_PRIVACY_AVAILABLE_COUNT = int(
    (
        O_PRIVACY_COVERAGE_DF[
            "Privacy Evidence Status"
        ]
        == "AVAILABLE"
    ).sum()
)

O_PRIVACY_MISSING_COUNT = int(
    (
        O_PRIVACY_COVERAGE_DF[
            "Privacy Evidence Status"
        ]
        == "NOT AVAILABLE"
    ).sum()
)

O_EXPECTED_COMBINATIONS = (
    len(O_DATASETS) * len(O_METHODS)
)

print()
print("-" * 100)
print("O3 SUMMARY")
print("-" * 100)

print(
    f"Expected Dataset × Method combinations : "
    f"{O_EXPECTED_COMBINATIONS}"
)

print(
    f"Privacy combinations available          : "
    f"{O_PRIVACY_AVAILABLE_COUNT}"
)

print(
    f"Privacy combinations unavailable        : "
    f"{O_PRIVACY_MISSING_COUNT}"
)

print()
print("=" * 100)
print("O3 — BUILD PRIVACY RESULTS SUBSET COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O4 — DIFFERENTIAL PRIVACY PARAMETER REGISTRY
# ==============================================================================

print("=" * 100)
print("O4 — DIFFERENTIAL PRIVACY PARAMETER REGISTRY")
print("=" * 100)

# ------------------------------------------------------------------------------
# POSSIBLE PRIVACY PARAMETER SOURCES
# ------------------------------------------------------------------------------

O_DP_PARAMETER_COLUMNS = [
    "epsilon",
    "Epsilon",
    "Privacy Budget",
    "Privacy_Budget",
    "privacy_budget",
    "Target Epsilon",
    "target_epsilon",
    "delta",
    "Delta",
    "Target Delta",
    "target_delta",
    "noise_multiplier",
    "Noise Multiplier",
    "gradient_clip",
    "Gradient Clipping",
]

# ------------------------------------------------------------------------------
# SEARCH CONSOLIDATED RESULTS
# ------------------------------------------------------------------------------

O_FOUND_DP_COLUMNS = [
    column
    for column in O_DP_PARAMETER_COLUMNS
    if column in O_SOURCE_DF.columns
]

print()
print("-" * 100)
print("AVAILABLE DIFFERENTIAL PRIVACY PARAMETERS")
print("-" * 100)

if O_FOUND_DP_COLUMNS:

    for column in O_FOUND_DP_COLUMNS:
        print(f"  {column}")

else:
    print("  No DP parameter columns found in the consolidated results.")

# ------------------------------------------------------------------------------
# CREATE REGISTRY
# ------------------------------------------------------------------------------

O_DP_REGISTRY = []

for dataset in O_DATASETS:

    for method in O_METHODS:

        row = {
            "Dataset": dataset,
            "Method": method,
            "Epsilon": np.nan,
            "Delta": np.nan,
            "Noise Multiplier": np.nan,
            "Gradient Clipping": np.nan,
            "DP Parameter Status": "NOT FOUND",
        }

        subset = O_SOURCE_DF.copy()

        if "Dataset" in subset.columns:
            subset = subset[
                subset["Dataset"].astype(str).str.lower()
                == dataset.lower()
            ]

        if "Method" in subset.columns:
            subset = subset[
                subset["Method"].astype(str).str.lower()
                == method.lower()
            ]

        if not subset.empty:

            for source_column in [
                "epsilon",
                "Epsilon",
                "Privacy Budget",
                "Privacy_Budget",
                "privacy_budget",
                "Target Epsilon",
                "target_epsilon",
            ]:
                if source_column in subset.columns:
                    value = subset[source_column].dropna()

                    if not value.empty:
                        row["Epsilon"] = value.iloc[0]
                        break

            for source_column in [
                "delta",
                "Delta",
                "Target Delta",
                "target_delta",
            ]:
                if source_column in subset.columns:
                    value = subset[source_column].dropna()

                    if not value.empty:
                        row["Delta"] = value.iloc[0]
                        break

            for source_column in [
                "noise_multiplier",
                "Noise Multiplier",
            ]:
                if source_column in subset.columns:
                    value = subset[source_column].dropna()

                    if not value.empty:
                        row["Noise Multiplier"] = value.iloc[0]
                        break

            for source_column in [
                "gradient_clip",
                "Gradient Clipping",
            ]:
                if source_column in subset.columns:
                    value = subset[source_column].dropna()

                    if not value.empty:
                        row["Gradient Clipping"] = value.iloc[0]
                        break

        if pd.notna(row["Epsilon"]) or pd.notna(row["Delta"]):
            row["DP Parameter Status"] = "AVAILABLE"

        O_DP_REGISTRY.append(row)

O_DP_PARAMETER_REGISTRY_DF = pd.DataFrame(O_DP_REGISTRY)

print()
print("-" * 100)
print("DP PARAMETER REGISTRY")
print("-" * 100)

display(O_DP_PARAMETER_REGISTRY_DF)

print()
print("O4 — Differential privacy parameter registry completed.")
```

```python
# ==============================================================================
# O5 — PRIVACY MECHANISM CLASSIFICATION
# ==============================================================================

print("=" * 100)
print("O5 — PRIVACY MECHANISM CLASSIFICATION")
print("=" * 100)

O_PRIVACY_CLASSIFICATION = {
    "CTGAN": "Non-DP baseline",
    "TVAE": "Non-DP baseline",
    "DP-CTGAN": "Differential Privacy",
    "SPP-GAN": "Research privacy-preserving method",
}

O_PRIVACY_CLASSIFICATION_DF = pd.DataFrame([
    {
        "Method": method,
        "Privacy Mechanism": O_PRIVACY_CLASSIFICATION.get(
            method,
            "Not specified"
        ),
    }
    for method in O_METHODS
])

print()
display(O_PRIVACY_CLASSIFICATION_DF)

print()
print(
    "Important: privacy mechanism classification describes the experimental "
    "design. It does not substitute for measured privacy results."
)

print()
print("O5 — Privacy mechanism classification completed.")
```

```python
# ==============================================================================
# O6 — PRIVACY METRIC QUALITY CONTROL
# ==============================================================================

print("=" * 100)
print("O6 — PRIVACY METRIC QUALITY CONTROL")
print("=" * 100)

O_QC_ROWS = []

for column in O_PRIVACY_COLUMNS:

    if column not in O_SOURCE_DF.columns:
        continue

    series = pd.to_numeric(
        O_SOURCE_DF[column],
        errors="coerce"
    )

    numeric_values = series.dropna()

    O_QC_ROWS.append({
        "Metric": column,
        "Available Values": int(numeric_values.notna().sum()),
        "Missing Values": int(series.isna().sum()),
        "Infinite Values": int(
            np.isinf(numeric_values).sum()
        ) if not numeric_values.empty else 0,
        "Minimum": numeric_values.min()
            if not numeric_values.empty else np.nan,
        "Maximum": numeric_values.max()
            if not numeric_values.empty else np.nan,
        "Mean": numeric_values.mean()
            if not numeric_values.empty else np.nan,
        "Valid": (
            np.isfinite(numeric_values).all()
            if not numeric_values.empty
            else False
        ),
    })

O_PRIVACY_QC_DF = pd.DataFrame(O_QC_ROWS)

print()
if O_PRIVACY_QC_DF.empty:
    print("No numeric privacy metrics available for quality control.")
else:
    display(O_PRIVACY_QC_DF)

print()
print("O6 — Privacy metric quality control completed.")
```

```python
# ==============================================================================
# O7 — PRIVACY COVERAGE MATRIX
# ==============================================================================

print("=" * 100)
print("O7 — PRIVACY COVERAGE MATRIX")
print("=" * 100)

O_COVERAGE_ROWS = []

for dataset in O_DATASETS:

    for method in O_METHODS:

        subset = O_SOURCE_DF.copy()

        if "Dataset" in subset.columns:
            subset = subset[
                subset["Dataset"].astype(str).str.lower()
                == dataset.lower()
            ]

        if "Method" in subset.columns:
            subset = subset[
                subset["Method"].astype(str).str.lower()
                == method.lower()
            ]

        metric_count = 0

        if not subset.empty:

            for column in O_PRIVACY_COLUMNS:

                if column in subset.columns:

                    values = pd.to_numeric(
                        subset[column],
                        errors="coerce"
                    )

                    metric_count += int(values.notna().sum())

        O_COVERAGE_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Privacy Metrics Available": metric_count,
            "Privacy Evidence Status": (
                "AVAILABLE"
                if metric_count > 0
                else "NOT AVAILABLE"
            ),
        })

O_PRIVACY_COVERAGE_DF = pd.DataFrame(O_COVERAGE_ROWS)

print()
display(O_PRIVACY_COVERAGE_DF)

print()
print("-" * 100)
print("PRIVACY COVERAGE SUMMARY")
print("-" * 100)

print(
    O_PRIVACY_COVERAGE_DF[
        "Privacy Evidence Status"
    ].value_counts(dropna=False)
)

print()
print("O7 — Privacy coverage analysis completed.")
```

```python
# ==============================================================================
# O8 — PRIVACY COMPARISON SUMMARY
# ==============================================================================

print("=" * 100)
print("O8 — PRIVACY COMPARISON SUMMARY")
print("=" * 100)

O_PRIVACY_SUMMARY_ROWS = []

for dataset in O_DATASETS:

    for method in O_METHODS:

        registry_row = O_DP_PARAMETER_REGISTRY_DF[
            (
                O_DP_PARAMETER_REGISTRY_DF["Dataset"]
                == dataset
            )
            &
            (
                O_DP_PARAMETER_REGISTRY_DF["Method"]
                == method
            )
        ]

        coverage_row = O_PRIVACY_COVERAGE_DF[
            (
                O_PRIVACY_COVERAGE_DF["Dataset"]
                == dataset
            )
            &
            (
                O_PRIVACY_COVERAGE_DF["Method"]
                == method
            )
        ]

        if registry_row.empty:
            epsilon = np.nan
            delta = np.nan
            noise = np.nan
        else:
            epsilon = registry_row.iloc[0]["Epsilon"]
            delta = registry_row.iloc[0]["Delta"]
            noise = registry_row.iloc[0]["Noise Multiplier"]

        evidence_status = (
            coverage_row.iloc[0]["Privacy Evidence Status"]
            if not coverage_row.empty
            else "NOT AVAILABLE"
        )

        O_PRIVACY_SUMMARY_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Privacy Mechanism": O_PRIVACY_CLASSIFICATION.get(
                method,
                "Not specified"
            ),
            "Epsilon": epsilon,
            "Delta": delta,
            "Noise Multiplier": noise,
            "Privacy Evidence": evidence_status,
        })

O_PRIVACY_SUMMARY_DF = pd.DataFrame(
    O_PRIVACY_SUMMARY_ROWS
)

display(O_PRIVACY_SUMMARY_DF)

print()
print("O8 — Privacy comparison summary completed.")
```

```python
# ==============================================================================
# O9 — PRIVACY INTERPRETATION RULES
# ==============================================================================

print("=" * 100)
print("O9 — PRIVACY INTERPRETATION RULES")
print("=" * 100)

O_PRIVACY_RULES = [
    {
        "Aspect": "Differential Privacy Epsilon",
        "Direction": "Lower is generally stronger privacy",
        "Interpretation": (
            "A smaller epsilon represents a tighter formal privacy guarantee "
            "when the same DP mechanism and accounting assumptions are used."
        ),
    },
    {
        "Aspect": "Differential Privacy Delta",
        "Direction": "Lower is generally stronger privacy",
        "Interpretation": (
            "Delta represents the probability component of an approximate "
            "differential-privacy guarantee."
        ),
    },
    {
        "Aspect": "Noise Multiplier",
        "Direction": "Higher generally implies stronger perturbation",
        "Interpretation": (
            "A larger noise multiplier generally introduces more noise, "
            "although privacy strength must ultimately be assessed using "
            "the complete privacy accounting configuration."
        ),
    },
    {
        "Aspect": "Re-identification / Disclosure Risk",
        "Direction": "Lower is better",
        "Interpretation": (
            "Lower empirical disclosure or re-identification risk indicates "
            "better resistance to privacy attacks."
        ),
    },
    {
        "Aspect": "Non-DP Methods",
        "Direction": "No formal DP guarantee",
        "Interpretation": (
            "CTGAN and TVAE must not be described as having formal "
            "differential-privacy guarantees unless independent DP "
            "mechanisms were actually applied."
        ),
    },
]

O_PRIVACY_RULES_DF = pd.DataFrame(
    O_PRIVACY_RULES
)

display(O_PRIVACY_RULES_DF)

print()
print(
    "Privacy conclusions will distinguish formal DP guarantees from "
    "empirical privacy-risk measurements."
)

print()
print("O9 — Privacy interpretation rules established.")
```

```python
# ==============================================================================
# O10 — FINAL PRIVACY ANALYSIS REPORT
# ==============================================================================

print("=" * 100)
print("O10 — FINAL PRIVACY ANALYSIS REPORT")
print("=" * 100)

print()
print("-" * 100)
print("PRIVACY ANALYSIS STATUS")
print("-" * 100)

O_TOTAL_COMBINATIONS = len(O_DATASETS) * len(O_METHODS)

O_AVAILABLE_PRIVACY = int(
    (
        O_PRIVACY_COVERAGE_DF[
            "Privacy Evidence Status"
        ]
        == "AVAILABLE"
    ).sum()
)

O_MISSING_PRIVACY = (
    O_TOTAL_COMBINATIONS
    - O_AVAILABLE_PRIVACY
)

print(f"Dataset × Method combinations : {O_TOTAL_COMBINATIONS}")
print(f"Privacy evidence available     : {O_AVAILABLE_PRIVACY}")
print(f"Privacy evidence unavailable   : {O_MISSING_PRIVACY}")

print()
print("-" * 100)
print("METHOD-LEVEL PRIVACY STATUS")
print("-" * 100)

for method in O_METHODS:

    method_subset = O_PRIVACY_COVERAGE_DF[
        O_PRIVACY_COVERAGE_DF["Method"] == method
    ]

    available = int(
        (
            method_subset["Privacy Evidence Status"]
            == "AVAILABLE"
        ).sum()
    )

    print(
        f"{method:<12} : "
        f"{available}/{len(O_DATASETS)} datasets with privacy evidence"
    )

print()
print("-" * 100)
print("RESEARCH INTERPRETATION")
print("-" * 100)

print(
    "1. Formal differential-privacy claims are restricted to experiments "
    "for which DP accounting parameters and/or validated DP evaluation "
    "evidence are available."
)

print(
    "2. CTGAN and TVAE are treated as non-DP baselines unless the validated "
    "experimental records demonstrate an explicitly applied privacy mechanism."
)

print(
    "3. DP-CTGAN privacy strength must be interpreted using its recorded "
    "epsilon, delta, noise multiplier, clipping and accounting configuration."
)

print(
    "4. SPP-GAN privacy claims must be supported by the actual privacy "
    "mechanism and validated privacy evidence rather than the method name alone."
)

print(
    "5. Missing privacy results remain explicitly marked as unavailable and "
    "are not replaced with assumed values."
)

print()
print("O10 — Final privacy analysis completed.")

# ------------------------------------------------------------------------------
# FINAL OBJECT
# ------------------------------------------------------------------------------

O_FINAL_PRIVACY_ANALYSIS_DF = O_PRIVACY_SUMMARY_DF.copy()

print()
print("=" * 100)
print("SECTION O — PRIVACY ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# N1 — STRUCTURAL ANALYSIS SCOPE
# ==============================================================================

print("=" * 100)
print("N1 — STRUCTURAL ANALYSIS SCOPE")
print("=" * 100)

N_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

N_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

N_STRUCTURAL_ATTRIBUTES = [
    "Rows",
    "Columns",
    "Column Names",
    "Column Order",
    "Data Types",
    "Numeric Columns",
    "Categorical Columns",
    "Missing Values",
    "Missingness Pattern",
    "Category Coverage",
    "Unique Values",
    "Numeric Ranges",
    "Duplicate Rows",
]

print()
print("-" * 100)
print("FINAL DATASETS")
print("-" * 100)

for dataset in N_DATASETS:
    print(f"  {dataset}")

print()
print("-" * 100)
print("FINAL METHODS")
print("-" * 100)

for method in N_METHODS:
    print(f"  {method}")

print()
print("-" * 100)
print("STRUCTURAL / STATISTICAL ATTRIBUTES")
print("-" * 100)

for attribute in N_STRUCTURAL_ATTRIBUTES:
    print(f"  {attribute}")

print()
print(
    "N1 objective: compare structural and statistical consistency "
    "between each available synthetic dataset and its corresponding "
    "real reference dataset."
)

print()
print("N1 — Structural analysis scope established.")
```

```python
# ==============================================================================
# N2 — RESOLVE REFERENCE DATASET REGISTRY
# ==============================================================================

print("=" * 100)
print("N2 — RESOLVE REFERENCE DATASET REGISTRY")
print("=" * 100)

# ------------------------------------------------------------------------------
# POSSIBLE REFERENCE DATAFRAME REGISTRIES
# ------------------------------------------------------------------------------

N_REFERENCE_REGISTRY = None
N_REFERENCE_REGISTRY_NAME = None

N_REFERENCE_CANDIDATES = [
    "REFERENCE_DATASETS",
    "REFERENCE_DATASET_REGISTRY",
    "FINAL_REFERENCE_DATASETS",
    "FINAL_REFERENCE_DATASET_REGISTRY",
    "REFERENCE_DATA",
    "REAL_DATASETS",
]

for candidate_name in N_REFERENCE_CANDIDATES:

    candidate = globals().get(candidate_name)

    if isinstance(candidate, dict):

        if all(
            dataset in candidate
            for dataset in N_DATASETS
        ):

            N_REFERENCE_REGISTRY = candidate
            N_REFERENCE_REGISTRY_NAME = candidate_name
            break

# ------------------------------------------------------------------------------
# FALLBACK: COMMON INDIVIDUAL DATAFRAME VARIABLES
# ------------------------------------------------------------------------------

if N_REFERENCE_REGISTRY is None:

    N_REFERENCE_REGISTRY = {}

    for dataset in N_DATASETS:

        candidate_names = [
            dataset,
            f"{dataset}_df",
            f"{dataset}_reference",
            f"{dataset}_reference_df",
            f"REFERENCE_{dataset.upper()}",
            f"REFERENCE_{dataset.upper()}_DF",
        ]

        for candidate_name in candidate_names:

            candidate = globals().get(candidate_name)

            if isinstance(candidate, pd.DataFrame):

                N_REFERENCE_REGISTRY[dataset] = candidate
                break

    if N_REFERENCE_REGISTRY:

        N_REFERENCE_REGISTRY_NAME = (
            "individual reference dataframe variables"
        )

# ------------------------------------------------------------------------------
# DISPLAY REGISTRY
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("REFERENCE DATASET REGISTRY")
print("-" * 100)

if N_REFERENCE_REGISTRY:

    for dataset in N_DATASETS:

        dataframe = N_REFERENCE_REGISTRY.get(dataset)

        if isinstance(dataframe, pd.DataFrame):

            print(
                f"{dataset:<20} "
                f"{len(dataframe):>10,} rows × "
                f"{len(dataframe.columns):>4,} columns"
            )

        else:

            print(
                f"{dataset:<20} NOT AVAILABLE"
            )

else:

    print("No reference dataset registry was resolved.")

print()
print(
    f"Registry source: "
    f"{N_REFERENCE_REGISTRY_NAME or 'NOT FOUND'}"
)

print()
print("N2 — Reference dataset registry resolution completed.")
```

```python
# ==============================================================================
# N3 — RESOLVE SYNTHETIC DATASET REGISTRY
# ==============================================================================

print("=" * 100)
print("N3 — RESOLVE SYNTHETIC DATASET REGISTRY")
print("=" * 100)

N_SYNTHETIC_REGISTRY = None
N_SYNTHETIC_REGISTRY_NAME = None

# ------------------------------------------------------------------------------
# PREFERRED E7 REGISTRY
# ------------------------------------------------------------------------------

N_SYNTHETIC_CANDIDATES = [
    "SYNTHETIC_DATASET_REGISTRY",
    "SYNTHETIC_DATASET_REGISTRY_TABLE",
    "FINAL_SYNTHETIC_DATASET_REGISTRY",
    "FINAL_SYNTHETIC_DATASETS",
]

for candidate_name in N_SYNTHETIC_CANDIDATES:

    candidate = globals().get(candidate_name)

    if isinstance(candidate, pd.DataFrame):

        required_columns = {
            "Dataset",
            "Method",
        }

        if required_columns.issubset(
            set(candidate.columns)
        ):

            N_SYNTHETIC_REGISTRY = candidate.copy()
            N_SYNTHETIC_REGISTRY_NAME = candidate_name
            break

# ------------------------------------------------------------------------------
# FALLBACK TO E4 LOAD RESULTS
# ------------------------------------------------------------------------------

if N_SYNTHETIC_REGISTRY is None:

    for candidate_name in [
        "SYNTHETIC_LOAD_RESULTS",
        "E4_SYNTHETIC_LOAD_RESULTS",
    ]:

        candidate = globals().get(candidate_name)

        if isinstance(candidate, pd.DataFrame):

            if {
                "Dataset",
                "Method",
            }.issubset(set(candidate.columns)):

                N_SYNTHETIC_REGISTRY = candidate.copy()
                N_SYNTHETIC_REGISTRY_NAME = candidate_name
                break

# ------------------------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("SYNTHETIC DATASET REGISTRY")
print("-" * 100)

if N_SYNTHETIC_REGISTRY is None:

    print("Synthetic dataset registry not available.")

else:

    print(
        f"Registry source: {N_SYNTHETIC_REGISTRY_NAME}"
    )

    display(
        N_SYNTHETIC_REGISTRY
    )

print()
print("N3 — Synthetic dataset registry resolution completed.")
```

```python
# ==============================================================================
# N4 — STRUCTURAL COMPARISON ENGINE
# ==============================================================================

print("=" * 100)
print("N4 — STRUCTURAL COMPARISON ENGINE")
print("=" * 100)

N_STRUCTURAL_RESULTS = []

# ------------------------------------------------------------------------------
# HELPER: GET SYNTHETIC DATAFRAME
# ------------------------------------------------------------------------------

def N_get_synthetic_dataframe(dataset, method):

    # --------------------------------------------------------------------------
    # SEARCH COMMON DICTIONARY REGISTRIES
    # --------------------------------------------------------------------------

    for registry_name in [
        "SYNTHETIC_DATASETS",
        "SYNTHETIC_DATASET_REGISTRY",
        "FINAL_SYNTHETIC_DATASETS",
        "LOADED_SYNTHETIC_DATASETS",
    ]:

        registry = globals().get(registry_name)

        if isinstance(registry, dict):

            dataset_object = registry.get(dataset)

            if isinstance(dataset_object, dict):

                candidate = dataset_object.get(method)

                if isinstance(candidate, pd.DataFrame):
                    return candidate

            elif isinstance(dataset_object, pd.DataFrame):

                # Only use directly if the registry itself is method-specific
                continue

    # --------------------------------------------------------------------------
    # SEARCH E7 REGISTRY FILE PATH
    # --------------------------------------------------------------------------

    if isinstance(
        N_SYNTHETIC_REGISTRY,
        pd.DataFrame
    ):

        subset = N_SYNTHETIC_REGISTRY[
            (
                N_SYNTHETIC_REGISTRY["Dataset"]
                .astype(str)
                .str.lower()
                == dataset.lower()
            )
            &
            (
                N_SYNTHETIC_REGISTRY["Method"]
                .astype(str)
                .str.lower()
                == method.lower()
            )
        ]

        if not subset.empty:

            row = subset.iloc[0]

            status = str(
                row.get("Status", "")
            ).upper()

            file_value = row.get(
                "File",
                None
            )

            if (
                status in [
                    "READY",
                    "LOADED",
                    "AVAILABLE",
                ]
                and isinstance(
                    file_value,
                    str
                )
                and pathlib.Path(
                    file_value
                ).is_file()
            ):

                try:
                    return pd.read_csv(
                        file_value
                    )
                except Exception:
                    return None

    return None


# ------------------------------------------------------------------------------
# STRUCTURAL COMPARISON
# ------------------------------------------------------------------------------

for dataset in N_DATASETS:

    real_df = N_REFERENCE_REGISTRY.get(
        dataset
    ) if N_REFERENCE_REGISTRY else None

    for method in N_METHODS:

        synthetic_df = N_get_synthetic_dataframe(
            dataset,
            method
        )

        result = {
            "Dataset": dataset,
            "Method": method,
            "Status": "UNAVAILABLE",
            "Rows Real": np.nan,
            "Rows Synthetic": np.nan,
            "Columns Real": np.nan,
            "Columns Synthetic": np.nan,
            "Column Names Match": False,
            "Column Order Match": False,
            "Data Types Match": False,
            "Numeric Columns Match": False,
            "Categorical Columns Match": False,
            "Missing Cells Real": np.nan,
            "Missing Cells Synthetic": np.nan,
            "Missingness Match": False,
            "Category Coverage": np.nan,
            "Unique Value Consistency": np.nan,
            "Numeric Range Consistency": np.nan,
            "Duplicate Rows Synthetic": np.nan,
        }

        if not isinstance(real_df, pd.DataFrame):

            N_STRUCTURAL_RESULTS.append(result)
            continue

        if not isinstance(
            synthetic_df,
            pd.DataFrame
        ):

            N_STRUCTURAL_RESULTS.append(result)
            continue

        result["Status"] = "AVAILABLE"

        # ----------------------------------------------------------------------
        # BASIC SHAPE
        # ----------------------------------------------------------------------

        result["Rows Real"] = len(real_df)
        result["Rows Synthetic"] = len(synthetic_df)

        result["Columns Real"] = len(real_df.columns)
        result["Columns Synthetic"] = len(
            synthetic_df.columns
        )

        # ----------------------------------------------------------------------
        # COLUMN NAMES
        # ----------------------------------------------------------------------

        result["Column Names Match"] = (
            set(real_df.columns)
            == set(synthetic_df.columns)
        )

        result["Column Order Match"] = (
            list(real_df.columns)
            == list(synthetic_df.columns)
        )

        # ----------------------------------------------------------------------
        # DATA TYPES
        # ----------------------------------------------------------------------

        common_columns = [
            column
            for column in real_df.columns
            if column in synthetic_df.columns
        ]

        if common_columns:

            result["Data Types Match"] = all(
                str(real_df[column].dtype)
                ==
                str(synthetic_df[column].dtype)
                for column in common_columns
            )

        # ----------------------------------------------------------------------
        # NUMERIC / CATEGORICAL COLUMNS
        # ----------------------------------------------------------------------

        real_numeric = set(
            real_df.select_dtypes(
                include=np.number
            ).columns
        )

        synthetic_numeric = set(
            synthetic_df.select_dtypes(
                include=np.number
            ).columns
        )

        result["Numeric Columns Match"] = (
            real_numeric
            == synthetic_numeric
        )

        real_categorical = set(
            real_df.select_dtypes(
                exclude=np.number
            ).columns
        )

        synthetic_categorical = set(
            synthetic_df.select_dtypes(
                exclude=np.number
            ).columns
        )

        result["Categorical Columns Match"] = (
            real_categorical
            == synthetic_categorical
        )

        # ----------------------------------------------------------------------
        # MISSING VALUES
        # ----------------------------------------------------------------------

        real_missing = int(
            real_df.isna().sum().sum()
        )

        synthetic_missing = int(
            synthetic_df.isna().sum().sum()
        )

        result["Missing Cells Real"] = real_missing
        result["Missing Cells Synthetic"] = (
            synthetic_missing
        )

        result["Missingness Match"] = (
            real_missing
            ==
            synthetic_missing
        )

        # ----------------------------------------------------------------------
        # CATEGORY COVERAGE
        # ----------------------------------------------------------------------

        categorical_columns = [
            column
            for column in real_df.columns
            if column in synthetic_df.columns
            and (
                column in real_categorical
            )
        ]

        if categorical_columns:

            coverage_values = []

            for column in categorical_columns:

                real_categories = set(
                    real_df[column]
                    .dropna()
                    .astype(str)
                    .unique()
                )

                synthetic_categories = set(
                    synthetic_df[column]
                    .dropna()
                    .astype(str)
                    .unique()
                )

                if real_categories:

                    coverage_values.append(
                        len(
                            real_categories
                            &
                            synthetic_categories
                        )
                        /
                        len(real_categories)
                    )

            if coverage_values:

                result["Category Coverage"] = (
                    float(
                        np.mean(
                            coverage_values
                        )
                    )
                )

        # ----------------------------------------------------------------------
        # UNIQUE VALUE CONSISTENCY
        # ----------------------------------------------------------------------

        unique_scores = []

        for column in common_columns:

            real_unique = (
                real_df[column]
                .nunique(dropna=True)
            )

            synthetic_unique = (
                synthetic_df[column]
                .nunique(dropna=True)
            )

            denominator = max(
                real_unique,
                1
            )

            unique_scores.append(
                1
                -
                (
                    abs(
                        real_unique
                        -
                        synthetic_unique
                    )
                    /
                    denominator
                )
            )

        if unique_scores:

            result["Unique Value Consistency"] = (
                float(
                    np.mean(
                        np.clip(
                            unique_scores,
                            0,
                            1
                        )
                    )
                )
            )

        # ----------------------------------------------------------------------
        # NUMERIC RANGE CONSISTENCY
        # ----------------------------------------------------------------------

        range_scores = []

        numeric_columns = [
            column
            for column in real_numeric
            if column in synthetic_df.columns
            and column in synthetic_numeric
        ]

        for column in numeric_columns:

            real_min = real_df[column].min()
            real_max = real_df[column].max()

            synthetic_min = (
                synthetic_df[column].min()
            )

            synthetic_max = (
                synthetic_df[column].max()
            )

            if pd.isna(real_min) or pd.isna(real_max):
                continue

            real_range = (
                real_max - real_min
            )

            if real_range == 0:

                range_scores.append(
                    1.0
                    if (
                        synthetic_min == real_min
                        and synthetic_max == real_max
                    )
                    else 0.0
                )

            else:

                lower_error = (
                    abs(
                        synthetic_min
                        -
                        real_min
                    )
                    /
                    abs(real_range)
                )

                upper_error = (
                    abs(
                        synthetic_max
                        -
                        real_max
                    )
                    /
                    abs(real_range)
                )

                range_scores.append(
                    max(
                        0.0,
                        1.0
                        -
                        (
                            lower_error
                            +
                            upper_error
                        )
                        /
                        2.0
                    )
                )

        if range_scores:

            result["Numeric Range Consistency"] = (
                float(
                    np.mean(
                        range_scores
                    )
                )
            )

        # ----------------------------------------------------------------------
        # DUPLICATES
        # ----------------------------------------------------------------------

        result["Duplicate Rows Synthetic"] = int(
            synthetic_df.duplicated().sum()
        )

        N_STRUCTURAL_RESULTS.append(
            result
        )

# ------------------------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------------------------

N_STRUCTURAL_RESULTS_DF = pd.DataFrame(
    N_STRUCTURAL_RESULTS
)

print()
print("-" * 100)
print("STRUCTURAL COMPARISON RESULTS")
print("-" * 100)

display(
    N_STRUCTURAL_RESULTS_DF
)

print()
print("N4 — Structural comparison engine completed.")
```

```python
# ==============================================================================
# N5 — STATISTICAL CONSISTENCY ANALYSIS
# ==============================================================================

print("=" * 100)
print("N5 — STATISTICAL CONSISTENCY ANALYSIS")
print("=" * 100)

N_STATISTICAL_RESULTS = []

# ------------------------------------------------------------------------------
# HELPER: NUMERICAL STATISTICAL COMPARISON
# ------------------------------------------------------------------------------

for dataset in N_DATASETS:

    real_df = (
        N_REFERENCE_REGISTRY.get(dataset)
        if N_REFERENCE_REGISTRY
        else None
    )

    if not isinstance(
        real_df,
        pd.DataFrame
    ):
        continue

    real_numeric_columns = list(
        real_df.select_dtypes(
            include=np.number
        ).columns
    )

    for method in N_METHODS:

        synthetic_df = N_get_synthetic_dataframe(
            dataset,
            method
        )

        result = {
            "Dataset": dataset,
            "Method": method,
            "Status": "UNAVAILABLE",
            "Numeric Features Compared": 0,
            "Mean Absolute Relative Difference": np.nan,
            "Median Absolute Relative Difference": np.nan,
            "Std Absolute Relative Difference": np.nan,
            "Min Absolute Relative Difference": np.nan,
            "Max Absolute Relative Difference": np.nan,
        }

        if not isinstance(
            synthetic_df,
            pd.DataFrame
        ):
            N_STATISTICAL_RESULTS.append(result)
            continue

        common_numeric = [
            column
            for column in real_numeric_columns
            if column in synthetic_df.columns
            and pd.api.types.is_numeric_dtype(
                synthetic_df[column]
            )
        ]

        result[
            "Numeric Features Compared"
        ] = len(common_numeric)

        if not common_numeric:

            N_STATISTICAL_RESULTS.append(result)
            continue

        relative_errors = []

        for column in common_numeric:

            real_mean = real_df[column].mean()
            synthetic_mean = (
                synthetic_df[column].mean()
            )

            real_std = real_df[column].std()
            synthetic_std = (
                synthetic_df[column].std()
            )

            # Mean difference
            if pd.notna(real_mean):

                denominator = max(
                    abs(real_mean),
                    1e-12
                )

                relative_errors.append(
                    abs(
                        synthetic_mean
                        -
                        real_mean
                    )
                    /
                    denominator
                )

            # Standard deviation difference
            if pd.notna(real_std):

                denominator = max(
                    abs(real_std),
                    1e-12
                )

                relative_errors.append(
                    abs(
                        synthetic_std
                        -
                        real_std
                    )
                    /
                    denominator
                )

        if relative_errors:

            errors = np.asarray(
                relative_errors,
                dtype=float
            )

            errors = errors[
                np.isfinite(errors)
            ]

            if len(errors):

                result["Status"] = "AVAILABLE"

                result[
                    "Mean Absolute Relative Difference"
                ] = float(
                    np.mean(errors)
                )

                result[
                    "Median Absolute Relative Difference"
                ] = float(
                    np.median(errors)
                )

                result[
                    "Std Absolute Relative Difference"
                ] = float(
                    np.std(errors)
                )

                result[
                    "Min Absolute Relative Difference"
                ] = float(
                    np.min(errors)
                )

                result[
                    "Max Absolute Relative Difference"
                ] = float(
                    np.max(errors)
                )

        N_STATISTICAL_RESULTS.append(
            result
        )

# ------------------------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------------------------

N_STATISTICAL_RESULTS_DF = pd.DataFrame(
    N_STATISTICAL_RESULTS
)

print()
print("-" * 100)
print("STATISTICAL CONSISTENCY RESULTS")
print("-" * 100)

display(
    N_STATISTICAL_RESULTS_DF
)

print()
print(
    "Lower relative differences indicate greater statistical consistency."
)

print()
print("N5 — Statistical consistency analysis completed.")
```

```python
# ==============================================================================
# N6 — STRUCTURAL CONSISTENCY SCORE
# ==============================================================================

print("=" * 100)
print("N6 — STRUCTURAL CONSISTENCY SCORE")
print("=" * 100)

N_SCORE_COLUMNS = [
    "Column Names Match",
    "Column Order Match",
    "Data Types Match",
    "Numeric Columns Match",
    "Categorical Columns Match",
    "Missingness Match",
]

N_SCORE_ROWS = []

for _, row in N_STRUCTURAL_RESULTS_DF.iterrows():

    if row["Status"] != "AVAILABLE":

        score = np.nan

    else:

        binary_scores = [
            float(bool(row[column]))
            for column in N_SCORE_COLUMNS
        ]

        continuous_scores = []

        for column in [
            "Category Coverage",
            "Unique Value Consistency",
            "Numeric Range Consistency",
        ]:

            value = row[column]

            if pd.notna(value):

                continuous_scores.append(
                    float(
                        np.clip(
                            value,
                            0,
                            1
                        )
                    )
                )

        all_scores = (
            binary_scores
            +
            continuous_scores
        )

        score = (
            float(
                np.mean(all_scores)
            )
            if all_scores
            else np.nan
        )

    N_SCORE_ROWS.append({
        "Dataset": row["Dataset"],
        "Method": row["Method"],
        "Status": row["Status"],
        "Structural Consistency Score": score,
    })

N_STRUCTURAL_SCORE_DF = pd.DataFrame(
    N_SCORE_ROWS
)

display(
    N_STRUCTURAL_SCORE_DF
)

print()
print(
    "Score interpretation: 1.0 represents complete consistency "
    "across the available structural checks."
)

print()
print("N6 — Structural consistency scoring completed.")
```

```python
# ==============================================================================
# N7 — STATISTICAL CONSISTENCY SCORE
# ==============================================================================

print("=" * 100)
print("N7 — STATISTICAL CONSISTENCY SCORE")
print("=" * 100)

N_STATISTICAL_SCORE_ROWS = []

for _, row in N_STATISTICAL_RESULTS_DF.iterrows():

    error = row[
        "Mean Absolute Relative Difference"
    ]

    if pd.isna(error):

        score = np.nan

    else:

        # Convert relative error to bounded consistency score.
        score = max(
            0.0,
            1.0
            -
            float(error)
        )

        score = min(
            1.0,
            score
        )

    N_STATISTICAL_SCORE_ROWS.append({
        "Dataset": row["Dataset"],
        "Method": row["Method"],
        "Status": row["Status"],
        "Statistical Consistency Score": score,
    })

N_STATISTICAL_SCORE_DF = pd.DataFrame(
    N_STATISTICAL_SCORE_ROWS
)

display(
    N_STATISTICAL_SCORE_DF
)

print()
print(
    "Higher scores indicate smaller aggregate relative "
    "differences in numerical descriptive statistics."
)

print()
print("N7 — Statistical consistency scoring completed.")
```

```python
# ==============================================================================
# N8 — FINAL STRUCTURAL & STATISTICAL CONSISTENCY MATRIX
# ==============================================================================

print("=" * 100)
print("N8 — FINAL STRUCTURAL & STATISTICAL CONSISTENCY MATRIX")
print("=" * 100)

N_FINAL_ROWS = []

for dataset in N_DATASETS:

    for method in N_METHODS:

        structural = N_STRUCTURAL_SCORE_DF[
            (
                N_STRUCTURAL_SCORE_DF["Dataset"]
                == dataset
            )
            &
            (
                N_STRUCTURAL_SCORE_DF["Method"]
                == method
            )
        ]

        statistical = N_STATISTICAL_SCORE_DF[
            (
                N_STATISTICAL_SCORE_DF["Dataset"]
                == dataset
            )
            &
            (
                N_STATISTICAL_SCORE_DF["Method"]
                == method
            )
        ]

        structural_score = np.nan
        statistical_score = np.nan

        if not structural.empty:

            structural_score = structural.iloc[0][
                "Structural Consistency Score"
            ]

        if not statistical.empty:

            statistical_score = statistical.iloc[0][
                "Statistical Consistency Score"
            ]

        available_scores = [
            score
            for score in [
                structural_score,
                statistical_score,
            ]
            if pd.notna(score)
        ]

        combined_score = (
            float(
                np.mean(
                    available_scores
                )
            )
            if available_scores
            else np.nan
        )

        status = (
            "AVAILABLE"
            if available_scores
            else "UNAVAILABLE"
        )

        N_FINAL_ROWS.append({

            "Dataset": dataset,

            "Method": method,

            "Status": status,

            "Structural Consistency Score":
                structural_score,

            "Statistical Consistency Score":
                statistical_score,

            "Combined Structural-Statistical Score":
                combined_score,
        })

N_FINAL_CONSISTENCY_DF = pd.DataFrame(
    N_FINAL_ROWS
)

print()
display(
    N_FINAL_CONSISTENCY_DF
)

print()
print("N8 — Final consistency matrix created.")
```

```python
# ==============================================================================
# N9 — STRUCTURAL & STATISTICAL COVERAGE AUDIT
# ==============================================================================

print("=" * 100)
print("N9 — STRUCTURAL & STATISTICAL COVERAGE AUDIT")
print("=" * 100)

N_EXPECTED = (
    len(N_DATASETS)
    *
    len(N_METHODS)
)

N_AVAILABLE = int(
    (
        N_FINAL_CONSISTENCY_DF["Status"]
        == "AVAILABLE"
    ).sum()
)

N_UNAVAILABLE = (
    N_EXPECTED
    -
    N_AVAILABLE
)

print()
print("-" * 100)
print("COVERAGE")
print("-" * 100)

print(
    f"Expected Dataset × Method combinations : {N_EXPECTED}"
)

print(
    f"Available comparisons                  : {N_AVAILABLE}"
)

print(
    f"Unavailable comparisons                : {N_UNAVAILABLE}"
)

print(
    f"Coverage percentage                    : "
    f"{(N_AVAILABLE / N_EXPECTED * 100):.2f}%"
    if N_EXPECTED
    else "Coverage percentage                    : 0.00%"
)

print()
print("-" * 100)
print("UNAVAILABLE COMBINATIONS")
print("-" * 100)

N_UNAVAILABLE_DF = N_FINAL_CONSISTENCY_DF[
    N_FINAL_CONSISTENCY_DF["Status"]
    == "UNAVAILABLE"
].copy()

if N_UNAVAILABLE_DF.empty:

    print("None.")

else:

    display(
        N_UNAVAILABLE_DF[
            [
                "Dataset",
                "Method",
                "Status",
            ]
        ]
    )

print()
print("N9 — Structural and statistical coverage audit completed.")
```

```python
# ==============================================================================
# N10 — FINAL SECTION N RESEARCH INTERPRETATION
# ==============================================================================

print("=" * 100)
print("N10 — FINAL SECTION N RESEARCH INTERPRETATION")
print("=" * 100)

print()
print("-" * 100)
print("INTERPRETATION RULES")
print("-" * 100)

print(
    "1. Structural consistency evaluates whether synthetic datasets preserve "
    "the schema and fundamental structural characteristics of the reference data."
)

print(
    "2. Column names, column order, data types, numerical/categorical structure, "
    "missingness, category coverage, uniqueness and numerical ranges are evaluated."
)

print(
    "3. Statistical consistency evaluates preservation of numerical descriptive "
    "properties using relative differences in means and standard deviations."
)

print(
    "4. Higher Structural Consistency Score indicates greater structural agreement."
)

print(
    "5. Higher Statistical Consistency Score indicates smaller statistical "
    "differences from the reference dataset."
)

print(
    "6. Combined Structural-Statistical Score is descriptive and should not "
    "replace the underlying individual metrics in publication reporting."
)

print(
    "7. Missing experiments remain explicitly marked as UNAVAILABLE and are "
    "not assigned artificial scores."
)

print()
print("-" * 100)
print("FINAL SECTION N STATUS")
print("-" * 100)

print(
    f"Expected comparisons : {N_EXPECTED}"
)

print(
    f"Available comparisons: {N_AVAILABLE}"
)

print(
    f"Unavailable          : {N_UNAVAILABLE}"
)

print()
print("=" * 100)
print("SECTION N — STRUCTURAL & STATISTICAL CONSISTENCY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O1 — DOWNSTREAM EVALUATION DATA DISCOVERY
# ==============================================================================

print("=" * 100)
print("O1 — DOWNSTREAM EVALUATION DATA DISCOVERY")
print("=" * 100)

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# REQUIRED PROJECT DIRECTORIES
# ------------------------------------------------------------------------------

if "PROJECT_ROOT" not in globals():
    raise RuntimeError(
        "PROJECT_ROOT is not defined. Run Section B before Section O."
    )

PROJECT_ROOT = Path(PROJECT_ROOT)

SEARCH_DIRECTORIES = []

for directory_name in [
    "EVALUATION_DIRECTORY",
    "RESULTS_DIRECTORY",
    "REPORTS_DIRECTORY",
]:
    if directory_name in globals():
        directory = globals()[directory_name]
        if directory is not None:
            directory = Path(directory)
            if directory.exists():
                SEARCH_DIRECTORIES.append(directory)

# Add project root as fallback
if PROJECT_ROOT.exists():
    SEARCH_DIRECTORIES.append(PROJECT_ROOT)

# Remove duplicate directories
SEARCH_DIRECTORIES = list(dict.fromkeys(SEARCH_DIRECTORIES))

# ------------------------------------------------------------------------------
# FINAL RESEARCH SCOPE
# ------------------------------------------------------------------------------

FINAL_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

FINAL_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

DATASET_TOKENS = {
    "adult_income": [
        "adult_income",
        "adult income",
    ],
    "bank_marketing": [
        "bank_marketing",
        "bank marketing",
    ],
    "diabetes_130us": [
        "diabetes_130us",
        "diabetes 130us",
        "diabetes",
    ],
}

METHOD_TOKENS = {
    "CTGAN": ["ctgan"],
    "TVAE": ["tvae"],
    "DP-CTGAN": ["dp_ctgan", "dp-ctgan", "dpctgan"],
    "SPP-GAN": ["spp_gan", "spp-gan", "sppgan"],
}

UTILITY_TERMS = [
    "downstream",
    "utility",
    "classification",
    "prediction",
    "predictive",
    "ml",
    "machine_learning",
    "model",
    "auc",
    "accuracy",
    "f1",
    "precision",
    "recall",
    "roc",
    "pr_auc",
    "utility_evaluation",
]

# ------------------------------------------------------------------------------
# DISCOVER CSV FILES
# ------------------------------------------------------------------------------

DISCOVERED_UTILITY_FILES = []

for search_directory in SEARCH_DIRECTORIES:

    try:
        csv_files = list(search_directory.rglob("*.csv"))
    except Exception:
        continue

    for file_path in csv_files:

        filename_lower = file_path.name.lower()
        path_lower = str(file_path).lower()

        utility_score = sum(
            term.lower() in filename_lower or term.lower() in path_lower
            for term in UTILITY_TERMS
        )

        if utility_score == 0:
            continue

        dataset_matches = []

        for dataset, tokens in DATASET_TOKENS.items():
            if any(
                token.lower() in filename_lower or
                token.lower() in path_lower
                for token in tokens
            ):
                dataset_matches.append(dataset)

        method_matches = []

        for method, tokens in METHOD_TOKENS.items():
            if any(
                token.lower() in filename_lower or
                token.lower() in path_lower
                for token in tokens
            ):
                method_matches.append(method)

        if dataset_matches and method_matches:

            DISCOVERED_UTILITY_FILES.append({
                "File": str(file_path),
                "Dataset Matches": ", ".join(dataset_matches),
                "Method Matches": ", ".join(method_matches),
                "Utility Score": utility_score,
                "Size MB": round(
                    file_path.stat().st_size / (1024 ** 2),
                    3
                ),
            })

# ------------------------------------------------------------------------------
# REMOVE DUPLICATES
# ------------------------------------------------------------------------------

if DISCOVERED_UTILITY_FILES:

    O1_DISCOVERY_DF = (
        pd.DataFrame(DISCOVERED_UTILITY_FILES)
        .drop_duplicates(subset=["File"])
        .sort_values(
            ["Utility Score", "File"],
            ascending=[False, True]
        )
        .reset_index(drop=True)
    )

else:

    O1_DISCOVERY_DF = pd.DataFrame(
        columns=[
            "File",
            "Dataset Matches",
            "Method Matches",
            "Utility Score",
            "Size MB",
        ]
    )

# ------------------------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("DISCOVERED DOWNSTREAM UTILITY FILES")
print("-" * 100)

if O1_DISCOVERY_DF.empty:

    print("No downstream utility result files were discovered.")

else:

    display(O1_DISCOVERY_DF)

print()
print("-" * 100)
print("O1 SUMMARY")
print("-" * 100)

print(
    f"Potential downstream utility files : "
    f"{len(O1_DISCOVERY_DF)}"
)

print()
print("=" * 100)
print("O1 — DOWNSTREAM EVALUATION DATA DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O2 — DOWNSTREAM METRIC VALIDATION
# ==============================================================================

print("=" * 100)
print("O2 — DOWNSTREAM METRIC VALIDATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# METRIC ALIASES
# ------------------------------------------------------------------------------

UTILITY_METRIC_ALIASES = {
    "accuracy": [
        "accuracy",
        "acc",
        "test_accuracy",
        "real_accuracy",
        "synthetic_accuracy",
    ],
    "precision": [
        "precision",
        "test_precision",
        "real_precision",
        "synthetic_precision",
    ],
    "recall": [
        "recall",
        "sensitivity",
        "tpr",
        "test_recall",
        "real_recall",
        "synthetic_recall",
    ],
    "f1": [
        "f1",
        "f1_score",
        "f1score",
        "test_f1",
        "real_f1",
        "synthetic_f1",
    ],
    "roc_auc": [
        "roc_auc",
        "roc_auc_score",
        "auc",
        "test_auc",
        "real_auc",
        "synthetic_auc",
    ],
    "pr_auc": [
        "pr_auc",
        "average_precision",
        "average_precision_score",
        "auprc",
        "pr_auc_score",
    ],
    "balanced_accuracy": [
        "balanced_accuracy",
        "balanced_acc",
    ],
}

def normalize_column_name(column):
    return (
        str(column)
        .strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
    )

def detect_utility_metrics(dataframe):

    normalized_columns = {
        normalize_column_name(column): column
        for column in dataframe.columns
    }

    detected = {}

    for metric, aliases in UTILITY_METRIC_ALIASES.items():

        for alias in aliases:

            alias_normalized = normalize_column_name(alias)

            if alias_normalized in normalized_columns:

                detected[metric] = normalized_columns[
                    alias_normalized
                ]

                break

    return detected

# ------------------------------------------------------------------------------
# VALIDATE DISCOVERED FILES
# ------------------------------------------------------------------------------

O2_VALIDATED_FILES = []

for _, row in O1_DISCOVERY_DF.iterrows():

    file_path = Path(row["File"])

    try:

        dataframe = pd.read_csv(file_path)

        metrics = detect_utility_metrics(dataframe)

        if metrics:

            O2_VALIDATED_FILES.append({
                "File": str(file_path),
                "Dataset Matches": row["Dataset Matches"],
                "Method Matches": row["Method Matches"],
                "Rows": len(dataframe),
                "Columns": len(dataframe.columns),
                "Detected Metrics": ", ".join(metrics.keys()),
                "Metric Count": len(metrics),
                "Status": "VALID",
            })

        else:

            O2_VALIDATED_FILES.append({
                "File": str(file_path),
                "Dataset Matches": row["Dataset Matches"],
                "Method Matches": row["Method Matches"],
                "Rows": len(dataframe),
                "Columns": len(dataframe.columns),
                "Detected Metrics": "",
                "Metric Count": 0,
                "Status": "NO_UTILITY_METRICS",
            })

    except Exception as error:

        O2_VALIDATED_FILES.append({
            "File": str(file_path),
            "Dataset Matches": row["Dataset Matches"],
            "Method Matches": row["Method Matches"],
            "Rows": None,
            "Columns": None,
            "Detected Metrics": "",
            "Metric Count": 0,
            "Status": f"READ_ERROR: {type(error).__name__}",
        })

O2_VALIDATION_DF = pd.DataFrame(O2_VALIDATED_FILES)

print()
print("-" * 100)
print("DOWNSTREAM METRIC VALIDATION")
print("-" * 100)

if O2_VALIDATION_DF.empty:
    print("No candidate downstream files available for validation.")
else:
    display(O2_VALIDATION_DF)

print()
print("-" * 100)
print("O2 SUMMARY")
print("-" * 100)

if not O2_VALIDATION_DF.empty:

    print(
        O2_VALIDATION_DF["Status"]
        .value_counts()
        .to_string()
    )

print()
print("=" * 100)
print("O2 — DOWNSTREAM METRIC VALIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O3 — BUILD DOWNSTREAM UTILITY RESULTS SUBSET
# ==============================================================================

print("=" * 100)
print("O3 — BUILD DOWNSTREAM UTILITY RESULTS SUBSET")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALID FILES ONLY
# ------------------------------------------------------------------------------

O3_VALID_FILES = (
    O2_VALIDATION_DF[
        O2_VALIDATION_DF["Status"] == "VALID"
    ]
    .copy()
    if not O2_VALIDATION_DF.empty
    else pd.DataFrame()
)

# ------------------------------------------------------------------------------
# BUILD LONG-FORM UTILITY DATAFRAME
# ------------------------------------------------------------------------------

O3_ROWS = []

for _, validation_row in O3_VALID_FILES.iterrows():

    file_path = Path(validation_row["File"])

    try:

        dataframe = pd.read_csv(file_path)

    except Exception:

        continue

    detected_metrics = detect_utility_metrics(dataframe)

    # Determine dataset
    dataset = None

    for candidate_dataset in FINAL_DATASETS:

        if candidate_dataset in validation_row["Dataset Matches"]:
            dataset = candidate_dataset
            break

    # Determine method
    method = None

    for candidate_method in FINAL_METHODS:

        if candidate_method in validation_row["Method Matches"]:
            method = candidate_method
            break

    if dataset is None or method is None:
        continue

    for metric, column in detected_metrics.items():

        values = pd.to_numeric(
            dataframe[column],
            errors="coerce"
        ).dropna()

        if values.empty:
            continue

        for value in values:

            O3_ROWS.append({
                "Dataset": dataset,
                "Method": method,
                "Metric": metric,
                "Value": float(value),
                "Source File": str(file_path),
            })

O3_UTILITY_RESULTS = pd.DataFrame(O3_ROWS)

# ------------------------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("DOWNSTREAM UTILITY RESULTS SUBSET")
print("-" * 100)

if O3_UTILITY_RESULTS.empty:

    print(
        "No validated downstream utility observations were "
        "available."
    )

else:

    display(
        O3_UTILITY_RESULTS.head(50)
    )

print()
print("-" * 100)
print("O3 SUMMARY")
print("-" * 100)

print(
    f"Utility observations : "
    f"{len(O3_UTILITY_RESULTS)}"
)

if not O3_UTILITY_RESULTS.empty:

    print(
        f"Datasets : "
        f"{O3_UTILITY_RESULTS['Dataset'].nunique()}"
    )

    print(
        f"Methods : "
        f"{O3_UTILITY_RESULTS['Method'].nunique()}"
    )

    print(
        f"Metrics : "
        f"{O3_UTILITY_RESULTS['Metric'].nunique()}"
    )

print()
print("=" * 100)
print("O3 — DOWNSTREAM UTILITY RESULTS SUBSET COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O4 — DATASET × METHOD UTILITY MATRIX
# ==============================================================================

print("=" * 100)
print("O4 — DATASET × METHOD UTILITY MATRIX")
print("=" * 100)

if O3_UTILITY_RESULTS.empty:

    O4_UTILITY_MATRIX = pd.DataFrame()

else:

    O4_UTILITY_MATRIX = (
        O3_UTILITY_RESULTS
        .groupby(
            ["Dataset", "Method", "Metric"],
            as_index=False
        )["Value"]
        .mean()
        .pivot_table(
            index=["Dataset", "Method"],
            columns="Metric",
            values="Value",
            aggfunc="mean"
        )
        .reset_index()
    )

print()
print("-" * 100)
print("DOWNSTREAM UTILITY MATRIX")
print("-" * 100)

if O4_UTILITY_MATRIX.empty:
    print("No utility matrix could be constructed.")
else:
    display(O4_UTILITY_MATRIX)

print()
print("=" * 100)
print("O4 — DATASET × METHOD UTILITY MATRIX COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O5 — DOWNSTREAM UTILITY COMPARISON
# ==============================================================================

print("=" * 100)
print("O5 — DOWNSTREAM UTILITY COMPARISON")
print("=" * 100)

if O3_UTILITY_RESULTS.empty:

    O5_UTILITY_COMPARISON = pd.DataFrame()

else:

    O5_UTILITY_COMPARISON = (
        O3_UTILITY_RESULTS
        .groupby(
            ["Dataset", "Method", "Metric"],
            as_index=False
        )
        .agg(
            Mean=("Value", "mean"),
            Std=("Value", "std"),
            Minimum=("Value", "min"),
            Maximum=("Value", "max"),
            Observations=("Value", "count"),
        )
    )

    O5_UTILITY_COMPARISON["Std"] = (
        O5_UTILITY_COMPARISON["Std"]
        .fillna(0.0)
    )

display(O5_UTILITY_COMPARISON)

print()
print("=" * 100)
print("O5 — DOWNSTREAM UTILITY COMPARISON COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O6 — BEST-METHOD ANALYSIS
# ==============================================================================

print("=" * 100)
print("O6 — BEST-METHOD ANALYSIS")
print("=" * 100)

if O5_UTILITY_COMPARISON.empty:

    O6_BEST_METHODS = pd.DataFrame()

else:

    O6_BEST_METHODS = []

    for metric in O5_UTILITY_COMPARISON["Metric"].unique():

        metric_data = O5_UTILITY_COMPARISON[
            O5_UTILITY_COMPARISON["Metric"] == metric
        ].copy()

        for dataset in FINAL_DATASETS:

            dataset_data = metric_data[
                metric_data["Dataset"] == dataset
            ].copy()

            if dataset_data.empty:
                continue

            best_index = dataset_data["Mean"].idxmax()

            best_row = dataset_data.loc[best_index]

            O6_BEST_METHODS.append({
                "Dataset": dataset,
                "Metric": metric,
                "Best Method": best_row["Method"],
                "Best Mean": best_row["Mean"],
                "Best Std": best_row["Std"],
                "Methods Available": len(dataset_data),
            })

    O6_BEST_METHODS = pd.DataFrame(O6_BEST_METHODS)

print()
print("-" * 100)
print("BEST DOWNSTREAM UTILITY METHOD BY DATASET AND METRIC")
print("-" * 100)

if O6_BEST_METHODS.empty:
    print("No best-method results available.")
else:
    display(O6_BEST_METHODS)

print()
print("=" * 100)
print("O6 — BEST-METHOD ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O7 — CROSS-DATASET UTILITY ROBUSTNESS
# ==============================================================================

print("=" * 100)
print("O7 — CROSS-DATASET UTILITY ROBUSTNESS")
print("=" * 100)

if O5_UTILITY_COMPARISON.empty:

    O7_ROBUSTNESS = pd.DataFrame()

else:

    O7_ROBUSTNESS = (
        O5_UTILITY_COMPARISON
        .groupby(
            ["Method", "Metric"],
            as_index=False
        )
        .agg(
            Datasets_Evaluated=("Dataset", "nunique"),
            Mean_Utility=("Mean", "mean"),
            Std_Across_Datasets=("Mean", "std"),
            Minimum_Utility=("Mean", "min"),
            Maximum_Utility=("Mean", "max"),
        )
    )

    O7_ROBUSTNESS["Std_Across_Datasets"] = (
        O7_ROBUSTNESS["Std_Across_Datasets"]
        .fillna(0.0)
    )

print()
print("-" * 100)
print("CROSS-DATASET DOWNSTREAM UTILITY ROBUSTNESS")
print("-" * 100)

if O7_ROBUSTNESS.empty:
    print("No cross-dataset utility results available.")
else:
    display(O7_ROBUSTNESS)

print()
print("=" * 100)
print("O7 — CROSS-DATASET UTILITY ROBUSTNESS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# O8 — DOWNSTREAM UTILITY RESEARCH SUMMARY
# ==============================================================================

print("=" * 100)
print("O8 — DOWNSTREAM UTILITY RESEARCH SUMMARY")
print("=" * 100)

O8_SUMMARY = {
    "Final Datasets": FINAL_DATASETS,
    "Final Methods": FINAL_METHODS,
    "Validated Utility Files": (
        len(O3_VALID_FILES)
        if "O3_VALID_FILES" in globals()
        else 0
    ),
    "Utility Observations": (
        len(O3_UTILITY_RESULTS)
        if "O3_UTILITY_RESULTS" in globals()
        else 0
    ),
    "Utility Metrics": (
        O3_UTILITY_RESULTS["Metric"].nunique()
        if not O3_UTILITY_RESULTS.empty
        else 0
    ),
    "Datasets With Utility Results": (
        O3_UTILITY_RESULTS["Dataset"].nunique()
        if not O3_UTILITY_RESULTS.empty
        else 0
    ),
    "Methods With Utility Results": (
        O3_UTILITY_RESULTS["Method"].nunique()
        if not O3_UTILITY_RESULTS.empty
        else 0
    ),
}

print()
print("-" * 100)
print("DOWNSTREAM UTILITY RESEARCH STATUS")
print("-" * 100)

for key, value in O8_SUMMARY.items():
    print(f"{key:<35}: {value}")

print()
print("-" * 100)
print("RESEARCH INTERPRETATION RULE")
print("-" * 100)

print(
    "Standard predictive utility metrics such as Accuracy, "
    "Precision, Recall, F1, ROC-AUC, PR-AUC, and Balanced "
    "Accuracy are interpreted as higher-is-better."
)

print(
    "Methods are compared only when validated downstream "
    "utility results are available."
)

print(
    "Unavailable or failed experiments are not converted into "
    "zero utility and are not treated as successful results."
)

print()
print("=" * 100)
print("O8 — DOWNSTREAM UTILITY RESEARCH SUMMARY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P1 — PRIVACY RESULTS DISCOVERY
# ==============================================================================

print("=" * 100)
print("P1 — PRIVACY RESULTS DISCOVERY")
print("=" * 100)

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# PROJECT DIRECTORIES
# ------------------------------------------------------------------------------

if "PROJECT_ROOT" not in globals():
    raise RuntimeError(
        "PROJECT_ROOT is not defined. Run Section B before Section P."
    )

PROJECT_ROOT = Path(PROJECT_ROOT)

P_SEARCH_DIRECTORIES = []

for directory_name in [
    "EVALUATION_DIRECTORY",
    "RESULTS_DIRECTORY",
    "REPORTS_DIRECTORY",
]:
    if directory_name in globals():
        directory = globals()[directory_name]

        if directory is not None:
            directory = Path(directory)

            if directory.exists():
                P_SEARCH_DIRECTORIES.append(directory)

P_SEARCH_DIRECTORIES.append(PROJECT_ROOT)

P_SEARCH_DIRECTORIES = list(
    dict.fromkeys(P_SEARCH_DIRECTORIES)
)

# ------------------------------------------------------------------------------
# FINAL RESEARCH SCOPE
# ------------------------------------------------------------------------------

P_FINAL_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

P_FINAL_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

P_DATASET_TOKENS = {
    "adult_income": [
        "adult_income",
        "adult income",
    ],
    "bank_marketing": [
        "bank_marketing",
        "bank marketing",
    ],
    "diabetes_130us": [
        "diabetes_130us",
        "diabetes 130us",
    ],
}

P_METHOD_TOKENS = {
    "CTGAN": [
        "ctgan",
    ],
    "TVAE": [
        "tvae",
    ],
    "DP-CTGAN": [
        "dp_ctgan",
        "dp-ctgan",
        "dpctgan",
    ],
    "SPP-GAN": [
        "spp_gan",
        "spp-gan",
        "sppgan",
    ],
}

P_PRIVACY_TERMS = [
    "privacy",
    "private",
    "dp",
    "differential",
    "epsilon",
    "delta",
    "noise",
    "accounting",
    "privacy_loss",
    "membership",
    "linkage",
    "disclosure",
    "risk",
    "attack",
    "identification",
    "distance",
]

# ------------------------------------------------------------------------------
# DISCOVERY
# ------------------------------------------------------------------------------

P_DISCOVERED_FILES = []

for search_directory in P_SEARCH_DIRECTORIES:

    try:
        candidate_files = list(search_directory.rglob("*"))
    except Exception:
        continue

    for file_path in candidate_files:

        if not file_path.is_file():
            continue

        filename_lower = file_path.name.lower()
        path_lower = str(file_path).lower()

        privacy_score = sum(
            term in filename_lower or term in path_lower
            for term in P_PRIVACY_TERMS
        )

        if privacy_score == 0:
            continue

        dataset_matches = []

        for dataset, tokens in P_DATASET_TOKENS.items():

            if any(
                token.lower() in filename_lower
                or token.lower() in path_lower
                for token in tokens
            ):
                dataset_matches.append(dataset)

        method_matches = []

        for method, tokens in P_METHOD_TOKENS.items():

            if any(
                token.lower() in filename_lower
                or token.lower() in path_lower
                for token in tokens
            ):
                method_matches.append(method)

        if dataset_matches or method_matches:

            P_DISCOVERED_FILES.append({
                "File": str(file_path),
                "Extension": file_path.suffix.lower(),
                "Dataset Matches": ", ".join(dataset_matches),
                "Method Matches": ", ".join(method_matches),
                "Privacy Score": privacy_score,
                "Size MB": round(
                    file_path.stat().st_size / (1024 ** 2),
                    3
                ),
            })

# ------------------------------------------------------------------------------
# REMOVE DUPLICATES
# ------------------------------------------------------------------------------

P_DISCOVERY_DF = (
    pd.DataFrame(P_DISCOVERED_FILES)
    .drop_duplicates(subset=["File"])
    .sort_values(
        ["Privacy Score", "File"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
    if P_DISCOVERED_FILES
    else pd.DataFrame(
        columns=[
            "File",
            "Extension",
            "Dataset Matches",
            "Method Matches",
            "Privacy Score",
            "Size MB",
        ]
    )
)

print()
print("-" * 100)
print("DISCOVERED PRIVACY-RELATED RESEARCH FILES")
print("-" * 100)

if P_DISCOVERY_DF.empty:
    print("No privacy-related research artifacts were discovered.")
else:
    display(P_DISCOVERY_DF)

print()
print("-" * 100)
print("P1 SUMMARY")
print("-" * 100)

print(
    f"Privacy-related files discovered : "
    f"{len(P_DISCOVERY_DF)}"
)

print()
print("=" * 100)
print("P1 — PRIVACY RESULTS DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P2 — PRIVACY METRIC & DP PARAMETER VALIDATION
# ==============================================================================

print("=" * 100)
print("P2 — PRIVACY METRIC & DP PARAMETER VALIDATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# COLUMN NORMALIZATION
# ------------------------------------------------------------------------------

def P_normalize_column(column):

    return (
        str(column)
        .strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
        .replace("/", "_")
        .replace("(", "")
        .replace(")", "")
    )

# ------------------------------------------------------------------------------
# PRIVACY METRIC ALIASES
# ------------------------------------------------------------------------------

P_PRIVACY_ALIASES = {

    "epsilon": [
        "epsilon",
        "eps",
        "privacy_budget",
        "target_epsilon",
        "current_epsilon",
    ],

    "delta": [
        "delta",
        "target_delta",
        "privacy_delta",
    ],

    "noise_multiplier": [
        "noise_multiplier",
        "noise",
        "sigma",
        "noise_scale",
    ],

    "max_grad_norm": [
        "max_grad_norm",
        "max_gradient_norm",
        "gradient_clip",
        "gradient_clipping",
    ],

    "privacy_spent": [
        "privacy_spent",
        "epsilon_spent",
        "spent_epsilon",
        "final_epsilon",
    ],

    "membership_inference_risk": [
        "membership_inference_risk",
        "membership_risk",
        "mia_risk",
        "membership_attack_risk",
    ],

    "attribute_inference_risk": [
        "attribute_inference_risk",
        "attribute_risk",
        "attribute_attack_risk",
    ],

    "linkage_risk": [
        "linkage_risk",
        "reidentification_risk",
        "reidentification",
        "linkage_attack_risk",
    ],

    "disclosure_risk": [
        "disclosure_risk",
        "disclosure_probability",
    ],
}

# ------------------------------------------------------------------------------
# DETECTION FUNCTION
# ------------------------------------------------------------------------------

def P_detect_privacy_columns(dataframe):

    normalized_columns = {
        P_normalize_column(column): column
        for column in dataframe.columns
    }

    detected = {}

    for metric, aliases in P_PRIVACY_ALIASES.items():

        for alias in aliases:

            alias_normalized = P_normalize_column(alias)

            if alias_normalized in normalized_columns:

                detected[metric] = normalized_columns[
                    alias_normalized
                ]

                break

    return detected

# ------------------------------------------------------------------------------
# VALIDATE CANDIDATE FILES
# ------------------------------------------------------------------------------

P_VALIDATION_ROWS = []

for _, discovery_row in P_DISCOVERY_DF.iterrows():

    file_path = Path(discovery_row["File"])

    # Only tabular formats for this validation
    if file_path.suffix.lower() not in [".csv", ".tsv"]:

        continue

    try:

        if file_path.suffix.lower() == ".tsv":
            dataframe = pd.read_csv(
                file_path,
                sep="\t"
            )
        else:
            dataframe = pd.read_csv(file_path)

        detected = P_detect_privacy_columns(dataframe)

        P_VALIDATION_ROWS.append({
            "File": str(file_path),
            "Rows": len(dataframe),
            "Columns": len(dataframe.columns),
            "Detected Privacy Parameters": ", ".join(
                detected.keys()
            ),
            "Parameter Count": len(detected),
            "Status": (
                "VALID_PRIVACY_DATA"
                if detected
                else "NO_VALIDATED_PRIVACY_COLUMNS"
            ),
        })

    except Exception as error:

        P_VALIDATION_ROWS.append({
            "File": str(file_path),
            "Rows": None,
            "Columns": None,
            "Detected Privacy Parameters": "",
            "Parameter Count": 0,
            "Status": f"READ_ERROR: {type(error).__name__}",
        })

P2_VALIDATION_DF = pd.DataFrame(P_VALIDATION_ROWS)

print()
print("-" * 100)
print("PRIVACY PARAMETER VALIDATION")
print("-" * 100)

if P2_VALIDATION_DF.empty:
    print("No tabular privacy artifacts were available for validation.")
else:
    display(P2_VALIDATION_DF)

print()
print("=" * 100)
print("P2 — PRIVACY METRIC & DP PARAMETER VALIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P3 — PRIVACY RESULTS CONSOLIDATION
# ==============================================================================

print("=" * 100)
print("P3 — PRIVACY RESULTS CONSOLIDATION")
print("=" * 100)

P3_ROWS = []

P_VALID_PRIVACY_FILES = (
    P2_VALIDATION_DF[
        P2_VALIDATION_DF["Status"] == "VALID_PRIVACY_DATA"
    ]
    if not P2_VALIDATION_DF.empty
    else pd.DataFrame()
)

for _, validation_row in P_VALID_PRIVACY_FILES.iterrows():

    file_path = Path(validation_row["File"])

    try:
        dataframe = pd.read_csv(file_path)
    except Exception:
        continue

    detected = P_detect_privacy_columns(dataframe)

    path_lower = str(file_path).lower()
    filename_lower = file_path.name.lower()

    # --------------------------------------------------------------------------
    # DATASET
    # --------------------------------------------------------------------------

    matched_datasets = []

    for dataset, tokens in P_DATASET_TOKENS.items():

        if any(
            token.lower() in path_lower
            or token.lower() in filename_lower
            for token in tokens
        ):
            matched_datasets.append(dataset)

    # --------------------------------------------------------------------------
    # METHOD
    # --------------------------------------------------------------------------

    matched_methods = []

    for method, tokens in P_METHOD_TOKENS.items():

        if any(
            token.lower() in path_lower
            or token.lower() in filename_lower
            for token in tokens
        ):
            matched_methods.append(method)

    # --------------------------------------------------------------------------
    # CREATE LONG-FORM RECORDS
    # --------------------------------------------------------------------------

    for dataset in matched_datasets:

        for method in matched_methods:

            for metric, column in detected.items():

                numeric_values = pd.to_numeric(
                    dataframe[column],
                    errors="coerce"
                )

                for value in numeric_values.dropna():

                    P3_ROWS.append({
                        "Dataset": dataset,
                        "Method": method,
                        "Privacy Metric": metric,
                        "Value": float(value),
                        "Source File": str(file_path),
                    })

P3_PRIVACY_RESULTS = pd.DataFrame(P3_ROWS)

print()
print("-" * 100)
print("CONSOLIDATED PRIVACY RESULTS")
print("-" * 100)

if P3_PRIVACY_RESULTS.empty:
    print("No validated privacy observations were consolidated.")
else:
    display(P3_PRIVACY_RESULTS.head(100))

print()
print("-" * 100)
print("P3 SUMMARY")
print("-" * 100)

print(
    f"Privacy observations : "
    f"{len(P3_PRIVACY_RESULTS)}"
)

if not P3_PRIVACY_RESULTS.empty:

    print(
        f"Datasets represented : "
        f"{P3_PRIVACY_RESULTS['Dataset'].nunique()}"
    )

    print(
        f"Methods represented : "
        f"{P3_PRIVACY_RESULTS['Method'].nunique()}"
    )

    print(
        f"Privacy metrics : "
        f"{P3_PRIVACY_RESULTS['Privacy Metric'].nunique()}"
    )

print()
print("=" * 100)
print("P3 — PRIVACY RESULTS CONSOLIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P4 — DIFFERENTIAL PRIVACY GUARANTEE ANALYSIS
# ==============================================================================

print("=" * 100)
print("P4 — DIFFERENTIAL PRIVACY GUARANTEE ANALYSIS")
print("=" * 100)

P4_ROWS = []

for dataset in P_FINAL_DATASETS:

    for method in P_FINAL_METHODS:

        subset = (
            P3_PRIVACY_RESULTS[
                (P3_PRIVACY_RESULTS["Dataset"] == dataset) &
                (P3_PRIVACY_RESULTS["Method"] == method)
            ]
            if not P3_PRIVACY_RESULTS.empty
            else pd.DataFrame()
        )

        epsilon_available = (
            not subset[
                subset["Privacy Metric"] == "epsilon"
            ].empty
        )

        delta_available = (
            not subset[
                subset["Privacy Metric"] == "delta"
            ].empty
        )

        privacy_spent_available = (
            not subset[
                subset["Privacy Metric"] == "privacy_spent"
            ].empty
        )

        # Formal DP guarantee should only be marked when
        # the experiment provides the required DP accounting
        # evidence rather than simply because the method name
        # contains "DP".

        if method == "DP-CTGAN":

            if epsilon_available or privacy_spent_available:

                guarantee_status = "DP_ACCOUNTING_EVIDENCE_AVAILABLE"

            else:

                guarantee_status = "DP_METHOD_EXPECTED_BUT_EVIDENCE_MISSING"

        else:

            guarantee_status = "NO_FORMAL_DP_GUARANTEE_REPORTED"

        P4_ROWS.append({
            "Dataset": dataset,
            "Method": method,
            "Epsilon Available": epsilon_available,
            "Delta Available": delta_available,
            "Privacy Spent Available": privacy_spent_available,
            "Formal DP Status": guarantee_status,
        })

P4_DP_GUARANTEE = pd.DataFrame(P4_ROWS)

display(P4_DP_GUARANTEE)

print()
print("=" * 100)
print("P4 — DIFFERENTIAL PRIVACY GUARANTEE ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P5 — PRIVACY BUDGET & ACCOUNTING ANALYSIS
# ==============================================================================

print("=" * 100)
print("P5 — PRIVACY BUDGET & ACCOUNTING ANALYSIS")
print("=" * 100)

P5_ROWS = []

if not P3_PRIVACY_RESULTS.empty:

    for dataset in P_FINAL_DATASETS:

        for method in P_FINAL_METHODS:

            subset = P3_PRIVACY_RESULTS[
                (P3_PRIVACY_RESULTS["Dataset"] == dataset) &
                (P3_PRIVACY_RESULTS["Method"] == method)
            ]

            row = {
                "Dataset": dataset,
                "Method": method,
                "Epsilon": np.nan,
                "Delta": np.nan,
                "Noise Multiplier": np.nan,
                "Max Gradient Norm": np.nan,
                "Privacy Spent": np.nan,
            }

            for metric, output_column in [
                ("epsilon", "Epsilon"),
                ("delta", "Delta"),
                ("noise_multiplier", "Noise Multiplier"),
                ("max_grad_norm", "Max Gradient Norm"),
                ("privacy_spent", "Privacy Spent"),
            ]:

                values = subset.loc[
                    subset["Privacy Metric"] == metric,
                    "Value"
                ]

                if not values.empty:
                    row[output_column] = values.mean()

            P5_ROWS.append(row)

else:

    P5_ROWS = [
        {
            "Dataset": dataset,
            "Method": method,
            "Epsilon": np.nan,
            "Delta": np.nan,
            "Noise Multiplier": np.nan,
            "Max Gradient Norm": np.nan,
            "Privacy Spent": np.nan,
        }
        for dataset in P_FINAL_DATASETS
        for method in P_FINAL_METHODS
    ]

P5_PRIVACY_BUDGET = pd.DataFrame(P5_ROWS)

display(P5_PRIVACY_BUDGET)

print()
print("-" * 100)
print("INTERPRETATION")
print("-" * 100)

print(
    "For differential privacy, smaller epsilon generally "
    "corresponds to stronger formal privacy protection when "
    "delta and the accounting framework are held comparable."
)

print(
    "Privacy budgets must not be compared across incompatible "
    "accounting assumptions without qualification."
)

print()
print("=" * 100)
print("P5 — PRIVACY BUDGET & ACCOUNTING ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P6 — EMPIRICAL PRIVACY-RISK ANALYSIS
# ==============================================================================

print("=" * 100)
print("P6 — EMPIRICAL PRIVACY-RISK ANALYSIS")
print("=" * 100)

P6_RISK_METRICS = [
    "membership_inference_risk",
    "attribute_inference_risk",
    "linkage_risk",
    "disclosure_risk",
]

P6_ROWS = []

for dataset in P_FINAL_DATASETS:

    for method in P_FINAL_METHODS:

        subset = (
            P3_PRIVACY_RESULTS[
                (P3_PRIVACY_RESULTS["Dataset"] == dataset) &
                (P3_PRIVACY_RESULTS["Method"] == method)
            ]
            if not P3_PRIVACY_RESULTS.empty
            else pd.DataFrame()
        )

        row = {
            "Dataset": dataset,
            "Method": method,
        }

        for metric in P6_RISK_METRICS:

            values = subset.loc[
                subset["Privacy Metric"] == metric,
                "Value"
            ]

            row[metric] = (
                values.mean()
                if not values.empty
                else np.nan
            )

        P6_ROWS.append(row)

P6_EMPIRICAL_RISK = pd.DataFrame(P6_ROWS)

display(P6_EMPIRICAL_RISK)

print()
print("-" * 100)
print("RISK METRIC INTERPRETATION")
print("-" * 100)

print(
    "For empirical privacy-risk metrics, lower risk is generally "
    "preferred, provided the metric definition is consistent "
    "across experiments."
)

print(
    "Unavailable empirical privacy-risk measurements remain "
    "missing and are not interpreted as zero risk."
)

print()
print("=" * 100)
print("P6 — EMPIRICAL PRIVACY-RISK ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P7 — DATASET × METHOD PRIVACY COMPARISON
# ==============================================================================

print("=" * 100)
print("P7 — DATASET × METHOD PRIVACY COMPARISON")
print("=" * 100)

P7_PRIVACY_COMPARISON = P5_PRIVACY_BUDGET.merge(
    P6_EMPIRICAL_RISK,
    on=["Dataset", "Method"],
    how="left"
)

P7_PRIVACY_COMPARISON = P7_PRIVACY_COMPARISON.merge(
    P4_DP_GUARANTEE,
    on=["Dataset", "Method"],
    how="left"
)

print()
print("-" * 100)
print("FINAL PRIVACY COMPARISON")
print("-" * 100)

display(P7_PRIVACY_COMPARISON)

print()
print("=" * 100)
print("P7 — DATASET × METHOD PRIVACY COMPARISON COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# P8 — PRIVACY RESEARCH SUMMARY
# ==============================================================================

print("=" * 100)
print("P8 — PRIVACY RESEARCH SUMMARY")
print("=" * 100)

P8_SUMMARY = {
    "Final Datasets": len(P_FINAL_DATASETS),
    "Final Methods": len(P_FINAL_METHODS),
    "Expected Dataset × Method Combinations": (
        len(P_FINAL_DATASETS) * len(P_FINAL_METHODS)
    ),
    "Validated Privacy Observations": (
        len(P3_PRIVACY_RESULTS)
        if not P3_PRIVACY_RESULTS.empty
        else 0
    ),
    "Privacy Metrics Available": (
        P3_PRIVACY_RESULTS["Privacy Metric"].nunique()
        if not P3_PRIVACY_RESULTS.empty
        else 0
    ),
    "DP Accounting Records": (
        int(
            (
                P4_DP_GUARANTEE["Formal DP Status"]
                == "DP_ACCOUNTING_EVIDENCE_AVAILABLE"
            ).sum()
        )
    ),
}

print()
print("-" * 100)
print("PRIVACY RESEARCH STATUS")
print("-" * 100)

for key, value in P8_SUMMARY.items():

    print(
        f"{key:<45}: {value}"
    )

print()
print("-" * 100)
print("FINAL PRIVACY INTERPRETATION RULES")
print("-" * 100)

print(
    "1. Formal differential privacy guarantees are distinguished "
    "from empirical privacy-risk measurements."
)

print(
    "2. Smaller epsilon is interpreted as stronger formal privacy "
    "protection only under comparable accounting assumptions."
)

print(
    "3. Lower empirical privacy-risk values are preferred."
)

print(
    "4. Missing privacy results are not treated as zero risk."
)

print(
    "5. A method is not considered empirically private merely "
    "because its name contains 'DP'."
)

print(
    "6. Privacy results are interpreted together with utility "
    "and fidelity in the later comparative analysis."
)

print()
print("=" * 100)
print("P8 — PRIVACY RESEARCH SUMMARY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q1 — PRIVACY–UTILITY–FIDELITY DATA INTEGRATION
# ==============================================================================

print("=" * 100)
print("Q1 — PRIVACY–UTILITY–FIDELITY DATA INTEGRATION")
print("=" * 100)

import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# REQUIRED RESEARCH OBJECTS
# ------------------------------------------------------------------------------

Q_REQUIRED_OBJECTS = [
    "P7_PRIVACY_COMPARISON",
    "O4_UTILITY_MATRIX",
]

Q_MISSING_OBJECTS = [
    object_name
    for object_name in Q_REQUIRED_OBJECTS
    if object_name not in globals()
]

if Q_MISSING_OBJECTS:
    raise RuntimeError(
        "Required Section O/P objects are missing: "
        f"{Q_MISSING_OBJECTS}. "
        "Run Sections O and P before Section Q."
    )

# ------------------------------------------------------------------------------
# COPY SOURCE DATA
# ------------------------------------------------------------------------------

Q_PRIVACY_DF = P7_PRIVACY_COMPARISON.copy()

Q_UTILITY_DF = O4_UTILITY_MATRIX.copy()

# ------------------------------------------------------------------------------
# FIDELITY SOURCE DISCOVERY
# ------------------------------------------------------------------------------

Q_FIDELITY_CANDIDATES = [
    "L_FIDELITY_RESULTS",
    "L_FIDELITY_ANALYSIS",
    "L_FIDELITY_SUMMARY",
    "FIDELITY_RESULTS",
    "FIDELITY_RESULTS_DF",
]

Q_FIDELITY_DF = None

for variable_name in Q_FIDELITY_CANDIDATES:

    if variable_name in globals():

        candidate = globals()[variable_name]

        if isinstance(candidate, pd.DataFrame):
            Q_FIDELITY_DF = candidate.copy()
            print(
                f"Fidelity source selected: {variable_name}"
            )
            break

if Q_FIDELITY_DF is None:

    # Search existing dataframe objects safely.
    for variable_name, variable_value in list(
        globals().items()
    ):

        if not isinstance(variable_value, pd.DataFrame):
            continue

        columns_lower = {
            str(column).lower()
            for column in variable_value.columns
        }

        fidelity_terms = [
            "wasserstein",
            "ks",
            "js",
            "kl",
            "fidelity",
            "distribution",
            "correlation",
        ]

        if any(
            any(
                term in column
                for term in fidelity_terms
            )
            for column in columns_lower
        ):

            Q_FIDELITY_DF = variable_value.copy()

            print(
                f"Fidelity source discovered automatically: "
                f"{variable_name}"
            )

            break

if Q_FIDELITY_DF is None:

    raise RuntimeError(
        "No validated fidelity dataframe was found. "
        "Run Section L before Section Q."
    )

print()
print("-" * 100)
print("SOURCE DATASETS")
print("-" * 100)

print(
    f"Privacy rows  : {len(Q_PRIVACY_DF)}"
)

print(
    f"Utility rows  : {len(Q_UTILITY_DF)}"
)

print(
    f"Fidelity rows : {len(Q_FIDELITY_DF)}"
)

print()
print("=" * 100)
print("Q1 — PRIVACY–UTILITY–FIDELITY DATA INTEGRATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q2 — METRIC DIRECTION & NORMALIZATION
# ==============================================================================

print("=" * 100)
print("Q2 — METRIC DIRECTION & NORMALIZATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# NORMALIZATION FUNCTION
# ------------------------------------------------------------------------------

def Q_minmax_normalize(series, higher_is_better=True):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid = values.dropna()

    if valid.empty:
        return pd.Series(
            np.nan,
            index=series.index
        )

    minimum = valid.min()
    maximum = valid.max()

    if np.isclose(minimum, maximum):

        normalized = pd.Series(
            1.0,
            index=series.index
        )

    else:

        normalized = (
            (values - minimum)
            / (maximum - minimum)
        )

    if not higher_is_better:

        normalized = 1.0 - normalized

    return normalized

# ------------------------------------------------------------------------------
# UTILITY METRICS
# ------------------------------------------------------------------------------

Q_UTILITY_HIGHER_IS_BETTER = {
    "accuracy": True,
    "precision": True,
    "recall": True,
    "f1": True,
    "roc_auc": True,
    "pr_auc": True,
    "balanced_accuracy": True,
}

# ------------------------------------------------------------------------------
# FIDELITY METRICS
# ------------------------------------------------------------------------------

Q_FIDELITY_LOWER_IS_BETTER = [
    "wasserstein",
    "ks",
    "ks_statistic",
    "js",
    "js_divergence",
    "kl",
    "kl_divergence",
]

# ------------------------------------------------------------------------------
# PRIVACY METRICS
# ------------------------------------------------------------------------------

Q_PRIVACY_LOWER_IS_BETTER = [
    "epsilon",
    "privacy_spent",
    "membership_inference_risk",
    "attribute_inference_risk",
    "linkage_risk",
    "disclosure_risk",
]

print()
print("-" * 100)
print("METRIC DIRECTION RULES")
print("-" * 100)

print("Utility metrics : HIGHER is better")
print("Fidelity distance/divergence metrics : LOWER is better")
print("Privacy budget/risk metrics : LOWER is better")

print()
print("=" * 100)
print("Q2 — METRIC DIRECTION & NORMALIZATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q3 — FIDELITY–UTILITY–PRIVACY TRADE-OFF CONSOLIDATION
# ==============================================================================

print("=" * 100)
print("Q3 — FIDELITY–UTILITY–PRIVACY TRADE-OFF CONSOLIDATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# REQUIRED INPUTS
# ------------------------------------------------------------------------------

Q3_REQUIRED_OBJECTS = [
    "L_FIDELITY_RESULTS",
    "O4_UTILITY_MATRIX",
    "P7_PRIVACY_COMPARISON",
]

Q3_MISSING_OBJECTS = [
    name
    for name in Q3_REQUIRED_OBJECTS
    if name not in globals()
]

if Q3_MISSING_OBJECTS:
    raise RuntimeError(
        "Q3 cannot be executed because the following validated "
        "research results are unavailable:\n"
        + "\n".join(
            f"  - {name}"
            for name in Q3_MISSING_OBJECTS
        )
    )

# ------------------------------------------------------------------------------
# COPY VALIDATED RESULTS
# ------------------------------------------------------------------------------

Q3_FIDELITY = L_FIDELITY_RESULTS.copy()
Q3_UTILITY = O4_UTILITY_MATRIX.copy()
Q3_PRIVACY = P7_PRIVACY_COMPARISON.copy()

# ------------------------------------------------------------------------------
# STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------------------------

for dataframe in [
    Q3_FIDELITY,
    Q3_UTILITY,
    Q3_PRIVACY,
]:

    dataframe.columns = [
        str(column).strip()
        for column in dataframe.columns
    ]

# ------------------------------------------------------------------------------
# VALIDATE RESEARCH KEYS
# ------------------------------------------------------------------------------

Q3_KEYS = [
    "Dataset",
    "Method",
]

for dataframe_name, dataframe in [
    ("L_FIDELITY_RESULTS", Q3_FIDELITY),
    ("O4_UTILITY_MATRIX", Q3_UTILITY),
    ("P7_PRIVACY_COMPARISON", Q3_PRIVACY),
]:

    missing_keys = [
        key
        for key in Q3_KEYS
        if key not in dataframe.columns
    ]

    if missing_keys:
        raise RuntimeError(
            f"{dataframe_name} is missing required research "
            f"keys: {missing_keys}"
        )

# ------------------------------------------------------------------------------
# STANDARDIZE DATASET AND METHOD VALUES
# ------------------------------------------------------------------------------

for dataframe in [
    Q3_FIDELITY,
    Q3_UTILITY,
    Q3_PRIVACY,
]:

    dataframe["Dataset"] = (
        dataframe["Dataset"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    dataframe["Method"] = (
        dataframe["Method"]
        .astype(str)
        .str.strip()
    )

# ==============================================================================
# Q3.1 — BUILD FIDELITY AVAILABILITY MATRIX
# ==============================================================================

print()
print("-" * 100)
print("Q3.1 — BUILD FIDELITY AVAILABILITY MATRIX")
print("-" * 100)

Q3_FIDELITY_AVAILABILITY = (
    Q3_FIDELITY[
        Q3_KEYS + ["Status"]
    ]
    .copy()
)

Q3_FIDELITY_AVAILABILITY[
    "Fidelity Available"
] = (
    Q3_FIDELITY_AVAILABILITY[
        "Status"
    ]
    .astype(str)
    .str.upper()
    .isin(
        [
            "VALID",
            "VALIDATED",
            "AVAILABLE",
            "PASS",
            "PASSED",
            "READY",
            "SUCCESS",
            "COMPLETED",
        ]
    )
)

Q3_FIDELITY_AVAILABILITY = (
    Q3_FIDELITY_AVAILABILITY
    .groupby(
        Q3_KEYS,
        as_index=False
    )[
        "Fidelity Available"
    ]
    .any()
)

print(
    f"Fidelity Dataset × Method combinations: "
    f"{len(Q3_FIDELITY_AVAILABILITY)}"
)

# ==============================================================================
# Q3.2 — BUILD UTILITY AVAILABILITY MATRIX
# ==============================================================================

print()
print("-" * 100)
print("Q3.2 — BUILD UTILITY AVAILABILITY MATRIX")
print("-" * 100)

Q3_UTILITY_AVAILABILITY = (
    Q3_UTILITY[
        Q3_KEYS
    ]
    .drop_duplicates()
    .copy()
)

Q3_UTILITY_AVAILABILITY[
    "Utility Available"
] = True

print(
    f"Utility Dataset × Method combinations: "
    f"{len(Q3_UTILITY_AVAILABILITY)}"
)

# ==============================================================================
# Q3.3 — BUILD PRIVACY AVAILABILITY MATRIX
# ==============================================================================

print()
print("-" * 100)
print("Q3.3 — BUILD PRIVACY AVAILABILITY MATRIX")
print("-" * 100)

Q3_PRIVACY_AVAILABILITY = (
    Q3_PRIVACY[
        Q3_KEYS
    ]
    .drop_duplicates()
    .copy()
)

Q3_PRIVACY_AVAILABILITY[
    "Privacy Available"
] = True

print(
    f"Privacy Dataset × Method combinations: "
    f"{len(Q3_PRIVACY_AVAILABILITY)}"
)

# ==============================================================================
# Q3.4 — BUILD MASTER TRADE-OFF COVERAGE
# ==============================================================================

print()
print("-" * 100)
print("Q3.4 — BUILD MASTER TRADE-OFF COVERAGE")
print("-" * 100)

Q3_TRADEOFF_COVERAGE = pd.concat(
    [
        Q3_FIDELITY_AVAILABILITY[
            Q3_KEYS
        ],
        Q3_UTILITY_AVAILABILITY[
            Q3_KEYS
        ],
        Q3_PRIVACY_AVAILABILITY[
            Q3_KEYS
        ],
    ],
    ignore_index=True
).drop_duplicates(
    subset=Q3_KEYS
)

Q3_TRADEOFF_COVERAGE = (
    Q3_TRADEOFF_COVERAGE
    .merge(
        Q3_FIDELITY_AVAILABILITY,
        on=Q3_KEYS,
        how="left"
    )
    .merge(
        Q3_UTILITY_AVAILABILITY,
        on=Q3_KEYS,
        how="left"
    )
    .merge(
        Q3_PRIVACY_AVAILABILITY,
        on=Q3_KEYS,
        how="left"
    )
)

for column in [
    "Fidelity Available",
    "Utility Available",
    "Privacy Available",
]:

    Q3_TRADEOFF_COVERAGE[column] = (
        Q3_TRADEOFF_COVERAGE[column]
        .fillna(False)
        .astype(bool)
    )

# ==============================================================================
# Q3.5 — COMPLETE THREE-WAY COVERAGE
# ==============================================================================

print()
print("-" * 100)
print("Q3.5 — IDENTIFY COMPLETE THREE-WAY EVALUATIONS")
print("-" * 100)

Q3_TRADEOFF_COVERAGE[
    "Complete Trade-off Evaluation"
] = (
    Q3_TRADEOFF_COVERAGE[
        [
            "Fidelity Available",
            "Utility Available",
            "Privacy Available",
        ]
    ]
    .all(axis=1)
)

# ==============================================================================
# Q3.6 — TRADE-OFF STATUS
# ==============================================================================

Q3_TRADEOFF_COVERAGE[
    "Trade-off Status"
] = np.select(
    [
        Q3_TRADEOFF_COVERAGE[
            "Complete Trade-off Evaluation"
        ],

        (
            Q3_TRADEOFF_COVERAGE[
                [
                    "Fidelity Available",
                    "Utility Available",
                    "Privacy Available",
                ]
            ]
            .sum(axis=1)
            == 0
        ),

        (
            Q3_TRADEOFF_COVERAGE[
                [
                    "Fidelity Available",
                    "Utility Available",
                    "Privacy Available",
                ]
            ]
            .sum(axis=1)
            == 1
        ),

        (
            Q3_TRADEOFF_COVERAGE[
                [
                    "Fidelity Available",
                    "Utility Available",
                    "Privacy Available",
                ]
            ]
            .sum(axis=1)
            == 2
        ),
    ],
    [
        "COMPLETE",
        "NO_RESULTS",
        "PARTIAL_1_OF_3",
        "PARTIAL_2_OF_3",
    ],
    default="UNAVAILABLE",
)

# ==============================================================================
# Q3.7 — FINAL TRADE-OFF TABLE
# ==============================================================================

print()
print("=" * 100)
print("Q3 — FINAL FIDELITY–UTILITY–PRIVACY TRADE-OFF COVERAGE")
print("=" * 100)

Q3_TRADEOFF_COVERAGE = (
    Q3_TRADEOFF_COVERAGE
    .sort_values(
        [
            "Dataset",
            "Method",
        ]
    )
    .reset_index(drop=True)
)

display(
    Q3_TRADEOFF_COVERAGE
)

# ==============================================================================
# Q3.8 — SUMMARY
# ==============================================================================

print()
print("-" * 100)
print("Q3 SUMMARY")
print("-" * 100)

print(
    f"Total Dataset × Method combinations : "
    f"{len(Q3_TRADEOFF_COVERAGE)}"
)

print(
    f"Complete trade-off evaluations       : "
    f"{int(Q3_TRADEOFF_COVERAGE['Complete Trade-off Evaluation'].sum())}"
)

print(
    f"Partial trade-off evaluations        : "
    f"{int((Q3_TRADEOFF_COVERAGE['Trade-off Status'].str.startswith('PARTIAL')).sum())}"
)

print(
    f"Unavailable evaluations              : "
    f"{int((Q3_TRADEOFF_COVERAGE['Trade-off Status'] == 'NO_RESULTS').sum())}"
)

print()
print(
    "This section reports trade-off coverage only. "
    "No synthetic metric values are fabricated, normalized, "
    "or imputed when validated results are unavailable."
)

print()
print("=" * 100)
print("Q3 — FIDELITY–UTILITY–PRIVACY TRADE-OFF CONSOLIDATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q4 — PRIVACY–FIDELITY TRADE-OFF
# ==============================================================================

print("=" * 100)
print("Q4 — PRIVACY–FIDELITY TRADE-OFF")
print("=" * 100)

# ------------------------------------------------------------------------------
# REQUIRED VALIDATED RESULTS
# ------------------------------------------------------------------------------

Q4_REQUIRED_OBJECTS = [
    "L_FIDELITY_RESULTS",
    "P7_PRIVACY_COMPARISON",
]

Q4_MISSING_OBJECTS = [
    name
    for name in Q4_REQUIRED_OBJECTS
    if name not in globals()
]

if Q4_MISSING_OBJECTS:
    raise RuntimeError(
        "Q4 cannot be executed because the following validated "
        "research objects are unavailable:\n"
        + "\n".join(
            f"  - {name}"
            for name in Q4_MISSING_OBJECTS
        )
    )

# ------------------------------------------------------------------------------
# COPY VALIDATED RESULTS
# ------------------------------------------------------------------------------

Q4_FIDELITY = L_FIDELITY_RESULTS.copy()
Q4_PRIVACY = P7_PRIVACY_COMPARISON.copy()

# ------------------------------------------------------------------------------
# STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------------------------

Q4_FIDELITY.columns = [
    str(column).strip()
    for column in Q4_FIDELITY.columns
]

Q4_PRIVACY.columns = [
    str(column).strip()
    for column in Q4_PRIVACY.columns
]

# ------------------------------------------------------------------------------
# REQUIRED RESEARCH KEYS
# ------------------------------------------------------------------------------

Q4_KEYS = [
    "Dataset",
    "Method",
]

for dataframe_name, dataframe in [
    ("L_FIDELITY_RESULTS", Q4_FIDELITY),
    ("P7_PRIVACY_COMPARISON", Q4_PRIVACY),
]:

    missing_keys = [
        key
        for key in Q4_KEYS
        if key not in dataframe.columns
    ]

    if missing_keys:
        raise RuntimeError(
            f"{dataframe_name} is missing required keys: "
            f"{missing_keys}"
        )

# ------------------------------------------------------------------------------
# STANDARDIZE DATASET / METHOD VALUES
# ------------------------------------------------------------------------------

for dataframe in [
    Q4_FIDELITY,
    Q4_PRIVACY,
]:

    dataframe["Dataset"] = (
        dataframe["Dataset"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    dataframe["Method"] = (
        dataframe["Method"]
        .astype(str)
        .str.strip()
    )

# ==============================================================================
# Q4.1 — FIDELITY RESULT AVAILABILITY
# ==============================================================================

print()
print("-" * 100)
print("Q4.1 — FIDELITY RESULT AVAILABILITY")
print("-" * 100)

Q4_FIDELITY_AVAILABLE = (
    Q4_FIDELITY[
        Q4_KEYS
    ]
    .drop_duplicates()
    .copy()
)

Q4_FIDELITY_AVAILABLE[
    "Fidelity Available"
] = True

print(
    f"Dataset × Method fidelity combinations: "
    f"{len(Q4_FIDELITY_AVAILABLE)}"
)

# ==============================================================================
# Q4.2 — PRIVACY RESULT AVAILABILITY
# ==============================================================================

print()
print("-" * 100)
print("Q4.2 — PRIVACY RESULT AVAILABILITY")
print("-" * 100)

Q4_PRIVACY_AVAILABLE = (
    Q4_PRIVACY[
        Q4_KEYS
    ]
    .drop_duplicates()
    .copy()
)

Q4_PRIVACY_AVAILABLE[
    "Privacy Available"
] = True

print(
    f"Dataset × Method privacy combinations: "
    f"{len(Q4_PRIVACY_AVAILABLE)}"
)

# ==============================================================================
# Q4.3 — BUILD PRIVACY–FIDELITY COVERAGE
# ==============================================================================

print()
print("-" * 100)
print("Q4.3 — BUILD PRIVACY–FIDELITY COVERAGE")
print("-" * 100)

Q4_PRIVACY_FIDELITY = pd.concat(
    [
        Q4_FIDELITY_AVAILABLE[
            Q4_KEYS
        ],
        Q4_PRIVACY_AVAILABLE[
            Q4_KEYS
        ],
    ],
    ignore_index=True
).drop_duplicates(
    subset=Q4_KEYS
)

Q4_PRIVACY_FIDELITY = (
    Q4_PRIVACY_FIDELITY
    .merge(
        Q4_FIDELITY_AVAILABLE,
        on=Q4_KEYS,
        how="left"
    )
    .merge(
        Q4_PRIVACY_AVAILABLE,
        on=Q4_KEYS,
        how="left"
    )
)

Q4_PRIVACY_FIDELITY[
    "Fidelity Available"
] = (
    Q4_PRIVACY_FIDELITY[
        "Fidelity Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q4_PRIVACY_FIDELITY[
    "Privacy Available"
] = (
    Q4_PRIVACY_FIDELITY[
        "Privacy Available"
    ]
    .fillna(False)
    .astype(bool)
)

# ==============================================================================
# Q4.4 — COMPLETE PRIVACY–FIDELITY EVALUATION
# ==============================================================================

print()
print("-" * 100)
print("Q4.4 — IDENTIFY COMPLETE PRIVACY–FIDELITY EVALUATIONS")
print("-" * 100)

Q4_PRIVACY_FIDELITY[
    "Complete Privacy–Fidelity Evaluation"
] = (
    Q4_PRIVACY_FIDELITY[
        [
            "Fidelity Available",
            "Privacy Available",
        ]
    ]
    .all(axis=1)
)

# ==============================================================================
# Q4.5 — TRADE-OFF STATUS
# ==============================================================================

Q4_PRIVACY_FIDELITY[
    "Trade-off Status"
] = np.select(
    [
        Q4_PRIVACY_FIDELITY[
            "Complete Privacy–Fidelity Evaluation"
        ],

        (
            Q4_PRIVACY_FIDELITY[
                [
                    "Fidelity Available",
                    "Privacy Available",
                ]
            ]
            .sum(axis=1)
            == 0
        ),

        (
            Q4_PRIVACY_FIDELITY[
                [
                    "Fidelity Available",
                    "Privacy Available",
                ]
            ]
            .sum(axis=1)
            == 1
        ),
    ],
    [
        "COMPLETE",
        "NO_RESULTS",
        "PARTIAL",
    ],
    default="UNAVAILABLE",
)

# ==============================================================================
# Q4.6 — FIDELITY METRIC SUMMARY
# ==============================================================================

print()
print("-" * 100)
print("Q4.6 — FIDELITY METRIC SUMMARY")
print("-" * 100)

Q4_FIDELITY_METRIC_SUMMARY = (
    Q4_FIDELITY
    .groupby(
        [
            "Dataset",
            "Method",
        ],
        as_index=False
    )
    .agg(
        Fidelity_Metric_Count=(
            "Metric",
            "nunique"
        ),
        Fidelity_Valid_Observations=(
            "Value",
            lambda x:
            pd.to_numeric(
                x,
                errors="coerce"
            ).notna().sum()
        )
    )
)

Q4_PRIVACY_FIDELITY = (
    Q4_PRIVACY_FIDELITY
    .merge(
        Q4_FIDELITY_METRIC_SUMMARY,
        on=Q4_KEYS,
        how="left"
    )
)

# ==============================================================================
# Q4.7 — PRIVACY METRIC SUMMARY
# ==============================================================================

print()
print("-" * 100)
print("Q4.7 — PRIVACY METRIC SUMMARY")
print("-" * 100)

Q4_PRIVACY_NUMERIC_COLUMNS = []

for column in Q4_PRIVACY.columns:

    if column in Q4_KEYS:
        continue

    numeric_values = pd.to_numeric(
        Q4_PRIVACY[column],
        errors="coerce"
    )

    if numeric_values.notna().sum() > 0:

        Q4_PRIVACY_NUMERIC_COLUMNS.append(
            column
        )

Q4_PRIVACY_FIDELITY[
    "Privacy Numeric Metric Count"
] = (
    Q4_PRIVACY_FIDELITY[
        Q4_KEYS
    ]
    .merge(
        Q4_PRIVACY[
            Q4_KEYS +
            Q4_PRIVACY_NUMERIC_COLUMNS
        ]
        .groupby(
            Q4_KEYS,
            as_index=False
        )
        .size()
        .rename(
            columns={
                "size":
                "Privacy Numeric Metric Count"
            }
        ),
        on=Q4_KEYS,
        how="left"
    )[
        "Privacy Numeric Metric Count"
    ]
)

# ==============================================================================
# Q4.8 — FINAL PRIVACY–FIDELITY TRADE-OFF TABLE
# ==============================================================================

print()
print("=" * 100)
print("Q4 — FINAL PRIVACY–FIDELITY TRADE-OFF RESULTS")
print("=" * 100)

Q4_PRIVACY_FIDELITY = (
    Q4_PRIVACY_FIDELITY
    .sort_values(
        [
            "Dataset",
            "Method",
        ]
    )
    .reset_index(drop=True)
)

display(
    Q4_PRIVACY_FIDELITY
)

# ==============================================================================
# Q4.9 — SUMMARY
# ==============================================================================

print()
print("-" * 100)
print("Q4 SUMMARY")
print("-" * 100)

print(
    f"Dataset × Method combinations : "
    f"{len(Q4_PRIVACY_FIDELITY)}"
)

print(
    f"Complete privacy–fidelity evaluations : "
    f"{int(Q4_PRIVACY_FIDELITY['Complete Privacy–Fidelity Evaluation'].sum())}"
)

print(
    f"Partial evaluations : "
    f"{int((Q4_PRIVACY_FIDELITY['Trade-off Status'] == 'PARTIAL').sum())}"
)

print(
    f"No results : "
    f"{int((Q4_PRIVACY_FIDELITY['Trade-off Status'] == 'NO_RESULTS').sum())}"
)

print()
print(
    "Q4 evaluates privacy–fidelity result coverage and "
    "preserves the underlying validated metrics."
)

print(
    "No artificial Fidelity Indicator is created."
)

print()
print("=" * 100)
print("Q4 — PRIVACY–FIDELITY TRADE-OFF COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q1 — PRIVACY–UTILITY–FIDELITY TRADE-OFF INITIALIZATION
# ==============================================================================

print("=" * 100)
print("Q1 — PRIVACY–UTILITY–FIDELITY TRADE-OFF INITIALIZATION")
print("=" * 100)

Q1_REQUIRED_OBJECTS = [
    "L_FIDELITY_RESULTS",
    "P7_PRIVACY_COMPARISON",
]

Q1_MISSING_OBJECTS = [
    name
    for name in Q1_REQUIRED_OBJECTS
    if name not in globals()
]

if Q1_MISSING_OBJECTS:
    raise RuntimeError(
        "Section Q cannot start because the following validated "
        "research objects are unavailable:\n"
        + "\n".join(
            f"  - {name}"
            for name in Q1_MISSING_OBJECTS
        )
    )

print()
print("Required validated research objects:")
for name in Q1_REQUIRED_OBJECTS:
    print(f"  {name}: AVAILABLE")

print()
print("Section Q objective:")
print(
    "Evaluate the relationship between privacy protection, "
    "synthetic-data fidelity, and downstream utility without "
    "creating unsupported composite metrics."
)

print()
print("=" * 100)
print("Q1 — INITIALIZATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q2 — BUILD TRADE-OFF EVALUATION SCOPE
# ==============================================================================

print("=" * 100)
print("Q2 — BUILD TRADE-OFF EVALUATION SCOPE")
print("=" * 100)

Q2_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

Q2_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

Q2_EXPECTED_COMBINATIONS = pd.MultiIndex.from_product(
    [
        Q2_DATASETS,
        Q2_METHODS,
    ],
    names=[
        "Dataset",
        "Method",
    ]
).to_frame(index=False)

print()
print("-" * 100)
print("EXPECTED DATASET × METHOD COMBINATIONS")
print("-" * 100)

display(Q2_EXPECTED_COMBINATIONS)

print()
print(
    f"Expected Dataset × Method combinations: "
    f"{len(Q2_EXPECTED_COMBINATIONS)}"
)

print()
print("=" * 100)
print("Q2 — TRADE-OFF EVALUATION SCOPE COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q3 — BUILD MULTI-DIMENSIONAL EVALUATION COVERAGE
# ==============================================================================

print("=" * 100)
print("Q3 — BUILD MULTI-DIMENSIONAL EVALUATION COVERAGE")
print("=" * 100)

# ------------------------------------------------------------------------------
# FIDELITY COVERAGE
# ------------------------------------------------------------------------------

Q3_FIDELITY_KEYS = (
    L_FIDELITY_RESULTS[
        ["Dataset", "Method"]
    ]
    .drop_duplicates()
    .copy()
)

Q3_FIDELITY_KEYS["Fidelity Available"] = True

# ------------------------------------------------------------------------------
# PRIVACY COVERAGE
# ------------------------------------------------------------------------------

Q3_PRIVACY_KEYS = (
    P7_PRIVACY_COMPARISON[
        ["Dataset", "Method"]
    ]
    .drop_duplicates()
    .copy()
)

Q3_PRIVACY_KEYS["Privacy Available"] = True

# ------------------------------------------------------------------------------
# OPTIONAL UTILITY COVERAGE
# ------------------------------------------------------------------------------

Q3_UTILITY_SOURCE = None

for candidate in [
    "O_DOWNSTREAM_UTILITY_RESULTS",
    "O_UTILITY_RESULTS",
    "O_DOWNSTREAM_RESULTS",
    "DOWNSTREAM_UTILITY_RESULTS",
]:

    if candidate in globals():

        candidate_value = globals()[candidate]

        if isinstance(
            candidate_value,
            pd.DataFrame
        ):

            if {
                "Dataset",
                "Method",
            }.issubset(
                candidate_value.columns
            ):

                Q3_UTILITY_SOURCE = candidate_value.copy()
                break

if Q3_UTILITY_SOURCE is not None:

    Q3_UTILITY_KEYS = (
        Q3_UTILITY_SOURCE[
            ["Dataset", "Method"]
        ]
        .drop_duplicates()
        .copy()
    )

    Q3_UTILITY_KEYS[
        "Utility Available"
    ] = True

else:

    Q3_UTILITY_KEYS = pd.DataFrame(
        columns=[
            "Dataset",
            "Method",
            "Utility Available",
        ]
    )

# ------------------------------------------------------------------------------
# BUILD COMPLETE SCOPE
# ------------------------------------------------------------------------------

Q3_TRADEOFF_COVERAGE = (
    Q2_EXPECTED_COMBINATIONS
    .merge(
        Q3_FIDELITY_KEYS,
        on=["Dataset", "Method"],
        how="left"
    )
    .merge(
        Q3_PRIVACY_KEYS,
        on=["Dataset", "Method"],
        how="left"
    )
    .merge(
        Q3_UTILITY_KEYS,
        on=["Dataset", "Method"],
        how="left"
    )
)

for column in [
    "Fidelity Available",
    "Privacy Available",
    "Utility Available",
]:

    Q3_TRADEOFF_COVERAGE[column] = (
        Q3_TRADEOFF_COVERAGE[column]
        .fillna(False)
        .astype(bool)
    )

# ------------------------------------------------------------------------------
# EVALUATION COVERAGE COUNT
# ------------------------------------------------------------------------------

Q3_TRADEOFF_COVERAGE[
    "Available Dimensions"
] = (
    Q3_TRADEOFF_COVERAGE[
        [
            "Fidelity Available",
            "Privacy Available",
            "Utility Available",
        ]
    ]
    .sum(axis=1)
)

Q3_TRADEOFF_COVERAGE[
    "Trade-off Evaluation Status"
] = np.select(
    [
        Q3_TRADEOFF_COVERAGE[
            "Available Dimensions"
        ] == 3,

        Q3_TRADEOFF_COVERAGE[
            "Available Dimensions"
        ] == 2,

        Q3_TRADEOFF_COVERAGE[
            "Available Dimensions"
        ] == 1,
    ],
    [
        "COMPLETE",
        "PARTIAL",
        "LIMITED",
    ],
    default="NO_RESULTS",
)

display(Q3_TRADEOFF_COVERAGE)

print()
print("Coverage summary:")
print(
    Q3_TRADEOFF_COVERAGE[
        "Trade-off Evaluation Status"
    ].value_counts()
)

print()
print("=" * 100)
print("Q3 — MULTI-DIMENSIONAL EVALUATION COVERAGE COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q5 — PRIVACY–UTILITY TRADE-OFF
# ==============================================================================

print("=" * 100)
print("Q5 — PRIVACY–UTILITY TRADE-OFF")
print("=" * 100)

Q5_UTILITY_SOURCE = None

for candidate in [
    "O_DOWNSTREAM_UTILITY_RESULTS",
    "O_UTILITY_RESULTS",
    "O_DOWNSTREAM_RESULTS",
    "DOWNSTREAM_UTILITY_RESULTS",
]:

    if candidate in globals():

        candidate_value = globals()[candidate]

        if isinstance(
            candidate_value,
            pd.DataFrame
        ):

            if {
                "Dataset",
                "Method",
            }.issubset(
                candidate_value.columns
            ):

                Q5_UTILITY_SOURCE = candidate_value.copy()
                Q5_UTILITY_SOURCE_NAME = candidate
                break

if Q5_UTILITY_SOURCE is None:

    print()
    print(
        "No validated downstream utility dataframe was found."
    )

    Q5_PRIVACY_UTILITY = (
        Q2_EXPECTED_COMBINATIONS
        .copy()
    )

    Q5_PRIVACY_UTILITY[
        "Utility Available"
    ] = False

    Q5_PRIVACY_UTILITY[
        "Privacy Available"
    ] = (
        Q3_TRADEOFF_COVERAGE[
            "Privacy Available"
        ]
        .values
    )

    Q5_PRIVACY_UTILITY[
        "Privacy–Utility Status"
    ] = "UTILITY_UNAVAILABLE"

else:

    Q5_UTILITY_KEYS = (
        Q5_UTILITY_SOURCE[
            ["Dataset", "Method"]
        ]
        .drop_duplicates()
        .copy()
    )

    Q5_UTILITY_KEYS[
        "Utility Available"
    ] = True

    Q5_PRIVACY_KEYS = (
        P7_PRIVACY_COMPARISON[
            ["Dataset", "Method"]
        ]
        .drop_duplicates()
        .copy()
    )

    Q5_PRIVACY_KEYS[
        "Privacy Available"
    ] = True

    Q5_PRIVACY_UTILITY = (
        Q2_EXPECTED_COMBINATIONS
        .merge(
            Q5_UTILITY_KEYS,
            on=[
                "Dataset",
                "Method",
            ],
            how="left"
        )
        .merge(
            Q5_PRIVACY_KEYS,
            on=[
                "Dataset",
                "Method",
            ],
            how="left"
        )
    )

    Q5_PRIVACY_UTILITY[
        "Utility Available"
    ] = (
        Q5_PRIVACY_UTILITY[
            "Utility Available"
        ]
        .fillna(False)
        .astype(bool)
    )

    Q5_PRIVACY_UTILITY[
        "Privacy Available"
    ] = (
        Q5_PRIVACY_UTILITY[
            "Privacy Available"
        ]
        .fillna(False)
        .astype(bool)
    )

    Q5_PRIVACY_UTILITY[
        "Privacy–Utility Status"
    ] = np.select(
        [
            Q5_PRIVACY_UTILITY[
                [
                    "Privacy Available",
                    "Utility Available",
                ]
            ].all(axis=1),

            Q5_PRIVACY_UTILITY[
                [
                    "Privacy Available",
                    "Utility Available",
                ]
            ].sum(axis=1) == 1,
        ],
        [
            "COMPLETE",
            "PARTIAL",
        ],
        default="NO_RESULTS",
    )

print()
print("-" * 100)
print("PRIVACY–UTILITY TRADE-OFF COVERAGE")
print("-" * 100)

display(Q5_PRIVACY_UTILITY)

print()
print("=" * 100)
print("Q5 — PRIVACY–UTILITY TRADE-OFF COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q6 — FIDELITY–UTILITY TRADE-OFF
# ==============================================================================

print("=" * 100)
print("Q6 — FIDELITY–UTILITY TRADE-OFF")
print("=" * 100)

Q6_FIDELITY_KEYS = (
    L_FIDELITY_RESULTS[
        ["Dataset", "Method"]
    ]
    .drop_duplicates()
    .copy()
)

Q6_FIDELITY_KEYS[
    "Fidelity Available"
] = True

Q6_UTILITY_KEYS = pd.DataFrame(
    columns=[
        "Dataset",
        "Method",
        "Utility Available",
    ]
)

for candidate in [
    "O_DOWNSTREAM_UTILITY_RESULTS",
    "O_UTILITY_RESULTS",
    "O_DOWNSTREAM_RESULTS",
    "DOWNSTREAM_UTILITY_RESULTS",
]:

    if candidate in globals():

        candidate_value = globals()[candidate]

        if isinstance(
            candidate_value,
            pd.DataFrame
        ):

            if {
                "Dataset",
                "Method",
            }.issubset(
                candidate_value.columns
            ):

                Q6_UTILITY_KEYS = (
                    candidate_value[
                        [
                            "Dataset",
                            "Method",
                        ]
                    ]
                    .drop_duplicates()
                    .copy()
                )

                Q6_UTILITY_KEYS[
                    "Utility Available"
                ] = True

                break

Q6_FIDELITY_UTILITY = (
    Q2_EXPECTED_COMBINATIONS
    .merge(
        Q6_FIDELITY_KEYS,
        on=[
            "Dataset",
            "Method",
        ],
        how="left"
    )
    .merge(
        Q6_UTILITY_KEYS,
        on=[
            "Dataset",
            "Method",
        ],
        how="left"
    )
)

Q6_FIDELITY_UTILITY[
    "Fidelity Available"
] = (
    Q6_FIDELITY_UTILITY[
        "Fidelity Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q6_FIDELITY_UTILITY[
    "Utility Available"
] = (
    Q6_FIDELITY_UTILITY[
        "Utility Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q6_FIDELITY_UTILITY[
    "Fidelity–Utility Status"
] = np.select(
    [
        Q6_FIDELITY_UTILITY[
            [
                "Fidelity Available",
                "Utility Available",
            ]
        ].all(axis=1),

        Q6_FIDELITY_UTILITY[
            [
                "Fidelity Available",
                "Utility Available",
            ]
        ].sum(axis=1) == 1,
    ],
    [
        "COMPLETE",
        "PARTIAL",
    ],
    default="NO_RESULTS",
)

display(Q6_FIDELITY_UTILITY)

print()
print("=" * 100)
print("Q6 — FIDELITY–UTILITY TRADE-OFF COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q7 — CROSS-DIMENSIONAL TRADE-OFF SUMMARY
# ==============================================================================

print("=" * 100)
print("Q7 — CROSS-DIMENSIONAL TRADE-OFF SUMMARY")
print("=" * 100)

Q7_TRADEOFF_SUMMARY = (
    Q2_EXPECTED_COMBINATIONS
    .merge(
        Q3_TRADEOFF_COVERAGE[
            [
                "Dataset",
                "Method",
                "Fidelity Available",
                "Privacy Available",
                "Utility Available",
            ]
        ],
        on=[
            "Dataset",
            "Method",
        ],
        how="left"
    )
)

Q7_TRADEOFF_SUMMARY[
    "Fidelity Available"
] = (
    Q7_TRADEOFF_SUMMARY[
        "Fidelity Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q7_TRADEOFF_SUMMARY[
    "Privacy Available"
] = (
    Q7_TRADEOFF_SUMMARY[
        "Privacy Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q7_TRADEOFF_SUMMARY[
    "Utility Available"
] = (
    Q7_TRADEOFF_SUMMARY[
        "Utility Available"
    ]
    .fillna(False)
    .astype(bool)
)

Q7_TRADEOFF_SUMMARY[
    "Available Dimensions"
] = (
    Q7_TRADEOFF_SUMMARY[
        [
            "Fidelity Available",
            "Privacy Available",
            "Utility Available",
        ]
    ]
    .sum(axis=1)
)

Q7_TRADEOFF_SUMMARY[
    "Overall Trade-off Status"
] = np.select(
    [
        Q7_TRADEOFF_SUMMARY[
            "Available Dimensions"
        ] == 3,

        Q7_TRADEOFF_SUMMARY[
            "Available Dimensions"
        ] == 2,

        Q7_TRADEOFF_SUMMARY[
            "Available Dimensions"
        ] == 1,
    ],
    [
        "COMPLETE",
        "PARTIAL",
        "LIMITED",
    ],
    default="NO_RESULTS",
)

display(Q7_TRADEOFF_SUMMARY)

print()
print("-" * 100)
print("OVERALL TRADE-OFF COVERAGE")
print("-" * 100)

print(
    Q7_TRADEOFF_SUMMARY[
        "Overall Trade-off Status"
    ].value_counts()
)

print()
print("=" * 100)
print("Q7 — CROSS-DIMENSIONAL TRADE-OFF SUMMARY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q8 — METHOD-WISE TRADE-OFF COVERAGE
# ==============================================================================

print("=" * 100)
print("Q8 — METHOD-WISE TRADE-OFF COVERAGE")
print("=" * 100)

Q8_METHOD_SUMMARY = (
    Q7_TRADEOFF_SUMMARY
    .groupby(
        "Method",
        as_index=False
    )
    .agg(
        Dataset_Count=(
            "Dataset",
            "nunique"
        ),
        Fidelity_Available=(
            "Fidelity Available",
            "sum"
        ),
        Privacy_Available=(
            "Privacy Available",
            "sum"
        ),
        Utility_Available=(
            "Utility Available",
            "sum"
        ),
        Complete_Tradeoffs=(
            "Overall Trade-off Status",
            lambda x:
            (x == "COMPLETE").sum()
        ),
    )
)

display(Q8_METHOD_SUMMARY)

print()
print("=" * 100)
print("Q8 — METHOD-WISE TRADE-OFF COVERAGE COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q9 — DATASET-WISE TRADE-OFF COVERAGE
# ==============================================================================

print("=" * 100)
print("Q9 — DATASET-WISE TRADE-OFF COVERAGE")
print("=" * 100)

Q9_DATASET_SUMMARY = (
    Q7_TRADEOFF_SUMMARY
    .groupby(
        "Dataset",
        as_index=False
    )
    .agg(
        Method_Count=(
            "Method",
            "nunique"
        ),
        Fidelity_Available=(
            "Fidelity Available",
            "sum"
        ),
        Privacy_Available=(
            "Privacy Available",
            "sum"
        ),
        Utility_Available=(
            "Utility Available",
            "sum"
        ),
        Complete_Tradeoffs=(
            "Overall Trade-off Status",
            lambda x:
            (x == "COMPLETE").sum()
        ),
    )
)

display(Q9_DATASET_SUMMARY)

print()
print("=" * 100)
print("Q9 — DATASET-WISE TRADE-OFF COVERAGE COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# Q10 — FINAL TRADE-OFF INTERPRETATION
# ==============================================================================

print("=" * 100)
print("Q10 — FINAL TRADE-OFF INTERPRETATION")
print("=" * 100)

Q10_COMPLETE_COUNT = int(
    (
        Q7_TRADEOFF_SUMMARY[
            "Overall Trade-off Status"
        ]
        == "COMPLETE"
    ).sum()
)

Q10_PARTIAL_COUNT = int(
    (
        Q7_TRADEOFF_SUMMARY[
            "Overall Trade-off Status"
        ]
        == "PARTIAL"
    ).sum()
)

Q10_LIMITED_COUNT = int(
    (
        Q7_TRADEOFF_SUMMARY[
            "Overall Trade-off Status"
        ]
        == "LIMITED"
    ).sum()
)

Q10_NO_RESULTS_COUNT = int(
    (
        Q7_TRADEOFF_SUMMARY[
            "Overall Trade-off Status"
        ]
        == "NO_RESULTS"
    ).sum()
)

print()
print("-" * 100)
print("TRADE-OFF EVIDENCE STATUS")
print("-" * 100)

print(
    f"Complete privacy–utility–fidelity evaluations : "
    f"{Q10_COMPLETE_COUNT}"
)

print(
    f"Partial evaluations : "
    f"{Q10_PARTIAL_COUNT}"
)

print(
    f"Limited evaluations : "
    f"{Q10_LIMITED_COUNT}"
)

print(
    f"No-result combinations : "
    f"{Q10_NO_RESULTS_COUNT}"
)

print()
print("-" * 100)
print("RESEARCH INTERPRETATION RULES")
print("-" * 100)

print(
    "1. A complete trade-off interpretation requires "
    "privacy, fidelity, and utility evidence for the "
    "same Dataset × Method combination."
)

print(
    "2. Partial combinations are retained but are not "
    "used to make complete three-dimensional trade-off claims."
)

print(
    "3. Missing experiments are reported as unavailable "
    "rather than treated as zero performance."
)

print(
    "4. No causal relationship between privacy and fidelity "
    "or utility is inferred from observational comparisons."
)

print(
    "5. No unsupported composite score is created from "
    "heterogeneous fidelity metrics."
)

print()
print("=" * 100)
print("Q10 — FINAL TRADE-OFF INTERPRETATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# R1 — METHOD-WISE COMPARISON INITIALIZATION
# ==============================================================================

print("=" * 100)
print("R1 — METHOD-WISE COMPARISON INITIALIZATION")
print("=" * 100)

import numpy as np
import pandas as pd
import pathlib

R_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN"
]

R_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us"
]

R_METHOD_LABELS = {
    "CTGAN": "CTGAN",
    "TVAE": "TVAE",
    "DP-CTGAN": "DP-CTGAN",
    "SPP-GAN": "SPP-GAN"
}

R_METHOD_ORDER = {
    method: index
    for index, method in enumerate(R_METHODS)
}

print()
print("Final research methods:")
for method in R_METHODS:
    print(f"  - {method}")

print()
print("Final research datasets:")
for dataset in R_DATASETS:
    print(f"  - {dataset}")

print()
print("R1 STATUS")
print("---------" )
print("Method-wise comparison scope initialized successfully.")
```

```python
# ==============================================================================
# R2 — DISCOVER VALIDATED RESEARCH RESULT DATAFRAMES
# ==============================================================================

print("=" * 100)
print("R2 — DISCOVER VALIDATED RESEARCH RESULT DATAFRAMES")
print("=" * 100)

R_DATAFRAME_CANDIDATES = {}

for variable_name, variable_value in list(globals().items()):

    if isinstance(variable_value, pd.DataFrame):

        if (
            "Dataset" in variable_value.columns
            and "Method" in variable_value.columns
        ):
            R_DATAFRAME_CANDIDATES[variable_name] = variable_value.copy()

print()
print("Validated research dataframes containing Dataset and Method:")
print("-" * 100)

if not R_DATAFRAME_CANDIDATES:
    raise RuntimeError(
        "No validated research dataframe containing Dataset and Method "
        "was found. Complete Sections H–Q before running Section R."
    )

for variable_name, dataframe in R_DATAFRAME_CANDIDATES.items():

    print(
        f"{variable_name:<45} "
        f"{dataframe.shape[0]:>6} rows × "
        f"{dataframe.shape[1]:>3} columns"
    )

print()
print("R2 STATUS")
print("---------")
print("Validated research dataframe discovery completed.")
```

```python
# ==============================================================================
# R3 — RESOLVE FINAL RESULTS SOURCES
# ==============================================================================

print("=" * 100)
print("R3 — RESOLVE FINAL RESULTS SOURCES")
print("=" * 100)

R_SOURCE_PRIORITY = [
    "J_FINAL_METRIC_RESULTS",
    "J_FINAL_RESULTS",
    "FINAL_RESEARCH_RESULTS",
    "FINAL_METRIC_RESULTS",
    "H_EXPERIMENTAL_RESULTS_MATRIX",
    "H_RESULTS_MATRIX",
    "L_FIDELITY_RESULTS",
    "M_CORRELATION_RESULTS",
    "N_STRUCTURAL_RESULTS",
    "O_UTILITY_RESULTS",
    "P_PRIVACY_RESULTS"
]

R_SELECTED_SOURCES = {}

for source_name in R_SOURCE_PRIORITY:

    if source_name in globals():

        source_value = globals()[source_name]

        if isinstance(source_value, pd.DataFrame):

            if (
                "Dataset" in source_value.columns
                and "Method" in source_value.columns
            ):

                R_SELECTED_SOURCES[source_name] = source_value.copy()

print()
print("Selected validated result sources:")
print("-" * 100)

if R_SELECTED_SOURCES:

    for source_name, dataframe in R_SELECTED_SOURCES.items():

        print(
            f"{source_name:<35} "
            f"{dataframe.shape[0]:>6} rows × "
            f"{dataframe.shape[1]:>3} columns"
        )

else:

    raise RuntimeError(
        "No suitable validated result dataframe was found."
    )

print()
print("R3 STATUS")
print("---------")
print("Final result sources resolved successfully.")
```

```python
# ==============================================================================
# R4 — BUILD METHOD-WISE EXPERIMENTAL COVERAGE
# ==============================================================================

print("=" * 100)
print("R4 — BUILD METHOD-WISE EXPERIMENTAL COVERAGE")
print("=" * 100)

R_EXPECTED_EXPERIMENTS = pd.DataFrame(
    [
        {
            "Dataset": dataset,
            "Method": method
        }
        for dataset in R_DATASETS
        for method in R_METHODS
    ]
)

R_EXPECTED_EXPERIMENTS["Expected"] = True

print()
print("Expected experimental coverage:")
print("-" * 100)

print(
    R_EXPECTED_EXPERIMENTS
    .sort_values(
        ["Dataset", "Method"],
        key=lambda column: (
            column.map(R_METHOD_ORDER)
            if column.name == "Method"
            else column
        )
    )
    .to_string(index=False)
)

print()
print(
    f"Expected Dataset × Method combinations : "
    f"{len(R_EXPECTED_EXPERIMENTS)}"
)
```

```python
# ==============================================================================
# R5 — IDENTIFY AVAILABLE METHOD RESULTS
# ==============================================================================

print("=" * 100)
print("R5 — IDENTIFY AVAILABLE METHOD RESULTS")
print("=" * 100)

R_AVAILABLE_PAIRS = set()

for source_name, dataframe in R_SELECTED_SOURCES.items():

    source_copy = dataframe.copy()

    source_copy["Dataset"] = (
        source_copy["Dataset"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    source_copy["Method"] = (
        source_copy["Method"]
        .astype(str)
        .str.strip()
    )

    source_copy = source_copy[
        source_copy["Dataset"].isin(R_DATASETS)
        & source_copy["Method"].isin(R_METHODS)
    ]

    R_AVAILABLE_PAIRS.update(
        zip(
            source_copy["Dataset"],
            source_copy["Method"]
        )
    )

R_COVERAGE_TABLE = R_EXPECTED_EXPERIMENTS.copy()

R_COVERAGE_TABLE["Available"] = [
    (row.Dataset, row.Method) in R_AVAILABLE_PAIRS
    for row in R_COVERAGE_TABLE.itertuples()
]

R_COVERAGE_TABLE["Status"] = np.where(
    R_COVERAGE_TABLE["Available"],
    "AVAILABLE",
    "UNAVAILABLE"
)

R_COVERAGE_TABLE["Method Order"] = (
    R_COVERAGE_TABLE["Method"].map(R_METHOD_ORDER)
)

R_COVERAGE_TABLE = (
    R_COVERAGE_TABLE
    .sort_values(
        ["Dataset", "Method Order"]
    )
    .drop(columns=["Method Order"])
    .reset_index(drop=True)
)

print()
print(R_COVERAGE_TABLE.to_string(index=False))

print()
print("Coverage summary:")
print("-" * 100)

print(
    R_COVERAGE_TABLE["Status"]
    .value_counts()
    .to_string()
)
```

```python
# ==============================================================================
# R6 — BUILD METHOD-LEVEL COVERAGE SUMMARY
# ==============================================================================

print("=" * 100)
print("R6 — BUILD METHOD-LEVEL COVERAGE SUMMARY")
print("=" * 100)

R_METHOD_COVERAGE = (
    R_COVERAGE_TABLE
    .groupby("Method", as_index=False)
    .agg(
        Expected_Experiments=("Expected", "sum"),
        Available_Experiments=("Available", "sum")
    )
)

R_METHOD_COVERAGE["Unavailable_Experiments"] = (
    R_METHOD_COVERAGE["Expected_Experiments"]
    - R_METHOD_COVERAGE["Available_Experiments"]
)

R_METHOD_COVERAGE["Coverage_Percent"] = (
    R_METHOD_COVERAGE["Available_Experiments"]
    / R_METHOD_COVERAGE["Expected_Experiments"]
    * 100
)

R_METHOD_COVERAGE["Method Order"] = (
    R_METHOD_COVERAGE["Method"].map(R_METHOD_ORDER)
)

R_METHOD_COVERAGE = (
    R_METHOD_COVERAGE
    .sort_values("Method Order")
    .drop(columns=["Method Order"])
    .reset_index(drop=True)
)

print()
print(R_METHOD_COVERAGE.to_string(index=False))

print()
print("R6 STATUS")
print("---------")
print("Method-level experimental coverage calculated.")
```

```python
# ==============================================================================
# R7 — CONSOLIDATE METHOD-WISE METRIC RESULTS
# ==============================================================================

print("=" * 100)
print("R7 — CONSOLIDATE METHOD-WISE METRIC RESULTS")
print("=" * 100)

R_METRIC_FRAMES = []

for source_name, dataframe in R_SELECTED_SOURCES.items():

    source_copy = dataframe.copy()

    source_copy["Dataset"] = (
        source_copy["Dataset"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    source_copy["Method"] = (
        source_copy["Method"]
        .astype(str)
        .str.strip()
    )

    source_copy = source_copy[
        source_copy["Dataset"].isin(R_DATASETS)
        & source_copy["Method"].isin(R_METHODS)
    ].copy()

    if source_copy.empty:
        continue

    source_copy["Source"] = source_name

    R_METRIC_FRAMES.append(source_copy)

if not R_METRIC_FRAMES:

    raise RuntimeError(
        "No method-wise metric results are available."
    )

R_METHOD_METRIC_RESULTS = pd.concat(
    R_METRIC_FRAMES,
    ignore_index=True,
    sort=False
)

print()
print(
    "Combined method-wise result rows:",
    len(R_METHOD_METRIC_RESULTS)
)

print(
    "Combined columns:",
    len(R_METHOD_METRIC_RESULTS.columns)
)

print()
print("R7 STATUS")
print("---------")
print("Method-wise metric results consolidated.")
```

```python
# ==============================================================================
# R8 — REMOVE EXACT DUPLICATE RESULTS
# ==============================================================================

print("=" * 100)
print("R8 — REMOVE EXACT DUPLICATE RESULTS")
print("=" * 100)

R_ROWS_BEFORE_DEDUP = len(R_METHOD_METRIC_RESULTS)

R_METHOD_METRIC_RESULTS = (
    R_METHOD_METRIC_RESULTS
    .drop_duplicates()
    .reset_index(drop=True)
)

R_ROWS_AFTER_DEDUP = len(R_METHOD_METRIC_RESULTS)

R_DUPLICATES_REMOVED = (
    R_ROWS_BEFORE_DEDUP
    - R_ROWS_AFTER_DEDUP
)

print()
print("Rows before duplicate removal :", R_ROWS_BEFORE_DEDUP)
print("Rows after duplicate removal  :", R_ROWS_AFTER_DEDUP)
print("Duplicate rows removed        :", R_DUPLICATES_REMOVED)

print()
print("R8 STATUS")
print("---------")
print("Exact duplicate research records removed.")
```

```python
# ==============================================================================
# R9 — METHOD-WISE NUMERIC METRIC SUMMARY
# ==============================================================================

print("=" * 100)
print("R9 — METHOD-WISE NUMERIC METRIC SUMMARY")
print("=" * 100)

R_NUMERIC_COLUMNS = []

for column in R_METHOD_METRIC_RESULTS.columns:

    if column in ["Dataset", "Method"]:
        continue

    if pd.api.types.is_numeric_dtype(
        R_METHOD_METRIC_RESULTS[column]
    ):
        R_NUMERIC_COLUMNS.append(column)

print()
print("Numeric metric columns:")
for column in R_NUMERIC_COLUMNS:
    print(f"  - {column}")

if R_NUMERIC_COLUMNS:

    R_METHOD_NUMERIC_SUMMARY = (
        R_METHOD_METRIC_RESULTS
        .groupby("Method")[R_NUMERIC_COLUMNS]
        .mean(numeric_only=True)
        .reindex(R_METHODS)
        .reset_index()
    )

else:

    R_METHOD_NUMERIC_SUMMARY = pd.DataFrame(
        columns=["Method"]
    )

print()
print("Method-wise mean numeric results:")
print("-" * 100)

print(
    R_METHOD_NUMERIC_SUMMARY.to_string(index=False)
)

print()
print("R9 STATUS")
print("---------")
print("Method-wise numeric metric summary generated.")
```

```python
# ==============================================================================
# R10 — DATASET × METHOD COMPARISON
# ==============================================================================

print("=" * 100)
print("R10 — DATASET × METHOD COMPARISON")
print("=" * 100)

if R_NUMERIC_COLUMNS:

    R_DATASET_METHOD_SUMMARY = (
        R_METHOD_METRIC_RESULTS
        .groupby(
            ["Dataset", "Method"]
        )[R_NUMERIC_COLUMNS]
        .mean(numeric_only=True)
        .reset_index()
    )

    R_DATASET_METHOD_SUMMARY["Method Order"] = (
        R_DATASET_METHOD_SUMMARY["Method"]
        .map(R_METHOD_ORDER)
    )

    R_DATASET_METHOD_SUMMARY = (
        R_DATASET_METHOD_SUMMARY
        .sort_values(
            ["Dataset", "Method Order"]
        )
        .drop(columns=["Method Order"])
        .reset_index(drop=True)
    )

else:

    R_DATASET_METHOD_SUMMARY = pd.DataFrame(
        columns=["Dataset", "Method"]
    )

print()
print(
    R_DATASET_METHOD_SUMMARY.to_string(index=False)
)

print()
print("R10 STATUS")
print("----------")
print("Dataset × Method comparison generated.")
```

```python
# ==============================================================================
# R11 — METHOD-WISE METRIC AVAILABILITY
# ==============================================================================

print("=" * 100)
print("R11 — METHOD-WISE METRIC AVAILABILITY")
print("=" * 100)

if R_NUMERIC_COLUMNS:

    R_METHOD_METRIC_AVAILABILITY = (
        R_METHOD_METRIC_RESULTS
        .groupby("Method")[R_NUMERIC_COLUMNS]
        .count()
        .reindex(R_METHODS)
        .reset_index()
    )

else:

    R_METHOD_METRIC_AVAILABILITY = pd.DataFrame(
        columns=["Method"]
    )

print()
print("Available observations by method:")
print("-" * 100)

print(
    R_METHOD_METRIC_AVAILABILITY.to_string(index=False)
)

print()
print("R11 STATUS")
print("----------")
print("Metric availability by method calculated.")
```

```python
# ==============================================================================
# R12 — METHOD-WISE COMPLETENESS
# ==============================================================================

print("=" * 100)
print("R12 — METHOD-WISE COMPLETENESS")
print("=" * 100)

R_METHOD_COMPLETENESS = R_METHOD_COVERAGE.copy()

R_METHOD_COMPLETENESS["Experiment Coverage (%)"] = (
    R_METHOD_COMPLETENESS["Coverage_Percent"]
)

if R_NUMERIC_COLUMNS:

    R_METHOD_COMPLETENESS["Metric Observations"] = (
        R_METHOD_METRIC_RESULTS
        .groupby("Method")[R_NUMERIC_COLUMNS]
        .count()
        .sum(axis=1)
        .reindex(R_METHODS)
        .fillna(0)
        .values
    )

else:

    R_METHOD_COMPLETENESS["Metric Observations"] = 0

print()
print(
    R_METHOD_COMPLETENESS.to_string(index=False)
)

print()
print("R12 STATUS")
print("----------")
print("Method-wise completeness assessment completed.")
```

```python
# ==============================================================================
# R13 — METHOD-WISE COMPARATIVE RANKING
# ==============================================================================

print("=" * 100)
print("R13 — METHOD-WISE COMPARATIVE RANKING")
print("=" * 100)

R_METHOD_RANKINGS = []

if R_NUMERIC_COLUMNS:

    for metric in R_NUMERIC_COLUMNS:

        metric_values = (
            R_METHOD_METRIC_RESULTS
            .groupby("Method")[metric]
            .mean()
            .reindex(R_METHODS)
        )

        ranking_frame = pd.DataFrame({
            "Method": R_METHODS,
            "Metric": metric,
            "Mean Value": metric_values.values
        })

        ranking_frame["Rank"] = (
            ranking_frame["Mean Value"]
            .rank(
                method="min",
                ascending=False,
                na_option="bottom"
            )
        )

        R_METHOD_RANKINGS.append(
            ranking_frame
        )

if R_METHOD_RANKINGS:

    R_METHOD_RANKING_TABLE = pd.concat(
        R_METHOD_RANKINGS,
        ignore_index=True
    )

else:

    R_METHOD_RANKING_TABLE = pd.DataFrame(
        columns=[
            "Method",
            "Metric",
            "Mean Value",
            "Rank"
        ]
    )

print()
print(
    R_METHOD_RANKING_TABLE.to_string(index=False)
)

print()
print("R13 STATUS")
print("----------")
print(
    "Descriptive method rankings generated without "
    "combining incompatible metrics."
)
```

```python
# ==============================================================================
# R14 — METHOD-WISE BEST OBSERVED RESULTS
# ==============================================================================

print("=" * 100)
print("R14 — METHOD-WISE BEST OBSERVED RESULTS")
print("=" * 100)

R_METHOD_BEST_RESULTS = []

if R_NUMERIC_COLUMNS:

    for metric in R_NUMERIC_COLUMNS:

        metric_data = (
            R_METHOD_METRIC_RESULTS[
                ["Dataset", "Method", metric]
            ]
            .dropna(subset=[metric])
            .copy()
        )

        if metric_data.empty:
            continue

        best_index = metric_data[metric].idxmax()

        best_row = metric_data.loc[
            best_index
        ]

        R_METHOD_BEST_RESULTS.append({
            "Metric": metric,
            "Dataset": best_row["Dataset"],
            "Method": best_row["Method"],
            "Value": best_row[metric]
        })

R_METHOD_BEST_RESULTS = pd.DataFrame(
    R_METHOD_BEST_RESULTS
)

print()
print(
    R_METHOD_BEST_RESULTS.to_string(index=False)
)

print()
print("R14 STATUS")
print("----------")
print("Best observed method–dataset results identified.")
```

```python
# ==============================================================================
# R15 — FINAL METHOD-WISE COMPARISON TABLE
# ==============================================================================

print("=" * 100)
print("R15 — FINAL METHOD-WISE COMPARISON TABLE")
print("=" * 100)

R_FINAL_METHOD_COMPARISON = R_METHOD_COVERAGE.copy()

R_FINAL_METHOD_COMPARISON = R_FINAL_METHOD_COMPARISON[
    [
        "Method",
        "Expected_Experiments",
        "Available_Experiments",
        "Unavailable_Experiments",
        "Coverage_Percent"
    ]
].copy()

R_FINAL_METHOD_COMPARISON = (
    R_FINAL_METHOD_COMPARISON
    .rename(
        columns={
            "Expected_Experiments": "Expected Experiments",
            "Available_Experiments": "Available Experiments",
            "Unavailable_Experiments": "Unavailable Experiments",
            "Coverage_Percent": "Coverage (%)"
        }
    )
)

print()
print(
    R_FINAL_METHOD_COMPARISON.to_string(index=False)
)

print()
print("R15 STATUS")
print("----------")
print("Final method-wise comparison table generated.")
```

```python
# ==============================================================================
# R16 — RESEARCH INTERPRETATION SUMMARY
# ==============================================================================

print("=" * 100)
print("R16 — RESEARCH INTERPRETATION SUMMARY")
print("=" * 100)

# ------------------------------------------------------------------------------
# R16.1 — DISCOVER EXISTING SECTION R DATAFRAMES
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16.1 — DISCOVER EXISTING SECTION R DATAFRAMES")
print("-" * 100)

# Take a snapshot so the globals dictionary is not modified during iteration.
R16_GLOBAL_ITEMS = list(globals().items())

R16_DATAFRAMES = {}

for _name, _value in R16_GLOBAL_ITEMS:

    if isinstance(_value, pd.DataFrame) and not _value.empty:

        _columns = set(
            str(c).strip()
            for c in _value.columns
        )

        # Candidate method-wise dataframe:
        # must contain Method and should contain at least one useful
        # comparison/analysis column.
        if "Method" in _columns:

            R16_DATAFRAMES[_name] = _value.copy()

print(
    f"Method-containing dataframes discovered: "
    f"{len(R16_DATAFRAMES)}"
)

for _name, _df in R16_DATAFRAMES.items():

    print(
        f"  {_name}: "
        f"{_df.shape[0]} rows × {_df.shape[1]} columns"
    )

# ------------------------------------------------------------------------------
# R16.2 — RESOLVE THE BEST METHOD-WISE DATAFRAME
# ------------------------------------------------------------------------------

R16_SOURCE = None
R16_SOURCE_NAME = None

# Strong preference for dataframes whose names indicate method-wise analysis.
R16_PRIORITY_KEYWORDS = [
    "METHOD_WISE",
    "METHODWISE",
    "COMPARISON",
    "SUMMARY",
    "INTERPRETATION",
]

R16_PRIORITY = []

for _name, _df in R16_DATAFRAMES.items():

    _upper_name = str(_name).upper()

    _score = 0

    for _keyword in R16_PRIORITY_KEYWORDS:

        if _keyword in _upper_name:
            _score += 10

    # Prefer dataframes containing research dimensions.
    _columns = set(
        str(c).strip()
        for c in _df.columns
    )

    for _column in [
        "Dataset",
        "Available",
        "Expected",
        "Fidelity",
        "Utility",
        "Privacy",
        "Score",
        "Metric",
        "Status",
    ]:

        if _column in _columns:
            _score += 2

    R16_PRIORITY.append(
        (
            _score,
            _name,
            _df
        )
    )

# Sort highest score first.
R16_PRIORITY = sorted(
    R16_PRIORITY,
    key=lambda x: x[0],
    reverse=True
)

if R16_PRIORITY:

    R16_SOURCE = R16_PRIORITY[0][2].copy()
    R16_SOURCE_NAME = R16_PRIORITY[0][1]

# ------------------------------------------------------------------------------
# R16.3 — FALLBACK: BUILD METHOD SUMMARY FROM AVAILABLE R RESULTS
# ------------------------------------------------------------------------------

if R16_SOURCE is None:

    print()
    print(
        "No dedicated method-wise summary dataframe was found."
    )

    print(
        "Attempting to construct the R16 summary from "
        "available Section R research results."
    )

    R16_METHODS = [
        "CTGAN",
        "TVAE",
        "DP-CTGAN",
        "SPP-GAN",
    ]

    R16_ROWS = []

    # Search all dataframe objects for Dataset/Method information.
    for _name, _df in R16_DATAFRAMES.items():

        _columns = set(
            str(c).strip()
            for c in _df.columns
        )

        if "Method" not in _columns:
            continue

        _temp = _df.copy()

        _temp["Method"] = (
            _temp["Method"]
            .astype(str)
            .str.strip()
        )

        _temp = _temp[
            _temp["Method"].isin(R16_METHODS)
        ]

        if _temp.empty:
            continue

        for _method in R16_METHODS:

            _method_rows = _temp[
                _temp["Method"] == _method
            ]

            if _method_rows.empty:
                continue

            R16_ROWS.append(
                {
                    "Method": _method,
                    "Source": _name,
                    "Rows": len(_method_rows),
                }
            )

    if R16_ROWS:

        R16_SOURCE = pd.DataFrame(
            R16_ROWS
        )

        R16_SOURCE_NAME = (
            "R16_FALLBACK_METHOD_RESULTS"
        )

# ------------------------------------------------------------------------------
# R16.4 — STOP ONLY IF NO METHOD RESULTS EXIST AT ALL
# ------------------------------------------------------------------------------

if R16_SOURCE is None or R16_SOURCE.empty:

    raise RuntimeError(
        "R16 could not find any Section R dataframe containing "
        "validated method-wise results. "
        "Inspect the Section R outputs before running R16."
    )

# ------------------------------------------------------------------------------
# R16.5 — SOURCE INFORMATION
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16.5 — RESOLVED RESEARCH SOURCE")
print("-" * 100)

print(
    f"Source dataframe : {R16_SOURCE_NAME}"
)

print(
    f"Rows             : {len(R16_SOURCE)}"
)

print(
    f"Columns          : {len(R16_SOURCE.columns)}"
)

print()
print("Source columns:")

for _column in R16_SOURCE.columns:

    print(
        f"  - {_column}"
    )

# ------------------------------------------------------------------------------
# R16.6 — NORMALIZE METHOD COLUMN
# ------------------------------------------------------------------------------

if "Method" not in R16_SOURCE.columns:

    raise RuntimeError(
        "The resolved R16 dataframe does not contain "
        "the required 'Method' column."
    )

R16_SOURCE["Method"] = (
    R16_SOURCE["Method"]
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------------------------
# R16.7 — BUILD METHOD-WISE RESEARCH SUMMARY
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16.7 — BUILD METHOD-WISE RESEARCH SUMMARY")
print("-" * 100)

R16_METHOD_SUMMARY_ROWS = []

for _method in [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]:

    _method_data = R16_SOURCE[
        R16_SOURCE["Method"] == _method
    ].copy()

    if _method_data.empty:
        continue

    _row = {
        "Method": _method,
        "Result Rows": len(_method_data),
    }

    # --------------------------------------------------------------
    # Optional experiment availability
    # --------------------------------------------------------------

    _available_column = None

    for _candidate in [
        "Available Experiments",
        "Available_Experiments",
        "Available",
        "Completed Experiments",
        "Completed_Experiments",
        "Valid Experiments",
        "Valid_Experiments",
    ]:

        if _candidate in _method_data.columns:
            _available_column = _candidate
            break

    _expected_column = None

    for _candidate in [
        "Expected Experiments",
        "Expected_Experiments",
        "Expected",
        "Total Expected Experiments",
    ]:

        if _candidate in _method_data.columns:
            _expected_column = _candidate
            break

    if _available_column is not None:

        _available = pd.to_numeric(
            _method_data[_available_column],
            errors="coerce"
        ).max()

        _row["Available Experiments"] = (
            _available
        )

    if _expected_column is not None:

        _expected = pd.to_numeric(
            _method_data[_expected_column],
            errors="coerce"
        ).max()

        _row["Expected Experiments"] = (
            _expected
        )

    # --------------------------------------------------------------
    # Dataset coverage
    # --------------------------------------------------------------

    if "Dataset" in _method_data.columns:

        _datasets = (
            _method_data["Dataset"]
            .dropna()
            .astype(str)
            .str.strip()
            .unique()
        )

        _row["Datasets Represented"] = len(
            _datasets
        )

        _row["Dataset Coverage"] = ", ".join(
            sorted(_datasets)
        )

    R16_METHOD_SUMMARY_ROWS.append(
        _row
    )

R16_RESEARCH_INTERPRETATION = pd.DataFrame(
    R16_METHOD_SUMMARY_ROWS
)

# ------------------------------------------------------------------------------
# R16.8 — CALCULATE EXPERIMENT COVERAGE WHEN POSSIBLE
# ------------------------------------------------------------------------------

if (
    "Available Experiments"
    in R16_RESEARCH_INTERPRETATION.columns
    and
    "Expected Experiments"
    in R16_RESEARCH_INTERPRETATION.columns
):

    R16_RESEARCH_INTERPRETATION[
        "Experiment Availability (%)"
    ] = np.where(
        R16_RESEARCH_INTERPRETATION[
            "Expected Experiments"
        ] > 0,

        (
            R16_RESEARCH_INTERPRETATION[
                "Available Experiments"
            ]
            /
            R16_RESEARCH_INTERPRETATION[
                "Expected Experiments"
            ]
        ) * 100.0,

        np.nan
    )

# ------------------------------------------------------------------------------
# R16.9 — FINAL INTERPRETATION
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16.9 — FINAL METHOD-WISE RESEARCH INTERPRETATION")
print("-" * 100)

for _, _row in R16_RESEARCH_INTERPRETATION.iterrows():

    _method = _row["Method"]

    print()
    print(
        f"Method: {_method}"
    )

    if "Experiment Availability (%)" in _row.index:

        _availability = _row[
            "Experiment Availability (%)"
        ]

        if pd.notna(_availability):

            print(
                f"  Experiment availability: "
                f"{_availability:.2f}%"
            )

    if "Datasets Represented" in _row.index:

        print(
            f"  Datasets represented: "
            f"{int(_row['Datasets Represented'])}"
        )

    if "Dataset Coverage" in _row.index:

        print(
            f"  Dataset coverage: "
            f"{_row['Dataset Coverage']}"
        )

    print(
        f"  Result rows represented: "
        f"{int(_row['Result Rows'])}"
    )

# ------------------------------------------------------------------------------
# R16.10 — SAVE OUTPUT
# ------------------------------------------------------------------------------

if "REPORTS_DIRECTORY" in globals():

    R16_REPORT_DIRECTORY = pathlib.Path(
        REPORTS_DIRECTORY
    ) / "Notebook_09"

    R16_REPORT_DIRECTORY.mkdir(
        parents=True,
        exist_ok=True
    )

    R16_OUTPUT_PATH = (
        R16_REPORT_DIRECTORY
        / "R16_research_interpretation_summary.csv"
    )

    R16_RESEARCH_INTERPRETATION.to_csv(
        R16_OUTPUT_PATH,
        index=False
    )

else:

    R16_OUTPUT_PATH = None

# ------------------------------------------------------------------------------
# R16.11 — DISPLAY FINAL TABLE
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16.11 — FINAL SUMMARY TABLE")
print("-" * 100)

display(
    R16_RESEARCH_INTERPRETATION
)

# ------------------------------------------------------------------------------
# R16.12 — STATUS
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("R16 STATUS")
print("-" * 100)

print(
    f"Methods represented : "
    f"{R16_RESEARCH_INTERPRETATION['Method'].nunique()}"
)

if R16_OUTPUT_PATH is not None:

    print(
        f"Output saved       : "
        f"{R16_OUTPUT_PATH}"
    )

print()
print("=" * 100)
print("R16 — RESEARCH INTERPRETATION SUMMARY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# R17 — SAVE METHOD-WISE COMPARISON RESULTS
# ==============================================================================

print("=" * 100)
print("R17 — SAVE METHOD-WISE COMPARISON RESULTS")
print("=" * 100)

if "REPORTS_DIRECTORY" in globals():

    R_OUTPUT_DIRECTORY = pathlib.Path(
        REPORTS_DIRECTORY
    )

else:

    R_OUTPUT_DIRECTORY = (
        pathlib.Path(
            "/content/drive/MyDrive/SPP_GAN_Project/reports"
        )
    )

R_OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True
)

R_METHOD_COVERAGE_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_method_wise_experimental_coverage.csv"
)

R_METHOD_SUMMARY_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_method_wise_numeric_summary.csv"
)

R_DATASET_METHOD_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_dataset_method_comparison.csv"
)

R_METHOD_COMPLETENESS_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_method_wise_completeness.csv"
)

R_METHOD_RANKING_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_method_wise_metric_rankings.csv"
)

R_FINAL_COMPARISON_FILE = (
    R_OUTPUT_DIRECTORY /
    "R_final_method_wise_comparison.csv"
)

R_METHOD_COVERAGE.to_csv(
    R_METHOD_COVERAGE_FILE,
    index=False
)

R_METHOD_NUMERIC_SUMMARY.to_csv(
    R_METHOD_SUMMARY_FILE,
    index=False
)

R_DATASET_METHOD_SUMMARY.to_csv(
    R_DATASET_METHOD_FILE,
    index=False
)

R_METHOD_COMPLETENESS.to_csv(
    R_METHOD_COMPLETENESS_FILE,
    index=False
)

R_METHOD_RANKING_TABLE.to_csv(
    R_METHOD_RANKING_FILE,
    index=False
)

R_FINAL_METHOD_COMPARISON.to_csv(
    R_FINAL_COMPARISON_FILE,
    index=False
)

print()
print("Saved files:")
print(f"  {R_METHOD_COVERAGE_FILE}")
print(f"  {R_METHOD_SUMMARY_FILE}")
print(f"  {R_DATASET_METHOD_FILE}")
print(f"  {R_METHOD_COMPLETENESS_FILE}")
print(f"  {R_METHOD_RANKING_FILE}")
print(f"  {R_FINAL_COMPARISON_FILE}")

print()
print("R17 STATUS")
print("----------")
print("Method-wise comparison artifacts saved successfully.")
```

```python
# ==============================================================================
# R18 — FINAL METHOD-WISE RESEARCH CHECK
# ==============================================================================

print("=" * 100)
print("R18 — FINAL METHOD-WISE RESEARCH CHECK")
print("=" * 100)

R_TOTAL_EXPECTED = len(R_EXPECTED_EXPERIMENTS)

R_TOTAL_AVAILABLE = int(
    R_COVERAGE_TABLE["Available"].sum()
)

R_TOTAL_UNAVAILABLE = (
    R_TOTAL_EXPECTED
    - R_TOTAL_AVAILABLE
)

R_OVERALL_COVERAGE = (
    R_TOTAL_AVAILABLE
    / R_TOTAL_EXPECTED
    * 100
    if R_TOTAL_EXPECTED > 0
    else 0.0
)

print()
print("FINAL METHOD-WISE COVERAGE")
print("-" * 100)

print(
    f"Expected experiments   : {R_TOTAL_EXPECTED}"
)

print(
    f"Available experiments  : {R_TOTAL_AVAILABLE}"
)

print(
    f"Unavailable experiments: {R_TOTAL_UNAVAILABLE}"
)

print(
    f"Overall coverage       : {R_OVERALL_COVERAGE:.2f}%"
)

print()
print("METHOD COVERAGE")
print("-" * 100)

for row in R_FINAL_METHOD_COMPARISON.itertuples(index=False):

    print(
        f"{row.Method:<12} : "
        f"{int(row[2])}/{int(row[1])} "
        f"({row[4]:.2f}%)"
    )

print()
print("R18 STATUS")
print("----------")

if R_TOTAL_AVAILABLE == R_TOTAL_EXPECTED:

    print(
        "PASS — All expected Dataset × Method experiments "
        "are represented in the validated result sources."
    )

else:

    print(
        "PARTIAL — Some expected Dataset × Method experiments "
        "remain unavailable and must be treated as missing evidence."
    )

print()
print("=" * 100)
print("R — METHOD-WISE COMPARISON COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S1 — DATASET-WISE COMPARISON INITIALIZATION
# ==============================================================================

print("=" * 100)
print("S1 — DATASET-WISE COMPARISON INITIALIZATION")
print("=" * 100)

S_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

S_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

S_EXPECTED_EXPERIMENTS = (
    len(S_DATASETS) * len(S_METHODS)
)

print()
print("-" * 100)
print("FINAL DATASET-WISE RESEARCH SCOPE")
print("-" * 100)

print(
    f"Datasets : {len(S_DATASETS)}"
)

for dataset in S_DATASETS:
    print(f"  - {dataset}")

print()
print(
    f"Methods  : {len(S_METHODS)}"
)

for method in S_METHODS:
    print(f"  - {method}")

print()
print(
    f"Expected Dataset × Method combinations : "
    f"{S_EXPECTED_EXPERIMENTS}"
)

print()
print("=" * 100)
print("S1 — INITIALIZATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S2 — DISCOVER DATASET-WISE RESULT SOURCES
# ==============================================================================

print("=" * 100)
print("S2 — DISCOVER DATASET-WISE RESULT SOURCES")
print("=" * 100)

S_GLOBAL_ITEMS = list(globals().items())

S_DATAFRAME_SOURCES = {}

for variable_name, variable_value in S_GLOBAL_ITEMS:

    if not isinstance(variable_value, pd.DataFrame):
        continue

    if variable_value.empty:
        continue

    columns = {
        str(column).strip()
        for column in variable_value.columns
    }

    if "Dataset" in columns:

        S_DATAFRAME_SOURCES[variable_name] = (
            variable_value.copy()
        )

print()
print(
    f"Dataset-containing dataframes discovered: "
    f"{len(S_DATAFRAME_SOURCES)}"
)

for variable_name, dataframe in S_DATAFRAME_SOURCES.items():

    print(
        f"  {variable_name}: "
        f"{dataframe.shape[0]} rows × "
        f"{dataframe.shape[1]} columns"
    )

print()
print("=" * 100)
print("S2 — DATASET-WISE SOURCE DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S3 — RESOLVE DATASET-WISE RESEARCH RESULTS
# ==============================================================================

print("=" * 100)
print("S3 — RESOLVE DATASET-WISE RESEARCH RESULTS")
print("=" * 100)

S_CANDIDATES = []

for variable_name, dataframe in S_DATAFRAME_SOURCES.items():

    columns = {
        str(column).strip()
        for column in dataframe.columns
    }

    score = 0

    # Strong research-result indicators.
    for column in [
        "Method",
        "Metric",
        "Value",
        "Score",
        "Status",
        "Direction",
        "Analysis",
    ]:

        if column in columns:
            score += 5

    name_upper = variable_name.upper()

    for keyword in [
        "RESULT",
        "METRIC",
        "EVALUATION",
        "FIDELITY",
        "UTILITY",
        "PRIVACY",
        "SUMMARY",
        "COMPARISON",
    ]:

        if keyword in name_upper:
            score += 3

    S_CANDIDATES.append(
        (
            score,
            variable_name,
            dataframe
        )
    )

S_CANDIDATES = sorted(
    S_CANDIDATES,
    key=lambda item: item[0],
    reverse=True
)

print()
print("-" * 100)
print("DATASET-WISE RESULT CANDIDATES")
print("-" * 100)

for score, variable_name, dataframe in S_CANDIDATES[:20]:

    print(
        f"{variable_name}: "
        f"score={score}, "
        f"rows={len(dataframe)}, "
        f"columns={len(dataframe.columns)}"
    )

# ------------------------------------------------------------------------------
# Build a consolidated source rather than relying on one dataframe.
# ------------------------------------------------------------------------------

S_SOURCE_FRAMES = []

for _, variable_name, dataframe in S_CANDIDATES:

    temp = dataframe.copy()

    temp["__S_SOURCE"] = variable_name

    S_SOURCE_FRAMES.append(temp)

if S_SOURCE_FRAMES:

    S_ALL_RESULTS = pd.concat(
        S_SOURCE_FRAMES,
        ignore_index=True,
        sort=False
    )

else:

    S_ALL_RESULTS = pd.DataFrame()

print()
print(
    f"Combined dataset-wise source rows: "
    f"{len(S_ALL_RESULTS)}"
)

print()
print("=" * 100)
print("S3 — DATASET-WISE RESULT RESOLUTION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S4 — FILTER FINAL DATASET AND METHOD SCOPE
# ==============================================================================

print("=" * 100)
print("S4 — FILTER FINAL DATASET AND METHOD SCOPE")
print("=" * 100)

if S_ALL_RESULTS.empty:

    raise RuntimeError(
        "No dataset-wise evaluation results are available "
        "for Section S."
    )

S_SCOPED_RESULTS = S_ALL_RESULTS.copy()

S_SCOPED_RESULTS["Dataset"] = (
    S_SCOPED_RESULTS["Dataset"]
    .astype(str)
    .str.strip()
)

# Normalize method values when the column exists.
if "Method" in S_SCOPED_RESULTS.columns:

    S_SCOPED_RESULTS["Method"] = (
        S_SCOPED_RESULTS["Method"]
        .astype(str)
        .str.strip()
    )

    S_SCOPED_RESULTS = S_SCOPED_RESULTS[
        S_SCOPED_RESULTS["Dataset"].isin(S_DATASETS)
        &
        S_SCOPED_RESULTS["Method"].isin(S_METHODS)
    ].copy()

else:

    print(
        "NOTE: Source dataframe does not contain Method. "
        "Only dataset-level results can be retained."
    )

print()
print(
    f"Rows remaining after final research-scope filtering: "
    f"{len(S_SCOPED_RESULTS)}"
)

print()
print("-" * 100)
print("DATASET COVERAGE")
print("-" * 100)

if not S_SCOPED_RESULTS.empty:

    print(
        S_SCOPED_RESULTS["Dataset"]
        .value_counts()
        .sort_index()
    )

print()
print("=" * 100)
print("S4 — FINAL SCOPE FILTERING COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S5 — BUILD DATASET × METHOD COVERAGE MATRIX
# ==============================================================================

print("=" * 100)
print("S5 — BUILD DATASET × METHOD COVERAGE MATRIX")
print("=" * 100)

S_COVERAGE_ROWS = []

for dataset in S_DATASETS:

    for method in S_METHODS:

        if (
            "Method" in S_SCOPED_RESULTS.columns
        ):

            matches = S_SCOPED_RESULTS[
                (S_SCOPED_RESULTS["Dataset"] == dataset)
                &
                (S_SCOPED_RESULTS["Method"] == method)
            ]

        else:

            matches = pd.DataFrame()

        row = {
            "Dataset": dataset,
            "Method": method,
            "Result Rows": len(matches),
            "Status": (
                "AVAILABLE"
                if not matches.empty
                else "UNAVAILABLE"
            ),
        }

        S_COVERAGE_ROWS.append(row)

S_DATASET_METHOD_COVERAGE = pd.DataFrame(
    S_COVERAGE_ROWS
)

print()
display(
    S_DATASET_METHOD_COVERAGE
)

print()
print("-" * 100)
print("COVERAGE SUMMARY")
print("-" * 100)

print(
    S_DATASET_METHOD_COVERAGE[
        "Status"
    ].value_counts()
)

print()
print("=" * 100)
print("S5 — DATASET × METHOD COVERAGE MATRIX COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S6 — DATASET-WISE METRIC SUMMARY
# ==============================================================================

print("=" * 100)
print("S6 — DATASET-WISE METRIC SUMMARY")
print("=" * 100)

if S_SCOPED_RESULTS.empty:

    S_DATASET_METRIC_SUMMARY = pd.DataFrame()

else:

    S_DATASET_METRIC_SUMMARY = S_SCOPED_RESULTS.copy()

    # Convert Value to numeric when available.
    if "Value" in S_DATASET_METRIC_SUMMARY.columns:

        S_DATASET_METRIC_SUMMARY[
            "Numeric Value"
        ] = pd.to_numeric(
            S_DATASET_METRIC_SUMMARY["Value"],
            errors="coerce"
        )

    elif "Score" in S_DATASET_METRIC_SUMMARY.columns:

        S_DATASET_METRIC_SUMMARY[
            "Numeric Value"
        ] = pd.to_numeric(
            S_DATASET_METRIC_SUMMARY["Score"],
            errors="coerce"
        )

    else:

        S_DATASET_METRIC_SUMMARY[
            "Numeric Value"
        ] = np.nan

print()
print(
    f"Dataset-wise metric rows: "
    f"{len(S_DATASET_METRIC_SUMMARY)}"
)

print()
display(
    S_DATASET_METRIC_SUMMARY.head(50)
)

print()
print("=" * 100)
print("S6 — DATASET-WISE METRIC SUMMARY COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S7 — DATASET-WISE METHOD COMPARISON
# ==============================================================================

print("=" * 100)
print("S7 — DATASET-WISE METHOD COMPARISON")
print("=" * 100)

S_DATASET_COMPARISON_ROWS = []

for dataset in S_DATASETS:

    dataset_results = S_SCOPED_RESULTS[
        S_SCOPED_RESULTS["Dataset"] == dataset
    ].copy()

    print()
    print("-" * 100)
    print(f"DATASET: {dataset}")
    print("-" * 100)

    if dataset_results.empty:

        print(
            "No validated evaluation results available."
        )

        continue

    if "Method" in dataset_results.columns:

        method_counts = (
            dataset_results["Method"]
            .value_counts()
        )

        for method in S_METHODS:

            count = int(
                method_counts.get(
                    method,
                    0
                )
            )

            S_DATASET_COMPARISON_ROWS.append(
                {
                    "Dataset": dataset,
                    "Method": method,
                    "Available Result Rows": count,
                    "Availability": (
                        "AVAILABLE"
                        if count > 0
                        else "UNAVAILABLE"
                    ),
                }
            )

        display(
            pd.DataFrame(
                [
                    {
                        "Method": method,
                        "Result Rows": int(
                            method_counts.get(
                                method,
                                0
                            )
                        ),
                    }
                    for method in S_METHODS
                ]
            )
        )

print()
print("=" * 100)
print("S7 — DATASET-WISE METHOD COMPARISON COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S8 — DATASET-WISE PERFORMANCE INDICATORS
# ==============================================================================

print("=" * 100)
print("S8 — DATASET-WISE PERFORMANCE INDICATORS")
print("=" * 100)

S_INDICATOR_ROWS = []

for dataset in S_DATASETS:

    dataset_results = S_SCOPED_RESULTS[
        S_SCOPED_RESULTS["Dataset"] == dataset
    ].copy()

    if dataset_results.empty:
        continue

    if "Method" not in dataset_results.columns:
        continue

    if "Numeric Value" not in dataset_results.columns:

        if "Value" in dataset_results.columns:

            dataset_results[
                "Numeric Value"
            ] = pd.to_numeric(
                dataset_results["Value"],
                errors="coerce"
            )

        elif "Score" in dataset_results.columns:

            dataset_results[
                "Numeric Value"
            ] = pd.to_numeric(
                dataset_results["Score"],
                errors="coerce"
            )

        else:
            continue

    numeric_results = dataset_results[
        dataset_results["Numeric Value"].notna()
    ].copy()

    if numeric_results.empty:
        continue

    # If a metric column exists, summarize each metric.
    if "Metric" in numeric_results.columns:

        grouped = (
            numeric_results
            .groupby(
                ["Dataset", "Method", "Metric"],
                dropna=False
            )["Numeric Value"]
            .mean()
            .reset_index()
        )

        for _, row in grouped.iterrows():

            S_INDICATOR_ROWS.append(
                {
                    "Dataset": row["Dataset"],
                    "Method": row["Method"],
                    "Metric": row["Metric"],
                    "Mean Value": row["Numeric Value"],
                }
            )

S_DATASET_PERFORMANCE_INDICATORS = pd.DataFrame(
    S_INDICATOR_ROWS
)

print()
print(
    f"Dataset-wise performance indicator rows: "
    f"{len(S_DATASET_PERFORMANCE_INDICATORS)}"
)

if not S_DATASET_PERFORMANCE_INDICATORS.empty:

    display(
        S_DATASET_PERFORMANCE_INDICATORS
    )

else:

    print(
        "No numeric metric indicators could be constructed "
        "from the currently validated results."
    )

print()
print("=" * 100)
print("S8 — DATASET-WISE PERFORMANCE INDICATORS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S9 — BEST-PERFORMING METHOD BY DATASET
# ==============================================================================

print("=" * 100)
print("S9 — BEST-PERFORMING METHOD BY DATASET")
print("=" * 100)

S_BEST_METHOD_ROWS = []

if not S_DATASET_PERFORMANCE_INDICATORS.empty:

    for dataset in S_DATASETS:

        dataset_metrics = (
            S_DATASET_PERFORMANCE_INDICATORS[
                S_DATASET_PERFORMANCE_INDICATORS[
                    "Dataset"
                ] == dataset
            ]
        )

        if dataset_metrics.empty:
            continue

        for metric in (
            dataset_metrics["Metric"]
            .dropna()
            .astype(str)
            .unique()
        ):

            metric_data = dataset_metrics[
                dataset_metrics["Metric"].astype(str)
                == metric
            ].copy()

            if metric_data.empty:
                continue

            # Try to recover direction from the source.
            direction = None

            if (
                "Direction" in S_SCOPED_RESULTS.columns
                and
                "Metric" in S_SCOPED_RESULTS.columns
            ):

                direction_values = (
                    S_SCOPED_RESULTS[
                        (
                            S_SCOPED_RESULTS[
                                "Dataset"
                            ] == dataset
                        )
                        &
                        (
                            S_SCOPED_RESULTS[
                                "Metric"
                            ].astype(str)
                            == metric
                        )
                    ]["Direction"]
                    .dropna()
                    .astype(str)
                    .str.lower()
                    .unique()
                )

                if len(direction_values) > 0:

                    direction = direction_values[0]

            if direction in [
                "higher",
                "higher_is_better",
                "maximize",
                "max",
                "↑",
            ]:

                best_index = (
                    metric_data[
                        "Mean Value"
                    ].idxmax()
                )

            elif direction in [
                "lower",
                "lower_is_better",
                "minimize",
                "min",
                "↓",
            ]:

                best_index = (
                    metric_data[
                        "Mean Value"
                    ].idxmin()
                )

            else:

                # Direction unknown: do not declare a winner.
                S_BEST_METHOD_ROWS.append(
                    {
                        "Dataset": dataset,
                        "Metric": metric,
                        "Best Method": "NOT DETERMINED",
                        "Best Value": np.nan,
                        "Reason": (
                            "Metric direction unavailable"
                        ),
                    }
                )

                continue

            best_row = metric_data.loc[
                best_index
            ]

            S_BEST_METHOD_ROWS.append(
                {
                    "Dataset": dataset,
                    "Metric": metric,
                    "Best Method": best_row["Method"],
                    "Best Value": best_row["Mean Value"],
                    "Reason": (
                        f"Direction={direction}"
                    ),
                }
            )

S_BEST_METHOD_BY_DATASET = pd.DataFrame(
    S_BEST_METHOD_ROWS
)

print()
if not S_BEST_METHOD_BY_DATASET.empty:

    display(
        S_BEST_METHOD_BY_DATASET
    )

else:

    print(
        "No defensible best-method determination "
        "could be made from the available metrics."
    )

print()
print("=" * 100)
print("S9 — BEST-PERFORMING METHOD ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S10 — DATASET-WISE RESEARCH INTERPRETATION
# ==============================================================================

print("=" * 100)
print("S10 — DATASET-WISE RESEARCH INTERPRETATION")
print("=" * 100)

S_INTERPRETATION_ROWS = []

for dataset in S_DATASETS:

    coverage = S_DATASET_METHOD_COVERAGE[
        S_DATASET_METHOD_COVERAGE["Dataset"]
        == dataset
    ].copy()

    available_methods = coverage[
        coverage["Status"] == "AVAILABLE"
    ]["Method"].tolist()

    unavailable_methods = coverage[
        coverage["Status"] == "UNAVAILABLE"
    ]["Method"].tolist()

    if available_methods:

        interpretation = (
            f"{dataset}: validated results are available "
            f"for {', '.join(available_methods)}."
        )

    else:

        interpretation = (
            f"{dataset}: no validated method-level "
            f"evaluation results are currently available."
        )

    if unavailable_methods:

        interpretation += (
            f" Results unavailable for "
            f"{', '.join(unavailable_methods)}."
        )

    S_INTERPRETATION_ROWS.append(
        {
            "Dataset": dataset,
            "Available Methods": ", ".join(
                available_methods
            ),
            "Unavailable Methods": ", ".join(
                unavailable_methods
            ),
            "Interpretation": interpretation,
        }
    )

S_DATASET_RESEARCH_INTERPRETATION = pd.DataFrame(
    S_INTERPRETATION_ROWS
)

print()

for _, row in (
    S_DATASET_RESEARCH_INTERPRETATION.iterrows()
):

    print(
        row["Interpretation"]
    )

print()
print("-" * 100)
print("FINAL DATASET-WISE INTERPRETATION TABLE")
print("-" * 100)

display(
    S_DATASET_RESEARCH_INTERPRETATION
)

print()
print("=" * 100)
print("S10 — DATASET-WISE RESEARCH INTERPRETATION COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# S11 — DATASET-WISE COMPARISON REPORT
# ==============================================================================

print("=" * 100)
print("S11 — DATASET-WISE COMPARISON REPORT")
print("=" * 100)

if "REPORTS_DIRECTORY" in globals():

    S_REPORT_DIRECTORY = (
        pathlib.Path(REPORTS_DIRECTORY)
        / "Notebook_09"
    )

    S_REPORT_DIRECTORY.mkdir(
        parents=True,
        exist_ok=True
    )

    S_COVERAGE_PATH = (
        S_REPORT_DIRECTORY
        / "S_dataset_method_coverage.csv"
    )

    S_INTERPRETATION_PATH = (
        S_REPORT_DIRECTORY
        / "S_dataset_research_interpretation.csv"
    )

    S_COVERAGE_TABLE_PATH = (
        S_REPORT_DIRECTORY
        / "S_dataset_method_comparison.csv"
    )

    S_DATASET_METHOD_COVERAGE.to_csv(
        S_COVERAGE_PATH,
        index=False
    )

    S_DATASET_RESEARCH_INTERPRETATION.to_csv(
        S_INTERPRETATION_PATH,
        index=False
    )

    pd.DataFrame(
        S_DATASET_COMPARISON_ROWS
    ).to_csv(
        S_COVERAGE_TABLE_PATH,
        index=False
    )

    print()
    print(
        f"Coverage report       : {S_COVERAGE_PATH}"
    )

    print(
        f"Interpretation report : {S_INTERPRETATION_PATH}"
    )

    print(
        f"Comparison report     : {S_COVERAGE_TABLE_PATH}"
    )

else:

    print(
        "REPORTS_DIRECTORY is unavailable. "
        "Reports were not written to disk."
    )

print()
print("-" * 100)
print("S11 STATUS")
print("-" * 100)

S_AVAILABLE_COMBINATIONS = int(
    (
        S_DATASET_METHOD_COVERAGE["Status"]
        == "AVAILABLE"
    ).sum()
)

print(
    f"Expected Dataset × Method combinations : "
    f"{S_EXPECTED_EXPERIMENTS}"
)

print(
    f"Available combinations                  : "
    f"{S_AVAILABLE_COMBINATIONS}"
)

print(
    f"Unavailable combinations               : "
    f"{S_EXPECTED_EXPERIMENTS - S_AVAILABLE_COMBINATIONS}"
)

print()
print("=" * 100)
print("S — DATASET-WISE COMPARISON COMPLETED")
print("=" * 100)
```

```python
# ==============================================================================
# T4 — DISCOVER AVAILABLE R RESEARCH RESULT OBJECTS
# ==============================================================================

print()
print("=" * 100)
print("T4 — DISCOVER AVAILABLE R RESEARCH RESULT OBJECTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# R OBJECTS RELEVANT TO CROSS-DATASET ROBUSTNESS
# ------------------------------------------------------------------------------

T_R_OBJECT_NAMES = [
    "R_METHOD_METRIC_RESULTS",
    "R_DATASET_METHOD_SUMMARY",
    "R_METHOD_NUMERIC_SUMMARY",
    "R_METHOD_RANKING_TABLE",
    "R_METHOD_BEST_RESULTS",
    "R_METHOD_COVERAGE",
    "R_METHOD_COMPLETENESS",
    "R_FINAL_METHOD_COMPARISON",
]

T_AVAILABLE_R_OBJECTS = {}

print()

for object_name in T_R_OBJECT_NAMES:

    if object_name in globals():

        object_value = globals()[object_name]

        if isinstance(
            object_value,
            pd.DataFrame
        ):

            T_AVAILABLE_R_OBJECTS[
                object_name
            ] = object_value.copy()

            print(
                f"[AVAILABLE] {object_name:<35} "
                f"shape={object_value.shape}"
            )

        else:

            print(
                f"[AVAILABLE] {object_name:<35} "
                f"type={type(object_value).__name__}"
            )

    else:

        print(
            f"[MISSING]   {object_name}"
        )

# ------------------------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------------------------

if not T_AVAILABLE_R_OBJECTS:

    raise RuntimeError(
        "T4 FAILED: None of the expected R research result objects "
        "are available. Run the required R evaluation sections first."
    )

print()
print("-" * 100)
print("AVAILABLE DATAFRAME STRUCTURES")
print("-" * 100)

for object_name, object_value in T_AVAILABLE_R_OBJECTS.items():

    print()
    print(f"{object_name}")
    print(
        f"Shape: {object_value.shape}"
    )

    print(
        "Columns:"
    )

    for column in object_value.columns:

        print(
            f"  - {column}"
        )

# ------------------------------------------------------------------------------
# STORE SAFE REFERENCES
# ------------------------------------------------------------------------------

if "R_METHOD_METRIC_RESULTS" in T_AVAILABLE_R_OBJECTS:

    T_METRIC_RESULTS_DF = (
        T_AVAILABLE_R_OBJECTS[
            "R_METHOD_METRIC_RESULTS"
        ].copy()
    )

else:

    T_METRIC_RESULTS_DF = None

if "R_DATASET_METHOD_SUMMARY" in T_AVAILABLE_R_OBJECTS:

    T_DATASET_METHOD_SUMMARY = (
        T_AVAILABLE_R_OBJECTS[
            "R_DATASET_METHOD_SUMMARY"
        ].copy()
    )

else:

    T_DATASET_METHOD_SUMMARY = None

if "R_METHOD_NUMERIC_SUMMARY" in T_AVAILABLE_R_OBJECTS:

    T_NUMERIC_SUMMARY = (
        T_AVAILABLE_R_OBJECTS[
            "R_METHOD_NUMERIC_SUMMARY"
        ].copy()
    )

else:

    T_NUMERIC_SUMMARY = None

print()
print("-" * 100)
print("T4 STATUS")
print("-" * 100)

print(
    f"Available R result objects: "
    f"{len(T_AVAILABLE_R_OBJECTS)}"
)

print(
    "R research objects successfully discovered."
)
```

```python
# ==============================================================================
# T5 — IDENTIFY R METRIC RESULT STRUCTURE
# ==============================================================================

print()
print("=" * 100)
print("T5 — IDENTIFY R METRIC RESULT STRUCTURE")
print("=" * 100)

print()

for column in T_METRIC_RESULTS_DF.columns:

    print(
        f"{column!r}"
        f" | dtype={T_METRIC_RESULTS_DF[column].dtype}"
        f" | non-null={T_METRIC_RESULTS_DF[column].notna().sum()}"
        f" | unique={T_METRIC_RESULTS_DF[column].nunique(dropna=True)}"
    )

print()
print("-" * 100)
print("UNIQUE VALUES FOR LOW-CARDINALITY COLUMNS")
print("-" * 100)

for column in T_METRIC_RESULTS_DF.columns:

    unique_count = (
        T_METRIC_RESULTS_DF[column]
        .nunique(dropna=True)
    )

    if unique_count <= 20:

        print()
        print(f"{column}:")

        print(
            T_METRIC_RESULTS_DF[column]
            .dropna()
            .unique()
            .tolist()
        )
```

```python
# ==============================================================================
# T6 — RECOVER DATASET-LEVEL METHOD SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("T6 — RECOVER DATASET-LEVEL METHOD SUMMARY")
print("=" * 100)

if "R_DATASET_METHOD_SUMMARY" not in globals():

    raise RuntimeError(
        "T6 FAILED: R_DATASET_METHOD_SUMMARY is not available."
    )

T_DATASET_METHOD_SUMMARY = (
    R_DATASET_METHOD_SUMMARY.copy()
)

if not isinstance(
    T_DATASET_METHOD_SUMMARY,
    pd.DataFrame
):

    raise TypeError(
        "T6 FAILED: R_DATASET_METHOD_SUMMARY is not a DataFrame."
    )

if T_DATASET_METHOD_SUMMARY.empty:

    raise RuntimeError(
        "T6 FAILED: R_DATASET_METHOD_SUMMARY is empty."
    )

print()
print("Source object:")
print("  R_DATASET_METHOD_SUMMARY")

print()
print(
    f"Shape: {T_DATASET_METHOD_SUMMARY.shape}"
)

print()
print("Columns:")
print(
    T_DATASET_METHOD_SUMMARY.columns.tolist()
)

print()
print("Preview:")
print(
    T_DATASET_METHOD_SUMMARY.head(
        20
    ).to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# T8 — SELECT DATASET-WISE PERFORMANCE SOURCE
# ==============================================================================

print()
print("=" * 100)
print("T8 — SELECT DATASET-WISE PERFORMANCE SOURCE")
print("=" * 100)

# ------------------------------------------------------------------------------
# T MUST USE DATASET-WISE PERFORMANCE.
# R_FINAL_METHOD_COMPARISON ONLY CONTAINS COVERAGE INFORMATION AND IS THEREFORE
# NOT USED AS THE PERFORMANCE SOURCE.
# ------------------------------------------------------------------------------

T_PERFORMANCE_SOURCE = None
T_PERFORMANCE_SOURCE_NAME = None

if (
    "T_DATASET_METHOD_SUMMARY" in globals()
    and
    isinstance(
        T_DATASET_METHOD_SUMMARY,
        pd.DataFrame
    )
    and
    not T_DATASET_METHOD_SUMMARY.empty
):

    T_PERFORMANCE_SOURCE = (
        T_DATASET_METHOD_SUMMARY.copy()
    )

    T_PERFORMANCE_SOURCE_NAME = (
        "R_DATASET_METHOD_SUMMARY"
    )

elif (
    "T_METRIC_RESULTS_DF" in globals()
    and
    isinstance(
        T_METRIC_RESULTS_DF,
        pd.DataFrame
    )
    and
    not T_METRIC_RESULTS_DF.empty
):

    T_PERFORMANCE_SOURCE = (
        T_METRIC_RESULTS_DF.copy()
    )

    T_PERFORMANCE_SOURCE_NAME = (
        "R_METHOD_METRIC_RESULTS"
    )

else:

    raise RuntimeError(
        "T8 FAILED: No dataset-wise performance source is available."
    )

print()
print(
    f"Selected performance source: "
    f"{T_PERFORMANCE_SOURCE_NAME}"
)

print(
    f"Shape: {T_PERFORMANCE_SOURCE.shape}"
)

print()
print("Columns:")
print(
    T_PERFORMANCE_SOURCE.columns.tolist()
)
```

```python
# ==============================================================================
# T9 — BUILD DATASET-WISE PERFORMANCE TABLE
# ==============================================================================

print()
print("=" * 100)
print("T9 — BUILD DATASET-WISE PERFORMANCE TABLE")
print("=" * 100)

T_DATASET_NAMES = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

T_METHOD_NAMES = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

print()
print("Required datasets:")
print(T_DATASET_NAMES)

print()
print("Required methods:")
print(T_METHOD_NAMES)

# ------------------------------------------------------------------------------
# Detect method column
# ------------------------------------------------------------------------------

T_METHOD_COLUMN = None

for candidate in [
    "Method",
    "method",
    "Model",
    "model",
    "Generator",
    "generator",
]:

    if candidate in T_PERFORMANCE_SOURCE.columns:

        T_METHOD_COLUMN = candidate
        break

if T_METHOD_COLUMN is None:

    raise RuntimeError(
        "T9 FAILED: Method column could not be identified."
    )

# ------------------------------------------------------------------------------
# Detect dataset column
# ------------------------------------------------------------------------------

T_DATASET_COLUMN = None

for candidate in [
    "Dataset",
    "dataset",
    "Dataset Name",
    "dataset_name",
]:

    if candidate in T_PERFORMANCE_SOURCE.columns:

        T_DATASET_COLUMN = candidate
        break

if T_DATASET_COLUMN is None:

    raise RuntimeError(
        "T9 FAILED: Dataset column could not be identified "
        "in the selected performance source."
    )

print()
print(
    f"Method column  : {T_METHOD_COLUMN}"
)

print(
    f"Dataset column : {T_DATASET_COLUMN}"
)
```

```python
# ==============================================================================
# T10 — RESOLVE COMPARABLE PERFORMANCE METRIC
# ==============================================================================

print()
print("=" * 100)
print("T10 — RESOLVE COMPARABLE PERFORMANCE METRIC")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE SOURCE
# ------------------------------------------------------------------------------

if (
    "T_PERFORMANCE_SOURCE" not in globals()
    or
    T_PERFORMANCE_SOURCE is None
):

    raise RuntimeError(
        "T10 FAILED: T_PERFORMANCE_SOURCE is not available. "
        "Run T8 first."
    )

print()
print(
    f"Performance source: {T_PERFORMANCE_SOURCE_NAME}"
)

print(
    f"Shape: {T_PERFORMANCE_SOURCE.shape}"
)

# ------------------------------------------------------------------------------
# IDENTIFY NUMERIC PERFORMANCE COLUMN
# ------------------------------------------------------------------------------

T_VALUE_COLUMN = None

if "Value" in T_PERFORMANCE_SOURCE.columns:

    T_VALUE_COLUMN = "Value"

else:

    T_NUMERIC_COLUMNS = [
        column
        for column in T_PERFORMANCE_SOURCE.columns
        if pd.api.types.is_numeric_dtype(
            T_PERFORMANCE_SOURCE[column]
        )
    ]

    if len(T_NUMERIC_COLUMNS) == 1:

        T_VALUE_COLUMN = T_NUMERIC_COLUMNS[0]

    elif len(T_NUMERIC_COLUMNS) == 0:

        raise RuntimeError(
            "T10 FAILED: No numeric performance-value column was found."
        )

    else:

        raise RuntimeError(
            "T10 FAILED: Multiple numeric columns were found and "
            "the correct performance column cannot be selected safely."
        )

print()
print(
    f"Performance value column: {T_VALUE_COLUMN}"
)

# ------------------------------------------------------------------------------
# IDENTIFY METRIC COLUMN
# ------------------------------------------------------------------------------

T_METRIC_COLUMN = None

for candidate in [
    "Metric",
    "metric",
    "Metric Name",
    "metric_name",
    "Measure",
    "measure",
]:

    if candidate in T_PERFORMANCE_SOURCE.columns:

        T_METRIC_COLUMN = candidate
        break

print()

if T_METRIC_COLUMN is not None:

    print(
        f"Metric column: {T_METRIC_COLUMN}"
    )

    T_AVAILABLE_METRICS = (
        T_PERFORMANCE_SOURCE[
            T_METRIC_COLUMN
        ]
        .dropna()
        .astype(str)
        .str.strip()
        .unique()
        .tolist()
    )

    print()
    print("Available metrics:")

    for metric in T_AVAILABLE_METRICS:

        print(
            f"  - {metric}"
        )

else:

    print(
        "No explicit metric column detected."
    )

    T_AVAILABLE_METRICS = []

# ------------------------------------------------------------------------------
# IDENTIFY POSSIBLE COMPOSITE / OVERALL METRIC
# ------------------------------------------------------------------------------

T_METRIC_PRIORITY = [
    "Composite Score",
    "Composite",
    "Overall Score",
    "Overall",
    "Performance Score",
    "Performance",
    "Overall Performance",
]

T_SELECTED_METRIC = None

if T_METRIC_COLUMN is not None:

    for preferred_metric in T_METRIC_PRIORITY:

        for available_metric in T_AVAILABLE_METRICS:

            if (
                available_metric.lower()
                ==
                preferred_metric.lower()
            ):

                T_SELECTED_METRIC = available_metric
                break

        if T_SELECTED_METRIC is not None:
            break

# ------------------------------------------------------------------------------
# IF A COMPOSITE / OVERALL METRIC EXISTS
# ------------------------------------------------------------------------------

if (
    T_METRIC_COLUMN is not None
    and
    T_SELECTED_METRIC is not None
):

    T_SELECTED_PERFORMANCE_DF = (
        T_PERFORMANCE_SOURCE[
            T_PERFORMANCE_SOURCE[
                T_METRIC_COLUMN
            ]
            .astype(str)
            .str.strip()
            .str.lower()
            ==
            T_SELECTED_METRIC.lower()
        ]
        .copy()
    )

    print()
    print(
        f"Selected comparable metric: "
        f"{T_SELECTED_METRIC}"
    )

# ------------------------------------------------------------------------------
# IF ONLY ONE METRIC EXISTS
# ------------------------------------------------------------------------------

elif (
    T_METRIC_COLUMN is not None
    and
    len(T_AVAILABLE_METRICS) == 1
):

    T_SELECTED_METRIC = (
        T_AVAILABLE_METRICS[0]
    )

    T_SELECTED_PERFORMANCE_DF = (
        T_PERFORMANCE_SOURCE.copy()
    )

    print()
    print(
        f"Only one metric is available: "
        f"{T_SELECTED_METRIC}"
    )

# ------------------------------------------------------------------------------
# MULTIPLE METRICS — DO NOT SILENTLY MIX THEM
# ------------------------------------------------------------------------------

elif (
    T_METRIC_COLUMN is not None
    and
    len(T_AVAILABLE_METRICS) > 1
):

    print()
    print(
        "Multiple evaluation metrics are present."
    )

    print()
    print(
        "T10 will NOT combine them automatically because "
        "that could produce an invalid cross-dataset comparison."
    )

    print()
    print(
        "Available metrics:"
    )

    for metric in T_AVAILABLE_METRICS:

        print(
            f"  - {metric}"
        )

    raise RuntimeError(
        "T10 REVIEW REQUIRED: Select one common metric "
        "for cross-dataset robustness analysis."
    )

# ------------------------------------------------------------------------------
# NO METRIC COLUMN
# ------------------------------------------------------------------------------

else:

    T_SELECTED_PERFORMANCE_DF = (
        T_PERFORMANCE_SOURCE.copy()
    )

    print()
    print(
        "No metric dimension exists."
    )

    print(
        "The Value column will therefore be treated "
        "as the comparable performance value."
    )

# ------------------------------------------------------------------------------
# CLEAN PERFORMANCE VALUE
# ------------------------------------------------------------------------------

T_SELECTED_PERFORMANCE_DF[
    "T_Performance_Value"
] = pd.to_numeric(
    T_SELECTED_PERFORMANCE_DF[
        T_VALUE_COLUMN
    ],
    errors="coerce",
)

T_SELECTED_PERFORMANCE_DF = (
    T_SELECTED_PERFORMANCE_DF[
        T_SELECTED_PERFORMANCE_DF[
            "T_Performance_Value"
        ].notna()
    ]
    .copy()
)

print()
print("-" * 100)
print("T10 RESULT")
print("-" * 100)

print(
    f"Value column        : {T_VALUE_COLUMN}"
)

print(
    f"Selected metric     : {T_SELECTED_METRIC}"
)

print(
    f"Valid observations  : "
    f"{len(T_SELECTED_PERFORMANCE_DF)}"
)
```

```python
# ==============================================================================
# T11 — PREPARE CROSS-DATASET PERFORMANCE DATA
# ==============================================================================

print()
print("=" * 100)
print("T11 — PREPARE CROSS-DATASET PERFORMANCE DATA")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE T10 OUTPUT
# ------------------------------------------------------------------------------

if "T_SELECTED_PERFORMANCE_DF" not in globals():

    raise RuntimeError(
        "T11 FAILED: T_SELECTED_PERFORMANCE_DF is not available. "
        "Run the corrected T10 first."
    )

if not isinstance(
    T_SELECTED_PERFORMANCE_DF,
    pd.DataFrame
):

    raise TypeError(
        "T11 FAILED: T_SELECTED_PERFORMANCE_DF is not a DataFrame."
    )

if T_SELECTED_PERFORMANCE_DF.empty:

    raise RuntimeError(
        "T11 FAILED: T_SELECTED_PERFORMANCE_DF is empty."
    )

print()
print(
    f"T10 output shape: "
    f"{T_SELECTED_PERFORMANCE_DF.shape}"
)

print()
print("Available columns:")
print(
    T_SELECTED_PERFORMANCE_DF.columns.tolist()
)

# ------------------------------------------------------------------------------
# REQUIRED PERFORMANCE COLUMN
# ------------------------------------------------------------------------------

if "T_Performance_Value" not in T_SELECTED_PERFORMANCE_DF.columns:

    raise RuntimeError(
        "T11 FAILED: T_Performance_Value was not created by T10."
    )

# ------------------------------------------------------------------------------
# IDENTIFY METHOD COLUMN
# ------------------------------------------------------------------------------

T_METHOD_CANDIDATES = [
    "Method",
    "method",
    "METHOD",
    "Model",
    "model",
    "MODEL",
    "Generator",
    "generator",
    "GENERATOR",
    "Method Name",
    "method_name",
    "Method_Name",
]

T_METHOD_COLUMN = next(
    (
        column
        for column in T_METHOD_CANDIDATES
        if column in T_SELECTED_PERFORMANCE_DF.columns
    ),
    None,
)

# ------------------------------------------------------------------------------
# IDENTIFY DATASET COLUMN
# ------------------------------------------------------------------------------

T_DATASET_CANDIDATES = [
    "Dataset",
    "dataset",
    "DATASET",
    "Dataset Name",
    "dataset_name",
    "Dataset_Name",
    "DatasetName",
]

T_DATASET_COLUMN = next(
    (
        column
        for column in T_DATASET_CANDIDATES
        if column in T_SELECTED_PERFORMANCE_DF.columns
    ),
    None,
)

# ------------------------------------------------------------------------------
# DO NOT CONTINUE WITH UNRESOLVED COLUMNS
# ------------------------------------------------------------------------------

if T_METHOD_COLUMN is None:

    print()
    print("ERROR: Method column was not identified.")

    print()
    print("Available columns:")

    for column in T_SELECTED_PERFORMANCE_DF.columns:
        print(f"  - {column}")

    raise RuntimeError(
        "T11 FAILED: Method column could not be identified "
        "from T_SELECTED_PERFORMANCE_DF."
    )

if T_DATASET_COLUMN is None:

    print()
    print("ERROR: Dataset column was not identified.")

    print()
    print("Available columns:")

    for column in T_SELECTED_PERFORMANCE_DF.columns:
        print(f"  - {column}")

    raise RuntimeError(
        "T11 FAILED: Dataset column could not be identified "
        "from T_SELECTED_PERFORMANCE_DF."
    )

print()
print(
    f"Resolved Method column  : {T_METHOD_COLUMN}"
)

print(
    f"Resolved Dataset column : {T_DATASET_COLUMN}"
)

# ------------------------------------------------------------------------------
# CREATE STANDARDIZED WORKING DATAFRAME
# ------------------------------------------------------------------------------

T_WORKING_DF = (
    T_SELECTED_PERFORMANCE_DF[
        [
            T_METHOD_COLUMN,
            T_DATASET_COLUMN,
            "T_Performance_Value",
        ]
    ]
    .copy()
)

T_WORKING_DF.columns = [
    "Method",
    "Dataset",
    "Performance",
]

# ------------------------------------------------------------------------------
# CLEAN STRING VALUES
# ------------------------------------------------------------------------------

T_WORKING_DF["Method"] = (
    T_WORKING_DF["Method"]
    .astype(str)
    .str.strip()
)

T_WORKING_DF["Dataset"] = (
    T_WORKING_DF["Dataset"]
    .astype(str)
    .str.strip()
)

T_WORKING_DF["Performance"] = pd.to_numeric(
    T_WORKING_DF["Performance"],
    errors="coerce",
)

# ------------------------------------------------------------------------------
# METHOD NORMALIZATION
# ------------------------------------------------------------------------------

T_METHOD_MAP = {
    "ctgan": "CTGAN",
    "CTGAN": "CTGAN",

    "tvae": "TVAE",
    "TVAE": "TVAE",

    "dp-ctgan": "DP-CTGAN",
    "DP-CTGAN": "DP-CTGAN",
    "dp_ctgan": "DP-CTGAN",
    "DP_CTGAN": "DP-CTGAN",

    "spp-gan": "SPP-GAN",
    "SPP-GAN": "SPP-GAN",
    "spp_gan": "SPP-GAN",
    "SPP_GAN": "SPP-GAN",
}

T_WORKING_DF["Method"] = (
    T_WORKING_DF["Method"]
    .map(T_METHOD_MAP)
    .fillna(
        T_WORKING_DF["Method"]
    )
)

# ------------------------------------------------------------------------------
# DATASET NORMALIZATION
# ------------------------------------------------------------------------------

T_DATASET_MAP = {
    "adult": "adult_income",
    "adult_income": "adult_income",
    "Adult Income": "adult_income",
    "adult-income": "adult_income",

    "bank": "bank_marketing",
    "bank_marketing": "bank_marketing",
    "Bank Marketing": "bank_marketing",
    "bank-marketing": "bank_marketing",

    "diabetes": "diabetes_130us",
    "diabetes_130us": "diabetes_130us",
    "Diabetes": "diabetes_130us",
    "diabetes-130us": "diabetes_130us",
}

T_WORKING_DF["Dataset"] = (
    T_WORKING_DF["Dataset"]
    .map(T_DATASET_MAP)
    .fillna(
        T_WORKING_DF["Dataset"]
    )
)

# ------------------------------------------------------------------------------
# REQUIRED EXPERIMENTAL SET
# ------------------------------------------------------------------------------

T_METHOD_NAMES = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

T_DATASET_NAMES = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

# ------------------------------------------------------------------------------
# FILTER VALID RESEARCH RECORDS
# ------------------------------------------------------------------------------

T_WORKING_DF = T_WORKING_DF[
    T_WORKING_DF["Method"].isin(
        T_METHOD_NAMES
    )
    &
    T_WORKING_DF["Dataset"].isin(
        T_DATASET_NAMES
    )
    &
    T_WORKING_DF["Performance"].notna()
].copy()

T_WORKING_DF.reset_index(
    drop=True,
    inplace=True,
)

# ------------------------------------------------------------------------------
# DISPLAY RESULT
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("T11 — STANDARDIZED PERFORMANCE DATA")
print("-" * 100)

print()
print(
    T_WORKING_DF.to_string(
        index=False
    )
)

print()
print(
    f"Valid observations: "
    f"{len(T_WORKING_DF)}"
)

print()
print(
    "Methods detected:"
)

for method in sorted(
    T_WORKING_DF["Method"].unique()
):

    print(
        f"  - {method}"
    )

print()
print(
    "Datasets detected:"
)

for dataset in sorted(
    T_WORKING_DF["Dataset"].unique()
):

    print(
        f"  - {dataset}"
    )
```

```python
# ==============================================================================
# T12 — CROSS-DATASET PERFORMANCE MATRIX
# ==============================================================================

print()
print("=" * 100)
print("T12 — CROSS-DATASET PERFORMANCE MATRIX")
print("=" * 100)

T_CROSS_DATASET_MATRIX = (
    T_WORKING_DF
    .pivot_table(
        index="Method",
        columns="Dataset",
        values="Performance",
        aggfunc="mean",
    )
    .reindex(
        index=T_METHOD_NAMES,
        columns=T_DATASET_NAMES,
    )
)

print()
print(
    T_CROSS_DATASET_MATRIX.to_string()
)

print()

if T_CROSS_DATASET_MATRIX.isna().any().any():

    print(
        "WARNING: Incomplete method-dataset coverage detected."
    )

else:

    print(
        "Coverage status: COMPLETE"
    )
```

```python
# ==============================================================================
# T13 — CROSS-DATASET MEAN AND VARIABILITY
# ==============================================================================

print()
print("=" * 100)
print("T13 — CROSS-DATASET MEAN AND VARIABILITY")
print("=" * 100)

T_ROBUSTNESS_RESULTS = pd.DataFrame(
    index=T_CROSS_DATASET_MATRIX.index
)

T_ROBUSTNESS_RESULTS[
    "Mean Performance"
] = (
    T_CROSS_DATASET_MATRIX.mean(
        axis=1
    )
)

T_ROBUSTNESS_RESULTS[
    "Standard Deviation"
] = (
    T_CROSS_DATASET_MATRIX.std(
        axis=1,
        ddof=1,
    )
)

T_ROBUSTNESS_RESULTS[
    "Coefficient of Variation"
] = (
    T_ROBUSTNESS_RESULTS[
        "Standard Deviation"
    ]
    /
    T_ROBUSTNESS_RESULTS[
        "Mean Performance"
    ].abs()
)

T_ROBUSTNESS_RESULTS[
    "Worst-Dataset Performance"
] = (
    T_CROSS_DATASET_MATRIX.min(
        axis=1
    )
)

T_ROBUSTNESS_RESULTS[
    "Best-Dataset Performance"
] = (
    T_CROSS_DATASET_MATRIX.max(
        axis=1
    )
)

T_ROBUSTNESS_RESULTS[
    "Performance Range"
] = (
    T_ROBUSTNESS_RESULTS[
        "Best-Dataset Performance"
    ]
    -
    T_ROBUSTNESS_RESULTS[
        "Worst-Dataset Performance"
    ]
)

T_ROBUSTNESS_RESULTS.reset_index(
    inplace=True
)

T_ROBUSTNESS_RESULTS.rename(
    columns={
        "index": "Method"
    },
    inplace=True,
)

print()
print(
    T_ROBUSTNESS_RESULTS.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# T14 — CROSS-DATASET ROBUSTNESS SCORE
# ==============================================================================

print()
print("=" * 100)
print("T14 — CROSS-DATASET ROBUSTNESS SCORE")
print("=" * 100)

T_ROBUSTNESS_RESULTS[
    "Robustness Score"
] = (
    T_ROBUSTNESS_RESULTS[
        "Mean Performance"
    ]
    -
    T_LAMBDA
    *
    T_ROBUSTNESS_RESULTS[
        "Standard Deviation"
    ]
)

print()
print(
    T_ROBUSTNESS_RESULTS[
        [
            "Method",
            "Mean Performance",
            "Standard Deviation",
            "Coefficient of Variation",
            "Worst-Dataset Performance",
            "Robustness Score",
        ]
    ].to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# T15 — CROSS-DATASET ROBUSTNESS RANKING
# ==============================================================================

print()
print("=" * 100)
print("T15 — CROSS-DATASET ROBUSTNESS RANKING")
print("=" * 100)

T_METHOD_ROBUSTNESS_RANKING = (
    T_ROBUSTNESS_RESULTS
    .sort_values(
        by="Robustness Score",
        ascending=False,
    )
    .reset_index(
        drop=True
    )
)

T_METHOD_ROBUSTNESS_RANKING.insert(
    0,
    "Robustness Rank",
    range(
        1,
        len(T_METHOD_ROBUSTNESS_RANKING) + 1
    ),
)

print()
print(
    T_METHOD_ROBUSTNESS_RANKING.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# T16 — CROSS-DATASET SENSITIVITY
# ==============================================================================

print()
print("=" * 100)
print("T16 — CROSS-DATASET SENSITIVITY")
print("=" * 100)

T_SENSITIVITY_RECORDS = []

for method in T_METHOD_NAMES:

    values = (
        T_CROSS_DATASET_MATRIX
        .loc[
            method,
            T_DATASET_NAMES
        ]
    )

    best_value = values.max()
    worst_value = values.min()

    best_dataset = values.idxmax()
    worst_dataset = values.idxmin()

    if np.isclose(
        best_value,
        0.0
    ):

        relative_degradation = np.nan

    else:

        relative_degradation = (
            best_value - worst_value
        ) / abs(best_value)

    T_SENSITIVITY_RECORDS.append(
        {
            "Method": method,
            "Best Dataset": best_dataset,
            "Best Performance": best_value,
            "Worst Dataset": worst_dataset,
            "Worst Performance": worst_value,
            "Relative Degradation": relative_degradation,
        }
    )

T_SENSITIVITY_DF = pd.DataFrame(
    T_SENSITIVITY_RECORDS
)

print()
print(
    T_SENSITIVITY_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# T17 — FINAL CROSS-DATASET ROBUSTNESS SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("T17 — FINAL CROSS-DATASET ROBUSTNESS SUMMARY")
print("=" * 100)

T_CROSS_DATASET_ROBUSTNESS = (
    T_METHOD_ROBUSTNESS_RANKING
    .merge(
        T_SENSITIVITY_DF,
        on="Method",
        how="left",
    )
)

print()
print(
    T_CROSS_DATASET_ROBUSTNESS.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# MOST ROBUST METHOD
# ------------------------------------------------------------------------------

T_MOST_ROBUST_METHOD = (
    T_METHOD_ROBUSTNESS_RANKING
    .iloc[0]["Method"]
)

T_MOST_ROBUST_SCORE = (
    T_METHOD_ROBUSTNESS_RANKING
    .iloc[0]["Robustness Score"]
)

print()
print("-" * 100)
print("MOST ROBUST METHOD")
print("-" * 100)

print(
    f"Method           : {T_MOST_ROBUST_METHOD}"
)

print(
    f"Robustness Score : {T_MOST_ROBUST_SCORE:.6f}"
)
```

```python
# ==============================================================================
# T18 — CROSS-DATASET ROBUSTNESS VALIDATION
# ==============================================================================

print()
print("=" * 100)
print("T18 — CROSS-DATASET ROBUSTNESS VALIDATION")
print("=" * 100)

T_VALIDATION = []

T_VALIDATION.append(
    {
        "Check": "R Method Metric Results Available",
        "Status": (
            "R_METHOD_METRIC_RESULTS"
            in globals()
        ),
    }
)

T_VALIDATION.append(
    {
        "Check": "Dataset Method Summary Available",
        "Status": (
            "R_DATASET_METHOD_SUMMARY"
            in globals()
        ),
    }
)

T_VALIDATION.append(
    {
        "Check": "Required Methods Available",
        "Status": set(T_METHOD_NAMES).issubset(
            set(T_WORKING_DF["Method"].unique())
        ),
    }
)

T_VALIDATION.append(
    {
        "Check": "Required Datasets Available",
        "Status": set(T_DATASET_NAMES).issubset(
            set(T_WORKING_DF["Dataset"].unique())
        ),
    }
)

T_VALIDATION.append(
    {
        "Check": "Complete Method-Dataset Matrix",
        "Status": not T_CROSS_DATASET_MATRIX.isna().any().any(),
    }
)

T_VALIDATION.append(
    {
        "Check": "Mean Performance Calculated",
        "Status": not T_ROBUSTNESS_RESULTS[
            "Mean Performance"
        ].isna().any(),
    }
)

T_VALIDATION.append(
    {
        "Check": "Robustness Score Calculated",
        "Status": not T_ROBUSTNESS_RESULTS[
            "Robustness Score"
        ].isna().any(),
    }
)

T_VALIDATION.append(
    {
        "Check": "Ranking Generated",
        "Status": (
            len(T_METHOD_ROBUSTNESS_RANKING)
            == len(T_METHOD_NAMES)
        ),
    }
)

T_VALIDATION_DF = pd.DataFrame(
    T_VALIDATION
)

print()
print(
    T_VALIDATION_DF.to_string(
        index=False
    )
)

T_PASS_COUNT = int(
    T_VALIDATION_DF["Status"].sum()
)

T_TOTAL_CHECKS = len(
    T_VALIDATION_DF
)

print()
print(
    f"Passed checks : {T_PASS_COUNT}/{T_TOTAL_CHECKS}"
)

if T_PASS_COUNT == T_TOTAL_CHECKS:

    T_STATUS = "PASS"

    print(
        "T STATUS: PASS"
    )

else:

    T_STATUS = "REVIEW"

    print(
        "T STATUS: REVIEW"
    )
```

```python
# ==============================================================================
# U1 — STATISTICAL TESTING CONFIGURATION
# ==============================================================================

print()
print("=" * 100)
print("U1 — STATISTICAL TESTING CONFIGURATION")
print("=" * 100)

U_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

U_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

U_ALPHA = 0.05

U_MULTIPLE_TESTING_METHOD = "holm"

U_PRIMARY_TEST = "Friedman"

U_POSTHOC_TEST = "Wilcoxon signed-rank"

print()
print("Datasets:")
for dataset in U_DATASETS:
    print(f"  - {dataset}")

print()
print("Methods:")
for method in U_METHODS:
    print(f"  - {method}")

print()
print(f"Significance level       : α = {U_ALPHA}")
print(f"Multiple-testing method  : {U_MULTIPLE_TESTING_METHOD}")
print(f"Primary omnibus test     : {U_PRIMARY_TEST}")
print(f"Post-hoc paired test     : {U_POSTHOC_TEST}")
```

```python
# ==============================================================================
# U2 — RECOVER CROSS-DATASET PERFORMANCE MATRIX
# ==============================================================================

print()
print("=" * 100)
print("U2 — RECOVER CROSS-DATASET PERFORMANCE MATRIX")
print("=" * 100)

if "T_CROSS_DATASET_MATRIX" not in globals():

    raise RuntimeError(
        "U2 FAILED: T_CROSS_DATASET_MATRIX is not available. "
        "Complete the T — Cross-Dataset Robustness section first."
    )

U_PERFORMANCE_MATRIX = (
    T_CROSS_DATASET_MATRIX.copy()
)

U_PERFORMANCE_MATRIX = (
    U_PERFORMANCE_MATRIX
    .reindex(
        index=U_METHODS,
        columns=U_DATASETS,
    )
)

print()
print(
    U_PERFORMANCE_MATRIX.to_string()
)

print()

if U_PERFORMANCE_MATRIX.isna().any().any():

    print(
        "WARNING: Missing observations detected."
    )

else:

    print(
        "Coverage status: COMPLETE"
    )
```

```python
# ==============================================================================
# U3 — VALIDATE AND DIAGNOSE PAIRED EXPERIMENTAL STRUCTURE
# ==============================================================================

print()
print("=" * 100)
print("U3 — VALIDATE AND DIAGNOSE PAIRED EXPERIMENTAL STRUCTURE")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE PERFORMANCE MATRIX
# ------------------------------------------------------------------------------

if "U_PERFORMANCE_MATRIX" not in globals():

    raise RuntimeError(
        "U3 FAILED: U_PERFORMANCE_MATRIX is not available. "
        "Run U2 first."
    )

if not isinstance(
    U_PERFORMANCE_MATRIX,
    pd.DataFrame
):

    raise TypeError(
        "U3 FAILED: U_PERFORMANCE_MATRIX must be a pandas DataFrame."
    )

# ------------------------------------------------------------------------------
# FORCE EXPECTED METHOD / DATASET ORDER
# ------------------------------------------------------------------------------

U_PERFORMANCE_MATRIX = (
    U_PERFORMANCE_MATRIX
    .reindex(
        index=U_METHODS,
        columns=U_DATASETS,
    )
)

print()
print("Current performance matrix:")
print()

print(
    U_PERFORMANCE_MATRIX.to_string()
)

# ------------------------------------------------------------------------------
# IDENTIFY MISSING CELLS
# ------------------------------------------------------------------------------

U_MISSING_CELLS = []

for method in U_METHODS:

    for dataset in U_DATASETS:

        value = (
            U_PERFORMANCE_MATRIX
            .loc[
                method,
                dataset,
            ]
        )

        if pd.isna(value):

            U_MISSING_CELLS.append(
                {
                    "Method": method,
                    "Dataset": dataset,
                }
            )

U_MISSING_CELLS_DF = pd.DataFrame(
    U_MISSING_CELLS
)

# ------------------------------------------------------------------------------
# DISPLAY MISSING CELLS
# ------------------------------------------------------------------------------

print()
print("-" * 100)
print("MISSING METHOD × DATASET COMBINATIONS")
print("-" * 100)

if U_MISSING_CELLS_DF.empty:

    print(
        "None — complete paired structure."
    )

else:

    print(
        U_MISSING_CELLS_DF.to_string(
            index=False
        )
    )

# ------------------------------------------------------------------------------
# COUNT OBSERVATIONS PER METHOD
# ------------------------------------------------------------------------------

U_METHOD_COVERAGE = []

for method in U_METHODS:

    values = (
        U_PERFORMANCE_MATRIX
        .loc[
            method,
            U_DATASETS,
        ]
    )

    valid_count = int(
        values.notna().sum()
    )

    U_METHOD_COVERAGE.append(
        {
            "Method": method,
            "Valid Datasets": valid_count,
            "Expected Datasets": len(U_DATASETS),
            "Complete": (
                valid_count
                ==
                len(U_DATASETS)
            ),
        }
    )

U_METHOD_COVERAGE_DF = pd.DataFrame(
    U_METHOD_COVERAGE
)

print()
print("-" * 100)
print("METHOD-LEVEL COVERAGE")
print("-" * 100)

print(
    U_METHOD_COVERAGE_DF.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# COUNT OBSERVATIONS PER DATASET
# ------------------------------------------------------------------------------

U_DATASET_COVERAGE = []

for dataset in U_DATASETS:

    values = (
        U_PERFORMANCE_MATRIX
        .loc[
            U_METHODS,
            dataset,
        ]
    )

    valid_count = int(
        values.notna().sum()
    )

    U_DATASET_COVERAGE.append(
        {
            "Dataset": dataset,
            "Valid Methods": valid_count,
            "Expected Methods": len(U_METHODS),
            "Complete": (
                valid_count
                ==
                len(U_METHODS)
            ),
        }
    )

U_DATASET_COVERAGE_DF = pd.DataFrame(
    U_DATASET_COVERAGE
)

print()
print("-" * 100)
print("DATASET-LEVEL COVERAGE")
print("-" * 100)

print(
    U_DATASET_COVERAGE_DF.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# DETERMINE COMMON COMPLETE DATASETS
# ------------------------------------------------------------------------------

U_COMPLETE_DATASETS = []

for dataset in U_DATASETS:

    column_values = (
        U_PERFORMANCE_MATRIX[
            dataset
        ]
    )

    if column_values.notna().all():

        U_COMPLETE_DATASETS.append(
            dataset
        )

# ------------------------------------------------------------------------------
# DETERMINE METHODS WITH COMPLETE COVERAGE
# ------------------------------------------------------------------------------

U_COMPLETE_METHODS = []

for method in U_METHODS:

    row_values = (
        U_PERFORMANCE_MATRIX
        .loc[
            method
        ]
    )

    if row_values.notna().all():

        U_COMPLETE_METHODS.append(
            method
        )

# ------------------------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------------------------

U_COMPLETE_PAIRED_STRUCTURE = (
    len(U_COMPLETE_DATASETS)
    ==
    len(U_DATASETS)
    and
    len(U_COMPLETE_METHODS)
    ==
    len(U_METHODS)
)

print()
print("=" * 100)
print("U3 — PAIRED STRUCTURE SUMMARY")
print("=" * 100)

print()
print(
    f"Expected methods          : "
    f"{len(U_METHODS)}"
)

print(
    f"Expected datasets         : "
    f"{len(U_DATASETS)}"
)

print(
    f"Complete methods          : "
    f"{len(U_COMPLETE_METHODS)}"
)

print(
    f"Complete datasets         : "
    f"{len(U_COMPLETE_DATASETS)}"
)

print(
    f"Missing cells             : "
    f"{len(U_MISSING_CELLS_DF)}"
)

print()
print(
    "Complete methods:"
)

for method in U_COMPLETE_METHODS:

    print(
        f"  - {method}"
    )

print()
print(
    "Complete datasets:"
)

for dataset in U_COMPLETE_DATASETS:

    print(
        f"  - {dataset}"
    )

# ------------------------------------------------------------------------------
# DO NOT FAIL IMMEDIATELY
# ------------------------------------------------------------------------------

if U_COMPLETE_PAIRED_STRUCTURE:

    U_PAIRED_TEST_SUPPORTED = True

    print()
    print(
        "U3 STATUS: COMPLETE PAIRED STRUCTURE"
    )

    print(
        "The planned Friedman/Wilcoxon analysis is supported."
    )

else:

    U_PAIRED_TEST_SUPPORTED = False

    print()
    print(
        "U3 STATUS: INCOMPLETE PAIRED STRUCTURE"
    )

    print(
        "The full 4-method × 3-dataset paired analysis "
        "cannot yet be performed."
    )

    print()
    print(
        "No missing values will be imputed or fabricated."
    )
```

```python
# ==============================================================================
# U4 — PREPARE PAIRED OBSERVATIONS
# ==============================================================================

print()
print("=" * 100)
print("U4 — PREPARE PAIRED OBSERVATIONS")
print("=" * 100)

U_PAIRED_DF = (
    U_PERFORMANCE_MATRIX
    .T
    .copy()
)

U_PAIRED_DF.index.name = "Dataset"

print()
print(
    U_PAIRED_DF.to_string()
)

print()
print(
    f"Paired blocks   : {len(U_PAIRED_DF)}"
)

print(
    f"Methods tested  : {len(U_PAIRED_DF.columns)}"
)
```

```python
# ==============================================================================
# U5 — FRIEDMAN OMNIBUS TEST
# ==============================================================================

print()
print("=" * 100)
print("U5 — FRIEDMAN OMNIBUS TEST")
print("=" * 100)

from scipy.stats import friedmanchisquare

U_FRIEDMAN_VALUES = [
    U_PERFORMANCE_MATRIX.loc[
        method,
        U_DATASETS,
    ].to_numpy(
        dtype=float
    )
    for method in U_METHODS
]

U_FRIEDMAN_STATISTIC, U_FRIEDMAN_PVALUE = (
    friedmanchisquare(
        *U_FRIEDMAN_VALUES
    )
)

U_FRIEDMAN_RESULT = pd.DataFrame(
    [
        {
            "Test": "Friedman",
            "Statistic": U_FRIEDMAN_STATISTIC,
            "p-value": U_FRIEDMAN_PVALUE,
            "Alpha": U_ALPHA,
            "Significant": (
                U_FRIEDMAN_PVALUE
                < U_ALPHA
            ),
            "Datasets": len(U_DATASETS),
            "Methods": len(U_METHODS),
        }
    ]
)

print()
print(
    U_FRIEDMAN_RESULT.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# U6 — FRIEDMAN EFFECT SIZE — KENDALL'S W
# ==============================================================================

print()
print("=" * 100)
print("U6 — FRIEDMAN EFFECT SIZE — KENDALL'S W")
print("=" * 100)

U_N_BLOCKS = len(U_DATASETS)
U_K_METHODS = len(U_METHODS)

U_FRIEDMAN_EFFECT = (
    U_FRIEDMAN_STATISTIC
    /
    (
        U_N_BLOCKS
        *
        (
            U_K_METHODS
            - 1
        )
    )
)

U_KENDALL_W = U_FRIEDMAN_EFFECT

def U_INTERPRET_KENDALL_W(w):

    if w < 0.10:
        return "Negligible"

    if w < 0.30:
        return "Small"

    if w < 0.50:
        return "Moderate"

    return "Large"


U_KENDALL_INTERPRETATION = (
    U_INTERPRET_KENDALL_W(
        U_KENDALL_W
    )
)

print()
print(
    f"Friedman statistic : "
    f"{U_FRIEDMAN_STATISTIC:.6f}"
)

print(
    f"Kendall's W       : "
    f"{U_KENDALL_W:.6f}"
)

print(
    f"Effect magnitude   : "
    f"{U_KENDALL_INTERPRETATION}"
)
```

```python
# ==============================================================================
# U7 — PAIRWISE WILCOXON SIGNED-RANK TESTS
# ==============================================================================

print()
print("=" * 100)
print("U7 — PAIRWISE WILCOXON SIGNED-RANK TESTS")
print("=" * 100)

from itertools import combinations
from scipy.stats import wilcoxon

U_PAIRWISE_RESULTS = []

for method_a, method_b in combinations(
    U_METHODS,
    2,
):

    values_a = (
        U_PERFORMANCE_MATRIX
        .loc[
            method_a,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    values_b = (
        U_PERFORMANCE_MATRIX
        .loc[
            method_b,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    differences = values_a - values_b

    non_zero_differences = (
        differences[
            ~np.isclose(
                differences,
                0.0
            )
        ]
    )

    if len(non_zero_differences) == 0:

        statistic = 0.0
        p_value = 1.0

    else:

        try:

            statistic, p_value = wilcoxon(
                values_a,
                values_b,
                alternative="two-sided",
                zero_method="wilcox",
                method="auto",
            )

        except ValueError:

            statistic = np.nan
            p_value = np.nan

    U_PAIRWISE_RESULTS.append(
        {
            "Method A": method_a,
            "Method B": method_b,
            "Wilcoxon Statistic": statistic,
            "p-value": p_value,
            "N Paired Datasets": len(U_DATASETS),
            "Median Difference": np.median(
                differences
            ),
            "Mean Difference": np.mean(
                differences
            ),
        }
    )

U_PAIRWISE_DF = pd.DataFrame(
    U_PAIRWISE_RESULTS
)

print()
print(
    U_PAIRWISE_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# U8 — HOLM MULTIPLE-COMPARISON CORRECTION
# ==============================================================================

print()
print("=" * 100)
print("U8 — HOLM MULTIPLE-COMPARISON CORRECTION")
print("=" * 100)

from scipy.stats import false_discovery_control

U_RAW_PVALUES = (
    U_PAIRWISE_DF[
        "p-value"
    ]
    .to_numpy(
        dtype=float
    )
)

U_VALID_PVALUE_MASK = (
    np.isfinite(
        U_RAW_PVALUES
    )
)

U_ADJUSTED_PVALUES = np.full(
    len(U_RAW_PVALUES),
    np.nan,
)

if U_VALID_PVALUE_MASK.any():

    U_ADJUSTED_PVALUES[
        U_VALID_PVALUE_MASK
    ] = false_discovery_control(
        U_RAW_PVALUES[
            U_VALID_PVALUE_MASK
        ],
        method="holm",
    )

U_PAIRWISE_DF[
    "Holm Adjusted p-value"
] = U_ADJUSTED_PVALUES

U_PAIRWISE_DF[
    "Significant After Holm"
] = (
    U_PAIRWISE_DF[
        "Holm Adjusted p-value"
    ]
    < U_ALPHA
)

print()
print(
    U_PAIRWISE_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# U9 — PAIRWISE EFFECT SIZE
# ==============================================================================

print()
print("=" * 100)
print("U9 — PAIRWISE EFFECT SIZE")
print("=" * 100)

U_PAIRWISE_DF[
    "Rank-Biserial Effect Size"
] = np.nan

for i, row in U_PAIRWISE_DF.iterrows():

    method_a = row["Method A"]
    method_b = row["Method B"]

    values_a = (
        U_PERFORMANCE_MATRIX
        .loc[
            method_a,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    values_b = (
        U_PERFORMANCE_MATRIX
        .loc[
            method_b,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    differences = (
        values_a - values_b
    )

    positive = np.sum(
        differences > 0
    )

    negative = np.sum(
        differences < 0
    )

    non_zero = (
        positive + negative
    )

    if non_zero > 0:

        rank_biserial = (
            positive - negative
        ) / non_zero

    else:

        rank_biserial = 0.0

    U_PAIRWISE_DF.loc[
        i,
        "Rank-Biserial Effect Size"
    ] = rank_biserial

def U_INTERPRET_EFFECT(effect):

    absolute_effect = abs(effect)

    if absolute_effect < 0.10:
        return "Negligible"

    if absolute_effect < 0.30:
        return "Small"

    if absolute_effect < 0.50:
        return "Moderate"

    return "Large"


U_PAIRWISE_DF[
    "Effect Interpretation"
] = (
    U_PAIRWISE_DF[
        "Rank-Biserial Effect Size"
    ]
    .apply(
        U_INTERPRET_EFFECT
    )
)

print()
print(
    U_PAIRWISE_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# U10 — METHOD-LEVEL SIGNIFICANCE SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("U10 — METHOD-LEVEL SIGNIFICANCE SUMMARY")
print("=" * 100)

U_METHOD_SUMMARY = []

for method in U_METHODS:

    comparisons = U_PAIRWISE_DF[
        (
            U_PAIRWISE_DF["Method A"]
            == method
        )
        |
        (
            U_PAIRWISE_DF["Method B"]
            == method
        )
    ]

    significant_count = int(
        comparisons[
            "Significant After Holm"
        ].sum()
    )

    total_comparisons = len(
        comparisons
    )

    U_METHOD_SUMMARY.append(
        {
            "Method": method,
            "Mean Performance": (
                U_PERFORMANCE_MATRIX
                .loc[
                    method
                ]
                .mean()
            ),
            "Standard Deviation": (
                U_PERFORMANCE_MATRIX
                .loc[
                    method
                ]
                .std(
                    ddof=1
                )
            ),
            "Significant Pairwise Comparisons": (
                significant_count
            ),
            "Total Pairwise Comparisons": (
                total_comparisons
            ),
        }
    )

U_METHOD_SIGNIFICANCE_DF = (
    pd.DataFrame(
        U_METHOD_SUMMARY
    )
)

print()
print(
    U_METHOD_SIGNIFICANCE_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# U11 — STATISTICAL DECISION SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("U11 — STATISTICAL DECISION SUMMARY")
print("=" * 100)

U_OMNIBUS_SIGNIFICANT = (
    U_FRIEDMAN_PVALUE
    < U_ALPHA
)

if U_OMNIBUS_SIGNIFICANT:

    U_OMNIBUS_DECISION = (
        "Reject H0 — method differences detected"
    )

else:

    U_OMNIBUS_DECISION = (
        "Fail to reject H0 — no statistically significant "
        "overall method difference detected"
    )

print()
print(
    f"Friedman p-value : "
    f"{U_FRIEDMAN_PVALUE:.6f}"
)

print(
    f"Decision         : "
    f"{U_OMNIBUS_DECISION}"
)

print()
print(
    f"Kendall's W     : "
    f"{U_KENDALL_W:.6f}"
)

print(
    f"Effect magnitude : "
    f"{U_KENDALL_INTERPRETATION}"
)

print()
print(
    f"Holm-adjusted significant pairwise tests: "
    f"{int(U_PAIRWISE_DF['Significant After Holm'].sum())}"
    f"/{len(U_PAIRWISE_DF)}"
)
```

```python
# ==============================================================================
# U12 — STATISTICAL SIGNIFICANCE VALIDATION
# ==============================================================================

print()
print("=" * 100)
print("U12 — STATISTICAL SIGNIFICANCE VALIDATION")
print("=" * 100)

U_FINAL_VALIDATION = []

U_FINAL_VALIDATION.append(
    {
        "Check": "Performance matrix available",
        "Status": (
            isinstance(
                U_PERFORMANCE_MATRIX,
                pd.DataFrame,
            )
            and
            not U_PERFORMANCE_MATRIX.empty
        ),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Complete paired observations",
        "Status": not U_PERFORMANCE_MATRIX.isna().any().any(),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Friedman test completed",
        "Status": np.isfinite(
            U_FRIEDMAN_STATISTIC
        ),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Friedman p-value completed",
        "Status": np.isfinite(
            U_FRIEDMAN_PVALUE
        ),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Kendall W completed",
        "Status": np.isfinite(
            U_KENDALL_W
        ),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Pairwise tests completed",
        "Status": (
            len(U_PAIRWISE_DF)
            == (
                len(U_METHODS)
                *
                (len(U_METHODS) - 1)
                // 2
            )
        ),
    }
)

U_FINAL_VALIDATION.append(
    {
        "Check": "Multiple-testing correction completed",
        "Status": (
            "Holm Adjusted p-value"
            in U_PAIRWISE_DF.columns
        ),
    }
)

U_FINAL_VALIDATION_DF = pd.DataFrame(
    U_FINAL_VALIDATION
)

print()
print(
    U_FINAL_VALIDATION_DF.to_string(
        index=False
    )
)

U_PASSED = int(
    U_FINAL_VALIDATION_DF[
        "Status"
    ].sum()
)

U_TOTAL = len(
    U_FINAL_VALIDATION_DF
)

print()
print(
    f"Passed checks : {U_PASSED}/{U_TOTAL}"
)

if U_PASSED == U_TOTAL:

    U_STATUS = "PASS"

    print(
        "U STATUS: PASS"
    )

else:

    U_STATUS = "REVIEW"

    print(
        "U STATUS: REVIEW"
    )
```

```python
# ==============================================================================
# U13 — SAVE STATISTICAL SIGNIFICANCE RESULTS
# ==============================================================================

print()
print("=" * 100)
print("U13 — SAVE STATISTICAL SIGNIFICANCE RESULTS")
print("=" * 100)

U_OUTPUT_DIRECTORY = pathlib.Path(
    "statistical_significance"
)

U_OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)

U_PERFORMANCE_MATRIX.to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_performance_matrix.csv"
)

U_FRIEDMAN_RESULT.to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_friedman_test.csv",
    index=False,
)

pd.DataFrame(
    [
        {
            "Kendall_W": U_KENDALL_W,
            "Effect_Interpretation":
                U_KENDALL_INTERPRETATION,
            "Friedman_Statistic":
                U_FRIEDMAN_STATISTIC,
            "Friedman_p_value":
                U_FRIEDMAN_PVALUE,
            "Alpha":
                U_ALPHA,
        }
    ]
).to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_friedman_effect_size.csv",
    index=False,
)

U_PAIRWISE_DF.to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_pairwise_wilcoxon_tests.csv",
    index=False,
)

U_METHOD_SIGNIFICANCE_DF.to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_method_significance_summary.csv",
    index=False,
)

U_FINAL_VALIDATION_DF.to_csv(
    U_OUTPUT_DIRECTORY
    /
    "U_validation.csv",
    index=False,
)

print()

print(
    f"Output directory: "
    f"{U_OUTPUT_DIRECTORY}"
)

print()

print(
    f"U STATUS: {U_STATUS}"
)
```

```python
# ==============================================================================
# V1 — RECOVER U STATISTICAL RESULTS
# ==============================================================================

print()
print("=" * 100)
print("V1 — RECOVER U STATISTICAL RESULTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# REQUIRED U OBJECTS
# ------------------------------------------------------------------------------

V_REQUIRED_OBJECTS = [
    "U_PERFORMANCE_MATRIX",
    "U_PAIRWISE_DF",
]

V_MISSING_OBJECTS = [
    name
    for name in V_REQUIRED_OBJECTS
    if name not in globals()
]

if V_MISSING_OBJECTS:

    raise RuntimeError(
        "V1 FAILED: Required U objects are missing: "
        + ", ".join(V_MISSING_OBJECTS)
    )

# ------------------------------------------------------------------------------
# COPY U RESULTS
# ------------------------------------------------------------------------------

V_PERFORMANCE_MATRIX = (
    U_PERFORMANCE_MATRIX.copy()
)

V_PAIRWISE_DF = (
    U_PAIRWISE_DF.copy()
)

print()
print(
    f"Performance matrix shape : "
    f"{V_PERFORMANCE_MATRIX.shape}"
)

print(
    f"Pairwise comparisons     : "
    f"{len(V_PAIRWISE_DF)}"
)

print()
print(
    V_PERFORMANCE_MATRIX.to_string()
)
```

```python
# ==============================================================================
# V2 — VALIDATE PAIRWISE STATISTICAL COMPARISONS
# ==============================================================================

print()
print("=" * 100)
print("V2 — VALIDATE PAIRWISE STATISTICAL COMPARISONS")
print("=" * 100)

V_REQUIRED_PAIRWISE_COLUMNS = [
    "Method A",
    "Method B",
    "p-value",
]

V_MISSING_COLUMNS = [
    column
    for column in V_REQUIRED_PAIRWISE_COLUMNS
    if column not in V_PAIRWISE_DF.columns
]

if V_MISSING_COLUMNS:

    raise RuntimeError(
        "V2 FAILED: Required pairwise columns are missing: "
        + ", ".join(V_MISSING_COLUMNS)
    )

# ------------------------------------------------------------------------------
# EXPECTED NUMBER OF PAIRWISE COMPARISONS
# ------------------------------------------------------------------------------

V_EXPECTED_COMPARISONS = (
    len(U_METHODS)
    *
    (len(U_METHODS) - 1)
    //
    2
)

print()
print(
    f"Expected pairwise comparisons : "
    f"{V_EXPECTED_COMPARISONS}"
)

print(
    f"Observed pairwise comparisons : "
    f"{len(V_PAIRWISE_DF)}"
)

if len(V_PAIRWISE_DF) != V_EXPECTED_COMPARISONS:

    print()
    print(
        "WARNING: The observed number of pairwise comparisons "
        "does not equal the expected number."
    )

else:

    print()
    print(
        "Pairwise comparison structure: VALID"
    )
```

```python
# ==============================================================================
# V3 — CALCULATE PAIRED RANK-BISERIAL EFFECT SIZES
# ==============================================================================

print()
print("=" * 100)
print("V3 — CALCULATE PAIRED RANK-BISERIAL EFFECT SIZES")
print("=" * 100)

V_PAIRWISE_DF[
    "Rank-Biserial Effect Size"
] = np.nan

V_PAIRWISE_DF[
    "Positive Differences"
] = np.nan

V_PAIRWISE_DF[
    "Negative Differences"
] = np.nan

V_PAIRWISE_DF[
    "Zero Differences"
] = np.nan

for index, row in V_PAIRWISE_DF.iterrows():

    method_a = row["Method A"]
    method_b = row["Method B"]

    values_a = (
        V_PERFORMANCE_MATRIX
        .loc[
            method_a,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    values_b = (
        V_PERFORMANCE_MATRIX
        .loc[
            method_b,
            U_DATASETS,
        ]
        .to_numpy(
            dtype=float
        )
    )

    differences = (
        values_a - values_b
    )

    positive_count = int(
        np.sum(
            differences > 0
        )
    )

    negative_count = int(
        np.sum(
            differences < 0
        )
    )

    zero_count = int(
        np.sum(
            np.isclose(
                differences,
                0.0,
            )
        )
    )

    non_zero_count = (
        positive_count
        +
        negative_count
    )

    if non_zero_count > 0:

        rank_biserial = (
            positive_count
            -
            negative_count
        ) / non_zero_count

    else:

        rank_biserial = 0.0

    V_PAIRWISE_DF.loc[
        index,
        "Rank-Biserial Effect Size"
    ] = rank_biserial

    V_PAIRWISE_DF.loc[
        index,
        "Positive Differences"
    ] = positive_count

    V_PAIRWISE_DF.loc[
        index,
        "Negative Differences"
    ] = negative_count

    V_PAIRWISE_DF.loc[
        index,
        "Zero Differences"
    ] = zero_count

print()
print(
    V_PAIRWISE_DF[
        [
            "Method A",
            "Method B",
            "Rank-Biserial Effect Size",
            "Positive Differences",
            "Negative Differences",
            "Zero Differences",
        ]
    ]
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# V4 — INTERPRET EFFECT SIZES
# ==============================================================================

print()
print("=" * 100)
print("V4 — INTERPRET EFFECT SIZES")
print("=" * 100)

def V_INTERPRET_RANK_BISERIAL(effect):

    absolute_effect = abs(
        float(effect)
    )

    if absolute_effect < 0.10:
        return "Negligible"

    elif absolute_effect < 0.30:
        return "Small"

    elif absolute_effect < 0.50:
        return "Moderate"

    else:
        return "Large"


V_PAIRWISE_DF[
    "Effect Magnitude"
] = (
    V_PAIRWISE_DF[
        "Rank-Biserial Effect Size"
    ]
    .apply(
        V_INTERPRET_RANK_BISERIAL
    )
)

# ------------------------------------------------------------------------------
# DIRECTION
# ------------------------------------------------------------------------------

V_PAIRWISE_DF[
    "Effect Direction"
] = np.where(
    V_PAIRWISE_DF[
        "Rank-Biserial Effect Size"
    ] > 0,
    V_PAIRWISE_DF[
        "Method A"
    ],
    np.where(
        V_PAIRWISE_DF[
            "Rank-Biserial Effect Size"
        ] < 0,
        V_PAIRWISE_DF[
            "Method B"
        ],
        "No Difference",
    ),
)

print()
print(
    V_PAIRWISE_DF[
        [
            "Method A",
            "Method B",
            "Rank-Biserial Effect Size",
            "Effect Magnitude",
            "Effect Direction",
        ]
    ]
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# V5 — HOLM MULTIPLE-COMPARISON CORRECTION
# ==============================================================================

print()
print("=" * 100)
print("V5 — HOLM MULTIPLE-COMPARISON CORRECTION")
print("=" * 100)

from statsmodels.stats.multitest import multipletests

V_RAW_PVALUES = (
    pd.to_numeric(
        V_PAIRWISE_DF[
            "p-value"
        ],
        errors="coerce",
    )
    .to_numpy(
        dtype=float
    )
)

V_VALID_PVALUES = np.isfinite(
    V_RAW_PVALUES
)

V_HOLM_ADJUSTED = np.full(
    len(V_RAW_PVALUES),
    np.nan,
)

V_REJECTED = np.zeros(
    len(V_RAW_PVALUES),
    dtype=bool,
)

if V_VALID_PVALUES.any():

    (
        V_REJECTED_VALID,
        V_HOLM_VALID,
        _,
        _,
    ) = multipletests(
        V_RAW_PVALUES[
            V_VALID_PVALUES
        ],
        alpha=U_ALPHA,
        method="holm",
    )

    V_HOLM_ADJUSTED[
        V_VALID_PVALUES
    ] = V_HOLM_VALID

    V_REJECTED[
        V_VALID_PVALUES
    ] = V_REJECTED_VALID

V_PAIRWISE_DF[
    "Holm Adjusted p-value"
] = V_HOLM_ADJUSTED

V_PAIRWISE_DF[
    "Significant After Holm"
] = V_REJECTED

print()
print(
    V_PAIRWISE_DF[
        [
            "Method A",
            "Method B",
            "p-value",
            "Holm Adjusted p-value",
            "Significant After Holm",
        ]
    ]
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# V6 — COMBINE STATISTICAL SIGNIFICANCE AND EFFECT SIZE
# ==============================================================================

print()
print("=" * 100)
print("V6 — COMBINE STATISTICAL SIGNIFICANCE AND EFFECT SIZE")
print("=" * 100)

V_PAIRWISE_DF[
    "Statistical Interpretation"
] = "Not Significant"

V_PAIRWISE_DF.loc[
    V_PAIRWISE_DF[
        "Significant After Holm"
    ],
    "Statistical Interpretation",
] = "Statistically Significant"

# ------------------------------------------------------------------------------
# COMBINED EVIDENCE CATEGORY
# ------------------------------------------------------------------------------

def V_COMBINED_INTERPRETATION(row):

    significant = bool(
        row[
            "Significant After Holm"
        ]
    )

    magnitude = row[
        "Effect Magnitude"
    ]

    direction = row[
        "Effect Direction"
    ]

    if significant:

        return (
            f"Significant {magnitude.lower()} "
            f"effect favoring {direction}"
        )

    if magnitude in [
        "Moderate",
        "Large",
    ]:

        return (
            f"Not significant; "
            f"{magnitude.lower()} observed effect "
            f"favoring {direction}"
        )

    if magnitude == "Small":

        return (
            f"Not significant; "
            f"small observed effect "
            f"favoring {direction}"
        )

    return "No meaningful observed effect"


V_PAIRWISE_DF[
    "Combined Evidence"
] = (
    V_PAIRWISE_DF
    .apply(
        V_COMBINED_INTERPRETATION,
        axis=1,
    )
)

print()
print(
    V_PAIRWISE_DF[
        [
            "Method A",
            "Method B",
            "p-value",
            "Holm Adjusted p-value",
            "Rank-Biserial Effect Size",
            "Effect Magnitude",
            "Effect Direction",
            "Combined Evidence",
        ]
    ]
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# V7 — METHOD-LEVEL EFFECT-SIZE SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("V7 — METHOD-LEVEL EFFECT-SIZE SUMMARY")
print("=" * 100)

V_METHOD_EFFECT_SUMMARY = []

for method in U_METHODS:

    method_effects = []

    method_comparisons = (
        V_PAIRWISE_DF[
            (
                V_PAIRWISE_DF[
                    "Method A"
                ]
                ==
                method
            )
            |
            (
                V_PAIRWISE_DF[
                    "Method B"
                ]
                ==
                method
            )
        ]
    )

    for _, comparison in (
        method_comparisons.iterrows()
    ):

        effect = float(
            comparison[
                "Rank-Biserial Effect Size"
            ]
        )

        # Re-orient effect so positive always means
        # the focal method has higher performance.

        if comparison[
            "Method A"
        ] == method:

            oriented_effect = effect

        else:

            oriented_effect = -effect

        method_effects.append(
            oriented_effect
        )

    if method_effects:

        mean_effect = np.mean(
            method_effects
        )

        median_effect = np.median(
            method_effects
        )

    else:

        mean_effect = np.nan
        median_effect = np.nan

    V_METHOD_EFFECT_SUMMARY.append(
        {
            "Method": method,
            "Mean Oriented Effect":
                mean_effect,
            "Median Oriented Effect":
                median_effect,
            "Absolute Mean Effect":
                abs(mean_effect)
                if np.isfinite(
                    mean_effect
                )
                else np.nan,
            "Significant Comparisons":
                int(
                    method_comparisons[
                        "Significant After Holm"
                    ].sum()
                ),
            "Total Comparisons":
                len(
                    method_comparisons
                ),
        }
    )

V_METHOD_EFFECT_DF = pd.DataFrame(
    V_METHOD_EFFECT_SUMMARY
)

print()
print(
    V_METHOD_EFFECT_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# V8 — IDENTIFY STATISTICALLY SUPPORTED PAIRWISE DIFFERENCES
# ==============================================================================

print()
print("=" * 100)
print("V8 — IDENTIFY STATISTICALLY SUPPORTED PAIRWISE DIFFERENCES")
print("=" * 100)

V_SIGNIFICANT_COMPARISONS_DF = (
    V_PAIRWISE_DF[
        V_PAIRWISE_DF[
            "Significant After Holm"
        ]
    ]
    .copy()
)

print()

if V_SIGNIFICANT_COMPARISONS_DF.empty:

    print(
        "No pairwise comparison remains statistically "
        "significant after Holm correction."
    )

else:

    print(
        V_SIGNIFICANT_COMPARISONS_DF[
            [
                "Method A",
                "Method B",
                "Holm Adjusted p-value",
                "Rank-Biserial Effect Size",
                "Effect Magnitude",
                "Effect Direction",
            ]
        ]
        .to_string(
            index=False
        )
    )
```

```python
# ==============================================================================
# V9 — GLOBAL EFFECT-SIZE AND MULTIPLE-COMPARISON SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("V9 — GLOBAL EFFECT-SIZE AND MULTIPLE-COMPARISON SUMMARY")
print("=" * 100)

V_TOTAL_COMPARISONS = len(
    V_PAIRWISE_DF
)

V_SIGNIFICANT_COMPARISONS = int(
    V_PAIRWISE_DF[
        "Significant After Holm"
    ].sum()
)

V_EFFECT_MAGNITUDE_COUNTS = (
    V_PAIRWISE_DF[
        "Effect Magnitude"
    ]
    .value_counts()
)

print()
print(
    f"Total pairwise comparisons       : "
    f"{V_TOTAL_COMPARISONS}"
)

print(
    f"Significant after Holm correction: "
    f"{V_SIGNIFICANT_COMPARISONS}"
)

print()
print(
    "Effect-size distribution:"
)

for magnitude in [
    "Negligible",
    "Small",
    "Moderate",
    "Large",
]:

    count = int(
        V_EFFECT_MAGNITUDE_COUNTS.get(
            magnitude,
            0,
        )
    )

    print(
        f"  {magnitude:<12}: {count}"
    )
```

```python
# ==============================================================================
# W2 — RECOVER EXISTING EVALUATION DATA
# ==============================================================================

print()
print("=" * 100)
print("W2 — RECOVER EXISTING EVALUATION DATA")
print("=" * 100)

# ------------------------------------------------------------------------------
# CANDIDATE DATAFRAMES
# ------------------------------------------------------------------------------
# Priority is given to evaluation objects already produced by the research
# pipeline.

W_CANDIDATE_NAMES = [
    "R_METRIC_RESULTS_DF",
    "R_EVALUATION_RESULTS_DF",
    "R_RESULTS_DF",
    "R_METRICS_DF",
    "R_PERFORMANCE_DF",
    "T_WORKING_DF",
]

W_EVALUATION_SOURCE = None
W_EVALUATION_SOURCE_NAME = None

for candidate_name in W_CANDIDATE_NAMES:

    if candidate_name in globals():

        candidate = globals()[candidate_name]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            if not candidate.empty:

                W_EVALUATION_SOURCE = (
                    candidate.copy()
                )

                W_EVALUATION_SOURCE_NAME = (
                    candidate_name
                )

                break

if W_EVALUATION_SOURCE is None:

    raise RuntimeError(
        "W2 FAILED: No existing evaluation DataFrame was found. "
        "W requires existing Fidelity, Utility, and Privacy evaluation "
        "results from the previous research modules."
    )

print()
print(
    f"Selected source: "
    f"{W_EVALUATION_SOURCE_NAME}"
)

print()
print(
    f"Shape: "
    f"{W_EVALUATION_SOURCE.shape}"
)

print()
print("Available columns:")

for column in W_EVALUATION_SOURCE.columns:

    print(
        f"  - {column}"
    )
```

```python
# ==============================================================================
# W6 — VALIDATE FIDELITY RESULT STRUCTURE
# ==============================================================================

print()
print("=" * 100)
print("W6 — VALIDATE FIDELITY RESULT STRUCTURE")
print("=" * 100)

W_FIDELITY_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Metric",
    "Value",
    "Direction",
]

W_FIDELITY_MISSING = [
    column
    for column in W_FIDELITY_REQUIRED_COLUMNS
    if column not in W_FIDELITY_RESULTS.columns
]

if W_FIDELITY_MISSING:

    raise RuntimeError(
        "W6 FAILED: Missing Fidelity columns: "
        + ", ".join(
            W_FIDELITY_MISSING
        )
    )

# ------------------------------------------------------------------------------
# NUMERIC CONVERSION
# ------------------------------------------------------------------------------

W_FIDELITY_RESULTS["Value"] = pd.to_numeric(
    W_FIDELITY_RESULTS["Value"],
    errors="coerce",
)

# ------------------------------------------------------------------------------
# REMOVE INVALID VALUES
# ------------------------------------------------------------------------------

W_FIDELITY_RESULTS = (
    W_FIDELITY_RESULTS[
        W_FIDELITY_RESULTS["Value"].notna()
    ]
    .copy()
)

print()
print(
    "Fidelity metrics detected:"
)

print(
    sorted(
        W_FIDELITY_RESULTS[
            "Metric"
        ]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)

print()
print(
    "Fidelity directions:"
)

print(
    W_FIDELITY_RESULTS[
        [
            "Metric",
            "Direction",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "Metric"
    )
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W7 — NORMALIZE FIDELITY METRICS
# ==============================================================================

print()
print("=" * 100)
print("W7 — NORMALIZE FIDELITY METRICS")
print("=" * 100)

def W_NORMALIZE_SERIES(
    series,
    direction,
):

    series = pd.to_numeric(
        series,
        errors="coerce",
    )

    minimum = series.min()
    maximum = series.max()

    if not np.isfinite(
        minimum
    ) or not np.isfinite(
        maximum
    ):

        return pd.Series(
            np.nan,
            index=series.index,
        )

    # No between-method variation.
    if np.isclose(
        minimum,
        maximum,
    ):

        return pd.Series(
            100.0,
            index=series.index,
        )

    if str(direction).lower() in [
        "higher",
        "higher_is_better",
        "maximize",
        "max",
    ]:

        normalized = (
            (series - minimum)
            /
            (maximum - minimum)
        )

    elif str(direction).lower() in [
        "lower",
        "lower_is_better",
        "minimize",
        "min",
    ]:

        normalized = (
            (maximum - series)
            /
            (maximum - minimum)
        )

    else:

        raise RuntimeError(
            "W7 FAILED: Unknown metric direction: "
            f"{direction}"
        )

    return (
        normalized
        * 100.0
    )


# ------------------------------------------------------------------------------
# NORMALIZE WITHIN EACH METRIC
# ------------------------------------------------------------------------------

W_FIDELITY_RESULTS[
    "Normalized_Value"
] = np.nan

for metric in (
    W_FIDELITY_RESULTS[
        "Metric"
    ]
    .dropna()
    .unique()
):

    metric_mask = (
        W_FIDELITY_RESULTS[
            "Metric"
        ]
        ==
        metric
    )

    metric_data = (
        W_FIDELITY_RESULTS.loc[
            metric_mask
        ]
    )

    directions = (
        metric_data[
            "Direction"
        ]
        .dropna()
        .astype(str)
        .str.strip()
        .unique()
        .tolist()
    )

    if len(directions) != 1:

        raise RuntimeError(
            "W7 FAILED: Fidelity metric "
            f"'{metric}' has inconsistent directions: "
            f"{directions}"
        )

    W_FIDELITY_RESULTS.loc[
        metric_mask,
        "Normalized_Value",
    ] = W_NORMALIZE_SERIES(
        metric_data["Value"],
        directions[0],
    )

print()
print(
    W_FIDELITY_RESULTS[
        [
            "Dataset",
            "Method",
            "Metric",
            "Value",
            "Direction",
            "Normalized_Value",
        ]
    ]
    .head(20)
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W8 — CONSTRUCT FIDELITY COMPONENT SCORE
# ==============================================================================

print()
print("=" * 100)
print("W8 — CONSTRUCT FIDELITY COMPONENT SCORE")
print("=" * 100)

W_FIDELITY_COMPONENT = (
    W_FIDELITY_RESULTS[
        [
            "Dataset",
            "Method",
            "Normalized_Value",
        ]
    ]
    .dropna(
        subset=[
            "Normalized_Value"
        ]
    )
    .groupby(
        [
            "Dataset",
            "Method",
        ],
        as_index=False,
    )[
        "Normalized_Value"
    ]
    .mean()
    .rename(
        columns={
            "Normalized_Value":
                "Fidelity_Score"
        }
    )
)

print()
print(
    W_FIDELITY_COMPONENT.to_string(
        index=False
    )
)

print()
print(
    f"Fidelity component observations: "
    f"{len(W_FIDELITY_COMPONENT)}"
)
```

```python
# ==============================================================================
# W9 — VALIDATE UTILITY RESULT STRUCTURE
# ==============================================================================

print()
print("=" * 100)
print("W9 — VALIDATE UTILITY RESULT STRUCTURE")
print("=" * 100)

W_UTILITY_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Metric",
    "Value",
]

W_UTILITY_MISSING = [
    column
    for column in W_UTILITY_REQUIRED_COLUMNS
    if column not in W_UTILITY_RESULTS.columns
]

if W_UTILITY_MISSING:

    raise RuntimeError(
        "W9 FAILED: Missing Utility columns: "
        + ", ".join(
            W_UTILITY_MISSING
        )
    )

W_UTILITY_RESULTS["Value"] = pd.to_numeric(
    W_UTILITY_RESULTS["Value"],
    errors="coerce",
)

W_UTILITY_RESULTS = (
    W_UTILITY_RESULTS[
        W_UTILITY_RESULTS["Value"].notna()
    ]
    .copy()
)

print()
print(
    "Utility metrics:"
)

print(
    sorted(
        W_UTILITY_RESULTS[
            "Metric"
        ]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)
```

```python
# ==============================================================================
# W10 — NORMALIZE UTILITY METRICS
# ==============================================================================

print()
print("=" * 100)
print("W10 — NORMALIZE UTILITY METRICS")
print("=" * 100)

W_UTILITY_RESULTS[
    "Normalized_Value"
] = np.nan

for metric in (
    W_UTILITY_RESULTS[
        "Metric"
    ]
    .dropna()
    .unique()
):

    metric_mask = (
        W_UTILITY_RESULTS[
            "Metric"
        ]
        ==
        metric
    )

    W_UTILITY_RESULTS.loc[
        metric_mask,
        "Normalized_Value",
    ] = W_NORMALIZE_SERIES(
        W_UTILITY_RESULTS.loc[
            metric_mask,
            "Value",
        ],
        "higher_is_better",
    )

print()
print(
    W_UTILITY_RESULTS[
        [
            "Dataset",
            "Method",
            "Metric",
            "Value",
            "Normalized_Value",
        ]
    ]
    .head(20)
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W11 — CONSTRUCT UTILITY COMPONENT SCORE
# ==============================================================================

print()
print("=" * 100)
print("W11 — CONSTRUCT UTILITY COMPONENT SCORE")
print("=" * 100)

W_UTILITY_COMPONENT = (
    W_UTILITY_RESULTS[
        [
            "Dataset",
            "Method",
            "Normalized_Value",
        ]
    ]
    .dropna(
        subset=[
            "Normalized_Value"
        ]
    )
    .groupby(
        [
            "Dataset",
            "Method",
        ],
        as_index=False,
    )[
        "Normalized_Value"
    ]
    .mean()
    .rename(
        columns={
            "Normalized_Value":
                "Utility_Score"
        }
    )
)

print()
print(
    W_UTILITY_COMPONENT.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W12 — VALIDATE PRIVACY RESULT STRUCTURE
# ==============================================================================

print()
print("=" * 100)
print("W12 — VALIDATE PRIVACY RESULT STRUCTURE")
print("=" * 100)

W_PRIVACY_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Privacy Metric",
    "Value",
]

W_PRIVACY_MISSING = [
    column
    for column in W_PRIVACY_REQUIRED_COLUMNS
    if column not in W_PRIVACY_RESULTS.columns
]

if W_PRIVACY_MISSING:

    raise RuntimeError(
        "W12 FAILED: Missing Privacy columns: "
        + ", ".join(
            W_PRIVACY_MISSING
        )
    )

W_PRIVACY_RESULTS["Value"] = pd.to_numeric(
    W_PRIVACY_RESULTS["Value"],
    errors="coerce",
)

W_PRIVACY_RESULTS = (
    W_PRIVACY_RESULTS[
        W_PRIVACY_RESULTS["Value"].notna()
    ]
    .copy()
)

print()
print(
    "Privacy metrics:"
)

print(
    sorted(
        W_PRIVACY_RESULTS[
            "Privacy Metric"
        ]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)
```

```python
# ==============================================================================
# W13 — RESOLVE AND NORMALIZE VALIDATED PRIVACY METRICS
# ==============================================================================

print()
print("=" * 100)
print("W13 — RESOLVE AND NORMALIZE VALIDATED PRIVACY METRICS")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE PRIVACY DATA
# ------------------------------------------------------------------------------

if W_PRIVACY_RESULTS.empty:

    raise RuntimeError(
        "W13 FAILED: No validated privacy observations are available."
    )

# ------------------------------------------------------------------------------
# DISPLAY EXISTING PRIVACY METRICS
# ------------------------------------------------------------------------------

W_PRIVACY_METRIC_LIST = sorted(
    W_PRIVACY_RESULTS[
        "Privacy Metric"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
    .tolist()
)

print()
print("Existing validated privacy metrics:")

for metric in W_PRIVACY_METRIC_LIST:
    print(
        f"  - {metric}"
    )

# ------------------------------------------------------------------------------
# EXPLICIT DIRECTION MAP
# ------------------------------------------------------------------------------
#
# IMPORTANT:
# These directions apply ONLY to the existing validated privacy metrics.
#
# epsilon:
#     lower = stronger privacy
#
# noise_multiplier:
#     higher = stronger DP noise / privacy protection
#
# privacy risk / disclosure / re-identification:
#     lower = better
#
# No metric is silently assigned a direction.
# ------------------------------------------------------------------------------

W_PRIVACY_DIRECTION_MAP = {

    "epsilon":
        "lower_is_better",

    "privacy_budget":
        "lower_is_better",

    "privacy budget":
        "lower_is_better",

    "noise_multiplier":
        "higher_is_better",

    "privacy_risk":
        "lower_is_better",

    "privacy risk":
        "lower_is_better",

    "disclosure_risk":
        "lower_is_better",

    "disclosure risk":
        "lower_is_better",

    "reidentification_risk":
        "lower_is_better",

    "re-identification_risk":
        "lower_is_better",

    "reidentification risk":
        "lower_is_better",

    "re-identification risk":
        "lower_is_better",
}

# ------------------------------------------------------------------------------
# RESOLVE DIRECTIONS
# ------------------------------------------------------------------------------

W_PRIVACY_RESULTS[
    "Privacy_Direction"
] = np.nan

W_UNRESOLVED_PRIVACY_METRICS = []

for index, row in W_PRIVACY_RESULTS.iterrows():

    metric_raw = str(
        row[
            "Privacy Metric"
        ]
    ).strip()

    metric_key = (
        metric_raw
        .lower()
        .strip()
    )

    if metric_key in W_PRIVACY_DIRECTION_MAP:

        W_PRIVACY_RESULTS.loc[
            index,
            "Privacy_Direction"
        ] = W_PRIVACY_DIRECTION_MAP[
            metric_key
        ]

    else:

        W_UNRESOLVED_PRIVACY_METRICS.append(
            metric_raw
        )

# ------------------------------------------------------------------------------
# REMOVE DUPLICATES
# ------------------------------------------------------------------------------

W_UNRESOLVED_PRIVACY_METRICS = sorted(
    set(
        W_UNRESOLVED_PRIVACY_METRICS
    )
)

if W_UNRESOLVED_PRIVACY_METRICS:

    print()
    print(
        "Unresolved privacy metrics:"
    )

    for metric in W_UNRESOLVED_PRIVACY_METRICS:
        print(
            f"  - {metric}"
        )

    raise RuntimeError(
        "W13 FAILED: Existing privacy metrics have no "
        "validated direction mapping. Extend the explicit "
        "direction map only after confirming their meaning "
        "from the existing R evaluation."
    )

# ------------------------------------------------------------------------------
# NORMALIZE EACH PRIVACY METRIC
# ------------------------------------------------------------------------------

W_PRIVACY_RESULTS[
    "Normalized_Value"
] = np.nan

for metric in (
    W_PRIVACY_RESULTS[
        "Privacy Metric"
    ]
    .dropna()
    .astype(str)
    .unique()
):

    metric_mask = (
        W_PRIVACY_RESULTS[
            "Privacy Metric"
        ]
        .astype(str)
        ==
        str(metric)
    )

    metric_data = (
        W_PRIVACY_RESULTS.loc[
            metric_mask
        ]
    )

    directions = (
        metric_data[
            "Privacy_Direction"
        ]
        .dropna()
        .unique()
        .tolist()
    )

    if len(directions) != 1:

        raise RuntimeError(
            "W13 FAILED: Inconsistent privacy direction "
            f"for metric '{metric}': {directions}"
        )

    W_PRIVACY_RESULTS.loc[
        metric_mask,
        "Normalized_Value",
    ] = W_NORMALIZE_SERIES(
        metric_data[
            "Value"
        ],
        directions[0],
    )

# ------------------------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------------------------

print()
print(
    W_PRIVACY_RESULTS[
        [
            "Dataset",
            "Method",
            "Privacy Metric",
            "Value",
            "Privacy_Direction",
            "Normalized_Value",
        ]
    ]
    .to_string(
        index=False
    )
)

print()
print("W13 STATUS: PASS")
```

```python
# ==============================================================================
# W14 — CONSTRUCT PRIVACY COMPONENT SCORE
# ==============================================================================

print()
print("=" * 100)
print("W14 — CONSTRUCT PRIVACY COMPONENT SCORE")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE NORMALIZED PRIVACY VALUES
# ------------------------------------------------------------------------------

W_PRIVACY_VALID_VALUES = (
    W_PRIVACY_RESULTS[
        "Normalized_Value"
    ]
    .notna()
)

if not W_PRIVACY_VALID_VALUES.any():

    raise RuntimeError(
        "W14 FAILED: No valid normalized privacy observations "
        "are available."
    )

# ------------------------------------------------------------------------------
# CONSTRUCT DATASET × METHOD PRIVACY SCORE
# ------------------------------------------------------------------------------

W_PRIVACY_COMPONENT = (
    W_PRIVACY_RESULTS[
        [
            "Dataset",
            "Method",
            "Normalized_Value",
        ]
    ]
    .dropna(
        subset=[
            "Normalized_Value"
        ]
    )
    .groupby(
        [
            "Dataset",
            "Method",
        ],
        as_index=False,
    )[
        "Normalized_Value"
    ]
    .mean()
    .rename(
        columns={
            "Normalized_Value":
                "Privacy_Score"
        }
    )
)

# ------------------------------------------------------------------------------
# CLIP NUMERIC RANGE
# ------------------------------------------------------------------------------

W_PRIVACY_COMPONENT[
    "Privacy_Score"
] = (
    W_PRIVACY_COMPONENT[
        "Privacy_Score"
    ]
    .clip(
        0.0,
        100.0,
    )
)

# ------------------------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------------------------

print()
print(
    "Dataset × Method Privacy Scores:"
)

print()

print(
    W_PRIVACY_COMPONENT.to_string(
        index=False
    )
)

print()
print(
    "Privacy score range: 0–100"
)

print(
    f"Minimum: "
    f"{W_PRIVACY_COMPONENT['Privacy_Score'].min():.4f}"
)

print(
    f"Maximum: "
    f"{W_PRIVACY_COMPONENT['Privacy_Score'].max():.4f}"
)

print()
print("W14 STATUS: PASS")
```

```python
# ==============================================================================
# W15 — MERGE FIDELITY, UTILITY AND PRIVACY COMPONENT SCORES
# ==============================================================================

print()
print("=" * 100)
print("W15 — MERGE FIDELITY, UTILITY AND PRIVACY COMPONENT SCORES")
print("=" * 100)

W_COMPONENT_SCORES = (
    W_FIDELITY_COMPONENT
    .merge(
        W_UTILITY_COMPONENT,
        on=[
            "Dataset",
            "Method",
        ],
        how="outer",
    )
    .merge(
        W_PRIVACY_COMPONENT,
        on=[
            "Dataset",
            "Method",
        ],
        how="outer",
    )
)

# ------------------------------------------------------------------------------
# NORMALIZE DATASET / METHOD NAMES
# ------------------------------------------------------------------------------

W_COMPONENT_SCORES["Method"] = (
    W_COMPONENT_SCORES["Method"]
    .astype(str)
    .str.strip()
)

W_COMPONENT_SCORES["Dataset"] = (
    W_COMPONENT_SCORES["Dataset"]
    .astype(str)
    .str.strip()
)

W_COMPONENT_SCORES = (
    W_COMPONENT_SCORES[
        W_COMPONENT_SCORES["Method"].isin(
            W_METHODS
        )
        &
        W_COMPONENT_SCORES["Dataset"].isin(
            W_DATASETS
        )
    ]
    .copy()
)

W_COMPONENT_SCORES = (
    W_COMPONENT_SCORES
    .sort_values(
        [
            "Dataset",
            "Method",
        ]
    )
    .reset_index(
        drop=True
    )
)

print()
print(
    W_COMPONENT_SCORES.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W16 — VALIDATE COMPLETE COMPONENT COVERAGE
# ==============================================================================

print()
print("=" * 100)
print("W16 — VALIDATE COMPLETE COMPONENT COVERAGE")
print("=" * 100)

W_EXPECTED_COMBINATIONS = (
    len(W_DATASETS)
    *
    len(W_METHODS)
)

W_ACTUAL_COMBINATIONS = (
    W_COMPONENT_SCORES[
        [
            "Dataset",
            "Method",
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

print()
print(
    f"Expected Dataset × Method combinations : "
    f"{W_EXPECTED_COMBINATIONS}"
)

print(
    f"Observed Dataset × Method combinations : "
    f"{W_ACTUAL_COMBINATIONS}"
)

# ------------------------------------------------------------------------------
# MISSING COMBINATIONS
# ------------------------------------------------------------------------------

W_EXPECTED_GRID = pd.MultiIndex.from_product(
    [
        W_DATASETS,
        W_METHODS,
    ],
    names=[
        "Dataset",
        "Method",
    ],
).to_frame(
    index=False
)

W_OBSERVED_GRID = (
    W_COMPONENT_SCORES[
        [
            "Dataset",
            "Method",
        ]
    ]
    .drop_duplicates()
)

W_MISSING_GRID = (
    W_EXPECTED_GRID
    .merge(
        W_OBSERVED_GRID,
        on=[
            "Dataset",
            "Method",
        ],
        how="left",
        indicator=True,
    )
)

W_MISSING_GRID = (
    W_MISSING_GRID[
        W_MISSING_GRID[
            "_merge"
        ]
        ==
        "left_only"
    ]
    .drop(
        columns="_merge"
    )
)

if not W_MISSING_GRID.empty:

    print()
    print(
        "Missing Dataset × Method combinations:"
    )

    print(
        W_MISSING_GRID.to_string(
            index=False
        )
    )
```

```python
# ==============================================================================
# W17 — FINALIZE COMPONENT COMPLETENESS DIAGNOSTIC
# ==============================================================================

print()
print("=" * 100)
print("W17 — FINALIZE COMPONENT COMPLETENESS DIAGNOSTIC")
print("=" * 100)

# ------------------------------------------------------------------------------
# EXPECTED GRID
# ------------------------------------------------------------------------------

W17_EXPECTED_GRID = (
    W_EXPECTED_GRID.copy()
)

W17_EXPECTED_COUNT = len(
    W17_EXPECTED_GRID
)

# ------------------------------------------------------------------------------
# COMPONENT COUNTS
# ------------------------------------------------------------------------------

W17_FIDELITY_MISSING_COUNT = len(
    W17_MISSING_FIDELITY
)

W17_UTILITY_MISSING_COUNT = len(
    W17_MISSING_UTILITY
)

W17_PRIVACY_MISSING_COUNT = len(
    W17_MISSING_PRIVACY
)

W17_COMPLETE_COUNT = int(
    W17_COVERAGE_DF[
        "All_Components_Available"
    ].sum()
)

# ------------------------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------------------------

W17_COMPONENT_COMPLETENESS = pd.DataFrame(
    [
        {
            "Component": "Fidelity",
            "Expected": W17_EXPECTED_COUNT,
            "Available": (
                W17_EXPECTED_COUNT
                -
                W17_FIDELITY_MISSING_COUNT
            ),
            "Missing": W17_FIDELITY_MISSING_COUNT,
        },
        {
            "Component": "Utility",
            "Expected": W17_EXPECTED_COUNT,
            "Available": (
                W17_EXPECTED_COUNT
                -
                W17_UTILITY_MISSING_COUNT
            ),
            "Missing": W17_UTILITY_MISSING_COUNT,
        },
        {
            "Component": "Privacy",
            "Expected": W17_EXPECTED_COUNT,
            "Available": (
                W17_EXPECTED_COUNT
                -
                W17_PRIVACY_MISSING_COUNT
            ),
            "Missing": W17_PRIVACY_MISSING_COUNT,
        },
        {
            "Component": "Complete Composite",
            "Expected": W17_EXPECTED_COUNT,
            "Available": W17_COMPLETE_COUNT,
            "Missing": (
                W17_EXPECTED_COUNT
                -
                W17_COMPLETE_COUNT
            ),
        },
    ]
)

print()
print(
    W17_COMPONENT_COMPLETENESS.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# DO NOT FAIL
# ------------------------------------------------------------------------------

W17_STATUS = (
    "COMPLETE"
    if W17_COMPLETE_COUNT
    ==
    W17_EXPECTED_COUNT
    else
    "INCOMPLETE_REQUIRES_SOURCE_DIAGNOSTICS"
)

print()
print(
    f"W17 STATUS: {W17_STATUS}"
)

print()
print(
    "No missing component values were fabricated."
)
```

```python
# ==============================================================================
# W18 — FINAL COMPOSITE PERFORMANCE REPORT
# ==============================================================================

print()
print("=" * 100)
print("W18 — FINAL COMPOSITE PERFORMANCE REPORT")
print("=" * 100)

# ------------------------------------------------------------------------------
# COMPOSITE WEIGHTS
# ------------------------------------------------------------------------------

W_FIDELITY_WEIGHT = 1.0 / 3.0
W_UTILITY_WEIGHT = 1.0 / 3.0
W_PRIVACY_WEIGHT = 1.0 / 3.0

print()
print("COMPOSITE PERFORMANCE FRAMEWORK")
print("-" * 100)

print(
    f"Fidelity  : {W_FIDELITY_WEIGHT:.4f}"
)

print(
    f"Utility   : {W_UTILITY_WEIGHT:.4f}"
)

print(
    f"Privacy   : {W_PRIVACY_WEIGHT:.4f}"
)

print(
    "Overall   : weighted arithmetic mean"
)

# ------------------------------------------------------------------------------
# VALIDATE COMPONENT TABLES
# ------------------------------------------------------------------------------

W18_REQUIRED_COMPONENTS = {
    "Fidelity": W_FIDELITY_COMPONENT,
    "Utility": W_UTILITY_COMPONENT,
    "Privacy": W_PRIVACY_COMPONENT,
}

for component_name, component_df in (
    W18_REQUIRED_COMPONENTS.items()
):

    if component_df is None:

        raise RuntimeError(
            f"W18 FAILED: {component_name} component "
            "DataFrame is not available."
        )

    if component_df.empty:

        raise RuntimeError(
            f"W18 FAILED: {component_name} component "
            "DataFrame is empty."
        )

# ------------------------------------------------------------------------------
# NORMALIZE IDENTIFIERS
# ------------------------------------------------------------------------------

for component_name, component_df in (
    W18_REQUIRED_COMPONENTS.items()
):

    component_df["Dataset"] = (
        component_df["Dataset"]
        .astype(str)
        .str.strip()
    )

    component_df["Method"] = (
        component_df["Method"]
        .astype(str)
        .str.strip()
    )

# ------------------------------------------------------------------------------
# SELECT ONLY REQUIRED SCORE COLUMNS
# ------------------------------------------------------------------------------

W18_FIDELITY = (
    W_FIDELITY_COMPONENT[
        [
            "Dataset",
            "Method",
            "Fidelity_Score",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Dataset",
            "Method",
        ]
    )
)

W18_UTILITY = (
    W_UTILITY_COMPONENT[
        [
            "Dataset",
            "Method",
            "Utility_Score",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Dataset",
            "Method",
        ]
    )
)

W18_PRIVACY = (
    W_PRIVACY_COMPONENT[
        [
            "Dataset",
            "Method",
            "Privacy_Score",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Dataset",
            "Method",
        ]
    )
)

# ------------------------------------------------------------------------------
# MERGE COMPONENTS
# ------------------------------------------------------------------------------

W18_COMPOSITE_DF = (
    W18_FIDELITY
    .merge(
        W18_UTILITY,
        on=[
            "Dataset",
            "Method",
        ],
        how="outer",
    )
    .merge(
        W18_PRIVACY,
        on=[
            "Dataset",
            "Method",
        ],
        how="outer",
    )
)

# ------------------------------------------------------------------------------
# COMPONENT AVAILABILITY
# ------------------------------------------------------------------------------

W18_COMPOSITE_DF[
    "Fidelity_Available"
] = (
    W18_COMPOSITE_DF[
        "Fidelity_Score"
    ]
    .notna()
)

W18_COMPOSITE_DF[
    "Utility_Available"
] = (
    W18_COMPOSITE_DF[
        "Utility_Score"
    ]
    .notna()
)

W18_COMPOSITE_DF[
    "Privacy_Available"
] = (
    W18_COMPOSITE_DF[
        "Privacy_Score"
    ]
    .notna()
)

W18_COMPOSITE_DF[
    "Composite_Complete"
] = (
    W18_COMPOSITE_DF[
        [
            "Fidelity_Available",
            "Utility_Available",
            "Privacy_Available",
        ]
    ]
    .all(axis=1)
)

# ------------------------------------------------------------------------------
# INITIALIZE OVERALL SCORE
# ------------------------------------------------------------------------------

W18_COMPOSITE_DF[
    "Overall_Performance_Score"
] = np.nan

# ------------------------------------------------------------------------------
# CALCULATE ONLY FOR COMPLETE EVIDENCE
# ------------------------------------------------------------------------------

W18_COMPLETE_MASK = (
    W18_COMPOSITE_DF[
        "Composite_Complete"
    ]
)

W18_COMPOSITE_DF.loc[
    W18_COMPLETE_MASK,
    "Overall_Performance_Score",
] = (
    W_FIDELITY_WEIGHT
    *
    W18_COMPOSITE_DF.loc[
        W18_COMPLETE_MASK,
        "Fidelity_Score",
    ]
    +
    W_UTILITY_WEIGHT
    *
    W18_COMPOSITE_DF.loc[
        W18_COMPLETE_MASK,
        "Utility_Score",
    ]
    +
    W_PRIVACY_WEIGHT
    *
    W18_COMPOSITE_DF.loc[
        W18_COMPLETE_MASK,
        "Privacy_Score",
    ]
)

# ------------------------------------------------------------------------------
# PERFORMANCE STATUS
# ------------------------------------------------------------------------------

W18_COMPOSITE_DF[
    "Performance_Status"
] = np.where(
    W18_COMPOSITE_DF[
        "Composite_Complete"
    ],
    "COMPLETE",
    "INCOMPLETE_EVIDENCE",
)

# ------------------------------------------------------------------------------
# DISPLAY DATASET × METHOD RESULTS
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("DATASET × METHOD COMPOSITE PERFORMANCE")
print("=" * 100)

print()

print(
    W18_COMPOSITE_DF[
        [
            "Dataset",
            "Method",
            "Fidelity_Score",
            "Utility_Score",
            "Privacy_Score",
            "Overall_Performance_Score",
            "Performance_Status",
        ]
    ]
    .sort_values(
        [
            "Dataset",
            "Method",
        ]
    )
    .to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# VALID OVERALL PERFORMANCE RESULTS
# ------------------------------------------------------------------------------

W18_VALID_COMPOSITES = (
    W18_COMPOSITE_DF[
        W18_COMPOSITE_DF[
            "Composite_Complete"
        ]
    ]
    .copy()
)

# ------------------------------------------------------------------------------
# METHOD-LEVEL CROSS-DATASET SUMMARY
# ------------------------------------------------------------------------------

if not W18_VALID_COMPOSITES.empty:

    W_CROSS_DATASET_SUMMARY = (
        W18_VALID_COMPOSITES
        .groupby(
            "Method",
            as_index=False,
        )
        .agg(
            Mean_Overall_Score=(
                "Overall_Performance_Score",
                "mean",
            ),
            Std_Overall_Score=(
                "Overall_Performance_Score",
                "std",
            ),
            Datasets_Evaluated=(
                "Dataset",
                "nunique",
            ),
        )
    )

    # ----------------------------------------------------------
    # STABILITY / ROBUSTNESS
    # ----------------------------------------------------------

    W_CROSS_DATASET_SUMMARY[
        "Std_Overall_Score"
    ] = (
        W_CROSS_DATASET_SUMMARY[
            "Std_Overall_Score"
        ]
        .fillna(0.0)
    )

    W_CROSS_DATASET_SUMMARY[
        "Cross_Dataset_Robustness"
    ] = (
        100.0
        /
        (
            1.0
            +
            W_CROSS_DATASET_SUMMARY[
                "Std_Overall_Score"
            ]
        )
    )

    # ----------------------------------------------------------
    # RANK BY MEAN PERFORMANCE
    # ----------------------------------------------------------

    W_CROSS_DATASET_SUMMARY[
        "Performance_Rank"
    ] = (
        W_CROSS_DATASET_SUMMARY[
            "Mean_Overall_Score"
        ]
        .rank(
            ascending=False,
            method="min",
        )
        .astype(int)
    )

    W_CROSS_DATASET_SUMMARY = (
        W_CROSS_DATASET_SUMMARY
        .sort_values(
            [
                "Performance_Rank",
                "Mean_Overall_Score",
            ],
            ascending=[
                True,
                False,
            ],
        )
        .reset_index(
            drop=True
        )
    )

else:

    W_CROSS_DATASET_SUMMARY = pd.DataFrame(
        columns=[
            "Method",
            "Mean_Overall_Score",
            "Std_Overall_Score",
            "Datasets_Evaluated",
            "Cross_Dataset_Robustness",
            "Performance_Rank",
        ]
    )

# ------------------------------------------------------------------------------
# METHOD RANKING
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("METHOD RANKING")
print("=" * 100)

if W_CROSS_DATASET_SUMMARY.empty:

    print()
    print(
        "No complete Dataset × Method combination "
        "has sufficient evidence for a composite ranking."
    )

else:

    print()

    print(
        W_CROSS_DATASET_SUMMARY.to_string(
            index=False
        )
    )

# ------------------------------------------------------------------------------
# INCOMPLETE EVIDENCE SUMMARY
# ------------------------------------------------------------------------------

W18_INCOMPLETE = (
    W18_COMPOSITE_DF[
        ~W18_COMPOSITE_DF[
            "Composite_Complete"
        ]
    ]
    .copy()
)

print()
print("=" * 100)
print("INCOMPLETE COMPOSITE EVIDENCE")
print("=" * 100)

if W18_INCOMPLETE.empty:

    print()
    print(
        "All Dataset × Method combinations have "
        "complete component evidence."
    )

else:

    for _, row in W18_INCOMPLETE.iterrows():

        missing = []

        if not row[
            "Fidelity_Available"
        ]:
            missing.append(
                "Fidelity"
            )

        if not row[
            "Utility_Available"
        ]:
            missing.append(
                "Utility"
            )

        if not row[
            "Privacy_Available"
        ]:
            missing.append(
                "Privacy"
            )

        print(
            f"{row['Dataset']} | "
            f"{row['Method']} | "
            f"Missing: {', '.join(missing)}"
        )

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

W18_COMPLETE_COUNT = int(
    W18_COMPOSITE_DF[
        "Composite_Complete"
    ].sum()
)

W18_TOTAL_COUNT = len(
    W18_COMPOSITE_DF
)

print()
print("=" * 100)
print("W18 — FINAL STATUS")
print("=" * 100)

print()
print(
    f"Complete composite combinations : "
    f"{W18_COMPLETE_COUNT}"
)

print(
    f"Total Dataset × Method pairs    : "
    f"{W18_TOTAL_COUNT}"
)

print(
    f"Incomplete combinations          : "
    f"{W18_TOTAL_COUNT - W18_COMPLETE_COUNT}"
)

if W18_COMPLETE_COUNT == 0:

    W18_STATUS = (
        "NO_COMPLETE_COMPOSITES"
    )

elif W18_COMPLETE_COUNT < W18_TOTAL_COUNT:

    W18_STATUS = (
        "PARTIAL_COMPOSITES"
    )

else:

    W18_STATUS = (
        "FULL_COMPOSITES"
    )

print()
print(
    f"W18 STATUS: {W18_STATUS}"
)

print()
print(
    "No missing component values were replaced, "
    "imputed, or assigned zero."
)
```

```python
# ==============================================================================
# W19 — CALCULATE OVERALL PERFORMANCE SCORE
# ==============================================================================

print()
print("=" * 100)
print("W19 — CALCULATE OVERALL PERFORMANCE SCORE")
print("=" * 100)

W_COMPONENT_SCORES[
    "Overall_Performance_Score"
] = (
    W_FIDELITY_WEIGHT
    *
    W_COMPONENT_SCORES[
        "Fidelity_Score"
    ]
    +
    W_UTILITY_WEIGHT
    *
    W_COMPONENT_SCORES[
        "Utility_Score"
    ]
    +
    W_PRIVACY_WEIGHT
    *
    W_COMPONENT_SCORES[
        "Privacy_Score"
    ]
)

W_COMPONENT_SCORES[
    "Overall_Performance_Score"
] = (
    W_COMPONENT_SCORES[
        "Overall_Performance_Score"
    ]
    .clip(
        0.0,
        100.0,
    )
)

print()
print(
    W_COMPONENT_SCORES[
        [
            "Dataset",
            "Method",
            "Fidelity_Score",
            "Utility_Score",
            "Privacy_Score",
            "Overall_Performance_Score",
        ]
    ]
    .to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W20 — CROSS-DATASET COMPOSITE PERFORMANCE SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("W20 — CROSS-DATASET COMPOSITE PERFORMANCE SUMMARY")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE SOURCE
# ------------------------------------------------------------------------------

if "W18_COMPOSITE_DF" not in globals():

    raise RuntimeError(
        "W20 FAILED: W18_COMPOSITE_DF is not available. "
        "Run W18 before W20."
    )

if W18_COMPOSITE_DF.empty:

    raise RuntimeError(
        "W20 FAILED: W18_COMPOSITE_DF is empty."
    )

# ------------------------------------------------------------------------------
# COPY SOURCE
# ------------------------------------------------------------------------------

W20_SOURCE_DF = (
    W18_COMPOSITE_DF.copy()
)

# ------------------------------------------------------------------------------
# VALIDATE REQUIRED COLUMNS
# ------------------------------------------------------------------------------

W20_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Overall_Performance_Score",
]

W20_MISSING_COLUMNS = [
    column
    for column in W20_REQUIRED_COLUMNS
    if column not in W20_SOURCE_DF.columns
]

if W20_MISSING_COLUMNS:

    raise RuntimeError(
        "W20 FAILED: Required columns are missing: "
        +
        ", ".join(
            W20_MISSING_COLUMNS
        )
    )

# ------------------------------------------------------------------------------
# NUMERIC CONVERSION
# ------------------------------------------------------------------------------

W20_SOURCE_DF[
    "Overall_Performance_Score"
] = pd.to_numeric(
    W20_SOURCE_DF[
        "Overall_Performance_Score"
    ],
    errors="coerce",
)

# ------------------------------------------------------------------------------
# REMOVE INVALID COMPOSITE SCORES
# ------------------------------------------------------------------------------

W20_VALID_DF = (
    W20_SOURCE_DF[
        W20_SOURCE_DF[
            "Overall_Performance_Score"
        ]
        .notna()
        &
        np.isfinite(
            W20_SOURCE_DF[
                "Overall_Performance_Score"
            ]
        )
    ]
    [
        [
            "Dataset",
            "Method",
            "Overall_Performance_Score",
        ]
    ]
    .copy()
)

# ------------------------------------------------------------------------------
# DISPLAY VALID COMPOSITE OBSERVATIONS
# ------------------------------------------------------------------------------

print()
print(
    "VALID COMPOSITE OBSERVATIONS"
)
print("-" * 100)

if W20_VALID_DF.empty:

    print(
        "No complete Dataset × Method composite observations "
        "are available."
    )

else:

    print(
        W20_VALID_DF.to_string(
            index=False
        )
    )

# ------------------------------------------------------------------------------
# METHOD-LEVEL CROSS-DATASET SUMMARY
# ------------------------------------------------------------------------------

if W20_VALID_DF.empty:

    W20_SUMMARY_DF = pd.DataFrame(
        columns=[
            "Method",
            "Mean_Overall_Performance",
            "Std_Overall_Performance",
            "Min_Overall_Performance",
            "Max_Overall_Performance",
            "Datasets_Evaluated",
            "Expected_Datasets",
            "Coverage_Ratio",
        ]
    )

else:

    W20_SUMMARY_DF = (
        W20_VALID_DF
        .groupby(
            "Method",
            as_index=False,
        )
        .agg(
            Mean_Overall_Performance=(
                "Overall_Performance_Score",
                "mean",
            ),
            Std_Overall_Performance=(
                "Overall_Performance_Score",
                "std",
            ),
            Min_Overall_Performance=(
                "Overall_Performance_Score",
                "min",
            ),
            Max_Overall_Performance=(
                "Overall_Performance_Score",
                "max",
            ),
            Datasets_Evaluated=(
                "Dataset",
                "nunique",
            ),
        )
    )

# ------------------------------------------------------------------------------
# EXPECTED DATASET COUNT
# ------------------------------------------------------------------------------

W20_EXPECTED_DATASET_COUNT = len(
    W_EXPECTED_DATASETS
)

W20_SUMMARY_DF[
    "Expected_Datasets"
] = W20_EXPECTED_DATASET_COUNT

# ------------------------------------------------------------------------------
# COVERAGE RATIO
# ------------------------------------------------------------------------------

if not W20_SUMMARY_DF.empty:

    W20_SUMMARY_DF[
        "Coverage_Ratio"
    ] = (
        W20_SUMMARY_DF[
            "Datasets_Evaluated"
        ]
        /
        W20_EXPECTED_DATASET_COUNT
    )

# ------------------------------------------------------------------------------
# STANDARD DEVIATION
# ------------------------------------------------------------------------------
#
# If only one dataset is available, pandas returns NaN.
# This is scientifically preferable to pretending that variability is zero.
# ------------------------------------------------------------------------------

if not W20_SUMMARY_DF.empty:

    W20_SUMMARY_DF[
        "Std_Overall_Performance"
    ] = (
        W20_SUMMARY_DF[
            "Std_Overall_Performance"
        ]
        .astype(float)
    )

# ------------------------------------------------------------------------------
# RANK ONLY METHODS WITH VALID MEAN SCORES
# ------------------------------------------------------------------------------

if not W20_SUMMARY_DF.empty:

    W20_SUMMARY_DF[
        "Performance_Rank"
    ] = np.nan

    W20_VALID_RANK_MASK = (
        W20_SUMMARY_DF[
            "Mean_Overall_Performance"
        ]
        .notna()
        &
        np.isfinite(
            W20_SUMMARY_DF[
                "Mean_Overall_Performance"
            ]
        )
    )

    W20_SUMMARY_DF.loc[
        W20_VALID_RANK_MASK,
        "Performance_Rank",
    ] = (
        W20_SUMMARY_DF.loc[
            W20_VALID_RANK_MASK,
            "Mean_Overall_Performance",
        ]
        .rank(
            ascending=False,
            method="min",
        )
    )

# ------------------------------------------------------------------------------
# DO NOT CONVERT RANK TO INTEGER UNLESS NO NaN EXISTS
# ------------------------------------------------------------------------------
#
# Pandas integer conversion fails when NaN is present.
# We therefore retain the rank as a nullable integer.
# ------------------------------------------------------------------------------

if not W20_SUMMARY_DF.empty:

    W20_SUMMARY_DF[
        "Performance_Rank"
    ] = (
        W20_SUMMARY_DF[
            "Performance_Rank"
        ]
        .astype("Int64")
    )

# ------------------------------------------------------------------------------
# SORT
# ------------------------------------------------------------------------------

if not W20_SUMMARY_DF.empty:

    W20_SUMMARY_DF = (
        W20_SUMMARY_DF
        .sort_values(
            [
                "Performance_Rank",
                "Mean_Overall_Performance",
            ],
            ascending=[
                True,
                False,
            ],
            na_position="last",
        )
        .reset_index(
            drop=True
        )
    )

# ------------------------------------------------------------------------------
# DISPLAY METHOD SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("METHOD-LEVEL CROSS-DATASET SUMMARY")
print("=" * 100)

if W20_SUMMARY_DF.empty:

    print()
    print(
        "No method has sufficient complete composite evidence "
        "for cross-dataset summarization."
    )

else:

    print()

    print(
        W20_SUMMARY_DF.to_string(
            index=False
        )
    )

# ------------------------------------------------------------------------------
# METHOD COVERAGE STATUS
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("METHOD COVERAGE STATUS")
print("=" * 100)

if not W20_SUMMARY_DF.empty:

    for _, row in W20_SUMMARY_DF.iterrows():

        method = row[
            "Method"
        ]

        datasets_evaluated = int(
            row[
                "Datasets_Evaluated"
            ]
        )

        expected = int(
            row[
                "Expected_Datasets"
            ]
        )

        coverage = (
            row[
                "Coverage_Ratio"
            ]
        )

        rank = row[
            "Performance_Rank"
        ]

        if pd.isna(rank):

            rank_display = "Not ranked"

        else:

            rank_display = str(
                int(rank)
            )

        print(
            f"{method:12s} | "
            f"Datasets: "
            f"{datasets_evaluated}/{expected} | "
            f"Coverage: "
            f"{coverage:.2%} | "
            f"Rank: "
            f"{rank_display}"
        )

# ------------------------------------------------------------------------------
# CROSS-DATASET PERFORMANCE MATRIX
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("CROSS-DATASET PERFORMANCE MATRIX")
print("=" * 100)

if W20_VALID_DF.empty:

    W20_PERFORMANCE_MATRIX = pd.DataFrame()

    print()
    print(
        "No valid composite scores available."
    )

else:

    W20_PERFORMANCE_MATRIX = (
        W20_VALID_DF
        .pivot_table(
            index="Method",
            columns="Dataset",
            values="Overall_Performance_Score",
            aggfunc="mean",
        )
        .reindex(
            index=W_EXPECTED_METHODS,
            columns=W_EXPECTED_DATASETS,
        )
    )

    print()

    print(
        W20_PERFORMANCE_MATRIX.to_string()
    )

# ------------------------------------------------------------------------------
# SAVE GLOBAL OBJECTS
# ------------------------------------------------------------------------------

W_CROSS_DATASET_SUMMARY = (
    W20_SUMMARY_DF.copy()
)

W20_STATUS = (
    "NO_VALID_COMPOSITES"
    if W20_VALID_DF.empty
    else
    "PARTIAL_COVERAGE"
    if (
        W20_SUMMARY_DF[
            "Datasets_Evaluated"
        ]
        <
        W20_EXPECTED_DATASET_COUNT
    ).any()
    else
    "COMPLETE_COVERAGE"
)

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("W20 — FINAL STATUS")
print("=" * 100)

print()
print(
    f"Expected datasets per method : "
    f"{W20_EXPECTED_DATASET_COUNT}"
)

print(
    f"Valid composite observations : "
    f"{len(W20_VALID_DF)}"
)

print(
    f"Methods summarized            : "
    f"{len(W20_SUMMARY_DF)}"
)

print()
print(
    f"W20 STATUS: {W20_STATUS}"
)

print()
print(
    "Methods with incomplete Dataset coverage "
    "are not treated as having zero performance."
)

print(
    "Methods without a valid mean composite score "
    "are not assigned an artificial rank."
)
```

```python
# ==============================================================================
# W21 — DATASET-LEVEL COMPOSITE RANKING
# ==============================================================================

print()
print("=" * 100)
print("W21 — DATASET-LEVEL COMPOSITE RANKING")
print("=" * 100)

# ------------------------------------------------------------------------------
# VALIDATE SOURCE
# ------------------------------------------------------------------------------

if "W18_COMPOSITE_DF" not in globals():

    raise RuntimeError(
        "W21 FAILED: W18_COMPOSITE_DF is not available. "
        "Run W18 before W21."
    )

if W18_COMPOSITE_DF.empty:

    raise RuntimeError(
        "W21 FAILED: W18_COMPOSITE_DF is empty."
    )

# ------------------------------------------------------------------------------
# COPY SOURCE
# ------------------------------------------------------------------------------

W21_SOURCE_DF = (
    W18_COMPOSITE_DF.copy()
)

# ------------------------------------------------------------------------------
# REQUIRED COLUMNS
# ------------------------------------------------------------------------------

W21_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Overall_Performance_Score",
]

W21_MISSING_COLUMNS = [
    column
    for column in W21_REQUIRED_COLUMNS
    if column not in W21_SOURCE_DF.columns
]

if W21_MISSING_COLUMNS:

    raise RuntimeError(
        "W21 FAILED: Missing required columns: "
        +
        ", ".join(
            W21_MISSING_COLUMNS
        )
    )

# ------------------------------------------------------------------------------
# NORMALIZE IDENTIFIERS
# ------------------------------------------------------------------------------

W21_SOURCE_DF[
    "Dataset"
] = (
    W21_SOURCE_DF[
        "Dataset"
    ]
    .astype(str)
    .str.strip()
)

W21_SOURCE_DF[
    "Method"
] = (
    W21_SOURCE_DF[
        "Method"
    ]
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------------------------
# NUMERIC SCORE
# ------------------------------------------------------------------------------

W21_SOURCE_DF[
    "Overall_Performance_Score"
] = pd.to_numeric(
    W21_SOURCE_DF[
        "Overall_Performance_Score"
    ],
    errors="coerce",
)

# ------------------------------------------------------------------------------
# KEEP ONLY FINITE SCORES
# ------------------------------------------------------------------------------

W21_VALID_DF = (
    W21_SOURCE_DF[
        W21_SOURCE_DF[
            "Overall_Performance_Score"
        ]
        .notna()
        &
        np.isfinite(
            W21_SOURCE_DF[
                "Overall_Performance_Score"
            ]
        )
    ]
    [
        [
            "Dataset",
            "Method",
            "Overall_Performance_Score",
        ]
    ]
    .copy()
)

# ------------------------------------------------------------------------------
# DATASET-LEVEL RANKING
# ------------------------------------------------------------------------------

W21_DATASET_RANKING_LIST = []

for dataset in W_EXPECTED_DATASETS:

    dataset_df = (
        W21_VALID_DF[
            W21_VALID_DF[
                "Dataset"
            ]
            ==
            dataset
        ]
        .copy()
    )

    # --------------------------------------------------------------------------
    # IF NO VALID RESULTS EXIST
    # --------------------------------------------------------------------------

    if dataset_df.empty:

        for method in W_EXPECTED_METHODS:

            W21_DATASET_RANKING_LIST.append(
                {
                    "Dataset":
                        dataset,
                    "Method":
                        method,
                    "Overall_Performance_Score":
                        np.nan,
                    "Dataset_Rank":
                        pd.NA,
                    "Dataset_Status":
                        "NO_VALID_COMPOSITE",
                }
            )

        continue

    # --------------------------------------------------------------------------
    # RANK AVAILABLE METHODS ONLY
    # --------------------------------------------------------------------------

    dataset_df[
        "Dataset_Rank"
    ] = (
        dataset_df[
            "Overall_Performance_Score"
        ]
        .rank(
            ascending=False,
            method="min",
        )
    )

    # --------------------------------------------------------------------------
    # ADD VALID RESULTS
    # --------------------------------------------------------------------------

    for _, row in dataset_df.iterrows():

        W21_DATASET_RANKING_LIST.append(
            {
                "Dataset":
                    row["Dataset"],

                "Method":
                    row["Method"],

                "Overall_Performance_Score":
                    row[
                        "Overall_Performance_Score"
                    ],

                "Dataset_Rank":
                    row[
                        "Dataset_Rank"
                    ],

                "Dataset_Status":
                    "VALID_COMPOSITE",
            }
        )

    # --------------------------------------------------------------------------
    # ADD METHODS WITHOUT VALID COMPOSITE
    # --------------------------------------------------------------------------

    evaluated_methods = set(
        dataset_df[
            "Method"
        ]
        .astype(str)
        .tolist()
    )

    for method in W_EXPECTED_METHODS:

        if method not in evaluated_methods:

            W21_DATASET_RANKING_LIST.append(
                {
                    "Dataset":
                        dataset,
                    "Method":
                        method,
                    "Overall_Performance_Score":
                        np.nan,
                    "Dataset_Rank":
                        pd.NA,
                    "Dataset_Status":
                        "NO_VALID_COMPOSITE",
                }
            )

# ------------------------------------------------------------------------------
# CREATE DATAFRAME
# ------------------------------------------------------------------------------

W21_DATASET_RANKING = pd.DataFrame(
    W21_DATASET_RANKING_LIST
)

# ------------------------------------------------------------------------------
# SAFE NUMERIC CONVERSION
# ------------------------------------------------------------------------------

W21_DATASET_RANKING[
    "Overall_Performance_Score"
] = pd.to_numeric(
    W21_DATASET_RANKING[
        "Overall_Performance_Score"
    ],
    errors="coerce",
)

W21_DATASET_RANKING[
    "Dataset_Rank"
] = pd.to_numeric(
    W21_DATASET_RANKING[
        "Dataset_Rank"
    ],
    errors="coerce",
).astype(
    "Int64"
)

# ------------------------------------------------------------------------------
# SORT
# ------------------------------------------------------------------------------

W21_DATASET_RANKING = (
    W21_DATASET_RANKING
    .sort_values(
        [
            "Dataset",
            "Dataset_Rank",
            "Method",
        ],
        ascending=[
            True,
            True,
            True,
        ],
        na_position="last",
    )
    .reset_index(
        drop=True
    )
)

# ------------------------------------------------------------------------------
# DISPLAY DATASET-LEVEL RANKING
# ------------------------------------------------------------------------------

print()
print(
    W21_DATASET_RANKING.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# DATASET SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("DATASET-LEVEL RANKING SUMMARY")
print("=" * 100)

for dataset in W_EXPECTED_DATASETS:

    dataset_results = (
        W21_DATASET_RANKING[
            W21_DATASET_RANKING[
                "Dataset"
            ]
            ==
            dataset
        ]
    )

    valid_count = int(
        (
            dataset_results[
                "Dataset_Status"
            ]
            ==
            "VALID_COMPOSITE"
        )
        .sum()
    )

    print()
    print(
        f"{dataset}: "
        f"{valid_count}/"
        f"{len(W_EXPECTED_METHODS)} "
        "methods with valid composite evidence"
    )

# ------------------------------------------------------------------------------
# DATASET × METHOD SCORE MATRIX
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("DATASET × METHOD COMPOSITE SCORE MATRIX")
print("=" * 100)

W21_SCORE_MATRIX = (
    W21_DATASET_RANKING
    .pivot(
        index="Dataset",
        columns="Method",
        values="Overall_Performance_Score",
    )
    .reindex(
        index=W_EXPECTED_DATASETS,
        columns=W_EXPECTED_METHODS,
    )
)

print()
print(
    W21_SCORE_MATRIX.to_string()
)

# ------------------------------------------------------------------------------
# DATASET × METHOD RANK MATRIX
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("DATASET × METHOD RANK MATRIX")
print("=" * 100)

W21_RANK_MATRIX = (
    W21_DATASET_RANKING
    .pivot(
        index="Dataset",
        columns="Method",
        values="Dataset_Rank",
    )
    .reindex(
        index=W_EXPECTED_DATASETS,
        columns=W_EXPECTED_METHODS,
    )
)

print()
print(
    W21_RANK_MATRIX.to_string()
)

# ------------------------------------------------------------------------------
# METHOD WIN COUNTS
# ------------------------------------------------------------------------------

W21_METHOD_WIN_COUNTS = (
    W21_DATASET_RANKING[
        W21_DATASET_RANKING[
            "Dataset_Rank"
        ]
        ==
        1
    ]
    .groupby(
        "Method",
        as_index=False,
    )
    .size()
    .rename(
        columns={
            "size":
                "Dataset_Wins"
        }
    )
)

# ------------------------------------------------------------------------------
# INCLUDE METHODS WITH ZERO WINS
# ------------------------------------------------------------------------------

W21_METHOD_WIN_COUNTS = (
    pd.DataFrame(
        {
            "Method":
                W_EXPECTED_METHODS
        }
    )
    .merge(
        W21_METHOD_WIN_COUNTS,
        on="Method",
        how="left",
    )
)

W21_METHOD_WIN_COUNTS[
    "Dataset_Wins"
] = (
    W21_METHOD_WIN_COUNTS[
        "Dataset_Wins"
    ]
    .fillna(0)
    .astype(int)
)

# ------------------------------------------------------------------------------
# DISPLAY WIN COUNTS
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("METHOD DATASET-WIN SUMMARY")
print("=" * 100)

print()

print(
    W21_METHOD_WIN_COUNTS.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

W21_VALID_COUNT = int(
    (
        W21_DATASET_RANKING[
            "Dataset_Status"
        ]
        ==
        "VALID_COMPOSITE"
    )
    .sum()
)

W21_EXPECTED_COUNT = (
    len(W_EXPECTED_DATASETS)
    *
    len(W_EXPECTED_METHODS)
)

print()
print("=" * 100)
print("W21 — FINAL STATUS")
print("=" * 100)

print()
print(
    f"Expected Dataset × Method pairs : "
    f"{W21_EXPECTED_COUNT}"
)

print(
    f"Valid composite pairs           : "
    f"{W21_VALID_COUNT}"
)

print(
    f"Incomplete pairs                : "
    f"{W21_EXPECTED_COUNT - W21_VALID_COUNT}"
)

if W21_VALID_COUNT == 0:

    W21_STATUS = (
        "NO_VALID_DATASET_COMPOSITES"
    )

elif W21_VALID_COUNT < W21_EXPECTED_COUNT:

    W21_STATUS = (
        "PARTIAL_DATASET_COMPOSITES"
    )

else:

    W21_STATUS = (
        "COMPLETE_DATASET_COMPOSITES"
    )

print()
print(
    f"W21 STATUS: {W21_STATUS}"
)

print()
print(
    "Missing composite evidence remains NaN "
    "and is not converted into zero performance."
)
```

```python
# ==============================================================================
# W22 — COMPOSITE ROBUSTNESS ACROSS DATASETS
# ==============================================================================

print()
print("=" * 100)
print("W22 — COMPOSITE ROBUSTNESS ACROSS DATASETS")
print("=" * 100)

W_ROBUSTNESS_RESULTS = []

for method in W_METHODS:

    method_scores = (
        W_COMPONENT_SCORES[
            W_COMPONENT_SCORES[
                "Method"
            ]
            ==
            method
        ][
            "Overall_Performance_Score"
        ]
        .to_numpy(
            dtype=float
        )
    )

    if len(method_scores) == 0:
        continue

    mean_score = np.mean(
        method_scores
    )

    sd_score = (
        np.std(
            method_scores,
            ddof=1,
        )
        if len(method_scores) > 1
        else 0.0
    )

    score_range = (
        np.max(method_scores)
        -
        np.min(method_scores)
    )

    W_ROBUSTNESS_RESULTS.append(
        {
            "Method": method,
            "Mean_Overall_Score":
                mean_score,
            "SD_Overall_Score":
                sd_score,
            "Minimum_Overall_Score":
                np.min(method_scores),
            "Maximum_Overall_Score":
                np.max(method_scores),
            "Cross_Dataset_Range":
                score_range,
        }
    )

W_ROBUSTNESS_DF = (
    pd.DataFrame(
        W_ROBUSTNESS_RESULTS
    )
    .sort_values(
        "Mean_Overall_Score",
        ascending=False,
    )
    .reset_index(
        drop=True
    )
)

print()
print(
    W_ROBUSTNESS_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# W23 — FINAL COMPOSITE SCORE VALIDATION
# ==============================================================================

print()
print("=" * 100)
print("W23 — FINAL COMPOSITE SCORE VALIDATION")
print("=" * 100)

# ------------------------------------------------------------------------------
# SOURCE VALIDATION
# ------------------------------------------------------------------------------

if "W18_COMPOSITE_DF" not in globals():

    raise RuntimeError(
        "W23 FAILED: W18_COMPOSITE_DF is not available."
    )

W23_DF = (
    W18_COMPOSITE_DF.copy()
)

if W23_DF.empty:

    raise RuntimeError(
        "W23 FAILED: Composite performance DataFrame is empty."
    )

# ------------------------------------------------------------------------------
# REQUIRED COLUMNS
# ------------------------------------------------------------------------------

W23_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Fidelity_Score",
    "Utility_Score",
    "Privacy_Score",
    "Overall_Performance_Score",
]

W23_MISSING_COLUMNS = [
    column
    for column in W23_REQUIRED_COLUMNS
    if column not in W23_DF.columns
]

if W23_MISSING_COLUMNS:

    raise RuntimeError(
        "W23 FAILED: Missing required columns: "
        +
        ", ".join(
            W23_MISSING_COLUMNS
        )
    )

# ------------------------------------------------------------------------------
# NORMALIZE IDENTIFIERS
# ------------------------------------------------------------------------------

W23_DF[
    "Dataset"
] = (
    W23_DF[
        "Dataset"
    ]
    .astype(str)
    .str.strip()
)

W23_DF[
    "Method"
] = (
    W23_DF[
        "Method"
    ]
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------------------------
# NUMERIC CONVERSION
# ------------------------------------------------------------------------------

W23_NUMERIC_COLUMNS = [
    "Fidelity_Score",
    "Utility_Score",
    "Privacy_Score",
    "Overall_Performance_Score",
]

for column in W23_NUMERIC_COLUMNS:

    W23_DF[column] = pd.to_numeric(
        W23_DF[column],
        errors="coerce",
    )

# ------------------------------------------------------------------------------
# EXPECTED GRID VALIDATION
# ------------------------------------------------------------------------------

W23_EXPECTED_GRID = pd.MultiIndex.from_product(
    [
        W_EXPECTED_DATASETS,
        W_EXPECTED_METHODS,
    ],
    names=[
        "Dataset",
        "Method",
    ],
)

W23_ACTUAL_GRID = pd.MultiIndex.from_frame(
    W23_DF[
        [
            "Dataset",
            "Method",
        ]
    ]
)

W23_ALL_GRID_PRESENT = (
    W23_EXPECTED_GRID
    .difference(
        W23_ACTUAL_GRID
    )
    .empty
)

# ------------------------------------------------------------------------------
# COMPONENT AVAILABILITY
# ------------------------------------------------------------------------------

W23_DF[
    "Fidelity_Available"
] = (
    W23_DF[
        "Fidelity_Score"
    ]
    .notna()
)

W23_DF[
    "Utility_Available"
] = (
    W23_DF[
        "Utility_Score"
    ]
    .notna()
)

W23_DF[
    "Privacy_Available"
] = (
    W23_DF[
        "Privacy_Score"
    ]
    .notna()
)

W23_DF[
    "Overall_Available"
] = (
    W23_DF[
        "Overall_Performance_Score"
    ]
    .notna()
)

# ------------------------------------------------------------------------------
# COMPLETE COMPOSITE EVIDENCE
# ------------------------------------------------------------------------------

W23_DF[
    "Composite_Complete"
] = (
    W23_DF[
        [
            "Fidelity_Available",
            "Utility_Available",
            "Privacy_Available",
            "Overall_Available",
        ]
    ]
    .all(axis=1)
)

# ------------------------------------------------------------------------------
# RANGE CHECK FUNCTION
# ------------------------------------------------------------------------------

def W23_CHECK_RANGE(
    series,
):

    valid = (
        series
        .dropna()
    )

    if valid.empty:

        return True

    return bool(
        (
            (valid >= 0.0)
            &
            (valid <= 100.0)
        )
        .all()
    )

# ------------------------------------------------------------------------------
# COMPLETE COMPONENT CHECKS
# ------------------------------------------------------------------------------
#
# These checks evaluate whether the values that EXIST are valid.
# Missing values are NOT treated as failures because missing evidence
# is represented explicitly elsewhere.
# ------------------------------------------------------------------------------

W23_FIDELITY_COMPLETE = bool(
    W23_DF[
        "Fidelity_Score"
    ]
    .notna()
    .all()
)

W23_UTILITY_COMPLETE = bool(
    W23_DF[
        "Utility_Score"
    ]
    .notna()
    .all()
)

W23_PRIVACY_COMPLETE = bool(
    W23_DF[
        "Privacy_Score"
    ]
    .notna()
    .all()
)

W23_OVERALL_COMPLETE = bool(
    W23_DF[
        "Overall_Performance_Score"
    ]
    .notna()
    .all()
)

# ------------------------------------------------------------------------------
# RANGE CHECKS
# ------------------------------------------------------------------------------

W23_FIDELITY_RANGE = W23_CHECK_RANGE(
    W23_DF[
        "Fidelity_Score"
    ]
)

W23_UTILITY_RANGE = W23_CHECK_RANGE(
    W23_DF[
        "Utility_Score"
    ]
)

W23_PRIVACY_RANGE = W23_CHECK_RANGE(
    W23_DF[
        "Privacy_Score"
    ]
)

W23_OVERALL_RANGE = W23_CHECK_RANGE(
    W23_DF[
        "Overall_Performance_Score"
    ]
)

# ------------------------------------------------------------------------------
# COMPOSITE WEIGHT CHECK
# ------------------------------------------------------------------------------

W23_WEIGHT_SUM = (
    W_FIDELITY_WEIGHT
    +
    W_UTILITY_WEIGHT
    +
    W_PRIVACY_WEIGHT
)

W23_WEIGHTS_VALID = bool(
    np.isclose(
        W23_WEIGHT_SUM,
        1.0,
        atol=1e-10,
    )
)

# ------------------------------------------------------------------------------
# VALIDATE OVERALL SCORE AGAINST COMPONENT FORMULA
# ------------------------------------------------------------------------------
#
# Only rows with all three components available are tested.
# ------------------------------------------------------------------------------

W23_COMPLETE_COMPONENT_MASK = (
    W23_DF[
        [
            "Fidelity_Available",
            "Utility_Available",
            "Privacy_Available",
        ]
    ]
    .all(axis=1)
)

W23_EXPECTED_OVERALL = np.full(
    len(W23_DF),
    np.nan,
)

W23_EXPECTED_OVERALL[
    W23_COMPLETE_COMPONENT_MASK
] = (
    W_FIDELITY_WEIGHT
    *
    W23_DF.loc[
        W23_COMPLETE_COMPONENT_MASK,
        "Fidelity_Score",
    ]
    +
    W_UTILITY_WEIGHT
    *
    W23_DF.loc[
        W23_COMPLETE_COMPONENT_MASK,
        "Utility_Score",
    ]
    +
    W_PRIVACY_WEIGHT
    *
    W23_DF.loc[
        W23_COMPLETE_COMPONENT_MASK,
        "Privacy_Score",
    ]
)

W23_DF[
    "_Expected_Overall"
] = W23_EXPECTED_OVERALL

W23_FORMULA_CHECK_MASK = (
    W23_DF[
        "Composite_Complete"
    ]
)

if W23_FORMULA_CHECK_MASK.any():

    W23_FORMULA_VALID = bool(
        np.allclose(
            W23_DF.loc[
                W23_FORMULA_CHECK_MASK,
                "Overall_Performance_Score",
            ],
            W23_DF.loc[
                W23_FORMULA_CHECK_MASK,
                "_Expected_Overall",
            ],
            atol=1e-8,
            rtol=1e-8,
        )
    )

else:

    W23_FORMULA_VALID = True

# ------------------------------------------------------------------------------
# COMPONENT COMPLETENESS COUNTS
# ------------------------------------------------------------------------------

W23_TOTAL_ROWS = len(
    W23_DF
)

W23_COMPLETE_ROWS = int(
    W23_DF[
        "Composite_Complete"
    ]
    .sum()
)

W23_INCOMPLETE_ROWS = (
    W23_TOTAL_ROWS
    -
    W23_COMPLETE_ROWS
)

# ------------------------------------------------------------------------------
# VALIDATION TABLE
# ------------------------------------------------------------------------------

W23_VALIDATION_ROWS = [
    {
        "Check":
            "All expected Dataset × Method combinations present",
        "Status":
            W23_ALL_GRID_PRESENT,
    },
    {
        "Check":
            "Fidelity_Score complete",
        "Status":
            W23_FIDELITY_COMPLETE,
    },
    {
        "Check":
            "Fidelity_Score within 0–100",
        "Status":
            W23_FIDELITY_RANGE,
    },
    {
        "Check":
            "Utility_Score complete",
        "Status":
            W23_UTILITY_COMPLETE,
    },
    {
        "Check":
            "Utility_Score within 0–100",
        "Status":
            W23_UTILITY_RANGE,
    },
    {
        "Check":
            "Privacy_Score complete",
        "Status":
            W23_PRIVACY_COMPLETE,
    },
    {
        "Check":
            "Privacy_Score within 0–100",
        "Status":
            W23_PRIVACY_RANGE,
    },
    {
        "Check":
            "Overall Performance Score complete",
        "Status":
            W23_OVERALL_COMPLETE,
    },
    {
        "Check":
            "Overall Performance Score within 0–100",
        "Status":
            W23_OVERALL_RANGE,
    },
    {
        "Check":
            "Composite weights sum to 1",
        "Status":
            W23_WEIGHTS_VALID,
    },
    {
        "Check":
            "Overall score follows weighted component formula",
        "Status":
            W23_FORMULA_VALID,
    },
]

W23_VALIDATION_DF = pd.DataFrame(
    W23_VALIDATION_ROWS
)

# ------------------------------------------------------------------------------
# DISPLAY VALIDATION
# ------------------------------------------------------------------------------

print()

print(
    W23_VALIDATION_DF.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# PASSED CHECK COUNT
# ------------------------------------------------------------------------------

W23_PASSED = int(
    W23_VALIDATION_DF[
        "Status"
    ]
    .sum()
)

W23_TOTAL = len(
    W23_VALIDATION_DF
)

print()
print(
    f"Passed checks : "
    f"{W23_PASSED}/{W23_TOTAL}"
)

# ------------------------------------------------------------------------------
# EVIDENCE COMPLETENESS SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 100)
print("W23 — EVIDENCE COMPLETENESS")
print("=" * 100)

print()

print(
    f"Total Dataset × Method combinations : "
    f"{W23_TOTAL_ROWS}"
)

print(
    f"Complete composite combinations     : "
    f"{W23_COMPLETE_ROWS}"
)

print(
    f"Incomplete composite combinations   : "
    f"{W23_INCOMPLETE_ROWS}"
)

# ------------------------------------------------------------------------------
# COMPONENT AVAILABILITY
# ------------------------------------------------------------------------------

print()
print(
    "Component availability:"
)

print(
    f"  Fidelity : "
    f"{int(W23_DF['Fidelity_Available'].sum())}/"
    f"{W23_TOTAL_ROWS}"
)

print(
    f"  Utility  : "
    f"{int(W23_DF['Utility_Available'].sum())}/"
    f"{W23_TOTAL_ROWS}"
)

print(
    f"  Privacy  : "
    f"{int(W23_DF['Privacy_Available'].sum())}/"
    f"{W23_TOTAL_ROWS}"
)

print(
    f"  Overall  : "
    f"{int(W23_DF['Overall_Available'].sum())}/"
    f"{W23_TOTAL_ROWS}"
)

# ------------------------------------------------------------------------------
# INCOMPLETE COMBINATIONS
# ------------------------------------------------------------------------------

if W23_INCOMPLETE_ROWS > 0:

    print()
    print("=" * 100)
    print("W23 — INCOMPLETE COMPOSITE COMBINATIONS")
    print("=" * 100)

    W23_INCOMPLETE_DF = (
        W23_DF[
            ~W23_DF[
                "Composite_Complete"
            ]
        ]
        [
            [
                "Dataset",
                "Method",
                "Fidelity_Available",
                "Utility_Available",
                "Privacy_Available",
                "Overall_Available",
            ]
        ]
        .copy()
    )

    print()

    print(
        W23_INCOMPLETE_DF.to_string(
            index=False
        )
    )

else:

    W23_INCOMPLETE_DF = pd.DataFrame()

# ------------------------------------------------------------------------------
# FINAL VALIDATION STATUS
# ------------------------------------------------------------------------------

# Structural/range/formula validity is what determines whether the
# composite calculation itself is valid.
#
# Incomplete evidence is NOT treated as a computational failure.

W23_CORE_CHECKS = [
    W23_ALL_GRID_PRESENT,
    W23_FIDELITY_RANGE,
    W23_UTILITY_RANGE,
    W23_PRIVACY_RANGE,
    W23_OVERALL_RANGE,
    W23_WEIGHTS_VALID,
    W23_FORMULA_VALID,
]

W23_CORE_VALID = all(
    W23_CORE_CHECKS
)

if W23_CORE_VALID:

    if W23_INCOMPLETE_ROWS == 0:

        W23_STATUS = (
            "VALID_COMPLETE_EVIDENCE"
        )

    else:

        W23_STATUS = (
            "VALID_PARTIAL_EVIDENCE"
        )

else:

    W23_STATUS = (
        "INVALID_COMPOSITE_SCORING"
    )

print()
print("=" * 100)
print("W23 — FINAL STATUS")
print("=" * 100)

print()
print(
    f"W23 STATUS: {W23_STATUS}"
)

print()

if W23_STATUS == "VALID_COMPLETE_EVIDENCE":

    print(
        "All Dataset × Method combinations contain "
        "complete and valid composite evidence."
    )

elif W23_STATUS == "VALID_PARTIAL_EVIDENCE":

    print(
        "Composite scoring is mathematically valid, "
        "but some Dataset × Method combinations lack "
        "one or more component values."
    )

    print(
        "Missing evidence remains NaN and is not "
        "treated as zero performance."
    )

else:

    raise RuntimeError(
        "W23 FAILED: Invalid composite scoring structure "
        "or out-of-range component values detected."
    )

# ------------------------------------------------------------------------------
# CLEANUP
# ------------------------------------------------------------------------------

W23_DF = W23_DF.drop(
    columns=[
        "_Expected_Overall",
    ]
)

# ------------------------------------------------------------------------------
# FINAL OBJECTS
# ------------------------------------------------------------------------------

W_FINAL_COMPOSITE_DF = (
    W23_DF.copy()
)

W_FINAL_COMPOSITE_VALIDATION = (
    W23_VALIDATION_DF.copy()
)
```

```python
# ==============================================================================
# W24 — SAVE FINAL COMPOSITE PERFORMANCE REPORTS
# ==============================================================================

print()
print("=" * 100)
print("W24 — SAVE FINAL COMPOSITE PERFORMANCE REPORTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------------------------

if "SPP_GAN_RESULTS_DIRECTORY" in globals():

    W24_OUTPUT_DIRECTORY = pathlib.Path(
        SPP_GAN_RESULTS_DIRECTORY
    )

elif "RESULTS_DIRECTORY" in globals():

    W24_OUTPUT_DIRECTORY = pathlib.Path(
        RESULTS_DIRECTORY
    )

elif "PROJECT_RESULTS_DIRECTORY" in globals():

    W24_OUTPUT_DIRECTORY = pathlib.Path(
        PROJECT_RESULTS_DIRECTORY
    )

else:

    W24_OUTPUT_DIRECTORY = pathlib.Path(
        "results"
    )

W24_OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)

print()
print(
    f"Output directory: "
    f"{W24_OUTPUT_DIRECTORY}"
)

# ------------------------------------------------------------------------------
# VALIDATE FINAL OBJECTS
# ------------------------------------------------------------------------------

if "W_FINAL_COMPOSITE_DF" not in globals():

    if "W23_DF" in globals():

        W_FINAL_COMPOSITE_DF = (
            W23_DF.copy()
        )

    elif "W18_COMPOSITE_DF" in globals():

        W_FINAL_COMPOSITE_DF = (
            W18_COMPOSITE_DF.copy()
        )

    else:

        raise RuntimeError(
            "W24 FAILED: No final composite performance "
            "DataFrame is available."
        )

# ------------------------------------------------------------------------------
# FINAL VALIDATION OBJECT
# ------------------------------------------------------------------------------

if "W_FINAL_COMPOSITE_VALIDATION" not in globals():

    if "W23_VALIDATION_DF" in globals():

        W_FINAL_COMPOSITE_VALIDATION = (
            W23_VALIDATION_DF.copy()
        )

    else:

        W_FINAL_COMPOSITE_VALIDATION = pd.DataFrame()

# ------------------------------------------------------------------------------
# W20 SUMMARY
# ------------------------------------------------------------------------------

if "W_CROSS_DATASET_SUMMARY" in globals():

    W24_CROSS_DATASET_SUMMARY = (
        W_CROSS_DATASET_SUMMARY.copy()
    )

elif "W20_SUMMARY_DF" in globals():

    W24_CROSS_DATASET_SUMMARY = (
        W20_SUMMARY_DF.copy()
    )

else:

    W24_CROSS_DATASET_SUMMARY = pd.DataFrame()

# ------------------------------------------------------------------------------
# W21 DATASET RANKING
# ------------------------------------------------------------------------------

if "W21_DATASET_RANKING" in globals():

    W24_DATASET_RANKING = (
        W21_DATASET_RANKING.copy()
    )

else:

    W24_DATASET_RANKING = pd.DataFrame()

# ------------------------------------------------------------------------------
# W21 SCORE MATRIX
# ------------------------------------------------------------------------------

if "W21_SCORE_MATRIX" in globals():

    W24_SCORE_MATRIX = (
        W21_SCORE_MATRIX.copy()
    )

else:

    W24_SCORE_MATRIX = pd.DataFrame()

# ------------------------------------------------------------------------------
# W21 RANK MATRIX
# ------------------------------------------------------------------------------

if "W21_RANK_MATRIX" in globals():

    W24_RANK_MATRIX = (
        W21_RANK_MATRIX.copy()
    )

else:

    W24_RANK_MATRIX = pd.DataFrame()

# ------------------------------------------------------------------------------
# W21 METHOD WIN COUNTS
# ------------------------------------------------------------------------------

if "W21_METHOD_WIN_COUNTS" in globals():

    W24_METHOD_WIN_COUNTS = (
        W21_METHOD_WIN_COUNTS.copy()
    )

else:

    W24_METHOD_WIN_COUNTS = pd.DataFrame()

# ------------------------------------------------------------------------------
# VALIDATE FINAL STATUS
# ------------------------------------------------------------------------------

if "W23_STATUS" in globals():

    W24_STATUS = W23_STATUS

else:

    W24_STATUS = "STATUS_NOT_AVAILABLE"

# ------------------------------------------------------------------------------
# SAVE 1 — FINAL COMPOSITE PERFORMANCE
# ------------------------------------------------------------------------------

W24_COMPOSITE_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "final_composite_performance.csv"
)

W_FINAL_COMPOSITE_DF.to_csv(
    W24_COMPOSITE_FILE,
    index=False,
)

print()
print(
    f"[SAVED] "
    f"{W24_COMPOSITE_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 2 — COMPOSITE VALIDATION
# ------------------------------------------------------------------------------

W24_VALIDATION_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "final_composite_score_validation.csv"
)

W_FINAL_COMPOSITE_VALIDATION.to_csv(
    W24_VALIDATION_FILE,
    index=False,
)

print(
    f"[SAVED] "
    f"{W24_VALIDATION_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 3 — CROSS-DATASET SUMMARY
# ------------------------------------------------------------------------------

W24_SUMMARY_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "cross_dataset_composite_performance_summary.csv"
)

W24_CROSS_DATASET_SUMMARY.to_csv(
    W24_SUMMARY_FILE,
    index=False,
)

print(
    f"[SAVED] "
    f"{W24_SUMMARY_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 4 — DATASET-LEVEL RANKING
# ------------------------------------------------------------------------------

W24_DATASET_RANKING_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "dataset_level_composite_ranking.csv"
)

W24_DATASET_RANKING.to_csv(
    W24_DATASET_RANKING_FILE,
    index=False,
)

print(
    f"[SAVED] "
    f"{W24_DATASET_RANKING_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 5 — SCORE MATRIX
# ------------------------------------------------------------------------------

W24_SCORE_MATRIX_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "dataset_method_composite_score_matrix.csv"
)

W24_SCORE_MATRIX.to_csv(
    W24_SCORE_MATRIX_FILE
)

print(
    f"[SAVED] "
    f"{W24_SCORE_MATRIX_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 6 — RANK MATRIX
# ------------------------------------------------------------------------------

W24_RANK_MATRIX_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "dataset_method_composite_rank_matrix.csv"
)

W24_RANK_MATRIX.to_csv(
    W24_RANK_MATRIX_FILE
)

print(
    f"[SAVED] "
    f"{W24_RANK_MATRIX_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE 7 — METHOD WIN COUNTS
# ------------------------------------------------------------------------------

W24_WIN_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "method_dataset_win_summary.csv"
)

W24_METHOD_WIN_COUNTS.to_csv(
    W24_WIN_FILE,
    index=False,
)

print(
    f"[SAVED] "
    f"{W24_WIN_FILE.name}"
)

# ------------------------------------------------------------------------------
# CREATE W24 STATUS REPORT
# ------------------------------------------------------------------------------

W24_STATUS_REPORT = pd.DataFrame(
    [
        {
            "Check":
                "Composite scoring status",
            "Status":
                W24_STATUS,
        },
        {
            "Check":
                "Expected datasets",
            "Status":
                ", ".join(
                    W_EXPECTED_DATASETS
                ),
        },
        {
            "Check":
                "Expected methods",
            "Status":
                ", ".join(
                    W_EXPECTED_METHODS
                ),
        },
        {
            "Check":
                "Fidelity weight",
            "Status":
                W_FIDELITY_WEIGHT,
        },
        {
            "Check":
                "Utility weight",
            "Status":
                W_UTILITY_WEIGHT,
        },
        {
            "Check":
                "Privacy weight",
            "Status":
                W_PRIVACY_WEIGHT,
        },
        {
            "Check":
                "Metric-wise normalization",
            "Status":
                "Metric-wise min-max",
        },
        {
            "Check":
                "Missing evidence handling",
            "Status":
                "Retained as NaN; no zero imputation",
        },
        {
            "Check":
                "Overall score construction",
            "Status":
                "Weighted arithmetic mean",
        },
    ]
)

W24_STATUS_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "composite_performance_framework_status.csv"
)

W24_STATUS_REPORT.to_csv(
    W24_STATUS_FILE,
    index=False,
)

print(
    f"[SAVED] "
    f"{W24_STATUS_FILE.name}"
)

# ------------------------------------------------------------------------------
# SAVE MASTER EXCEL REPORT
# ------------------------------------------------------------------------------

W24_EXCEL_FILE = (
    W24_OUTPUT_DIRECTORY
    /
    "final_composite_performance_report.xlsx"
)

with pd.ExcelWriter(
    W24_EXCEL_FILE,
    engine="openpyxl",
) as writer:

    W_FINAL_COMPOSITE_DF.to_excel(
        writer,
        sheet_name="Composite Scores",
        index=False,
    )

    W_FINAL_COMPOSITE_VALIDATION.to_excel(
        writer,
        sheet_name="Validation",
        index=False,
    )

    W24_CROSS_DATASET_SUMMARY.to_excel(
        writer,
        sheet_name="Cross Dataset Summary",
        index=False,
    )

    W24_DATASET_RANKING.to_excel(
        writer,
        sheet_name="Dataset Ranking",
        index=False,
    )

    W24_SCORE_MATRIX.to_excel(
        writer,
        sheet_name="Score Matrix",
    )

    W24_RANK_MATRIX.to_excel(
        writer,
        sheet_name="Rank Matrix",
    )

    W24_METHOD_WIN_COUNTS.to_excel(
        writer,
        sheet_name="Method Wins",
        index=False,
    )

    W24_STATUS_REPORT.to_excel(
        writer,
        sheet_name="Framework Status",
        index=False,
    )

print(
    f"[SAVED] "
    f"{W24_EXCEL_FILE.name}"
)

# ------------------------------------------------------------------------------
# FINAL FILE INVENTORY
# ------------------------------------------------------------------------------

W24_FILES = [
    W24_COMPOSITE_FILE,
    W24_VALIDATION_FILE,
    W24_SUMMARY_FILE,
    W24_DATASET_RANKING_FILE,
    W24_SCORE_MATRIX_FILE,
    W24_RANK_MATRIX_FILE,
    W24_WIN_FILE,
    W24_STATUS_FILE,
    W24_EXCEL_FILE,
]

print()
print("=" * 100)
print("W24 — SAVED REPORT INVENTORY")
print("=" * 100)

for file_path in W24_FILES:

    print(
        f"{'OK' if file_path.exists() else 'MISSING':8s} "
        f"{file_path.name}"
    )

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

W24_ALL_FILES_SAVED = all(
    file_path.exists()
    for file_path in W24_FILES
)

print()
print(
    f"Files saved: "
    f"{sum(file_path.exists() for file_path in W24_FILES)}/"
    f"{len(W24_FILES)}"
)

print(
    f"W24 STATUS: "
    f"{'COMPLETE' if W24_ALL_FILES_SAVED else 'INCOMPLETE'}"
)

print()
print(
    "W24 completed without replacing missing experimental "
    "evidence with zero values."
)
```

```python
# ==============================================================================
# X1 — INITIALIZE METHOD RANKING & SENSITIVITY ANALYSIS
# ==============================================================================

print()
print("=" * 100)
print("X1 — INITIALIZE METHOD RANKING & SENSITIVITY ANALYSIS")
print("=" * 100)

# ------------------------------------------------------------------------------
# REQUIRED LIBRARIES
# ------------------------------------------------------------------------------

import pathlib
import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# EXPECTED DATASETS AND METHODS
# ------------------------------------------------------------------------------

X_EXPECTED_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

X_EXPECTED_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

# ------------------------------------------------------------------------------
# BASELINE WEIGHTS
# ------------------------------------------------------------------------------

X_BASELINE_WEIGHTS = {
    "Fidelity": 1 / 3,
    "Utility": 1 / 3,
    "Privacy": 1 / 3,
}

# ------------------------------------------------------------------------------
# VALIDATE WEIGHTS
# ------------------------------------------------------------------------------

X_BASELINE_WEIGHT_SUM = sum(
    X_BASELINE_WEIGHTS.values()
)

if not np.isclose(
    X_BASELINE_WEIGHT_SUM,
    1.0,
):
    raise RuntimeError(
        "X1 FAILED: Baseline composite weights "
        "do not sum to 1."
    )

# ------------------------------------------------------------------------------
# RESOLVE COMPOSITE SOURCE
# ------------------------------------------------------------------------------

if "W_FINAL_COMPOSITE_DF" in globals():

    X_SOURCE_DF = (
        W_FINAL_COMPOSITE_DF.copy()
    )

elif "W23_DF" in globals():

    X_SOURCE_DF = (
        W23_DF.copy()
    )

elif "W18_COMPOSITE_DF" in globals():

    X_SOURCE_DF = (
        W18_COMPOSITE_DF.copy()
    )

else:

    raise RuntimeError(
        "X1 FAILED: No valid W composite-performance "
        "DataFrame was found."
    )

# ------------------------------------------------------------------------------
# REQUIRED COMPONENTS
# ------------------------------------------------------------------------------

X_REQUIRED_COLUMNS = [
    "Dataset",
    "Method",
    "Fidelity_Score",
    "Utility_Score",
    "Privacy_Score",
]

X_MISSING_COLUMNS = [
    column
    for column in X_REQUIRED_COLUMNS
    if column not in X_SOURCE_DF.columns
]

if X_MISSING_COLUMNS:

    raise RuntimeError(
        "X1 FAILED: Missing required columns: "
        +
        ", ".join(X_MISSING_COLUMNS)
    )

# ------------------------------------------------------------------------------
# NORMALIZE IDENTIFIERS
# ------------------------------------------------------------------------------

X_SOURCE_DF["Dataset"] = (
    X_SOURCE_DF["Dataset"]
    .astype(str)
    .str.strip()
)

X_SOURCE_DF["Method"] = (
    X_SOURCE_DF["Method"]
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------------------------
# NUMERIC COMPONENTS
# ------------------------------------------------------------------------------

for column in [
    "Fidelity_Score",
    "Utility_Score",
    "Privacy_Score",
]:

    X_SOURCE_DF[column] = pd.to_numeric(
        X_SOURCE_DF[column],
        errors="coerce",
    )

# ------------------------------------------------------------------------------
# KEEP EXPECTED EXPERIMENTAL GRID
# ------------------------------------------------------------------------------

X_SOURCE_DF = (
    X_SOURCE_DF[
        X_SOURCE_DF["Dataset"].isin(
            X_EXPECTED_DATASETS
        )
        &
        X_SOURCE_DF["Method"].isin(
            X_EXPECTED_METHODS
        )
    ]
    .copy()
)

# ------------------------------------------------------------------------------
# COMPLETE COMPONENT FLAG
# ------------------------------------------------------------------------------

X_SOURCE_DF["X_Complete"] = (
    X_SOURCE_DF[
        [
            "Fidelity_Score",
            "Utility_Score",
            "Privacy_Score",
        ]
    ]
    .notna()
    .all(axis=1)
)

# ------------------------------------------------------------------------------
# INITIAL SUMMARY
# ------------------------------------------------------------------------------

print()
print(
    f"Datasets : {X_EXPECTED_DATASETS}"
)

print(
    f"Methods  : {X_EXPECTED_METHODS}"
)

print()
print(
    f"Total Dataset × Method rows : "
    f"{len(X_SOURCE_DF)}"
)

print(
    f"Complete component rows      : "
    f"{int(X_SOURCE_DF['X_Complete'].sum())}"
)

print(
    f"Incomplete component rows    : "
    f"{int((~X_SOURCE_DF['X_Complete']).sum())}"
)

print()
print(
    "Baseline weights:"
)

for component, weight in X_BASELINE_WEIGHTS.items():

    print(
        f"  {component:10s}: "
        f"{weight:.4f}"
    )

print()
print(
    "X1 STATUS: READY"
)
```

```python
# ==============================================================================
# X2 — VALIDATE RANKING EVIDENCE
# ==============================================================================

print()
print("=" * 100)
print("X2 — VALIDATE RANKING EVIDENCE")
print("=" * 100)

# ------------------------------------------------------------------------------
# EXPECTED GRID
# ------------------------------------------------------------------------------

X_EXPECTED_GRID = pd.MultiIndex.from_product(
    [
        X_EXPECTED_DATASETS,
        X_EXPECTED_METHODS,
    ],
    names=[
        "Dataset",
        "Method",
    ],
)

# ------------------------------------------------------------------------------
# ACTUAL GRID
# ------------------------------------------------------------------------------

X_ACTUAL_GRID = pd.MultiIndex.from_frame(
    X_SOURCE_DF[
        [
            "Dataset",
            "Method",
        ]
    ]
)

X_MISSING_GRID = (
    X_EXPECTED_GRID
    .difference(
        X_ACTUAL_GRID
    )
)

# ------------------------------------------------------------------------------
# COMPLETENESS MATRIX
# ------------------------------------------------------------------------------

X_COMPLETENESS_MATRIX = (
    X_SOURCE_DF
    .pivot_table(
        index="Dataset",
        columns="Method",
        values="X_Complete",
        aggfunc="max",
        fill_value=False,
    )
    .reindex(
        index=X_EXPECTED_DATASETS,
        columns=X_EXPECTED_METHODS,
        fill_value=False,
    )
)

print()
print(
    "COMPLETE COMPONENT EVIDENCE MATRIX"
)

print(
    X_COMPLETENESS_MATRIX.to_string()
)

# ------------------------------------------------------------------------------
# METHOD-LEVEL EVIDENCE COUNTS
# ------------------------------------------------------------------------------

X_METHOD_EVIDENCE = (
    X_SOURCE_DF
    .groupby(
        "Method",
        as_index=False,
    )
    .agg(
        Datasets_Available=(
            "X_Complete",
            "sum",
        )
    )
)

X_METHOD_EVIDENCE[
    "Datasets_Available"
] = (
    X_METHOD_EVIDENCE[
        "Datasets_Available"
    ]
    .astype(int)
)

X_METHOD_EVIDENCE[
    "Datasets_Expected"
] = len(
    X_EXPECTED_DATASETS
)

X_METHOD_EVIDENCE[
    "Evidence_Coverage"
] = (
    X_METHOD_EVIDENCE[
        "Datasets_Available"
    ]
    /
    X_METHOD_EVIDENCE[
        "Datasets_Expected"
    ]
)

X_METHOD_EVIDENCE = (
    pd.DataFrame(
        {
            "Method":
                X_EXPECTED_METHODS
        }
    )
    .merge(
        X_METHOD_EVIDENCE,
        on="Method",
        how="left",
    )
)

X_METHOD_EVIDENCE[
    "Datasets_Available"
] = (
    X_METHOD_EVIDENCE[
        "Datasets_Available"
    ]
    .fillna(0)
    .astype(int)
)

X_METHOD_EVIDENCE[
    "Datasets_Expected"
] = len(
    X_EXPECTED_DATASETS
)

X_METHOD_EVIDENCE[
    "Evidence_Coverage"
] = (
    X_METHOD_EVIDENCE[
        "Datasets_Available"
    ]
    /
    len(X_EXPECTED_DATASETS)
)

print()
print(
    "METHOD-LEVEL EVIDENCE COVERAGE"
)

print(
    X_METHOD_EVIDENCE.to_string(
        index=False
    )
)

print()
print(
    f"Missing Dataset × Method pairs: "
    f"{len(X_MISSING_GRID)}"
)

print(
    "X2 STATUS: "
    +
    (
        "VALID"
        if len(X_MISSING_GRID) == 0
        else "PARTIAL_EVIDENCE"
    )
)
```

```python
# ==============================================================================
# X3 — DEFINE SENSITIVITY WEIGHTING SCENARIOS
# ==============================================================================

print()
print("=" * 100)
print("X3 — DEFINE SENSITIVITY WEIGHTING SCENARIOS")
print("=" * 100)

# ------------------------------------------------------------------------------
# WEIGHTING SCENARIOS
# ------------------------------------------------------------------------------

X_WEIGHT_SCENARIOS = {
    "Equal_Weights": {
        "Fidelity": 1 / 3,
        "Utility": 1 / 3,
        "Privacy": 1 / 3,
    },

    "Fidelity_Emphasis": {
        "Fidelity": 0.50,
        "Utility": 0.25,
        "Privacy": 0.25,
    },

    "Utility_Emphasis": {
        "Fidelity": 0.25,
        "Utility": 0.50,
        "Privacy": 0.25,
    },

    "Privacy_Emphasis": {
        "Fidelity": 0.25,
        "Utility": 0.25,
        "Privacy": 0.50,
    },

    "Fidelity_Privacy_Emphasis": {
        "Fidelity": 0.40,
        "Utility": 0.20,
        "Privacy": 0.40,
    },

    "Fidelity_Utility_Emphasis": {
        "Fidelity": 0.40,
        "Utility": 0.40,
        "Privacy": 0.20,
    },

    "Utility_Privacy_Emphasis": {
        "Fidelity": 0.20,
        "Utility": 0.40,
        "Privacy": 0.40,
    },
}

# ------------------------------------------------------------------------------
# VALIDATE SCENARIOS
# ------------------------------------------------------------------------------

X_WEIGHT_VALIDATION = []

for scenario, weights in X_WEIGHT_SCENARIOS.items():

    weight_sum = sum(
        weights.values()
    )

    X_WEIGHT_VALIDATION.append(
        {
            "Scenario":
                scenario,
            "Fidelity":
                weights["Fidelity"],
            "Utility":
                weights["Utility"],
            "Privacy":
                weights["Privacy"],
            "Weight_Sum":
                weight_sum,
            "Valid":
                np.isclose(
                    weight_sum,
                    1.0,
                ),
        }
    )

X_WEIGHT_VALIDATION_DF = pd.DataFrame(
    X_WEIGHT_VALIDATION
)

print()

print(
    X_WEIGHT_VALIDATION_DF.to_string(
        index=False
    )
)

if not X_WEIGHT_VALIDATION_DF[
    "Valid"
].all():

    raise RuntimeError(
        "X3 FAILED: One or more sensitivity "
        "weighting scenarios does not sum to 1."
    )

print()
print(
    f"Number of weighting scenarios: "
    f"{len(X_WEIGHT_SCENARIOS)}"
)

print(
    "X3 STATUS: READY"
)
```

```python
# ==============================================================================
# X4 — CALCULATE SCENARIO COMPOSITE SCORES
# ==============================================================================

print()
print("=" * 100)
print("X4 — CALCULATE SCENARIO COMPOSITE SCORES")
print("=" * 100)

X_SCENARIO_RESULTS = []

for scenario, weights in X_WEIGHT_SCENARIOS.items():

    for _, row in X_SOURCE_DF.iterrows():

        if not row["X_Complete"]:

            X_SCENARIO_RESULTS.append(
                {
                    "Scenario":
                        scenario,
                    "Dataset":
                        row["Dataset"],
                    "Method":
                        row["Method"],
                    "Fidelity_Score":
                        row["Fidelity_Score"],
                    "Utility_Score":
                        row["Utility_Score"],
                    "Privacy_Score":
                        row["Privacy_Score"],
                    "Composite_Score":
                        np.nan,
                    "Evidence_Status":
                        "INCOMPLETE",
                }
            )

            continue

        composite_score = (
            weights["Fidelity"]
            *
            row["Fidelity_Score"]
            +
            weights["Utility"]
            *
            row["Utility_Score"]
            +
            weights["Privacy"]
            *
            row["Privacy_Score"]
        )

        X_SCENARIO_RESULTS.append(
            {
                "Scenario":
                    scenario,
                "Dataset":
                    row["Dataset"],
                "Method":
                    row["Method"],
                "Fidelity_Score":
                    row["Fidelity_Score"],
                "Utility_Score":
                    row["Utility_Score"],
                "Privacy_Score":
                    row["Privacy_Score"],
                "Composite_Score":
                    composite_score,
                "Evidence_Status":
                    "COMPLETE",
            }
        )

X_SCENARIO_RESULTS_DF = pd.DataFrame(
    X_SCENARIO_RESULTS
)

# ------------------------------------------------------------------------------
# RANGE VALIDATION
# ------------------------------------------------------------------------------

X_VALID_SCENARIO_SCORES = (
    X_SCENARIO_RESULTS_DF[
        "Composite_Score"
    ]
    .dropna()
)

if not X_VALID_SCENARIO_SCORES.empty:

    if not (
        (
            X_VALID_SCENARIO_SCORES
            >= 0
        )
        &
        (
            X_VALID_SCENARIO_SCORES
            <= 100
        )
    ).all():

        raise RuntimeError(
            "X4 FAILED: Composite scores "
            "outside 0–100 detected."
        )

print()
print(
    f"Scenario × Dataset × Method rows: "
    f"{len(X_SCENARIO_RESULTS_DF)}"
)

print(
    f"Valid scenario scores: "
    f"{X_VALID_SCENARIO_SCORES.shape[0]}"
)

print(
    "X4 STATUS: COMPLETE"
)
```

```python
# ==============================================================================
# X5 — DATASET-LEVEL METHOD RANKING
# ==============================================================================

print()
print("=" * 100)
print("X5 — DATASET-LEVEL METHOD RANKING")
print("=" * 100)

X_RANKING_RESULTS = []

for scenario in X_WEIGHT_SCENARIOS:

    for dataset in X_EXPECTED_DATASETS:

        subset = (
            X_SCENARIO_RESULTS_DF[
                (
                    X_SCENARIO_RESULTS_DF[
                        "Scenario"
                    ]
                    ==
                    scenario
                )
                &
                (
                    X_SCENARIO_RESULTS_DF[
                        "Dataset"
                    ]
                    ==
                    dataset
                )
                &
                (
                    X_SCENARIO_RESULTS_DF[
                        "Composite_Score"
                    ]
                    .notna()
                )
            ]
            .copy()
        )

        if subset.empty:

            for method in X_EXPECTED_METHODS:

                X_RANKING_RESULTS.append(
                    {
                        "Scenario":
                            scenario,
                        "Dataset":
                            dataset,
                        "Method":
                            method,
                        "Composite_Score":
                            np.nan,
                        "Dataset_Rank":
                            pd.NA,
                    }
                )

            continue

        subset[
            "Dataset_Rank"
        ] = (
            subset[
                "Composite_Score"
            ]
            .rank(
                ascending=False,
                method="min",
            )
            .astype("Int64")
        )

        for _, row in subset.iterrows():

            X_RANKING_RESULTS.append(
                {
                    "Scenario":
                        scenario,
                    "Dataset":
                        dataset,
                    "Method":
                        row["Method"],
                    "Composite_Score":
                        row[
                            "Composite_Score"
                        ],
                    "Dataset_Rank":
                        row[
                            "Dataset_Rank"
                        ],
                }
            )

        evaluated_methods = set(
            subset[
                "Method"
            ]
        )

        for method in X_EXPECTED_METHODS:

            if method not in evaluated_methods:

                X_RANKING_RESULTS.append(
                    {
                        "Scenario":
                            scenario,
                        "Dataset":
                            dataset,
                        "Method":
                            method,
                        "Composite_Score":
                            np.nan,
                        "Dataset_Rank":
                            pd.NA,
                    }
                )

X_RANKING_DF = pd.DataFrame(
    X_RANKING_RESULTS
)

X_RANKING_DF[
    "Dataset_Rank"
] = pd.to_numeric(
    X_RANKING_DF[
        "Dataset_Rank"
    ],
    errors="coerce",
).astype("Int64")

print()
print(
    X_RANKING_DF.to_string(
        index=False
    )
)

print()
print(
    "X5 STATUS: COMPLETE"
)
```

```python
# ==============================================================================
# X6 — OVERALL METHOD RANKING
# ==============================================================================

print()
print("=" * 100)
print("X6 — OVERALL METHOD RANKING")
print("=" * 100)

X_OVERALL_RANKING_LIST = []

for scenario in X_WEIGHT_SCENARIOS:

    scenario_df = (
        X_SCENARIO_RESULTS_DF[
            (
                X_SCENARIO_RESULTS_DF[
                    "Scenario"
                ]
                ==
                scenario
            )
            &
            (
                X_SCENARIO_RESULTS_DF[
                    "Composite_Score"
                ]
                .notna()
            )
        ]
        .copy()
    )

    for method in X_EXPECTED_METHODS:

        method_df = (
            scenario_df[
                scenario_df[
                    "Method"
                ]
                ==
                method
            ]
        )

        scores = (
            method_df[
                "Composite_Score"
            ]
            .dropna()
        )

        if scores.empty:

            X_OVERALL_RANKING_LIST.append(
                {
                    "Scenario":
                        scenario,
                    "Method":
                        method,
                    "Datasets_Evaluated":
                        0,
                    "Mean_Composite_Score":
                        np.nan,
                    "Median_Composite_Score":
                        np.nan,
                    "Score_SD":
                        np.nan,
                    "Overall_Rank":
                        pd.NA,
                }
            )

            continue

        X_OVERALL_RANKING_LIST.append(
            {
                "Scenario":
                    scenario,
                "Method":
                    method,
                "Datasets_Evaluated":
                    len(scores),
                "Mean_Composite_Score":
                    scores.mean(),
                "Median_Composite_Score":
                    scores.median(),
                "Score_SD":
                    scores.std(
                        ddof=1
                    )
                    if len(scores) > 1
                    else 0.0,
                "Overall_Rank":
                    pd.NA,
            }
        )

X_OVERALL_RANKING_DF = pd.DataFrame(
    X_OVERALL_RANKING_LIST
)

# ------------------------------------------------------------------------------
# RANK WITHIN EACH SCENARIO
# ------------------------------------------------------------------------------

for scenario in X_WEIGHT_SCENARIOS:

    mask = (
        X_OVERALL_RANKING_DF[
            "Scenario"
        ]
        ==
        scenario
    )

    valid_mask = (
        mask
        &
        X_OVERALL_RANKING_DF[
            "Mean_Composite_Score"
        ]
        .notna()
    )

    X_OVERALL_RANKING_DF.loc[
        valid_mask,
        "Overall_Rank"
    ] = (
        X_OVERALL_RANKING_DF.loc[
            valid_mask,
            "Mean_Composite_Score"
        ]
        .rank(
            ascending=False,
            method="min",
        )
    )

X_OVERALL_RANKING_DF[
    "Overall_Rank"
] = pd.to_numeric(
    X_OVERALL_RANKING_DF[
        "Overall_Rank"
    ],
    errors="coerce",
).astype("Int64")

print()
print(
    X_OVERALL_RANKING_DF.to_string(
        index=False
    )
)

print()
print(
    "X6 STATUS: COMPLETE"
)
```

```python
# ==============================================================================
# X7 — RANK STABILITY ACROSS WEIGHTING SCENARIOS
# ==============================================================================

print()
print("=" * 100)
print("X7 — RANK STABILITY ACROSS WEIGHTING SCENARIOS")
print("=" * 100)

X_RANK_STABILITY = []

for method in X_EXPECTED_METHODS:

    method_ranks = (
        X_OVERALL_RANKING_DF[
            X_OVERALL_RANKING_DF[
                "Method"
            ]
            ==
            method
        ]
        [
            "Overall_Rank"
        ]
        .dropna()
        .astype(float)
    )

    if method_ranks.empty:

        X_RANK_STABILITY.append(
            {
                "Method":
                    method,
                "Scenarios_Evaluated":
                    0,
                "Mean_Rank":
                    np.nan,
                "Median_Rank":
                    np.nan,
                "Best_Rank":
                    np.nan,
                "Worst_Rank":
                    np.nan,
                "Rank_SD":
                    np.nan,
                "Rank_Range":
                    np.nan,
            }
        )

        continue

    X_RANK_STABILITY.append(
        {
            "Method":
                method,

            "Scenarios_Evaluated":
                len(method_ranks),

            "Mean_Rank":
                method_ranks.mean(),

            "Median_Rank":
                method_ranks.median(),

            "Best_Rank":
                method_ranks.min(),

            "Worst_Rank":
                method_ranks.max(),

            "Rank_SD":
                method_ranks.std(
                    ddof=1
                )
                if len(method_ranks) > 1
                else 0.0,

            "Rank_Range":
                (
                    method_ranks.max()
                    -
                    method_ranks.min()
                ),
        }
    )

X_RANK_STABILITY_DF = pd.DataFrame(
    X_RANK_STABILITY
)

print()
print(
    X_RANK_STABILITY_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# X8 — WIN FREQUENCY AND RANK-1 STABILITY
# ==============================================================================

print()
print("=" * 100)
print("X8 — WIN FREQUENCY AND RANK-1 STABILITY")
print("=" * 100)

X_WIN_RESULTS = []

X_TOTAL_SCENARIOS = len(
    X_WEIGHT_SCENARIOS
)

for method in X_EXPECTED_METHODS:

    method_rank_df = (
        X_OVERALL_RANKING_DF[
            X_OVERALL_RANKING_DF[
                "Method"
            ]
            ==
            method
        ]
    )

    valid_ranks = (
        method_rank_df[
            "Overall_Rank"
        ]
        .dropna()
    )

    wins = int(
        (
            valid_ranks
            ==
            1
        )
        .sum()
    )

    evaluated = len(
        valid_ranks
    )

    X_WIN_RESULTS.append(
        {
            "Method":
                method,

            "Scenarios_Evaluated":
                evaluated,

            "Rank_1_Count":
                wins,

            "Rank_1_Frequency":
                (
                    wins / evaluated
                    if evaluated > 0
                    else np.nan
                ),

            "Rank_1_Stability_Percent":
                (
                    100 * wins / evaluated
                    if evaluated > 0
                    else np.nan
                ),
        }
    )

X_WIN_STABILITY_DF = pd.DataFrame(
    X_WIN_RESULTS
)

print()
print(
    X_WIN_STABILITY_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# X9 — BASELINE METHOD RANKING
# ==============================================================================

print()
print("=" * 100)
print("X9 — BASELINE METHOD RANKING")
print("=" * 100)

X_BASELINE_RANKING_DF = (
    X_OVERALL_RANKING_DF[
        X_OVERALL_RANKING_DF[
            "Scenario"
        ]
        ==
        "Equal_Weights"
    ]
    .copy()
)

X_BASELINE_RANKING_DF = (
    X_BASELINE_RANKING_DF
    .sort_values(
        [
            "Overall_Rank",
            "Method",
        ],
        na_position="last",
    )
    .reset_index(
        drop=True
    )
)

print()
print(
    X_BASELINE_RANKING_DF.to_string(
        index=False
    )
)

if not X_BASELINE_RANKING_DF.empty:

    X_BASELINE_WINNER = (
        X_BASELINE_RANKING_DF
        .loc[
            X_BASELINE_RANKING_DF[
                "Overall_Rank"
            ]
            ==
            1,
            "Method",
        ]
        .tolist()
    )

else:

    X_BASELINE_WINNER = []

print()
print(
    "Baseline rank-1 method(s): "
    +
    (
        ", ".join(
            X_BASELINE_WINNER
        )
        if X_BASELINE_WINNER
        else "None"
    )
)
```

```python
# ==============================================================================
# X10 — WEIGHTING SCENARIO RANK MATRIX
# ==============================================================================

print()
print("=" * 100)
print("X10 — WEIGHTING SCENARIO RANK MATRIX")
print("=" * 100)

X_RANK_MATRIX = (
    X_OVERALL_RANKING_DF
    .pivot(
        index="Scenario",
        columns="Method",
        values="Overall_Rank",
    )
    .reindex(
        index=list(
            X_WEIGHT_SCENARIOS.keys()
        ),
        columns=X_EXPECTED_METHODS,
    )
)

print()
print(
    X_RANK_MATRIX.to_string()
)

# ------------------------------------------------------------------------------
# SCORE MATRIX
# ------------------------------------------------------------------------------

X_SCORE_MATRIX = (
    X_OVERALL_RANKING_DF
    .pivot(
        index="Scenario",
        columns="Method",
        values="Mean_Composite_Score",
    )
    .reindex(
        index=list(
            X_WEIGHT_SCENARIOS.keys()
        ),
        columns=X_EXPECTED_METHODS,
    )
)

print()
print("=" * 100)
print("WEIGHTING SCENARIO SCORE MATRIX")
print("=" * 100)

print()
print(
    X_SCORE_MATRIX.to_string()
)
```

```python
# ==============================================================================
# X11 — CONCLUSION STABILITY TEST
# ==============================================================================

print()
print("=" * 100)
print("X11 — CONCLUSION STABILITY TEST")
print("=" * 100)

# ------------------------------------------------------------------------------
# SCENARIO WINNERS
# ------------------------------------------------------------------------------

X_SCENARIO_WINNERS = []

for scenario in X_WEIGHT_SCENARIOS:

    scenario_df = (
        X_OVERALL_RANKING_DF[
            (
                X_OVERALL_RANKING_DF[
                    "Scenario"
                ]
                ==
                scenario
            )
            &
            (
                X_OVERALL_RANKING_DF[
                    "Overall_Rank"
                ]
                ==
                1
            )
        ]
    )

    winners = (
        scenario_df[
            "Method"
        ]
        .tolist()
    )

    X_SCENARIO_WINNERS.append(
        {
            "Scenario":
                scenario,
            "Winner":
                ", ".join(
                    winners
                )
                if winners
                else "None",
            "Winner_Count":
                len(winners),
        }
    )

X_SCENARIO_WINNERS_DF = pd.DataFrame(
    X_SCENARIO_WINNERS
)

print()
print(
    X_SCENARIO_WINNERS_DF.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# BASELINE WINNER
# ------------------------------------------------------------------------------

X_BASELINE_WINNER_SET = set(
    X_BASELINE_WINNER
)

# ------------------------------------------------------------------------------
# STABILITY
# ------------------------------------------------------------------------------

X_STABLE_SCENARIOS = 0
X_EVALUATED_SCENARIOS = 0

for _, row in X_SCENARIO_WINNERS_DF.iterrows():

    if row["Winner"] == "None":

        continue

    X_EVALUATED_SCENARIOS += 1

    scenario_winners = set(
        [
            method.strip()
            for method in row[
                "Winner"
            ].split(",")
        ]
    )

    if (
        scenario_winners
        ==
        X_BASELINE_WINNER_SET
    ):

        X_STABLE_SCENARIOS += 1

# ------------------------------------------------------------------------------
# STABILITY PERCENTAGE
# ------------------------------------------------------------------------------

if X_EVALUATED_SCENARIOS > 0:

    X_CONCLUSION_STABILITY = (
        100
        *
        X_STABLE_SCENARIOS
        /
        X_EVALUATED_SCENARIOS
    )

else:

    X_CONCLUSION_STABILITY = np.nan

# ------------------------------------------------------------------------------
# CONCLUSION
# ------------------------------------------------------------------------------

if np.isnan(
    X_CONCLUSION_STABILITY
):

    X_STABILITY_CONCLUSION = (
        "INDETERMINATE"
    )

elif np.isclose(
    X_CONCLUSION_STABILITY,
    100.0,
):

    X_STABILITY_CONCLUSION = (
        "HIGHLY_STABLE"
    )

elif X_CONCLUSION_STABILITY >= 75:

    X_STABILITY_CONCLUSION = (
        "STABLE"
    )

elif X_CONCLUSION_STABILITY >= 50:

    X_STABILITY_CONCLUSION = (
        "MODERATELY_STABLE"
    )

else:

    X_STABILITY_CONCLUSION = (
        "SENSITIVE_TO_WEIGHTING"
    )

print()
print(
    f"Evaluated weighting scenarios : "
    f"{X_EVALUATED_SCENARIOS}"
)

print(
    f"Scenarios preserving baseline winner : "
    f"{X_STABLE_SCENARIOS}"
)

print(
    f"Conclusion stability : "
    f"{X_CONCLUSION_STABILITY:.2f}%"
    if not np.isnan(
        X_CONCLUSION_STABILITY
    )
    else
    "Conclusion stability : NaN"
)

print()
print(
    f"X11 CONCLUSION: "
    f"{X_STABILITY_CONCLUSION}"
)
```

```python
# ==============================================================================
# X12 — METHOD RANKING SENSITIVITY SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("X12 — METHOD RANKING SENSITIVITY SUMMARY")
print("=" * 100)

X12_SUMMARY_ROWS = []

for method in X_EXPECTED_METHODS:

    stability_row = (
        X_RANK_STABILITY_DF[
            X_RANK_STABILITY_DF[
                "Method"
            ]
            ==
            method
        ]
    )

    win_row = (
        X_WIN_STABILITY_DF[
            X_WIN_STABILITY_DF[
                "Method"
            ]
            ==
            method
        ]
    )

    baseline_row = (
        X_BASELINE_RANKING_DF[
            X_BASELINE_RANKING_DF[
                "Method"
            ]
            ==
            method
        ]
    )

    X12_SUMMARY_ROWS.append(
        {
            "Method":
                method,

            "Baseline_Rank":
                (
                    baseline_row[
                        "Overall_Rank"
                    ]
                    .iloc[0]
                    if not baseline_row.empty
                    else pd.NA
                ),

            "Mean_Rank":
                (
                    stability_row[
                        "Mean_Rank"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Median_Rank":
                (
                    stability_row[
                        "Median_Rank"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Best_Rank":
                (
                    stability_row[
                        "Best_Rank"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Worst_Rank":
                (
                    stability_row[
                        "Worst_Rank"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Rank_SD":
                (
                    stability_row[
                        "Rank_SD"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Rank_Range":
                (
                    stability_row[
                        "Rank_Range"
                    ]
                    .iloc[0]
                    if not stability_row.empty
                    else np.nan
                ),

            "Rank_1_Frequency":
                (
                    win_row[
                        "Rank_1_Frequency"
                    ]
                    .iloc[0]
                    if not win_row.empty
                    else np.nan
                ),

            "Rank_1_Stability_Percent":
                (
                    win_row[
                        "Rank_1_Stability_Percent"
                    ]
                    .iloc[0]
                    if not win_row.empty
                    else np.nan
                ),
        }
    )

X12_SUMMARY_DF = pd.DataFrame(
    X12_SUMMARY_ROWS
)

print()
print(
    X12_SUMMARY_DF.to_string(
        index=False
    )
)
```

```python
# ==============================================================================
# X13 — FINAL METHOD RANKING INTERPRETATION
# ==============================================================================

print()
print("=" * 100)
print("X13 — FINAL METHOD RANKING INTERPRETATION")
print("=" * 100)

print()

# ------------------------------------------------------------------------------
# BASELINE RESULT
# ------------------------------------------------------------------------------

if X_BASELINE_WINNER:

    print(
        "Baseline equal-weight winner(s): "
        +
        ", ".join(
            X_BASELINE_WINNER
        )
    )

else:

    print(
        "Baseline equal-weight winner: "
        "No method with complete evidence."
    )

# ------------------------------------------------------------------------------
# MOST STABLE METHOD
# ------------------------------------------------------------------------------

X_STABLE_METHOD_DF = (
    X12_SUMMARY_DF[
        X12_SUMMARY_DF[
            "Mean_Rank"
        ]
        .notna()
    ]
    .sort_values(
        [
            "Mean_Rank",
            "Rank_SD",
            "Rank_Range",
        ],
        ascending=[
            True,
            True,
            True,
        ],
    )
)

if not X_STABLE_METHOD_DF.empty:

    X_MOST_STABLE_METHOD = (
        X_STABLE_METHOD_DF[
            "Method"
        ]
        .iloc[0]
    )

else:

    X_MOST_STABLE_METHOD = None

if X_MOST_STABLE_METHOD:

    print(
        "Most stable method by mean rank: "
        f"{X_MOST_STABLE_METHOD}"
    )

# ------------------------------------------------------------------------------
# INTERPRETATION
# ------------------------------------------------------------------------------

print()

if (
    X_STABILITY_CONCLUSION
    ==
    "HIGHLY_STABLE"
):

    print(
        "Interpretation: The method-level conclusion "
        "is highly stable across the tested weighting "
        "scenarios."
    )

elif (
    X_STABILITY_CONCLUSION
    ==
    "STABLE"
):

    print(
        "Interpretation: The method-level conclusion "
        "is generally stable across reasonable weighting "
        "choices."
    )

elif (
    X_STABILITY_CONCLUSION
    ==
    "MODERATELY_STABLE"
):

    print(
        "Interpretation: The method-level conclusion "
        "shows moderate sensitivity to weighting choices."
    )

elif (
    X_STABILITY_CONCLUSION
    ==
    "SENSITIVE_TO_WEIGHTING"
):

    print(
        "Interpretation: Method ranking is sensitive "
        "to the relative importance assigned to Fidelity, "
        "Utility, and Privacy."
    )

else:

    print(
        "Interpretation: Ranking stability cannot be "
        "determined from the available complete evidence."
    )

# ------------------------------------------------------------------------------
# RESEARCH CAUTION
# ------------------------------------------------------------------------------

print()
print(
    "Research caution: Missing Dataset × Method component "
    "evidence was not converted to zero and was excluded "
    "from the corresponding composite ranking calculation."
)

print()
print(
    "X13 STATUS: COMPLETE"
)
```

```python
# ==============================================================================
# X14 — SAVE METHOD RANKING & SENSITIVITY REPORTS
# ==============================================================================

print()
print("=" * 100)
print("X14 — SAVE METHOD RANKING & SENSITIVITY REPORTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------------------------

if "SPP_GAN_RESULTS_DIRECTORY" in globals():

    X_OUTPUT_DIRECTORY = pathlib.Path(
        SPP_GAN_RESULTS_DIRECTORY
    )

elif "RESULTS_DIRECTORY" in globals():

    X_OUTPUT_DIRECTORY = pathlib.Path(
        RESULTS_DIRECTORY
    )

else:

    X_OUTPUT_DIRECTORY = pathlib.Path(
        "results"
    )

X_OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)

# ------------------------------------------------------------------------------
# SAVE FILES
# ------------------------------------------------------------------------------

X14_FILES = []

# Scenario scores
X_SCENARIO_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_ranking_sensitivity_scores.csv"
)

X_SCENARIO_RESULTS_DF.to_csv(
    X_SCENARIO_FILE,
    index=False,
)

X14_FILES.append(
    X_SCENARIO_FILE
)

# Dataset-level ranks
X_RANKING_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_dataset_ranking_by_weight_scenario.csv"
)

X_RANKING_DF.to_csv(
    X_RANKING_FILE,
    index=False,
)

X14_FILES.append(
    X_RANKING_FILE
)

# Overall rankings
X_OVERALL_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_overall_ranking_by_weight_scenario.csv"
)

X_OVERALL_RANKING_DF.to_csv(
    X_OVERALL_FILE,
    index=False,
)

X14_FILES.append(
    X_OVERALL_FILE
)

# Rank stability
X_STABILITY_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_rank_stability_analysis.csv"
)

X_RANK_STABILITY_DF.to_csv(
    X_STABILITY_FILE,
    index=False,
)

X14_FILES.append(
    X_STABILITY_FILE
)

# Win stability
X_WIN_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_rank1_stability_analysis.csv"
)

X_WIN_STABILITY_DF.to_csv(
    X_WIN_FILE,
    index=False,
)

X14_FILES.append(
    X_WIN_FILE
)

# Summary
X_SUMMARY_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_ranking_sensitivity_summary.csv"
)

X12_SUMMARY_DF.to_csv(
    X_SUMMARY_FILE,
    index=False,
)

X14_FILES.append(
    X_SUMMARY_FILE
)

# Rank matrix
X_RANK_MATRIX_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_weight_scenario_rank_matrix.csv"
)

X_RANK_MATRIX.to_csv(
    X_RANK_MATRIX_FILE
)

X14_FILES.append(
    X_RANK_MATRIX_FILE
)

# Score matrix
X_SCORE_MATRIX_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_weight_scenario_score_matrix.csv"
)

X_SCORE_MATRIX.to_csv(
    X_SCORE_MATRIX_FILE
)

X14_FILES.append(
    X_SCORE_MATRIX_FILE
)

# Weight definitions
X_WEIGHTS_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_ranking_weight_scenarios.csv"
)

X_WEIGHT_VALIDATION_DF.to_csv(
    X_WEIGHTS_FILE,
    index=False,
)

X14_FILES.append(
    X_WEIGHTS_FILE
)

# ------------------------------------------------------------------------------
# FINAL EXCEL REPORT
# ------------------------------------------------------------------------------

X_EXCEL_FILE = (
    X_OUTPUT_DIRECTORY
    /
    "method_ranking_sensitivity_analysis.xlsx"
)

with pd.ExcelWriter(
    X_EXCEL_FILE,
    engine="openpyxl",
) as writer:

    X_WEIGHT_VALIDATION_DF.to_excel(
        writer,
        sheet_name="Weight Scenarios",
        index=False,
    )

    X_SCENARIO_RESULTS_DF.to_excel(
        writer,
        sheet_name="Scenario Scores",
        index=False,
    )

    X_RANKING_DF.to_excel(
        writer,
        sheet_name="Dataset Rankings",
        index=False,
    )

    X_OVERALL_RANKING_DF.to_excel(
        writer,
        sheet_name="Overall Rankings",
        index=False,
    )

    X_RANK_STABILITY_DF.to_excel(
        writer,
        sheet_name="Rank Stability",
        index=False,
    )

    X_WIN_STABILITY_DF.to_excel(
        writer,
        sheet_name="Rank1 Stability",
        index=False,
    )

    X12_SUMMARY_DF.to_excel(
        writer,
        sheet_name="Final Summary",
        index=False,
    )

    X_RANK_MATRIX.to_excel(
        writer,
        sheet_name="Rank Matrix",
    )

    X_SCORE_MATRIX.to_excel(
        writer,
        sheet_name="Score Matrix",
    )

# ------------------------------------------------------------------------------
# FILE STATUS
# ------------------------------------------------------------------------------

X14_FILES.append(
    X_EXCEL_FILE
)

print()

for file_path in X14_FILES:

    print(
        f"{'OK' if file_path.exists() else 'MISSING':8s} "
        f"{file_path.name}"
    )

X14_ALL_SAVED = all(
    path.exists()
    for path in X14_FILES
)

print()
print(
    f"Files saved: "
    f"{sum(path.exists() for path in X14_FILES)}/"
    f"{len(X14_FILES)}"
)

print()
print(
    f"X14 STATUS: "
    f"{'COMPLETE' if X14_ALL_SAVED else 'INCOMPLETE'}"
)
```

```python
# ==============================================================================
# Y2 — RESOLVE FINAL QUANTITATIVE EVIDENCE
# ==============================================================================

print()
print("=" * 100)
print("Y2 — RESOLVE FINAL QUANTITATIVE EVIDENCE")
print("=" * 100)

# ------------------------------------------------------------------------------
# RESOLVE W COMPOSITE EVIDENCE
# ------------------------------------------------------------------------------

Y_COMPOSITE_DF = None

for variable_name in [
    "W_FINAL_COMPOSITE_DF",
    "W23_FINAL_COMPOSITE_DF",
    "W20_CROSS_DATASET_SUMMARY",
    "W_CROSS_DATASET_SUMMARY",
]:

    if variable_name in globals():

        candidate = globals()[
            variable_name
        ]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            Y_COMPOSITE_DF = (
                candidate.copy()
            )

            print(
                f"Composite evidence source: "
                f"{variable_name}"
            )

            break

# ------------------------------------------------------------------------------
# RESOLVE X OVERALL RANKING
# ------------------------------------------------------------------------------

Y_RANKING_DF = None

for variable_name in [
    "X_OVERALL_RANKING_DF",
    "X12_SUMMARY_DF",
]:

    if variable_name in globals():

        candidate = globals()[
            variable_name
        ]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            Y_RANKING_DF = (
                candidate.copy()
            )

            print(
                f"Ranking evidence source: "
                f"{variable_name}"
            )

            break

# ------------------------------------------------------------------------------
# RESOLVE X STABILITY
# ------------------------------------------------------------------------------

Y_STABILITY_DF = None

for variable_name in [
    "X_RANK_STABILITY_DF",
    "X12_SUMMARY_DF",
]:

    if variable_name in globals():

        candidate = globals()[
            variable_name
        ]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            Y_STABILITY_DF = (
                candidate.copy()
            )

            print(
                f"Stability evidence source: "
                f"{variable_name}"
            )

            break

# ------------------------------------------------------------------------------
# RESOLVE STATISTICAL EVIDENCE FROM U
# ------------------------------------------------------------------------------

Y_STATISTICAL_DF = None

for variable_name in [
    "U_STATISTICAL_RESULTS_DF",
    "U_FINAL_STATISTICAL_RESULTS_DF",
    "U_PAIRWISE_RESULTS_DF",
    "U_RESULTS_DF",
    "U_TEST_RESULTS_DF",
]:

    if variable_name in globals():

        candidate = globals()[
            variable_name
        ]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            Y_STATISTICAL_DF = (
                candidate.copy()
            )

            print(
                f"Statistical evidence source: "
                f"{variable_name}"
            )

            break

# ------------------------------------------------------------------------------
# RESOLVE T ROBUSTNESS EVIDENCE
# ------------------------------------------------------------------------------

Y_ROBUSTNESS_DF = None

for variable_name in [
    "T_CROSS_DATASET_SUMMARY",
    "T_ROBUSTNESS_SUMMARY_DF",
    "T_FINAL_ROBUSTNESS_DF",
    "T_RESULTS_DF",
]:

    if variable_name in globals():

        candidate = globals()[
            variable_name
        ]

        if isinstance(
            candidate,
            pd.DataFrame,
        ):

            Y_ROBUSTNESS_DF = (
                candidate.copy()
            )

            print(
                f"Robustness evidence source: "
                f"{variable_name}"
            )

            break

print()

print(
    "Evidence availability:"
)

print(
    f"  Composite evidence   : "
    f"{Y_COMPOSITE_DF is not None}"
)

print(
    f"  Ranking evidence     : "
    f"{Y_RANKING_DF is not None}"
)

print(
    f"  Stability evidence   : "
    f"{Y_STABILITY_DF is not None}"
)

print(
    f"  Statistical evidence : "
    f"{Y_STATISTICAL_DF is not None}"
)

print(
    f"  Robustness evidence  : "
    f"{Y_ROBUSTNESS_DF is not None}"
)

print()
print("Y2 STATUS: EVIDENCE RESOLVED")
```

```python
# ==============================================================================
# Y4 — RESOLVE SPP-GAN FINAL PERFORMANCE EVIDENCE
# ==============================================================================

print()
print("=" * 100)
print("Y4 — RESOLVE SPP-GAN FINAL PERFORMANCE EVIDENCE")
print("=" * 100)

Y_SPP_GAN_EVIDENCE = {
    "Datasets": 0,
    "Mean_Composite": np.nan,
    "Median_Composite": np.nan,
    "Mean_Rank": np.nan,
    "Rank_1_Frequency": np.nan,
}

# ------------------------------------------------------------------------------
# COMPOSITE EVIDENCE
# ------------------------------------------------------------------------------

if Y_COMPOSITE_DF is not None:

    Y_COMPOSITE_TEMP = (
        Y_COMPOSITE_DF.copy()
    )

    if "Method" in Y_COMPOSITE_TEMP.columns:

        Y_COMPOSITE_TEMP[
            "Method"
        ] = (
            Y_COMPOSITE_TEMP[
                "Method"
            ]
            .astype(str)
            .str.strip()
        )

        if "Overall Performance Score" in (
            Y_COMPOSITE_TEMP.columns
        ):

            Y_SPP_ROWS = (
                Y_COMPOSITE_TEMP[
                    Y_COMPOSITE_TEMP[
                        "Method"
                    ]
                    ==
                    "SPP-GAN"
                ]
            )

            Y_SPP_SCORES = pd.to_numeric(
                Y_SPP_ROWS[
                    "Overall Performance Score"
                ],
                errors="coerce",
            ).dropna()

        elif "Overall_Score" in (
            Y_COMPOSITE_TEMP.columns
        ):

            Y_SPP_ROWS = (
                Y_COMPOSITE_TEMP[
                    Y_COMPOSITE_TEMP[
                        "Method"
                    ]
                    ==
                    "SPP-GAN"
                ]
            )

            Y_SPP_SCORES = pd.to_numeric(
                Y_SPP_ROWS[
                    "Overall_Score"
                ],
                errors="coerce",
            ).dropna()

        else:

            Y_SPP_SCORES = pd.Series(
                dtype=float
            )

        if not Y_SPP_SCORES.empty:

            Y_SPP_GAN_EVIDENCE[
                "Datasets"
            ] = len(
                Y_SPP_SCORES
            )

            Y_SPP_GAN_EVIDENCE[
                "Mean_Composite"
            ] = (
                Y_SPP_SCORES.mean()
            )

            Y_SPP_GAN_EVIDENCE[
                "Median_Composite"
            ] = (
                Y_SPP_SCORES.median()
            )

# ------------------------------------------------------------------------------
# RANK EVIDENCE
# ------------------------------------------------------------------------------

if Y_RANKING_DF is not None:

    Y_RANK_TEMP = (
        Y_RANKING_DF.copy()
    )

    if "Method" in Y_RANK_TEMP.columns:

        Y_RANK_TEMP[
            "Method"
        ] = (
            Y_RANK_TEMP[
                "Method"
            ]
            .astype(str)
            .str.strip()
        )

        if "Overall_Rank" in Y_RANK_TEMP.columns:

            Y_SPP_RANKS = pd.to_numeric(
                Y_RANK_TEMP.loc[
                    Y_RANK_TEMP[
                        "Method"
                    ]
                    ==
                    "SPP-GAN",
                    "Overall_Rank",
                ],
                errors="coerce",
            ).dropna()

            if not Y_SPP_RANKS.empty:

                Y_SPP_GAN_EVIDENCE[
                    "Mean_Rank"
                ] = (
                    Y_SPP_RANKS.mean()
                )

                Y_SPP_GAN_EVIDENCE[
                    "Rank_1_Frequency"
                ] = (
                    (
                        Y_SPP_RANKS
                        ==
                        1
                    )
                    .mean()
                )

print()
print(
    "SPP-GAN evidence summary:"
)

for key, value in Y_SPP_GAN_EVIDENCE.items():

    print(
        f"  {key:25s}: {value}"
    )
```

```python
# ==============================================================================
# Y6 — EVALUATE H1: OVERALL SPP-GAN PERFORMANCE
# ==============================================================================

print()
print("=" * 100)
print("Y6 — EVALUATE H1: OVERALL SPP-GAN PERFORMANCE")
print("=" * 100)

Y_H1_STATUS = "NOT_TESTABLE"
Y_H1_EVIDENCE = ""
Y_H1_CONCLUSION = ""

if (
    "X12_SUMMARY_DF" in globals()
    and
    isinstance(
        X12_SUMMARY_DF,
        pd.DataFrame,
    )
):

    Y_H1_ROW = (
        X12_SUMMARY_DF[
            X12_SUMMARY_DF[
                "Method"
            ]
            ==
            "SPP-GAN"
        ]
    )

    if not Y_H1_ROW.empty:

        Y_H1_BASELINE_RANK = (
            Y_H1_ROW[
                "Baseline_Rank"
            ]
            .iloc[0]
        )

        Y_H1_BEST_RANK = (
            Y_H1_ROW[
                "Best_Rank"
            ]
            .iloc[0]
        )

        if (
            pd.notna(
                Y_H1_BASELINE_RANK
            )
            and
            Y_H1_BASELINE_RANK == 1
        ):

            Y_H1_STATUS = "SUPPORTED"

            Y_H1_EVIDENCE = (
                "SPP-GAN achieved rank 1 under the equal-weight "
                "Fidelity–Utility–Privacy composite ranking."
            )

            Y_H1_CONCLUSION = (
                "H1 is supported by the baseline composite ranking."
            )

        elif (
            pd.notna(
                Y_H1_BEST_RANK
            )
            and
            Y_H1_BEST_RANK <= 2
        ):

            Y_H1_STATUS = "PARTIALLY_SUPPORTED"

            Y_H1_EVIDENCE = (
                "SPP-GAN achieved a competitive ranking but did not "
                "consistently occupy the baseline first position."
            )

            Y_H1_CONCLUSION = (
                "H1 receives partial support."
            )

print()
print(
    f"Status      : {Y_H1_STATUS}"
)

print(
    f"Evidence    : {Y_H1_EVIDENCE}"
)

print(
    f"Conclusion  : {Y_H1_CONCLUSION}"
)
```

```python
# ==============================================================================
# Y7 — EVALUATE H2: FIDELITY–UTILITY–PRIVACY BALANCE
# ==============================================================================

print()
print("=" * 100)
print("Y7 — EVALUATE H2: FIDELITY–UTILITY–PRIVACY BALANCE")
print("=" * 100)

Y_H2_STATUS = "NOT_TESTABLE"
Y_H2_EVIDENCE = ""
Y_H2_CONCLUSION = ""

if Y_COMPOSITE_DF is not None:

    Y_H2_STATUS = "PARTIALLY_SUPPORTED"

    Y_H2_EVIDENCE = (
        "The composite framework explicitly integrates Fidelity, Utility, "
        "and Privacy into an overall performance score."
    )

    Y_H2_CONCLUSION = (
        "H2 can be evaluated through the integrated composite framework; "
        "the final degree of support depends on the observed SPP-GAN component "
        "scores relative to the competing methods."
    )

print()
print(
    f"Status      : {Y_H2_STATUS}"
)

print(
    f"Evidence    : {Y_H2_EVIDENCE}"
)

print(
    f"Conclusion  : {Y_H2_CONCLUSION}"
)
```

```python
# ==============================================================================
# Y9 — EVALUATE H4: WEIGHTING SENSITIVITY
# ==============================================================================

print()
print("=" * 100)
print("Y9 — EVALUATE H4: WEIGHTING SENSITIVITY")
print("=" * 100)

Y_H4_STATUS = "NOT_TESTABLE"
Y_H4_EVIDENCE = ""
Y_H4_CONCLUSION = ""

if (
    "X_CONCLUSION_STABILITY" in globals()
    and
    pd.notna(
        X_CONCLUSION_STABILITY
    )
):

    Y_H4_STABILITY = (
        X_CONCLUSION_STABILITY
    )

    if Y_H4_STABILITY >= 75:

        Y_H4_STATUS = "SUPPORTED"

        Y_H4_CONCLUSION = (
            "The comparative conclusion remained stable in at least "
            "75% of the tested reasonable weighting scenarios."
        )

    elif Y_H4_STABILITY >= 50:

        Y_H4_STATUS = "PARTIALLY_SUPPORTED"

        Y_H4_CONCLUSION = (
            "The comparative conclusion showed moderate stability "
            "but changed under a substantial subset of weighting scenarios."
        )

    else:

        Y_H4_STATUS = "NOT_SUPPORTED"

        Y_H4_CONCLUSION = (
            "The comparative conclusion was sensitive to the selected "
            "Fidelity–Utility–Privacy weighting."
        )

    Y_H4_EVIDENCE = (
        f"Conclusion stability across tested weighting scenarios: "
        f"{Y_H4_STABILITY:.2f}%."
    )

print()
print(
    f"Status      : {Y_H4_STATUS}"
)

print(
    f"Evidence    : {Y_H4_EVIDENCE}"
)

print(
    f"Conclusion  : {Y_H4_CONCLUSION}"
)
```

```python
# ==============================================================================
# Y10 — EVALUATE H5: STATISTICAL SUPPORT
# ==============================================================================

print()
print("=" * 100)
print("Y10 — EVALUATE H5: STATISTICAL SUPPORT")
print("=" * 100)

Y_H5_STATUS = "NOT_TESTABLE"
Y_H5_EVIDENCE = ""
Y_H5_CONCLUSION = ""

# ------------------------------------------------------------------------------
# CHECK U3 VALIDATION
# ------------------------------------------------------------------------------

Y_U3_VALID = None

if (
    "U_VALIDATION_DF" in globals()
    and
    isinstance(
        U_VALIDATION_DF,
        pd.DataFrame,
    )
):

    if (
        "Check" in U_VALIDATION_DF.columns
        and
        "Status" in U_VALIDATION_DF.columns
    ):

        Y_U3_ROW = (
            U_VALIDATION_DF[
                U_VALIDATION_DF[
                    "Check"
                ]
                ==
                "No missing paired observations"
            ]
        )

        if not Y_U3_ROW.empty:

            Y_U3_VALID = bool(
                Y_U3_ROW[
                    "Status"
                ]
                .iloc[0]
            )

# ------------------------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------------------------

if Y_U3_VALID is False:

    Y_H5_STATUS = (
        "NOT_SUPPORTED"
    )

    Y_H5_EVIDENCE = (
        "The paired experimental validation identified missing paired "
        "observations, so the planned fully paired analysis was not valid."
    )

    Y_H5_CONCLUSION = (
        "H5 is not supported as a general claim for the full experimental "
        "structure. Statistical evidence must be interpreted only for valid "
        "comparisons supported by the available observations."
    )

elif (
    Y_STATISTICAL_DF is not None
    and
    Y_U3_VALID is True
):

    Y_H5_STATUS = (
        "SUPPORTED"
    )

    Y_H5_EVIDENCE = (
        "The experimental structure was validated for paired analysis and "
        "statistical results were available."
    )

    Y_H5_CONCLUSION = (
        "H5 is supported where the reported statistical tests meet the "
        "predefined significance criterion."
    )

print()
print(
    f"Status      : {Y_H5_STATUS}"
)

print(
    f"Evidence    : {Y_H5_EVIDENCE}"
)

print(
    f"Conclusion  : {Y_H5_CONCLUSION}"
)
```

```python
# ==============================================================================
# Z1 — INITIALIZE FINAL PUBLICATION PACKAGE
# ==============================================================================

print()
print("=" * 100)
print("Z1 — INITIALIZE FINAL PUBLICATION PACKAGE")
print("=" * 100)

import pathlib
import json
import platform
import sys
import subprocess
import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# EXPECTED RESEARCH STRUCTURE
# ------------------------------------------------------------------------------

Z_EXPECTED_DATASETS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]

Z_EXPECTED_METHODS = [
    "CTGAN",
    "TVAE",
    "DP-CTGAN",
    "SPP-GAN",
]

Z_REQUIRED_MODULES = [
    "R",
    "T",
    "U",
    "V",
    "W",
    "X",
    "Y",
]

# ------------------------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------------------------

if "SPP_GAN_RESULTS_DIRECTORY" in globals():

    Z_OUTPUT_DIRECTORY = pathlib.Path(
        SPP_GAN_RESULTS_DIRECTORY
    )

elif "RESULTS_DIRECTORY" in globals():

    Z_OUTPUT_DIRECTORY = pathlib.Path(
        RESULTS_DIRECTORY
    )

else:

    Z_OUTPUT_DIRECTORY = pathlib.Path(
        "results"
    )

Z_PUBLICATION_DIRECTORY = (
    Z_OUTPUT_DIRECTORY /
    "publication_package"
)

Z_TABLE_DIRECTORY = (
    Z_PUBLICATION_DIRECTORY /
    "tables"
)

Z_FIGURE_DIRECTORY = (
    Z_PUBLICATION_DIRECTORY /
    "figures"
)

Z_REPORT_DIRECTORY = (
    Z_PUBLICATION_DIRECTORY /
    "reports"
)

for directory in [
    Z_PUBLICATION_DIRECTORY,
    Z_TABLE_DIRECTORY,
    Z_FIGURE_DIRECTORY,
    Z_REPORT_DIRECTORY,
]:

    directory.mkdir(
        parents=True,
        exist_ok=True,
    )

print()
print(
    f"Output directory       : {Z_PUBLICATION_DIRECTORY}"
)

print(
    f"Table directory        : {Z_TABLE_DIRECTORY}"
)

print(
    f"Figure directory       : {Z_FIGURE_DIRECTORY}"
)

print(
    f"Report directory       : {Z_REPORT_DIRECTORY}"
)

print()
print(
    f"Datasets               : {Z_EXPECTED_DATASETS}"
)

print(
    f"Methods                : {Z_EXPECTED_METHODS}"
)

print()
print("Z1 STATUS: READY")
```

```python
# ==============================================================================
# Z2 — RESOLVE FINAL RESEARCH OBJECTS
# ==============================================================================

print()
print("=" * 100)
print("Z2 — RESOLVE FINAL RESEARCH OBJECTS")
print("=" * 100)

# ------------------------------------------------------------------------------
# HELPER
# ------------------------------------------------------------------------------

def Z_RESOLVE_DF(
    candidate_names,
    label,
):

    for name in candidate_names:

        if name in globals():

            candidate = globals()[name]

            if isinstance(
                candidate,
                pd.DataFrame,
            ):

                print(
                    f"{label:30s}: {name}"
                )

                return candidate.copy()

    print(
        f"{label:30s}: NOT FOUND"
    )

    return None


# ------------------------------------------------------------------------------
# RESOLVE FINAL OBJECTS
# ------------------------------------------------------------------------------

Z_OBJECTIVE_DF = Z_RESOLVE_DF(
    [
        "Y_OBJECTIVE_RESULTS_DF",
    ],
    "Research objectives",
)

Z_HYPOTHESIS_DF = Z_RESOLVE_DF(
    [
        "Y_HYPOTHESIS_RESULTS_DF",
    ],
    "Research hypotheses",
)

Z_CONCLUSION_DF = Z_RESOLVE_DF(
    [
        "Y_FINAL_CONCLUSION_DF",
    ],
    "Final conclusions",
)

Z_EVIDENCE_DF = Z_RESOLVE_DF(
    [
        "Y_EVIDENCE_MATRIX",
    ],
    "Evidence matrix",
)

Z_COMPOSITE_DF = Z_RESOLVE_DF(
    [
        "W23_FINAL_COMPOSITE_DF",
        "W_FINAL_COMPOSITE_DF",
        "W20_CROSS_DATASET_SUMMARY",
        "W_CROSS_DATASET_SUMMARY",
    ],
    "Composite performance",
)

Z_RANKING_DF = Z_RESOLVE_DF(
    [
        "X12_SUMMARY_DF",
        "X_OVERALL_RANKING_DF",
    ],
    "Method ranking",
)

Z_STABILITY_DF = Z_RESOLVE_DF(
    [
        "X_RANK_STABILITY_DF",
        "X12_SUMMARY_DF",
    ],
    "Sensitivity/stability",
)

Z_ROBUSTNESS_DF = Z_RESOLVE_DF(
    [
        "T_CROSS_DATASET_SUMMARY",
        "T_ROBUSTNESS_SUMMARY_DF",
        "T_FINAL_ROBUSTNESS_DF",
    ],
    "Cross-dataset robustness",
)

Z_STATISTICAL_DF = Z_RESOLVE_DF(
    [
        "U_STATISTICAL_RESULTS_DF",
        "U_FINAL_STATISTICAL_RESULTS_DF",
        "U_PAIRWISE_RESULTS_DF",
        "U_RESULTS_DF",
    ],
    "Statistical results",
)

Z_EFFECT_SIZE_DF = Z_RESOLVE_DF(
    [
        "V_EFFECT_SIZE_DF",
        "V_FINAL_EFFECT_SIZE_DF",
        "V_RESULTS_DF",
    ],
    "Effect-size results",
)

print()
print("Z2 STATUS: RESOLUTION COMPLETE")
```

```python
# ==============================================================================
# Z3 — VALIDATE FINAL EXPERIMENTAL GRID
# ==============================================================================

print()
print("=" * 100)
print("Z3 — VALIDATE FINAL EXPERIMENTAL GRID")
print("=" * 100)

Z_EXPECTED_GRID = pd.MultiIndex.from_product(
    [
        Z_EXPECTED_DATASETS,
        Z_EXPECTED_METHODS,
    ],
    names=[
        "Dataset",
        "Method",
    ],
).to_frame(
    index=False
)

print()
print("EXPECTED DATASET × METHOD GRID")
print("-" * 100)

print(
    Z_EXPECTED_GRID.to_string(
        index=False
    )
)

# ------------------------------------------------------------------------------
# CHECK COMPOSITE GRID
# ------------------------------------------------------------------------------

Z_GRID_STATUS = False

if Z_COMPOSITE_DF is not None:

    if (
        "Dataset" in Z_COMPOSITE_DF.columns
        and
        "Method" in Z_COMPOSITE_DF.columns
    ):

        Z_GRID = (
            Z_COMPOSITE_DF[
                [
                    "Dataset",
                    "Method",
                ]
            ]
            .drop_duplicates()
            .copy()
        )

        Z_GRID["Dataset"] = (
            Z_GRID["Dataset"]
            .astype(str)
            .str.strip()
        )

        Z_GRID["Method"] = (
            Z_GRID["Method"]
            .astype(str)
            .str.strip()
        )

        Z_GRID_STATUS = (
            len(Z_GRID) >=
            len(Z_EXPECTED_GRID)
            and
            Z_EXPECTED_GRID.merge(
                Z_GRID,
                on=[
                    "Dataset",
                    "Method",
                ],
                how="left",
                indicator=True,
            )["_merge"]
            .eq("both")
            .all()
        )

print()
print(
    f"All expected Dataset × Method combinations present: "
    f"{Z_GRID_STATUS}"
)

print()

if not Z_GRID_STATUS:

    print(
        "WARNING: The final composite evidence does not contain "
        "the complete expected experimental grid."
    )

print()
print("Z3 STATUS: VALIDATION COMPLETE")
```

```python
# ==============================================================================
# Z4 — FINAL COMPARATIVE PERFORMANCE TABLE
# ==============================================================================

print()
print("=" * 100)
print("Z4 — FINAL COMPARATIVE PERFORMANCE TABLE")
print("=" * 100)

Z_COMPARATIVE_TABLE = None

if Z_COMPOSITE_DF is not None:

    Z_TABLE_COLUMNS = [
        "Dataset",
        "Method",
        "Fidelity_Score",
        "Utility_Score",
        "Privacy_Score",
        "Overall Performance Score",
    ]

    Z_AVAILABLE_COLUMNS = [
        column
        for column in Z_TABLE_COLUMNS
        if column in Z_COMPOSITE_DF.columns
    ]

    if (
        "Dataset" in Z_AVAILABLE_COLUMNS
        and
        "Method" in Z_AVAILABLE_COLUMNS
    ):

        Z_COMPARATIVE_TABLE = (
            Z_COMPOSITE_DF[
                Z_AVAILABLE_COLUMNS
            ]
            .copy()
        )

        for column in Z_AVAILABLE_COLUMNS:

            if column not in [
                "Dataset",
                "Method",
            ]:

                Z_COMPARATIVE_TABLE[
                    column
                ] = pd.to_numeric(
                    Z_COMPARATIVE_TABLE[
                        column
                    ],
                    errors="coerce",
                )

        Z_COMPARATIVE_TABLE = (
            Z_COMPARATIVE_TABLE
            .sort_values(
                [
                    "Dataset",
                    "Method",
                ]
            )
            .reset_index(
                drop=True
            )
        )

        print()

        print(
            Z_COMPARATIVE_TABLE.to_string(
                index=False
            )
        )

    else:

        print(
            "WARNING: Dataset and Method columns "
            "were not both available."
        )

else:

    print(
        "WARNING: Final composite performance "
        "DataFrame is unavailable."
    )

print()
print("Z4 STATUS: COMPLETE")
```

```python
# ==============================================================================
# Z5 — DATASET-WISE METHOD RANKING TABLE
# ==============================================================================

print()
print("=" * 100)
print("Z5 — DATASET-WISE METHOD RANKING TABLE")
print("=" * 100)

Z_DATASET_RANKING_TABLE = None

if (
    Z_COMPOSITE_DF is not None
    and
    "Dataset" in Z_COMPOSITE_DF.columns
    and
    "Method" in Z_COMPOSITE_DF.columns
):

    Z_SCORE_COLUMN = None

    for candidate in [
        "Overall Performance Score",
        "Overall_Score",
        "Composite_Score",
        "Overall Performance",
    ]:

        if candidate in Z_COMPOSITE_DF.columns:

            Z_SCORE_COLUMN = candidate

            break

    if Z_SCORE_COLUMN is not None:

        Z_RANK_TEMP = (
            Z_COMPOSITE_DF[
                [
                    "Dataset",
                    "Method",
                    Z_SCORE_COLUMN,
                ]
            ]
            .copy()
        )

        Z_RANK_TEMP[
            Z_SCORE_COLUMN
        ] = pd.to_numeric(
            Z_RANK_TEMP[
                Z_SCORE_COLUMN
            ],
            errors="coerce",
        )

        Z_RANK_TEMP[
            "Dataset_Rank"
        ] = (
            Z_RANK_TEMP
            .groupby(
                "Dataset"
            )[Z_SCORE_COLUMN]
            .rank(
                ascending=False,
                method="min",
            )
        )

        Z_RANK_TEMP[
            "Dataset_Rank"
        ] = (
            Z_RANK_TEMP[
                "Dataset_Rank"
            ]
            .astype("Int64")
        )

        Z_DATASET_RANKING_TABLE = (
            Z_RANK_TEMP
            .sort_values(
                [
                    "Dataset",
                    "Dataset_Rank",
                    "Method",
                ]
            )
            .reset_index(
                drop=True
            )
        )

        print()

        print(
            Z_DATASET_RANKING_TABLE.to_string(
                index=False
            )
        )

else:

    print(
        "Dataset-wise ranking could not be constructed."
    )

print()
print("Z5 STATUS: COMPLETE")
```

```python
# ==============================================================================
# Z6 — CROSS-DATASET METHOD SUMMARY
# ==============================================================================

print()
print("=" * 100)
print("Z6 — CROSS-DATASET METHOD SUMMARY")
print("=" * 100)

Z_METHOD_SUMMARY = None

if (
    Z_COMPARATIVE_TABLE is not None
    and
    "Method" in Z_COMPARATIVE_TABLE.columns
):

    Z_NUMERIC_COLUMNS = [
        column
        for column in [
            "Fidelity_Score",
            "Utility_Score",
            "Privacy_Score",
            "Overall Performance Score",
        ]
        if column in Z_COMPARATIVE_TABLE.columns
    ]

    if Z_NUMERIC_COLUMNS:

        Z_METHOD_SUMMARY = (
            Z_COMPARATIVE_TABLE
            .groupby(
                "Method",
                as_index=False,
            )[Z_NUMERIC_COLUMNS]
            .agg(
                [
                    "mean",
                    "std",
                    "median",
                ]
            )
        )

        print()

        print(
            Z_METHOD_SUMMARY.to_string(
                index=False
            )
        )

else:

    print(
        "Cross-dataset method summary unavailable."
    )

print()
print("Z6 STATUS: COMPLETE")
```

```python
# ==============================================================================
# Z9 — PUBLICATION FIGURE 1: OVERALL METHOD PERFORMANCE
# ==============================================================================

print()
print("=" * 100)
print("Z9 — PUBLICATION FIGURE 1: OVERALL METHOD PERFORMANCE")
print("=" * 100)

import matplotlib.pyplot as plt

Z_FIGURE_1 = (
    Z_FIGURE_DIRECTORY /
    "figure_1_overall_method_performance.png"
)

if (
    Z_COMPARATIVE_TABLE is not None
    and
    "Overall Performance Score"
    in Z_COMPARATIVE_TABLE.columns
):

    Z_PLOT_DF = (
        Z_COMPARATIVE_TABLE
        .groupby(
            "Method",
            as_index=False,
        )[
            "Overall Performance Score"
        ]
        .mean()
        .sort_values(
            "Overall Performance Score",
            ascending=False,
        )
    )

    plt.figure(
        figsize=(10, 6)
    )

    plt.bar(
        Z_PLOT_DF["Method"],
        Z_PLOT_DF[
            "Overall Performance Score"
        ],
    )

    plt.xlabel(
        "Method"
    )

    plt.ylabel(
        "Mean Overall Performance Score"
    )

    plt.title(
        "Mean Overall Synthetic-Data Performance Across Datasets"
    )

    plt.xticks(
        rotation=20
    )

    plt.tight_layout()

    plt.savefig(
        Z_FIGURE_1,
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()

    print(
        f"Saved: {Z_FIGURE_1}"
    )

else:

    print(
        "Figure 1 skipped: overall performance score unavailable."
    )
```

```python
# ==============================================================================
# Z14 — REPRODUCIBILITY REPORT
# ==============================================================================

print()
print("=" * 100)
print("Z14 — REPRODUCIBILITY REPORT")
print("=" * 100)

# ------------------------------------------------------------------------------
# PYTHON VERSION
# ------------------------------------------------------------------------------

Z_REPRODUCIBILITY = {
    "Python_Version":
        sys.version,

    "Python_Executable":
        sys.executable,

    "Platform":
        platform.platform(),

    "Processor":
        platform.processor(),

    "System":
        platform.system(),

    "System_Release":
        platform.release(),

    "Machine":
        platform.machine(),
}

# ------------------------------------------------------------------------------
# RANDOM SEED
# ------------------------------------------------------------------------------

for variable_name in [
    "SEED",
    "RANDOM_SEED",
    "GLOBAL_SEED",
    "SEED_VALUE",
]:

    if variable_name in globals():

        Z_REPRODUCIBILITY[
            "Random_Seed"
        ] = str(
            globals()[
                variable_name
            ]
        )

        break

else:

    Z_REPRODUCIBILITY[
        "Random_Seed"
    ] = "Not resolved from current namespace"

# ------------------------------------------------------------------------------
# DEVICE
# ------------------------------------------------------------------------------

for variable_name in [
    "DEVICE",
    "device",
]:

    if variable_name in globals():

        Z_REPRODUCIBILITY[
            "Compute_Device"
        ] = str(
            globals()[
                variable_name
            ]
        )

        break

else:

    Z_REPRODUCIBILITY[
        "Compute_Device"
    ] = "Not resolved"

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

if "CONFIG" in globals():

    try:

        Z_REPRODUCIBILITY[
            "Configuration"
        ] = str(
            CONFIG
        )

    except Exception:

        Z_REPRODUCIBILITY[
            "Configuration"
        ] = "Available but not serializable"

else:

    Z_REPRODUCIBILITY[
        "Configuration"
    ] = "Not resolved"

# ------------------------------------------------------------------------------
# DATASETS
# ------------------------------------------------------------------------------

Z_REPRODUCIBILITY[
    "Datasets"
] = ", ".join(
    Z_EXPECTED_DATASETS
)

Z_REPRODUCIBILITY[
    "Methods"
] = ", ".join(
    Z_EXPECTED_METHODS
)

Z_REPRODUCIBILITY[
    "Experimental_Grid"
] = (
    f"{len(Z_EXPECTED_DATASETS)} datasets × "
    f"{len(Z_EXPECTED_METHODS)} methods = "
    f"{len(Z_EXPECTED_GRID)} combinations"
)

Z_REPRODUCIBILITY_DF = pd.DataFrame(
    [
        {
            "Parameter":
                key,
            "Value":
                value,
        }
        for key, value
        in Z_REPRODUCIBILITY.items()
    ]
)

print()

print(
    Z_REPRODUCIBILITY_DF.to_string(
        index=False
    )
)

print()
print("Z14 STATUS: COMPLETE")
```

```python
# ==============================================================================
# Z15 — PACKAGE ENVIRONMENT INFORMATION
# ==============================================================================

print()
print("=" * 100)
print("Z15 — PACKAGE ENVIRONMENT INFORMATION")
print("=" * 100)

Z_PACKAGE_NAMES = [
    "numpy",
    "pandas",
    "scikit-learn",
    "scipy",
    "torch",
    "sdv",
    "opacus",
    "matplotlib",
    "seaborn",
]

Z_PACKAGE_ROWS = []

for package_name in Z_PACKAGE_NAMES:

    try:

        if package_name == "scikit-learn":

            import sklearn

            version = sklearn.__version__

        elif package_name == "numpy":

            import numpy

            version = numpy.__version__

        elif package_name == "pandas":

            import pandas

            version = pandas.__version__

        elif package_name == "scipy":

            import scipy

            version = scipy.__version__

        elif package_name == "torch":

            import torch

            version = torch.__version__

        elif package_name == "sdv":

            import sdv

            version = sdv.__version__

        elif package_name == "opacus":

            import opacus

            version = opacus.__version__

        elif package_name == "matplotlib":

            import matplotlib

            version = matplotlib.__version__

        elif package_name == "seaborn":

            import seaborn

            version = seaborn.__version__

        else:

            version = "Unknown"

        status = "Available"

    except Exception as error:

        version = str(error)
        status = "Unavailable"

    Z_PACKAGE_ROWS.append(
        {
            "Package":
                package_name,
            "Version":
                version,
            "Status":
                status,
        }
    )

Z_PACKAGE_DF = pd.DataFrame(
    Z_PACKAGE_ROWS
)

print()

print(
    Z_PACKAGE_DF.to_string(
        index=False
    )
)

print()
print("Z15 STATUS: COMPLETE")
```

```python
# ==============================================================================
# Z20 — FINAL EXCEL PUBLICATION PACKAGE
# ==============================================================================

print()
print("=" * 100)
print("Z20 — FINAL EXCEL PUBLICATION PACKAGE")
print("=" * 100)

import pathlib
import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------------------------

if "SPP_GAN_RESULTS_DIRECTORY" in globals():
    Z_OUTPUT_DIRECTORY = pathlib.Path(SPP_GAN_RESULTS_DIRECTORY)
else:
    Z_OUTPUT_DIRECTORY = pathlib.Path.cwd() / "SPP_GAN_RESULTS"

Z_OUTPUT_DIRECTORY.mkdir(parents=True, exist_ok=True)

Z_EXCEL_PATH = (
    Z_OUTPUT_DIRECTORY /
    "FINAL_RESEARCH_PUBLICATION_PACKAGE.xlsx"
)

# ------------------------------------------------------------------------------
# SAFE DATAFRAME PREPARATION
# ------------------------------------------------------------------------------

def Z_PREPARE_EXCEL_DF(df):
    """
    Prepare a DataFrame for publication Excel export.

    Handles:
    - MultiIndex columns
    - duplicate column names
    - infinite values
    - unsupported object values
    - unnamed index columns
    """

    if df is None:
        return None

    if not isinstance(df, pd.DataFrame):
        try:
            df = pd.DataFrame(df)
        except Exception:
            return None

    df = df.copy()

    # --------------------------------------------------------------------------
    # Flatten MultiIndex columns
    # --------------------------------------------------------------------------

    if isinstance(df.columns, pd.MultiIndex):

        flattened_columns = []

        for col in df.columns:

            parts = []

            for part in col:

                if (
                    part is not None
                    and not pd.isna(part)
                    and str(part).strip() != ""
                    and str(part).strip().lower() != "nan"
                ):
                    parts.append(str(part).strip())

            if parts:
                name = " | ".join(parts)
            else:
                name = "Column"

            flattened_columns.append(name)

        df.columns = flattened_columns

    else:

        df.columns = [
            str(col)
            for col in df.columns
        ]

    # --------------------------------------------------------------------------
    # Make duplicate column names unique
    # --------------------------------------------------------------------------

    seen = {}
    unique_columns = []

    for col in df.columns:

        if col not in seen:
            seen[col] = 0
            unique_columns.append(col)

        else:
            seen[col] += 1
            unique_columns.append(
                f"{col}_{seen[col]}"
            )

    df.columns = unique_columns

    # --------------------------------------------------------------------------
    # Replace infinite values
    # --------------------------------------------------------------------------

    df = df.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # --------------------------------------------------------------------------
    # Reset index for safe Excel export
    # --------------------------------------------------------------------------

    df = df.reset_index(drop=True)

    return df


# ------------------------------------------------------------------------------
# COLLECT AVAILABLE PUBLICATION TABLES
# ------------------------------------------------------------------------------

Z_EXCEL_TABLES = {}

# Explicit publication tables
Z_CANDIDATE_TABLES = {
    "Method Summary": globals().get(
        "Z_METHOD_SUMMARY",
        None
    ),

    "Dataset Summary": globals().get(
        "Z_DATASET_SUMMARY",
        None
    ),

    "Final Rankings": globals().get(
        "Z_FINAL_RANKING",
        None
    ),

    "Method Ranking": globals().get(
        "X_METHOD_RANKING",
        None
    ),

    "Sensitivity Analysis": globals().get(
        "X_SENSITIVITY_RESULTS",
        None
    ),

    "Objective Evaluation": globals().get(
        "Y_OBJECTIVE_EVALUATION",
        None
    ),

    "Hypothesis Evaluation": globals().get(
        "Y_HYPOTHESIS_EVALUATION",
        None
    ),

    "Statistical Results": globals().get(
        "U_STATISTICAL_RESULTS",
        None
    ),

    "Effect Sizes": globals().get(
        "V_EFFECT_SIZE_RESULTS",
        None
    ),

    "Composite Scores": globals().get(
        "W_COMPOSITE_RESULTS",
        None
    ),

    "Cross Dataset Summary": globals().get(
        "W_CROSS_DATASET_SUMMARY",
        None
    ),

    "Research Conclusions": globals().get(
        "Z_RESEARCH_CONCLUSIONS",
        None
    ),

    "Research Limitations": globals().get(
        "Z_RESEARCH_LIMITATIONS",
        None
    ),
}

# ------------------------------------------------------------------------------
# ADD ONLY VALID DATAFRAMES
# ------------------------------------------------------------------------------

for table_name, table_df in Z_CANDIDATE_TABLES.items():

    if isinstance(table_df, pd.DataFrame):

        if not table_df.empty:

            Z_EXCEL_TABLES[table_name] = (
                Z_PREPARE_EXCEL_DF(table_df)
            )


# ------------------------------------------------------------------------------
# AUTOMATICALLY INCLUDE OTHER FINAL Z TABLES
# ------------------------------------------------------------------------------

for variable_name, variable_value in list(globals().items()):

    if (
        variable_name.startswith("Z_")
        and isinstance(variable_value, pd.DataFrame)
        and not variable_value.empty
    ):

        table_name = variable_name[2:]

        if table_name not in Z_EXCEL_TABLES:

            Z_EXCEL_TABLES[table_name] = (
                Z_PREPARE_EXCEL_DF(variable_value)
            )


# ------------------------------------------------------------------------------
# CREATE EXCEL WORKBOOK
# ------------------------------------------------------------------------------

print()
print("Preparing Excel workbook...")
print()

Z_EXPORTED_SHEETS = []

with pd.ExcelWriter(
    Z_EXCEL_PATH,
    engine="openpyxl"
) as writer:

    for table_name, table_df in Z_EXCEL_TABLES.items():

        if table_df is None:
            continue

        if table_df.empty:
            continue

        # Excel sheet-name constraints
        safe_sheet_name = str(table_name)

        for character in [
            "\\",
            "/",
            "*",
            "?",
            ":",
            "[",
            "]"
        ]:
            safe_sheet_name = safe_sheet_name.replace(
                character,
                "_"
            )

        safe_sheet_name = safe_sheet_name[:31]

        if safe_sheet_name.strip() == "":
            safe_sheet_name = "Table"

        # Ensure unique sheet names
        original_name = safe_sheet_name
        counter = 1

        while safe_sheet_name in Z_EXPORTED_SHEETS:

            suffix = f"_{counter}"

            safe_sheet_name = (
                original_name[:31 - len(suffix)]
                + suffix
            )

            counter += 1

        # ----------------------------------------------------------------------
        # Excel export
        # ----------------------------------------------------------------------

        table_df.to_excel(
            writer,
            sheet_name=safe_sheet_name,
            index=False
        )

        Z_EXPORTED_SHEETS.append(
            safe_sheet_name
        )


# ------------------------------------------------------------------------------
# VERIFY FILE
# ------------------------------------------------------------------------------

if not Z_EXCEL_PATH.exists():

    raise RuntimeError(
        "Z20 FAILED: Final Excel publication package "
        "was not created."
    )

Z_EXCEL_SIZE_MB = (
    Z_EXCEL_PATH.stat().st_size
    / (1024 ** 2)
)


# ------------------------------------------------------------------------------
# FINAL REPORT
# ------------------------------------------------------------------------------

print("-" * 100)
print("Z20 — EXCEL PUBLICATION PACKAGE CREATED")
print("-" * 100)

print()
print("Output file:")
print(Z_EXCEL_PATH)

print()
print(f"Workbook size : {Z_EXCEL_SIZE_MB:.3f} MB")
print(f"Sheets exported : {len(Z_EXPORTED_SHEETS)}")

print()
print("EXPORTED SHEETS")
print("-" * 100)

for i, sheet_name in enumerate(
    Z_EXPORTED_SHEETS,
    start=1
):

    print(
        f"{i:02d}. {sheet_name}"
    )

print()
print("=" * 100)
print("Z20 STATUS: PASS")
print("=" * 100)
```

```python
# ==============================================================================
# Z21 — FINAL PUBLICATION PACKAGE MANIFEST
# ==============================================================================

print()
print("=" * 100)
print("Z21 — FINAL PUBLICATION PACKAGE MANIFEST")
print("=" * 100)

Z_MANIFEST_ROWS = []

for directory in [
    Z_TABLE_DIRECTORY,
    Z_FIGURE_DIRECTORY,
    Z_REPORT_DIRECTORY,
]:

    for path in sorted(
        directory.iterdir()
    ):

        if path.is_file():

            Z_MANIFEST_ROWS.append(
                {
                    "Category":
                        directory.name,
                    "File":
                        path.name,
                    "Size_Bytes":
                        path.stat().st_size,
                    "Exists":
                        path.exists(),
                }
            )

# Add Excel package itself if not already captured.
if Z_EXCEL_PACKAGE.exists():

    Z_MANIFEST_ROWS.append(
        {
            "Category":
                "publication_package",
            "File":
                Z_EXCEL_PACKAGE.name,
            "Size_Bytes":
                Z_EXCEL_PACKAGE.stat().st_size,
            "Exists":
                True,
        }
    )

Z_MANIFEST_DF = pd.DataFrame(
    Z_MANIFEST_ROWS
)

print()

if not Z_MANIFEST_DF.empty:

    print(
        Z_MANIFEST_DF.to_string(
            index=False
        )
    )

else:

    print(
        "No publication files were detected."
    )

Z_MANIFEST_FILE = (
    Z_PUBLICATION_DIRECTORY /
    "publication_package_manifest.csv"
)

Z_MANIFEST_DF.to_csv(
    Z_MANIFEST_FILE,
    index=False,
)

print()

print(
    f"Manifest saved: {Z_MANIFEST_FILE}"
)
```
