## Research Background

High-quality synthetic tabular data that maintains the statistical features of real-world datasets while safeguarding sensitive information is becoming increasingly necessary due to the growing need for data-driven applications. Conventional anonymization methods can drastically decrease the usefulness of data and frequently fall short of offering sufficient privacy protection.

This research proposes a unified framework that integrates statistical data generation techniques with deep generative models to produce high-fidelity synthetic tabular datasets while incorporating privacy-preserving mechanisms. The framework aims to achieve an optimal balance between data utility, statistical fidelity, and privacy protection.

## Research Objectives

The primary objectives of this research are:

1. Develop a unified framework integrating statistical and deep learning approaches for synthetic tabular data generation.

2. Compare traditional statistical generators with modern deep generative models.

3. Incorporate privacy-preserving mechanisms into the synthetic data generation process.

4. Evaluate synthetic datasets using fidelity, utility, and privacy metrics.

5. Identify the most effective model for generating high-quality privacy-preserving synthetic tabular data.

## Datasets Used in This Research

The experimental evaluation will be conducted using three privacy-sensitive benchmark tabular datasets representing diverse real-world domains: demographic, financial, and healthcare applications.

These datasets were selected because they represent heterogeneous tabular data characteristics, including numerical and categorical attributes, high-dimensional feature spaces, and privacy-sensitive information. The selected datasets enable comprehensive evaluation of the proposed Privacy-Preserving Synthetic Generative Adversarial Network (SPP-GAN) framework across multiple application domains.

The Adult Income dataset provides a widely used benchmark for synthetic tabular data generation and contains demographic and socioeconomic attributes. The Bank Marketing dataset represents a financial domain where privacy protection is critical for secure data sharing and analytics. The Diabetes 130-US Hospitals dataset introduces a large-scale healthcare scenario with sensitive patient information, enabling evaluation of privacy-preserving synthetic data generation in a highly regulated domain.

Together, these datasets provide a comprehensive experimental foundation for assessing synthetic data quality, downstream machine learning utility, and privacy preservation capabilities of the proposed SPP-GAN framework.

```python
# ============================================================
# K2. Save YAML Configuration
# ============================================================

import os
import json
import yaml
from pathlib import PurePath

CONFIG_FILE = os.path.join(CONFIG_DIR, "config.yaml")


def convert_for_yaml(obj):
    """
    Recursively convert unsupported objects into YAML-safe types.
    """

    # Dictionaries
    if isinstance(obj, dict):
        return {
            str(k): convert_for_yaml(v)
            for k, v in obj.items()
        }

    # Lists / Tuples / Sets
    if isinstance(obj, (list, tuple, set)):
        return [convert_for_yaml(v) for v in obj]

    # Any pathlib object
    if isinstance(obj, PurePath):
        return str(obj)

    # Primitive types
    if isinstance(obj, (str, int, float, bool)) or obj is None:
        return obj

    # Everything else
    return str(obj)


config_to_save = convert_for_yaml(MASTER_CONFIG)

# Extra safety: force everything into native Python objects
config_to_save = json.loads(json.dumps(config_to_save))

with open(CONFIG_FILE, "w", encoding="utf-8") as file:

    yaml.safe_dump(
        config_to_save,
        file,
        sort_keys=False,
        default_flow_style=False,
        allow_unicode=True
    )

print("=" * 70)
print("CONFIGURATION SAVED")
print("=" * 70)
print(f"Location : {CONFIG_FILE}")
```
